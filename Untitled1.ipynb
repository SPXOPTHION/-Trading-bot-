````python
#!/usr/bin/env python3
# -*- coding: utf-8 -*-

# بوت توصيات الأسهم - نسخة شاكر 1.0 (نسخة كاملة)
# آخر تحديث: فبراير 2026

import yfinance as yf
import pandas as pd
import numpy as np
import ta
import logging
import time
import schedule
from datetime import datetime, timedelta
from telegram import Bot, InlineKeyboardButton, InlineKeyboardMarkup, Update
from telegram.ext import Application, CommandHandler, CallbackQueryHandler, ContextTypes, JobQueue, ChatMemberHandler # أضفنا ChatMemberHandler
import asyncio
import os

# ---------------------------
# 1. إعدادات البوت والتوكن
# ---------------------------
TELEGRAM_TOKEN = "8699821370:AAEQUSbLTgf7MmWqo5vV5LHPOz30wfqOfqw"      # ضع التوكن الخاص بك
CHAT_ID = "7854020427"                # ضع معرف المستخدم أو القناة

# تفعيل السجلات لمشاهدة الأخطاء
logging.basicConfig(level=logging.INFO, format='%(asctime)s - %(name)s - %(levelname)s - %(message)s')

# ---------------------------
# 2. قائمة الأسهم الكاملة (مدمجة)
# ---------------------------
TICKERS = [
    # --- أسهمك القديمة (من الصور) ---
    "A", "AA", "ABEV", "ABNB", "ABT", "ABVX", "ACGL", "ACM", "ACN",
    "ADBE", "ADI", "ADM", "ADP", "ADSK", "AEE", "AEG", "AEIS", "AEM",
    "AEP", "AER", "AES", "AFG", "AFL", "AFRM", "AG", "AGI", "AGNC",
    "AIG", "AIT", "AIZ", "AJG", "AKAM", "ALAB", "ALB", "ALC", "ALGN",
    "ALL", "ALLE", "ALLY", "ALNY", "ALSN", "AM", "AMCR", "AME", "AMH",
    "AMKR", "AMP", "AMRZ", "AMT", "AMX",

    # --- الشركات العملاقة (> 1 تريليون دولار) ---
    "NVDA", "AAPL", "GOOGL", "MSFT", "AMZN", "META", "TSM", "AVGO", "TSLA",
    "BRK.A", "WMT",

    # --- الشركات الكبرى (200 مليار - 1 تريليون) ---
    "LLY", "JPM", "XOM", "V", "JNJ", "MA", "MU", "ORCL", "COST", "AMD",
    "NFLX", "PLTR", "CSCO", "MRK", "TMO", "ASML", "ABBV", "HD", "CVX",
    "PG", "CAT", "KO", "PEP", "INTC", "CMCSA", "MCD", "TXN", "VZ", "AMGN"
]

# ---------------------------
# 3. دوال التحليل الأساسي (الفلاتر)
# ---------------------------

def get_stock_info(ticker):
    """جلب معلومات الشركة الأساسية"""
    try:
        stock = yf.Ticker(ticker)
        info = stock.info
        return info
    except Exception as e:
        logging.error(f"خطأ في جلب معلومات {ticker}: {e}")
        return None

def passes_basic_filters(ticker):
    """
    تطبيق الفلتر الأساسي:
    - Market Cap > 10B (ملاحظة: أغلب الشركات في القائمة أكبر من ذلك بكثير، لكن الفلتر يضمن ذلك)
    - Average Volume > 1M (تقريبي)
    - Spread < 5%
    """
    info = get_stock_info(ticker)
    if not info:
        return False

    # 1. القيمة السوقية
    mcap = info.get('marketCap', 0)
    if mcap < 10e9:
        # print(f"{ticker} فشل فلتر Market Cap: {mcap}") # Changed from logging.debug
        return False

    # 2. متوسط الحجم (نأخذ آخر 20 يوم)
    try:
        stock = yf.Ticker(ticker)
        hist = stock.history(period="1mo", interval="1d")
        if hist.empty:
            return False
        avg_vol = hist['Volume'].rolling(20).mean().iloc[-1]
        if avg_vol < 1_000_000:   # أقل من مليون سهم يومياً
            # print(f"{ticker} فشل فلتر الحجم: {avg_vol}") # Changed from logging.debug
            return False
    except:
        return False

    # 3. Spread (bid/ask)
    bid = info.get('bid', 0)
    ask = info.get('ask', 0)
    if bid and ask:
        spread = (ask - bid) / bid if bid != 0 else 1
        if spread > 0.05:   # أكبر من 5%
            # print(f"{ticker} فشل فلتر Spread: {spread:.2%}") # Changed from logging.debug
            return False

    return True

# ---------------------------
# 4. تحليل SPX (مؤشر S&P 500)
# ---------------------------

def analyze_spx(timeframe="15m"):
    """تحليل مؤشر SPX على الإطار المطلوب"""
    try:
        spx = yf.Ticker("^GSPC")  # أو "SPY"
        hist = spx.history(period="5d", interval=timeframe)
        if len(hist) < 50:
            return "neutral"

        # متوسطات بسيطة لتحديد الاتجاه
        hist['SMA20'] = hist['Close'].rolling(20).mean()
        last = hist.iloc[-1]
        prev = hist.iloc[-2]

        if last['Close'] > last['SMA20'] and last['Close'] > prev['Close']:
            return "uptrend"
        elif last['Close'] < last['SMA20'] and last['Close'] < prev['Close']:
            return "downtrend"
        else:
            return "neutral"
    except:
        return "neutral"

# ---------------------------
# 5. تحليل السهم (فريم ساعتين)
# ---------------------------

def analyze_stock_2h(ticker):
    """
    تحليل فني كامل على فريم الساعتين
    يعيد: {'signal': 'CALL'/'PUT'/None, 'confidence': 0-100, 'details': {...}}
    """
    try:
        # print(f"Analyzing {ticker} for 2h interval.") # Added print statement
        stock = yf.Ticker(ticker)
        hist = stock.history(period="1mo", interval="2h")  # آخر شهر، فريم ساعتين
        if len(hist) < 200:
            # print(f"Not enough historical data for {ticker} in 2h interval.") # Added print statement
            return None

        # حساب المتوسطات
        hist['SMA50'] = hist['Close'].rolling(50).mean()
        hist['SMA100'] = hist['Close'].rolling(100).mean()
        hist['SMA200'] = hist['Close'].rolling(200).mean()
        hist['Volume_SMA20'] = hist['Volume'].rolling(20).mean()

        last = hist.iloc[-1]
        current_price = last['Close']
        sma50 = last['SMA50']
        sma100 = last['SMA100']
        sma200 = last['SMA200']

        # قمم وقيعان آخر 3 شموع
        high_last_3 = hist['High'].iloc[-3:].max()
        low_last_3 = hist['Low'].iloc[-3:].min()

        # قمة/قاع اليوم (آخر 24 ساعة = 12 شمعة)
        high_today = hist['High'].iloc[-12:].max()
        low_today = hist['Low'].iloc[-12:].min()

        volume_ok = last['Volume'] > last['Volume_SMA20']

        # شروط CALL
        call_condition1 = (sma50 > sma100) or (sma50 > sma200)
        call_condition2 = current_price > sma200
        call_condition3 = current_price > high_last_3 or current_price > high_today
        call_condition4 = volume_ok

        # شروط PUT
        put_condition1 = (sma50 < sma100) or (sma50 < sma200)
        put_condition2 = current_price < sma200
        put_condition3 = current_price < low_last_3 or current_price < low_today
        put_condition4 = volume_ok

        signal = None
        confidence = 0
        details = {}

        if call_condition1 and call_condition2 and call_condition3 and call_condition4:
            signal = "CALL"
            # حساب الثقة: نضيف 25 لكل شرط متحقق
            confidence = 25 * sum([call_condition1, call_condition2, call_condition3, call_condition4])
            details = {
                "sma50": sma50,
                "sma100": sma100,
                "sma200": sma200,
                "price": current_price,
                "high_3": high_last_3,
                "volume_ratio": last['Volume'] / last['Volume_SMA20']
            }
        elif put_condition1 and put_condition2 and put_condition3 and put_condition4:
            signal = "PUT"
            confidence = 25 * sum([put_condition1, put_condition2, put_condition3, put_condition4])
            details = {
                "sma50": sma50,
                "sma100": sma100,
                "sma200": sma200,
                "price": current_price,
                "low_3": low_last_3,
                "volume_ratio": last['Volume'] / last['Volume_SMA20']
            }

        # if signal: # Added print statement
            # print(f"Signal for {ticker}: {signal}, Confidence: {confidence}")
        return {"signal": signal, "confidence": int(confidence), "details": details}

    except Exception as e:
        logging.error(f"خطأ في تحليل {ticker}: {e}")
        return None

# ---------------------------
# 6. تصنيف الصفقة (سكالب، قصيرة، سوينق)
# ---------------------------

def classify_trade(signal, confidence, details):
    """
    تصنيف حسب قوة الإشارة وترتيب المتوسطات
    """
    if not signal:
        return None

    sma50 = details.get('sma50')
    sma100 = details.get('sma100')
    sma200 = details.get('sma200')

    # ترتيب المتوسطات
    if signal == "CALL":
        if sma50 > sma100 > sma200:
            return "📈 سوينق 3-7 أيام"
        elif confidence >= 75:
            return "⏳ صفقة قصيرة 1-2 يوم"
        else:
            return "⚡ سكالب"
    else:  # PUT
        if sma50 < sma100 < sma200:
            return "📈 سوينق 3-7 أيام"
        elif confidence >= 75:
            return "⏳ صفقة قصيرة 1-2 يوم"
        else:
            return "⚡ سكالب"

# ---------------------------
# 7. اختيار العقد المناسب (بيانات الخيارات)
# ---------------------------

def get_option_contract(ticker, signal, classification):
    """
    البحث عن عقد خيار مناسب حسب التصنيف.
    """
    try:
        stock = yf.Ticker(ticker)
        # جلب تواريخ انتهاء الصلاحية المتاحة
        expirations = stock.options
        if not expirations:
            return "غير متوفر"

        # اختيار التاريخ المناسب حسب التصنيف
        today = datetime.now().date()
        if "سكالب" in classification:
            days = 7
        elif "قصيرة" in classification:
            days = 14
        else:  # سوينق
            days = 30

        target_date = today + timedelta(days=days)
        # البحث عن أقرب تاريخ انتهاء للعدد المطلوب من الأيام
        exp_date = None
        for exp in expirations:
            exp_dt = datetime.strptime(exp, '%Y-%m-%d').date()
            if exp_dt >= target_date:
                exp_date = exp
                break
        if not exp_date:
            exp_date = expirations[0]  # أقرب تاريخ

        # جلب عقود الخيارات لذلك التاريخ
        opt_chain = stock.option_chain(exp_date)
        if signal == "CALL":
            contracts = opt_chain.calls
        else:
            contracts = opt_chain.puts

        if contracts.empty:
            return "غير متوفر"

        # فلترة العقود: Open Interest عالي (أعلى 10%)، Spread ضيق
        current_price = stock.history(period="1d")['Close'].iloc[-1]
        contracts['strike_diff'] = abs(contracts['strike'] - current_price)
        # فلتر OI
        oi_threshold = contracts['openInterest'].quantile(0.7)  # أعلى 30%
        eligible = contracts[contracts['openInterest'] > oi_threshold]

        if eligible.empty:
            eligible = contracts

        # نختار العقد الأقرب للسعر الحالي (للدلتا التقريبية)
        best = eligible.nsmallest(1, 'strike_diff').iloc[0]

        # حساب Spread تقريبي (ask-bid)/ask
        spread = (best['ask'] - best['bid']) / best['ask'] if best['ask'] > 0 else 1
        if spread > 0.05:   # أكبر من 5%
            return "سبريد واسع"

        # بناء نص العقد المقترح
        delta_approx = 0.5 - (best['strike_diff'] / current_price)  # تقريب بسيط
        delta_approx = max(0.1, min(0.9, delta_approx))

        return f"Delta {delta_approx:.2f} – Exp {exp_date} (أيام: {days})"

    except Exception as e:
        logging.error(f"خطأ في جلب الخيارات لـ {ticker}: {e}")
        return "غير متوفر"

# ---------------------------
# 8. بناء رسالة التنبيه النهائية
# ---------------------------

def build_alert_message(ticker, signal, confidence, classification, details, spx_trend, option_text):
    """
    تنسيق الرسالة كما طلبت تماماً.
    """
    # ترجمة الاتجاه
    if signal == "CALL":
        direction = "Call"
        cross = "50 > 100" if details.get('sma50', 0) > details.get('sma100', 0) else "50 > 200"
        price_status = f"السعر فوق {details.get('sma200', 0):.2f}"
        event = "اختراق قمة 2H"
    else:
        direction = "Put"
        cross = "50 < 100" if details.get('sma50', 0) < details.get('sma100', 0) else "50 < 200"
        price_status = f"السعر تحت {details.get('sma200', 0):.2f}"
        event = "كسر قاع 2H"

    # توافق SPX
    if "uptrend" in spx_trend and signal == "CALL":
        spx_compat = "✅ متوافق"
    elif "downtrend" in spx_trend and signal == "PUT":
        spx_compat = "✅ متوافق"
    else:
        spx_compat = "⚠️ غير متوافق"

    message = f"""
🔔 {ticker}
الاتجاه: {direction}
التصنيف: {classification}
التقاطع: {cross} (إيجابي)
الحالة: {price_status}
الحدث: {event}
SPX: {spx_compat}
الثقة: {confidence}/100
العقد المقترح: {option_text}
"""
    return message

# ---------------------------
# 9. دالة المسح الرئيسية
# ---------------------------

def scan_all_stocks():
    """تمر على كل الأسهم وتفحصها وترسل التنبيهات"""
    alerts = []
    print("Starting stock scan...") # Added print statement
    for ticker in TICKERS:
        logging.info(f"فحص {ticker}...")
        time.sleep(0.5)  # تأخير نصف ثانية لتجنب حظر Yahoo

        # 1. الفلتر الأساسي
        if not passes_basic_filters(ticker):
            print(f"Ticker {ticker} did not pass basic filters.") # Added print statement
            continue

        # 2. تحليل السهم
        result = analyze_stock_2h(ticker)
        if not result or not result['signal']:
            print(f"No signal found for {ticker} after 2h analysis.") # Added print statement
            continue

        signal = result['signal']
        confidence = result['confidence']
        details = result['details']

        # 3. تحليل SPX
        spx_15m = analyze_spx("15m")

        # 4. تصنيف الصفقة
        classification = classify_trade(signal, confidence, details)

        # 5. اختيار العقد
        option = get_option_contract(ticker, signal, classification)

        # 6. بناء الرسالة
        msg = build_alert_message(ticker, signal, confidence, classification, details, spx_15m, option)
        alerts.append(msg)
    print(f"Scan finished. Found {len(alerts)} alerts.") # Added print statement
    return alerts

# ---------------------------
# 10. دوال بوت تليجرام
# ---------------------------

async def send_alerts(context):
    """إرسال التنبيهات إلى تليجرام"""
    print("send_alerts function called.") # Added print statement
    alerts = scan_all_stocks()
    if not alerts:
        print("No alerts found to send.") # Added print statement
        await context.bot.send_message(chat_id=CHAT_ID, text="لا توجد إشارات جديدة حالياً.")
    else:
        print(f"Sending {len(alerts)} alerts.") # Added print statement
        for alert in alerts:
            await context.bot.send_message(chat_id=CHAT_ID, text=alert, parse_mode='Markdown')
            await asyncio.sleep(1)  # تجنب سبام تليجرام

# دالة للترحيب بالأعضاء الجدد وإرسال إخلاء المسؤولية
async def greet_chat_members(update: Update, context: ContextTypes.DEFAULT_TYPE) -> None:
    print(f"New chat member event: {update.chat_member}") # Added print
    if update.chat_member.new_chat_member.status == 'member':
        user = update.chat_member.new_chat_member.user
        chat_title = update.chat_member.chat.title or "هذه المجموعة"
        welcome_new_member_message = f"""
أهلاً وسهلاً بك {user.full_name} في {chat_title}! 👋

⚠️ **إخلاء مسؤولية:**
توصيات هذا البوت هي لأغراض تعليمية وتحليلية فقط، ولا تشكل نصيحة استثمارية بأي شكل من الأشكال. تداول العقود والمشتقات المالية ينطوي على مخاطر عالية وقد يؤدي إلى خسارة رأس المال. أنت وحدك المسؤول عن قراراتك الاستثمارية. يرجى استشارة مستشار مالي متخصص قبل اتخاذ أي قرارات استثمارية.
"""
        print(f"Sending welcome message to {user.full_name} in chat {update.chat_member.chat.id}") # Added print
        await context.bot.send_message(chat_id=update.chat_member.chat.id, text=welcome_new_member_message, parse_mode='Markdown')


async def start(update: Update, context: ContextTypes.DEFAULT_TYPE) -> None:
    """معالج أمر /start"""
    print(f"Start command received from {update.effective_user.id} in chat {update.effective_chat.id}") # Added print
    welcome_message = """
👋 مرحباً بك في بوت توصيات الأسهم!

هذا البوت يقدم توصيات تحليل فني للأسهم بناءً على استراتيجيات محددة.

⚠️ **إخلاء مسؤولية:**
توصيات هذا البوت هي لأغراض تعليمية وتحليلية فقط، ولا تشكل نصيحة استثمارية بأي شكل من الأشكال. تداول العقود والمشتقات المالية ينطوي على مخاطر عالية وقد يؤدي إلى خسارة رأس المال. أنت وحدك المسؤول عن قراراتك الاستثمارية. يرجى استشارة مستشار مالي متخصص قبل اتخاذ أي قرارات استثمارية.

اختر نوع التوصية التي ترغب في الحصول عليها:
"""

    keyboard = [
        [InlineKeyboardButton("📈 عقود CALL (شراء)", callback_data='recommend_call')],
        [InlineKeyboardButton("📉 عقود PUT (بيع)", callback_data='recommend_put')],
        [InlineKeyboardButton("🔄 مسح الآن", callback_data='scan_now_manual')]
    ]
    reply_markup = InlineKeyboardMarkup(keyboard)

    await update.message.reply_text(welcome_message, reply_markup=reply_markup, parse_mode='Markdown')

async def button_callback(update: Update, context: ContextTypes.DEFAULT_TYPE) -> None:
    query = update.callback_query
    print(f"Callback query received: {query.data} from {query.from_user.id}") # Added print
    await query.answer()  # Always answer callback queries

    if query.data == 'recommend_call':
        await query.edit_message_text(text="جاري البحث عن توصيات CALL...")
        # Logic to scan for CALL recommendations
        # For now, let's just send a placeholder message
        await asyncio.sleep(2) # Simulate work
        await query.edit_message_text(text="لم يتم العثور على توصيات CALL حالياً.") # Replace with actual results
    elif query.data == 'recommend_put':
        await query.edit_message_text(text="جاري البحث عن توصيات PUT...")
        # Logic to scan for PUT recommendations
        # For now, let's just send a placeholder message
        await asyncio.sleep(2) # Simulate work
        await query.edit_message_text(text="لم يتم العثور على توصيات PUT حالياً.") # Replace with actual results
    elif query.data == 'scan_now_manual':
        await query.edit_message_text(text="جاري المسح الفوري، انتظر قليلاً...")
        alerts = scan_all_stocks()
        if not alerts:
            await query.edit_message_text("لا توجد إشارات جديدة.")
        else:
            for alert in alerts:
                await query.message.reply_text(alert, parse_mode='Markdown')
                await asyncio.sleep(1)
            await query.edit_message_text(text="تم الانتهاء من المسح الفوري.")


async def scan_now(update, context):
    """معالج أمر /scan لمسح فوري"""
    print("scan_now command called.") # Added print statement
    await update.message.reply_text("جاري المسح، انتظر قليلاً...")
    alerts = scan_all_stocks()
    if not alerts:
        print("No alerts found for manual scan.") # Added print statement
        await update.message.reply_text("لا توجد إشارات جديدة.")
    else:
        print(f"Sending {len(alerts)} alerts for manual scan.") # Added print statement
        for alert in alerts:
            await update.message.reply_text(alert, parse_mode='Markdown')
            await asyncio.sleep(1)

def main():
    """الدالة الرئيسية لتشغيل البوت"""
    # إنشاء تطبيق البوت
    # Create a JobQueue instance first
    job_queue_instance = JobQueue()

    # Pass the JobQueue instance to the ApplicationBuilder
    application = Application.builder().token(TELEGRAM_TOKEN).job_queue(job_queue_instance).build()

    # إضافة المعالجات للأوامر والرسائل
    application.add_handler(CommandHandler("start", start))
    application.add_handler(CommandHandler("scan", scan_now))
    application.add_handler(CallbackQueryHandler(button_callback)) # إضافة معالج لأزرار الـ Inline Keyboard
    application.add_handler(ChatMemberHandler(greet_chat_members, ChatMemberHandler.CHAT_MEMBER)) # إضافة معالج للأعضاء الجدد

    # جدولة المسح التلقائي كل ساعة
    job_queue_instance.run_repeating(send_alerts, interval=3600, first=10)  # كل ساعة، أول مرة بعد 10 ثواني

    # تشغيل البوت
    print("البوت يعمل الآن... اذهب لتجربته!")
    application.run_polling()
    # application.idle() is not necessary after run_polling() as it's a blocking call

# استدعاء الدالة الرئيسية
if __name__ == '__main__':
    main()
````

````python
#!/usr/bin/env python3
# -*- coding: utf-8 -*-

# بوت توصيات الأسهم - نسخة شاكر 1.0 (نسخة كاملة)
# آخر تحديث: فبراير 2026

import yfinance as yf
import pandas as pd
import numpy as np
import ta
import logging
import time
import schedule
from datetime import datetime, timedelta
from telegram import Bot, InlineKeyboardButton, InlineKeyboardMarkup
from telegram.ext import Application, CommandHandler, CallbackQueryHandler
import asyncio
import os

# ---------------------------
# 1. إعدادات البوت والتوكن
# ---------------------------
TELEGRAM_TOKEN = "YOUR_BOT_TOKEN_HERE"      # ضع التوكن الخاص بك
CHAT_ID = "YOUR_CHAT_ID_HERE"                # ضع معرف المستخدم أو القناة

# تفعيل السجلات لمشاهدة الأخطاء
logging.basicConfig(level=logging.INFO, format='%(asctime)s - %(name)s - %(levelname)s - %(message)s')

# ---------------------------
# 2. قائمة الأسهم الكاملة (مدمجة)
# ---------------------------
TICKERS = [
    # --- أسهمك القديمة (من الصور) ---
    "A", "AA", "ABEV", "ABNB", "ABT", "ABVX", "ACGL", "ACM", "ACN",
    "ADBE", "ADI", "ADM", "ADP", "ADSK", "AEE", "AEG", "AEIS", "AEM",
    "AEP", "AER", "AES", "AFG", "AFL", "AFRM", "AG", "AGI", "AGNC",
    "AIG", "AIT", "AIZ", "AJG", "AKAM", "ALAB", "ALB", "ALC", "ALGN",
    "ALL", "ALLE", "ALLY", "ALNY", "ALSN", "AM", "AMCR", "AME", "AMH",
    "AMKR", "AMP", "AMRZ", "AMT", "AMX",

    # --- الشركات العملاقة (> 1 تريليون دولار) ---
    "NVDA", "AAPL", "GOOGL", "MSFT", "AMZN", "META", "TSM", "AVGO", "TSLA",
    "BRK.A", "WMT",

    # --- الشركات الكبرى (200 مليار - 1 تريليون) ---
    "LLY", "JPM", "XOM", "V", "JNJ", "MA", "MU", "ORCL", "COST", "AMD",
    "NFLX", "PLTR", "CSCO", "MRK", "TMO", "ASML", "ABBV", "HD", "CVX",
    "PG", "CAT", "KO", "PEP", "INTC", "CMCSA", "MCD", "TXN", "VZ", "AMGN"
]

# ---------------------------
# 3. دوال التحليل الأساسي (الفلاتر)
# ---------------------------

def get_stock_info(ticker):
    """جلب معلومات الشركة الأساسية"""
    try:
        stock = yf.Ticker(ticker)
        info = stock.info
        return info
    except Exception as e:
        logging.error(f"خطأ في جلب معلومات {ticker}: {e}")
        return None

def passes_basic_filters(ticker):
    """
    تطبيق الفلتر الأساسي:
    - Market Cap > 10B (ملاحظة: أغلب الشركات في القائمة أكبر من ذلك بكثير، لكن الفلتر يضمن ذلك)
    - Average Volume > 1M (تقريبي)
    - Spread < 5%
    """
    info = get_stock_info(ticker)
    if not info:
        return False

    # 1. القيمة السوقية
    mcap = info.get('marketCap', 0)
    if mcap < 10e9:
        logging.debug(f"{ticker} فشل فلتر Market Cap: {mcap}")
        return False

    # 2. متوسط الحجم (نأخذ آخر 20 يوم)
    try:
        stock = yf.Ticker(ticker)
        hist = stock.history(period="1mo", interval="1d")
        if hist.empty:
            return False
        avg_vol = hist['Volume'].rolling(20).mean().iloc[-1]
        if avg_vol < 1_000_000:   # أقل من مليون سهم يومياً
            logging.debug(f"{ticker} فشل فلتر الحجم: {avg_vol}")
            return False
    except:
        return False

    # 3. Spread (bid/ask)
    bid = info.get('bid', 0)
    ask = info.get('ask', 0)
    if bid and ask:
        spread = (ask - bid) / bid if bid != 0 else 1
        if spread > 0.05:   # أكبر من 5%
            logging.debug(f"{ticker} فشل فلتر Spread: {spread:.2%}")
            return False

    return True

# ---------------------------
# 4. تحليل SPX (مؤشر S&P 500)
# ---------------------------

def analyze_spx(timeframe="15m"):
    """تحليل مؤشر SPX على الإطار المطلوب"""
    try:
        spx = yf.Ticker("^GSPC")  # أو "SPY"
        hist = spx.history(period="5d", interval=timeframe)
        if len(hist) < 50:
            return "neutral"

        # متوسطات بسيطة لتحديد الاتجاه
        hist['SMA20'] = hist['Close'].rolling(20).mean()
        last = hist.iloc[-1]
        prev = hist.iloc[-2]

        if last['Close'] > last['SMA20'] and last['Close'] > prev['Close']:
            return "uptrend"
        elif last['Close'] < last['SMA20'] and last['Close'] < prev['Close']:
            return "downtrend"
        else:
            return "neutral"
    except:
        return "neutral"

# ---------------------------
# 5. تحليل السهم (فريم ساعتين)
# ---------------------------

def analyze_stock_2h(ticker):
    """
    تحليل فني كامل على فريم الساعتين
    يعيد: {'signal': 'CALL'/'PUT'/None, 'confidence': 0-100, 'details': {...}}
    """
    try:
        stock = yf.Ticker(ticker)
        hist = stock.history(period="1mo", interval="2h")  # آخر شهر، فريم ساعتين
        if len(hist) < 200:
            return None

        # حساب المتوسطات
        hist['SMA50'] = hist['Close'].rolling(50).mean()
        hist['SMA100'] = hist['Close'].rolling(100).mean()
        hist['SMA200'] = hist['Close'].rolling(200).mean()
        hist['Volume_SMA20'] = hist['Volume'].rolling(20).mean()

        last = hist.iloc[-1]
        current_price = last['Close']
        sma50 = last['SMA50']
        sma100 = last['SMA100']
        sma200 = last['SMA200']

        # قمم وقيعان آخر 3 شموع
        high_last_3 = hist['High'].iloc[-3:].max()
        low_last_3 = hist['Low'].iloc[-3:].min()

        # قمة/قاع اليوم (آخر 24 ساعة = 12 شمعة)
        high_today = hist['High'].iloc[-12:].max()
        low_today = hist['Low'].iloc[-12:].min()

        volume_ok = last['Volume'] > last['Volume_SMA20']

        # شروط CALL
        call_condition1 = (sma50 > sma100) or (sma50 > sma200)
        call_condition2 = current_price > sma200
        call_condition3 = current_price > high_last_3 or current_price > high_today
        call_condition4 = volume_ok

        # شروط PUT
        put_condition1 = (sma50 < sma100) or (sma50 < sma200)
        put_condition2 = current_price < sma200
        put_condition3 = current_price < low_last_3 or current_price < low_today
        put_condition4 = volume_ok

        signal = None
        confidence = 0
        details = {}

        if call_condition1 and call_condition2 and call_condition3 and call_condition4:
            signal = "CALL"
            # حساب الثقة: نضيف 25 لكل شرط متحقق
            confidence = 25 * sum([call_condition1, call_condition2, call_condition3, call_condition4])
            details = {
                "sma50": sma50,
                "sma100": sma100,
                "sma200": sma200,
                "price": current_price,
                "high_3": high_last_3,
                "volume_ratio": last['Volume'] / last['Volume_SMA20']
            }
        elif put_condition1 and put_condition2 and put_condition3 and put_condition4:
            signal = "PUT"
            confidence = 25 * sum([put_condition1, put_condition2, put_condition3, put_condition4])
            details = {
                "sma50": sma50,
                "sma100": sma100,
                "sma200": sma200,
                "price": current_price,
                "low_3": low_last_3,
                "volume_ratio": last['Volume'] / last['Volume_SMA20']
            }

        return {"signal": signal, "confidence": int(confidence), "details": details}

    except Exception as e:
        logging.error(f"خطأ في تحليل {ticker}: {e}")
        return None

# ---------------------------
# 6. تصنيف الصفقة (سكالب، قصيرة، سوينق)
# ---------------------------

def classify_trade(signal, confidence, details):
    """
    تصنيف حسب قوة الإشارة وترتيب المتوسطات
    """
    if not signal:
        return None

    sma50 = details.get('sma50')
    sma100 = details.get('sma100')
    sma200 = details.get('sma200')

    # ترتيب المتوسطات
    if signal == "CALL":
        if sma50 > sma100 > sma200:
            return "📈 سوينق 3-7 أيام"
        elif confidence >= 75:
            return "⏳ صفقة قصيرة 1-2 يوم"
        else:
            return "⚡ سكالب"
    else:  # PUT
        if sma50 < sma100 < sma200:
            return "📈 سوينق 3-7 أيام"
        elif confidence >= 75:
            return "⏳ صفقة قصيرة 1-2 يوم"
        else:
            return "⚡ سكالب"

# ---------------------------
# 7. اختيار العقد المناسب (بيانات الخيارات)
# ---------------------------

def get_option_contract(ticker, signal, classification):
    """
    البحث عن عقد خيار مناسب حسب التصنيف.
    """
    try:
        stock = yf.Ticker(ticker)
        # جلب تواريخ انتهاء الصلاحية المتاحة
        expirations = stock.options
        if not expirations:
            return "غير متوفر"

        # اختيار التاريخ المناسب حسب التصنيف
        today = datetime.now().date()
        if "سكالب" in classification:
            days = 7
        elif "قصيرة" in classification:
            days = 14
        else:  # سوينق
            days = 30

        target_date = today + timedelta(days=days)
        # البحث عن أقرب تاريخ انتهاء للعدد المطلوب من الأيام
        exp_date = None
        for exp in expirations:
            exp_dt = datetime.strptime(exp, '%Y-%m-%d').date()
            if exp_dt >= target_date:
                exp_date = exp
                break
        if not exp_date:
            exp_date = expirations[0]  # أقرب تاريخ

        # جلب عقود الخيارات لذلك التاريخ
        opt_chain = stock.option_chain(exp_date)
        if signal == "CALL":
            contracts = opt_chain.calls
        else:
            contracts = opt_chain.puts

        if contracts.empty:
            return "غير متوفر"

        # فلترة العقود: Open Interest عالي (أعلى 10%)، Spread ضيق
        current_price = stock.history(period="1d")['Close'].iloc[-1]
        contracts['strike_diff'] = abs(contracts['strike'] - current_price)
        # فلتر OI
        oi_threshold = contracts['openInterest'].quantile(0.7)  # أعلى 30%
        eligible = contracts[contracts['openInterest'] > oi_threshold]

        if eligible.empty:
            eligible = contracts

        # نختار العقد الأقرب للسعر الحالي (للدلتا التقريبية)
        best = eligible.nsmallest(1, 'strike_diff').iloc[0]

        # حساب Spread تقريبي (ask-bid)/ask
        spread = (best['ask'] - best['bid']) / best['ask'] if best['ask'] > 0 else 1
        if spread > 0.5:  # spread كبير جداً
            return "سبريد واسع"

        # بناء نص العقد المقترح
        delta_approx = 0.5 - (best['strike_diff'] / current_price)  # تقريب بسيط
        delta_approx = max(0.1, min(0.9, delta_approx))

        return f"Delta {delta_approx:.2f} – Exp {exp_date} (أيام: {days})"

    except Exception as e:
        logging.error(f"خطأ في جلب الخيارات لـ {ticker}: {e}")
        return "غير متوفر"

# ---------------------------
# 8. بناء رسالة التنبيه النهائية
# ---------------------------

def build_alert_message(ticker, signal, confidence, classification, details, spx_trend, option_text):
    """
    تنسيق الرسالة كما طلبت تماماً.
    """
    # ترجمة الاتجاه
    if signal == "CALL":
        direction = "Call"
        cross = "50 > 100" if details.get('sma50', 0) > details.get('sma100', 0) else "50 > 200"
        price_status = f"السعر فوق {details.get('sma200', 0):.2f}"
        event = "اختراق قمة 2H"
    else:
        direction = "Put"
        cross = "50 < 100" if details.get('sma50', 0) < details.get('sma100', 0) else "50 < 200"
        price_status = f"السعر تحت {details.get('sma200', 0):.2f}"
        event = "كسر قاع 2H"

    # توافق SPX
    if "uptrend" in spx_trend and signal == "CALL":
        spx_compat = "✅ متوافق"
    elif "downtrend" in spx_trend and signal == "PUT":
        spx_compat = "✅ متوافق"
    else:
        spx_compat = "⚠️ غير متوافق"

    message = f"""
🔔 {ticker}
الاتجاه: {direction}
التصنيف: {classification}
التقاطع: {cross} (إيجابي)
الحالة: {price_status}
الحدث: {event}
SPX: {spx_compat}
الثقة: {confidence}/100
العقد المقترح: {option_text}
"""
    return message

# ---------------------------
# 9. دالة المسح الرئيسية
# ---------------------------

def scan_all_stocks():
    """تمر على كل الأسهم وتفحصها وترسل التنبيهات"""
    alerts = []
    for ticker in TICKERS:
        logging.info(f"فحص {ticker}...")
        time.sleep(0.5)  # تأخير نصف ثانية لتجنب حظر Yahoo

        # 1. الفلتر الأساسي
        if not passes_basic_filters(ticker):
            continue

        # 2. تحليل السهم
        result = analyze_stock_2h(ticker)
        if not result or not result['signal']:
            continue

        signal = result['signal']
        confidence = result['confidence']
        details = result['details']

        # 3. تحليل SPX
        spx_15m = analyze_spx("15m")

        # 4. تصنيف الصفقة
        classification = classify_trade(signal, confidence, details)

        # 5. اختيار العقد
        option = get_option_contract(ticker, signal, classification)

        # 6. بناء الرسالة
        msg = build_alert_message(ticker, signal, confidence, classification, details, spx_15m, option)
        alerts.append(msg)

    return alerts

# ---------------------------
# 10. دوال بوت تليجرام
# ---------------------------

async def send_alerts(context):
    """إرسال التنبيهات إلى تليجرام"""
    alerts = scan_all_stocks()
    if not alerts:
        await context.bot.send_message(chat_id=CHAT_ID, text="لا توجد إشارات جديدة حالياً.")
    else:
        for alert in alerts:
            await context.bot.send_message(chat_id=CHAT_ID, text=alert, parse_mode='Markdown')
            await asyncio.sleep(1)  # تجنب سبام تليجرام

async def start(update, context):
    """معالج أمر /start"""
    await update.message.reply_text(
        "مرحباً! بوت توصيات شاكر 1.0 يعمل.\n"
        "سأرسل لك التنبيهات تلقائياً كل ساعة.\n"
        "لبدء مسح فوري أرسل /scan"
    )

async def scan_now(update, context):
    """معالج أمر /scan لمسح فوري"""
    await update.message.reply_text("جاري المسح، انتظر قليلاً...")
    alerts = scan_all_stocks()
    if not alerts:
        await update.message.reply_text("لا توجد إشارات جديدة.")
    else:
        for alert in alerts:
            await update.message.reply_text(alert, parse_mode='Markdown')
            await asyncio.sleep(1)

def main():
    """الدالة الرئيسية لتشغيل البوت"""
    # إنشاء تطبيق البوت
    application = Application.builder().token(TELEGRAM_TOKEN).build()

    # إضافة الأوامر
    application.add_handler(CommandHandler("start", start))
    application.add_handler(CommandHandler("scan", scan_now))

    # جدولة المسح التلقائي كل ساعة
    job_queue = application.job_queue
    job_queue.run_repeating(send_alerts, interval=3600, first=10)  # كل ساعة، أول مرة بعد 10 ثواني

    # تشغيل البوت
    print("البوت يعمل الآن...")
    application.run_polling()

if __name__ == "__main__":
    main()
````

````python
#!/usr/bin/env python3
# -*- coding: utf-8 -*-

# بوت توصيات الأسهم - نسخة شاكر 1.0 (نسخة كاملة)
# آخر تحديث: فبراير 2026

import yfinance as yf
import pandas as pd
import numpy as np
import ta
import logging
import time
import schedule
from datetime import datetime, timedelta
from telegram import Bot, InlineKeyboardButton, InlineKeyboardMarkup, Update
from telegram.ext import Application, CommandHandler, CallbackQueryHandler, ContextTypes, JobQueue, ChatMemberHandler # أضفنا ChatMemberHandler
import asyncio
import os
import nest_asyncio # ADDED

# Apply nest_asyncio to allow nested event loops in environments like Colab
nest_asyncio.apply() # ADDED

# ---------------------------
# 1. إعدادات البوت والتوكن
# ---------------------------
TELEGRAM_TOKEN = "8699821370:AAEQUSbLTgf7MmWqo5vV5LHPOz30wfqOfqw"      # ضع التوكن الخاص بك
CHAT_ID = "7854020427"                # ضع معرف المستخدم أو القناة

# تفعيل السجلات لمشاهدة الأخطاء
logging.basicConfig(level=logging.INFO, format='%(asctime)s - %(name)s - %(levelname)s - %(message)s')

# ---------------------------
# 2. قائمة الأسهم الكاملة (مدمجة)
# ---------------------------
TICKERS = [
    # --- أسهمك القديمة (من الصور) ---
    "A", "AA", "ABEV", "ABNB", "ABT", "ABVX", "ACGL", "ACM", "ACN",
    "ADBE", "ADI", "ADM", "ADP", "ADSK", "AEE", "AEG", "AEIS", "AEM",
    "AEP", "AER", "AES", "AFG", "AFL", "AFRM", "AG", "AGI", "AGNC",
    "AIG", "AIT", "AIZ", "AJG", "AKAM", "ALAB", "ALB", "ALC", "ALGN",
    "ALL", "ALLE", "ALLY", "ALNY", "ALSN", "AM", "AMCR", "AME", "AMH",
    "AMKR", "AMP", "AMRZ", "AMT", "AMX",

    # --- الشركات العملاقة (> 1 تريليون دولار) ---
    "NVDA", "AAPL", "GOOGL", "MSFT", "AMZN", "META", "TSM", "AVGO", "TSLA",
    "BRK.A", "WMT",

    # --- الشركات الكبرى (200 مليار - 1 تريليون) ---
    "LLY", "JPM", "XOM", "V", "JNJ", "MA", "MU", "ORCL", "COST", "AMD",
    "NFLX", "PLTR", "CSCO", "MRK", "TMO", "ASML", "ABBV", "HD", "CVX",
    "PG", "CAT", "KO", "PEP", "INTC", "CMCSA", "MCD", "TXN", "VZ", "AMGN"
]

# ---------------------------
# 3. دوال التحليل الأساسي (الفلاتر)
# ---------------------------

def get_stock_info(ticker):
    """جلب معلومات الشركة الأساسية"""
    try:
        stock = yf.Ticker(ticker)
        info = stock.info
        return info
    except Exception as e:
        logging.error(f"خطأ في جلب معلومات {ticker}: {e}")
        return None

def passes_basic_filters(ticker):
    """
    تطبيق الفلتر الأساسي:
    - Market Cap > 10B (ملاحظة: أغلب الشركات في القائمة أكبر من ذلك بكثير، لكن الفلتر يضمن ذلك)
    - Average Volume > 1M (تقريبي)
    - Spread < 5%
    """
    info = get_stock_info(ticker)
    if not info:
        return False

    # 1. القيمة السوقية
    mcap = info.get('marketCap', 0)
    if mcap < 10e9:
        print(f"{ticker} فشل فلتر Market Cap: {mcap}") # Changed from logging.debug
        return False

    # 2. متوسط الحجم (نأخذ آخر 20 يوم)
    try:
        stock = yf.Ticker(ticker)
        hist = stock.history(period="1mo", interval="1d")
        if hist.empty:
            return False
        avg_vol = hist['Volume'].rolling(20).mean().iloc[-1]
        if avg_vol < 1_000_000:   # أقل من مليون سهم يومياً
            print(f"{ticker} فشل فلتر الحجم: {avg_vol}") # Changed from logging.debug
            return False
    except:
        return False

    # 3. Spread (bid/ask)
    bid = info.get('bid', 0)
    ask = info.get('ask', 0)
    if bid and ask:
        spread = (ask - bid) / bid if bid != 0 else 1
        if spread > 0.05:   # أكبر من 5%
            print(f"{ticker} فشل فلتر Spread: {spread:.2%}") # Changed from logging.debug
            return False

    return True

# ---------------------------
# 4. تحليل SPX (مؤشر S&P 500)
# ---------------------------

def analyze_spx(timeframe="15m"):
    """تحليل مؤشر SPX على الإطار المطلوب"""
    try:
        spx = yf.Ticker("^GSPC")  # أو "SPY"
        hist = spx.history(period="5d", interval=timeframe)
        if len(hist) < 50:
            return "neutral"

        # متوسطات بسيطة لتحديد الاتجاه
        hist['SMA20'] = hist['Close'].rolling(20).mean()
        last = hist.iloc[-1]
        prev = hist.iloc[-2]

        if last['Close'] > last['SMA20'] and last['Close'] > prev['Close']:
            return "uptrend"
        elif last['Close'] < last['SMA20'] and last['Close'] < prev['Close']:
            return "downtrend"
        else:
            return "neutral"
    except:
        return "neutral"

# ---------------------------
# 5. تحليل السهم (فريم ساعتين)
# ---------------------------

def analyze_stock_2h(ticker):
    """
    تحليل فني كامل على فريم الساعتين
    يعيد: {'signal': 'CALL'/'PUT'/None, 'confidence': 0-100, 'details': {...}}
    """
    try:
        print(f"Analyzing {ticker} for 2h interval.") # Added print statement
        stock = yf.Ticker(ticker)
        hist = stock.history(period="1mo", interval="2h")  # آخر شهر، فريم ساعتين
        if len(hist) < 200:
            print(f"Not enough historical data for {ticker} in 2h interval.") # Added print statement
            return None

        # حساب المتوسطات
        hist['SMA50'] = hist['Close'].rolling(50).mean()
        hist['SMA100'] = hist['Close'].rolling(100).mean()
        hist['SMA200'] = hist['Close'].rolling(200).mean()
        hist['Volume_SMA20'] = hist['Volume'].rolling(20).mean()

        last = hist.iloc[-1]
        current_price = last['Close']
        sma50 = last['SMA50']
        sma100 = last['SMA100']
        sma200 = last['SMA200']

        # قمم وقيعان آخر 3 شموع
        high_last_3 = hist['High'].iloc[-3:].max()
        low_last_3 = hist['Low'].iloc[-3:].min()

        # قمة/قاع اليوم (آخر 24 ساعة = 12 شمعة)
        high_today = hist['High'].iloc[-12:].max()
        low_today = hist['Low'].iloc[-12:].min()

        volume_ok = last['Volume'] > last['Volume_SMA20']

        # شروط CALL
        call_condition1 = (sma50 > sma100) or (sma50 > sma200)
        call_condition2 = current_price > sma200
        call_condition3 = current_price > high_last_3 or current_price > high_today
        call_condition4 = volume_ok

        # شروط PUT
        put_condition1 = (sma50 < sma100) or (sma50 < sma200)
        put_condition2 = current_price < sma200
        put_condition3 = current_price < low_last_3 or current_price < low_today
        put_condition4 = volume_ok

        signal = None
        confidence = 0
        details = {}

        if call_condition1 and call_condition2 and call_condition3 and call_condition4:
            signal = "CALL"
            # حساب الثقة: نضيف 25 لكل شرط متحقق
            confidence = 25 * sum([call_condition1, call_condition2, call_condition3, call_condition4])
            details = {
                "sma50": sma50,
                "sma100": sma100,
                "sma200": sma200,
                "price": current_price,
                "high_3": high_last_3,
                "volume_ratio": last['Volume'] / last['Volume_SMA20']
            }
        elif put_condition1 and put_condition2 and put_condition3 and put_condition4:
            signal = "PUT"
            confidence = 25 * sum([put_condition1, put_condition2, put_condition3, put_condition4])
            details = {
                "sma50": sma50,
                "sma100": sma100,
                "sma200": sma200,
                "price": current_price,
                "low_3": low_last_3,
                "volume_ratio": last['Volume'] / last['Volume_SMA20']
            }

        if signal: # Added print statement
            print(f"Signal for {ticker}: {signal}, Confidence: {confidence}")
        return {"signal": signal, "confidence": int(confidence), "details": details}

    except Exception as e:
        logging.error(f"خطأ في تحليل {ticker}: {e}")
        return None

# ---------------------------
# 6. تصنيف الصفقة (سكالب، قصيرة، سوينق)
# ---------------------------

def classify_trade(signal, confidence, details):
    """
    تصنيف حسب قوة الإشارة وترتيب المتوسطات
    """
    if not signal:
        return None

    sma50 = details.get('sma50')
    sma100 = details.get('sma100')
    sma200 = details.get('sma200')

    # ترتيب المتوسطات
    if signal == "CALL":
        if sma50 > sma100 > sma200:
            return "📈 سوينق 3-7 أيام"
        elif confidence >= 75:
            return "⏳ صفقة قصيرة 1-2 يوم"
        else:
            return "⚡ سكالب"
    else:  # PUT
        if sma50 < sma100 < sma200:
            return "📈 سوينق 3-7 أيام"
        elif confidence >= 75:
            return "⏳ صفقة قصيرة 1-2 يوم"
        else:
            return "⚡ سكالب"

# ---------------------------
# 7. اختيار العقد المناسب (بيانات الخيارات)
# ---------------------------

def get_option_contract(ticker, signal, classification):
    """
    البحث عن عقد خيار مناسب حسب التصنيف.
    """
    try:
        stock = yf.Ticker(ticker)
        # جلب تواريخ انتهاء الصلاحية المتاحة
        expirations = stock.options
        if not expirations:
            return "غير متوفر"

        # اختيار التاريخ المناسب حسب التصنيف
        today = datetime.now().date()
        if "سكالب" in classification:
            days = 7
        elif "قصيرة" in classification:
            days = 14
        else:  # سوينق
            days = 30

        target_date = today + timedelta(days=days)
        # البحث عن أقرب تاريخ انتهاء للعدد المطلوب من الأيام
        exp_date = None
        for exp in expirations:
            exp_dt = datetime.strptime(exp, '%Y-%m-%d').date()
            if exp_dt >= target_date:
                exp_date = exp
                break
        if not exp_date:
            exp_date = expirations[0]  # أقرب تاريخ

        # جلب عقود الخيارات لذلك التاريخ
        opt_chain = stock.option_chain(exp_date)
        if signal == "CALL":
            contracts = opt_chain.calls
        else:
            contracts = opt_chain.puts

        if contracts.empty:
            return "غير متوفر"

        # فلترة العقود: Open Interest عالي (أعلى 10%)، Spread ضيق
        current_price = stock.history(period="1d")['Close'].iloc[-1]
        contracts['strike_diff'] = abs(contracts['strike'] - current_price)
        # فلتر OI
        oi_threshold = contracts['openInterest'].quantile(0.7)  # أعلى 30%
        eligible = contracts[contracts['openInterest'] > oi_threshold]

        if eligible.empty:
            eligible = contracts

        # نختار العقد الأقرب للسعر الحالي (للدلتا التقريبية)
        best = eligible.nsmallest(1, 'strike_diff').iloc[0]

        # حساب Spread تقريبي (ask-bid)/ask
        spread = (best['ask'] - best['bid']) / best['ask'] if best['ask'] > 0 else 1
        if spread > 0.05:   # أكبر من 5%
            return "سبريد واسع"

        # بناء نص العقد المقترح
        delta_approx = 0.5 - (best['strike_diff'] / current_price)  # تقريب بسيط
        delta_approx = max(0.1, min(0.9, delta_approx))

        return f"Delta {delta_approx:.2f} – Exp {exp_date} (أيام: {days})"

    except Exception as e:
        logging.error(f"خطأ في جلب الخيارات لـ {ticker}: {e}")
        return "غير متوفر"

# ---------------------------
# 8. بناء رسالة التنبيه النهائية
# ---------------------------

def build_alert_message(ticker, signal, confidence, classification, details, spx_trend, option_text):
    """
    تنسيق الرسالة كما طلبت تماماً.
    """
    # ترجمة الاتجاه
    if signal == "CALL":
        direction = "Call"
        cross = "50 > 100" if details.get('sma50', 0) > details.get('sma100', 0) else "50 > 200"
        price_status = f"السعر فوق {details.get('sma200', 0):.2f}"
        event = "اختراق قمة 2H"
    else:
        direction = "Put"
        cross = "50 < 100" if details.get('sma50', 0) < details.get('sma100', 0) else "50 < 200"
        price_status = f"السعر تحت {details.get('sma200', 0):.2f}"
        event = "كسر قاع 2H"

    # توافق SPX
    if "uptrend" in spx_trend and signal == "CALL":
        spx_compat = "✅ متوافق"
    elif "downtrend" in spx_trend and signal == "PUT":
        spx_compat = "✅ متوافق"
    else:
        spx_compat = "⚠️ غير متوافق"

    message = f"""
🔔 {ticker}
الاتجاه: {direction}
التصنيف: {classification}
التقاطع: {cross} (إيجابي)
الحالة: {price_status}
الحدث: {event}
SPX: {spx_compat}
الثقة: {confidence}/100
العقد المقترح: {option_text}
"""
    return message

# ---------------------------
# 9. دالة المسح الرئيسية
# ---------------------------

def scan_all_stocks():
    """تمر على كل الأسهم وتفحصها وترسل التنبيهات"""
    alerts = []
    print("Starting stock scan...") # Added print statement
    for ticker in TICKERS:
        logging.info(f"فحص {ticker}...")
        time.sleep(0.5)  # تأخير نصف ثانية لتجنب حظر Yahoo

        # 1. الفلتر الأساسي
        if not passes_basic_filters(ticker):
            print(f"Ticker {ticker} did not pass basic filters.") # Added print statement
            continue

        # 2. تحليل السهم
        result = analyze_stock_2h(ticker)
        if not result or not result['signal']:
            print(f"No signal found for {ticker} after 2h analysis.") # Added print statement
            continue

        signal = result['signal']
        confidence = result['confidence']
        details = result['details']

        # 3. تحليل SPX
        spx_15m = analyze_spx("15m")

        # 4. تصنيف الصفقة
        classification = classify_trade(signal, confidence, details)

        # 5. اختيار العقد
        option = get_option_contract(ticker, signal, classification)

        # 6. بناء الرسالة
        msg = build_alert_message(ticker, signal, confidence, classification, details, spx_15m, option)
        alerts.append(msg)
    print(f"Scan finished. Found {len(alerts)} alerts.") # Added print statement
    return alerts

# ---------------------------
# 10. دوال بوت تليجرام
# ---------------------------

async def send_alerts(context):
    """إرسال التنبيهات إلى تليجرام"""
    print("send_alerts function called.") # Added print statement
    alerts = scan_all_stocks()
    if not alerts:
        print("No alerts found to send.") # Added print statement
        await context.bot.send_message(chat_id=CHAT_ID, text="لا توجد إشارات جديدة حالياً.")
    else:
        print(f"Sending {len(alerts)} alerts.") # Added print statement
        for alert in alerts:
            await context.bot.send_message(chat_id=CHAT_ID, text=alert, parse_mode='Markdown')
            await asyncio.sleep(1)  # تجنب سبام تليجرام

# دالة للترحيب بالأعضاء الجدد وإرسال إخلاء المسؤولية
async def greet_chat_members(update: Update, context: ContextTypes.DEFAULT_TYPE) -> None:
    print(f"New chat member event: {update.chat_member}") # Added print
    if update.chat_member.new_chat_member.status == 'member':
        user = update.chat_member.new_chat_member.user
        chat_title = update.chat_member.chat.title or "هذه المجموعة"
        welcome_new_member_message = f"""
أهلاً وسهلاً بك {user.full_name} في {chat_title}! 👋

⚠️ **إخلاء مسؤولية:**
توصيات هذا البوت هي لأغراض تعليمية وتحليلية فقط، ولا تشكل نصيحة استثمارية بأي شكل من الأشكال. تداول العقود والمشتقات المالية ينطوي على مخاطر عالية وقد يؤدي إلى خسارة رأس المال. أنت وحدك المسؤول عن قراراتك الاستثمارية. يرجى استشارة مستشار مالي متخصص قبل اتخاذ أي قرارات استثمارية.
"""
        print(f"Sending welcome message to {user.full_name} in chat {update.chat_member.chat.id}") # Added print
        await context.bot.send_message(chat_id=update.chat_member.chat.id, text=welcome_new_member_message, parse_mode='Markdown')


async def start(update: Update, context: ContextTypes.DEFAULT_TYPE) -> None:
    """معالج أمر /start"""
    print(f"Start command received from {update.effective_user.id} in chat {update.effective_chat.id}") # Added print
    welcome_message = """
👋 مرحباً بك في بوت توصيات الأسهم!

هذا البوت يقدم توصيات تحليل فني للأسهم بناءً على استراتيجيات محددة.

⚠️ **إخلاء مسؤولية:**
توصيات هذا البوت هي لأغراض تعليمية وتحليلية فقط، ولا تشكل نصيحة استثمارية بأي شكل من الأشكال. تداول العقود والمشتقات المالية ينطوي على مخاطر عالية وقد يؤدي إلى خسارة رأس المال. أنت وحدك المسؤول عن قراراتك الاستثمارية. يرجى استشارة مستشار مالي متخصص قبل اتخاذ أي قرارات استثمارية.

اختر نوع التوصية التي ترغب في الحصول عليها:
"""

    keyboard = [
        [InlineKeyboardButton("📈 عقود CALL (شراء)", callback_data='recommend_call')],
        [InlineKeyboardButton("📉 عقود PUT (بيع)", callback_data='recommend_put')],
        [InlineKeyboardButton("🔄 مسح الآن", callback_data='scan_now_manual')]
    ]
    reply_markup = InlineKeyboardMarkup(keyboard)

    await update.message.reply_text(welcome_message, reply_markup=reply_markup, parse_mode='Markdown')

async def button_callback(update: Update, context: ContextTypes.DEFAULT_TYPE) -> None:
    query = update.callback_query
    print(f"Callback query received: {query.data} from {query.from_user.id}") # Added print
    await query.answer()  # Always answer callback queries

    if query.data == 'recommend_call':
        await query.edit_message_text(text="جاري البحث عن توصيات CALL...")
        # Logic to scan for CALL recommendations
        # For now, let's just send a placeholder message
        await asyncio.sleep(2) # Simulate work
        await query.edit_message_text(text="لم يتم العثور على توصيات CALL حالياً.") # Replace with actual results
    elif query.data == 'recommend_put':
        await query.edit_message_text(text="جاري البحث عن توصيات PUT...")
        # Logic to scan for PUT recommendations
        # For now, let's just send a placeholder message
        await asyncio.sleep(2) # Simulate work
        await query.edit_message_text(text="لم يتم العثور على توصيات PUT حالياً.") # Replace with actual results
    elif query.data == 'scan_now_manual':
        await query.edit_message_text(text="جاري المسح الفوري، انتظر قليلاً...")
        alerts = scan_all_stocks()
        if not alerts:
            await query.edit_message_text("لا توجد إشارات جديدة.")
        else:
            for alert in alerts:
                await query.message.reply_text(alert, parse_mode='Markdown')
                await asyncio.sleep(1)
            await query.edit_message_text(text="تم الانتهاء من المسح الفوري.")


async def scan_now(update, context):
    """معالج أمر /scan لمسح فوري"""
    print("scan_now command called.") # Added print statement
    await update.message.reply_text("جاري المسح، انتظر قليلاً...")
    alerts = scan_all_stocks()
    if not alerts:
        print("No alerts found for manual scan.") # Added print statement
        await update.message.reply_text("لا توجد إشارات جديدة.")
    else:
        print(f"Sending {len(alerts)} alerts for manual scan.") # Added print statement
        for alert in alerts:
            await update.message.reply_text(alert, parse_mode='Markdown')
            await asyncio.sleep(1)

def main():
    """الدالة الرئيسية لتشغيل البوت"""
    # إنشاء تطبيق البوت
    # Create a JobQueue instance first
    job_queue_instance = JobQueue()

    # Pass the JobQueue instance to the ApplicationBuilder
    application = Application.builder().token(TELEGRAM_TOKEN).job_queue(job_queue_instance).build()

    # إضافة المعالجات للأوامر والرسائل
    application.add_handler(CommandHandler("start", start))
    application.add_handler(CommandHandler("scan", scan_now))
    application.add_handler(CallbackQueryHandler(button_callback)) # إضافة معالج لأزرار الـ Inline Keyboard
    application.add_handler(ChatMemberHandler(greet_chat_members, ChatMemberHandler.CHAT_MEMBER)) # إضافة معالج للأعضاء الجدد

    # جدولة المسح التلقائي كل ساعة
    job_queue_instance.run_repeating(send_alerts, interval=3600, first=10)  # كل ساعة، أول مرة بعد 10 ثواني

    # تشغيل البوت
    print("البوت يعمل الآن... اذهب لتجربته!")
    application.run_polling()
    # application.idle() is not necessary after run_polling() as it's a blocking call

# استدعاء الدالة الرئيسية
if __name__ == '__main__':
    main()
````

````python
#!/usr/bin/env python3
# -*- coding: utf-8 -*-

# بوت توصيات الأسهم - نسخة شاكر 1.0 (نسخة كاملة)
# آخر تحديث: فبراير 2026

import yfinance as yf
import pandas as pd
import numpy as np
import ta
import logging
import time
import schedule
from datetime import datetime, timedelta
from telegram import Bot, InlineKeyboardButton, InlineKeyboardMarkup
from telegram.ext import Application, CommandHandler, CallbackQueryHandler
import asyncio
import os

# ---------------------------
# 1. إعدادات البوت والتوكن
# ---------------------------
TELEGRAM_TOKEN = "YOUR_BOT_TOKEN_HERE"      # ضع التوكن الخاص بك
CHAT_ID = "YOUR_CHAT_ID_HERE"                # ضع معرف المستخدم أو القناة

# تفعيل السجلات لمشاهدة الأخطاء
logging.basicConfig(level=logging.INFO, format='%(asctime)s - %(name)s - %(levelname)s - %(message)s')

# ---------------------------
# 2. قائمة الأسهم الكاملة (مدمجة)
# ---------------------------
TICKERS = [
    # --- أسهمك القديمة (من الصور) ---
    "A", "AA", "ABEV", "ABNB", "ABT", "ABVX", "ACGL", "ACM", "ACN",
    "ADBE", "ADI", "ADM", "ADP", "ADSK", "AEE", "AEG", "AEIS", "AEM",
    "AEP", "AER", "AES", "AFG", "AFL", "AFRM", "AG", "AGI", "AGNC",
    "AIG", "AIT", "AIZ", "AJG", "AKAM", "ALAB", "ALB", "ALC", "ALGN",
    "ALL", "ALLE", "ALLY", "ALNY", "ALSN", "AM", "AMCR", "AME", "AMH",
    "AMKR", "AMP", "AMRZ", "AMT", "AMX",

    # --- الشركات العملاقة (> 1 تريليون دولار) ---
    "NVDA", "AAPL", "GOOGL", "MSFT", "AMZN", "META", "TSM", "AVGO", "TSLA",
    "BRK.A", "WMT",

    # --- الشركات الكبرى (200 مليار - 1 تريليون) ---
    "LLY", "JPM", "XOM", "V", "JNJ", "MA", "MU", "ORCL", "COST", "AMD",
    "NFLX", "PLTR", "CSCO", "MRK", "TMO", "ASML", "ABBV", "HD", "CVX",
    "PG", "CAT", "KO", "PEP", "INTC", "CMCSA", "MCD", "TXN", "VZ", "AMGN"
]

# ---------------------------
# 3. دوال التحليل الأساسي (الفلاتر)
# ---------------------------

def get_stock_info(ticker):
    """جلب معلومات الشركة الأساسية"""
    try:
        stock = yf.Ticker(ticker)
        info = stock.info
        return info
    except Exception as e:
        logging.error(f"خطأ في جلب معلومات {ticker}: {e}")
        return None

def passes_basic_filters(ticker):
    """
    تطبيق الفلتر الأساسي:
    - Market Cap > 10B (ملاحظة: أغلب الشركات في القائمة أكبر من ذلك بكثير، لكن الفلتر يضمن ذلك)
    - Average Volume > 1M (تقريبي)
    - Spread < 5%
    """
    info = get_stock_info(ticker)
    if not info:
        return False

    # 1. القيمة السوقية
    mcap = info.get('marketCap', 0)
    if mcap < 10e9:
        logging.debug(f"{ticker} فشل فلتر Market Cap: {mcap}")
        return False

    # 2. متوسط الحجم (نأخذ آخر 20 يوم)
    try:
        stock = yf.Ticker(ticker)
        hist = stock.history(period="1mo", interval="1d")
        if hist.empty:
            return False
        avg_vol = hist['Volume'].rolling(20).mean().iloc[-1]
        if avg_vol < 1_000_000:   # أقل من مليون سهم يومياً
            logging.debug(f"{ticker} فشل فلتر الحجم: {avg_vol}")
            return False
    except:
        return False

    # 3. Spread (bid/ask)
    bid = info.get('bid', 0)
    ask = info.get('ask', 0)
    if bid and ask:
        spread = (ask - bid) / bid if bid != 0 else 1
        if spread > 0.05:   # أكبر من 5%
            logging.debug(f"{ticker} فشل فلتر Spread: {spread:.2%}")
            return False

    return True

# ---------------------------
# 4. تحليل SPX (مؤشر S&P 500)
# ---------------------------

def analyze_spx(timeframe="15m"):
    """تحليل مؤشر SPX على الإطار المطلوب"""
    try:
        spx = yf.Ticker("^GSPC")  # أو "SPY"
        hist = spx.history(period="5d", interval=timeframe)
        if len(hist) < 50:
            return "neutral"

        # متوسطات بسيطة لتحديد الاتجاه
        hist['SMA20'] = hist['Close'].rolling(20).mean()
        last = hist.iloc[-1]
        prev = hist.iloc[-2]

        if last['Close'] > last['SMA20'] and last['Close'] > prev['Close']:
            return "uptrend"
        elif last['Close'] < last['SMA20'] and last['Close'] < prev['Close']:
            return "downtrend"
        else:
            return "neutral"
    except:
        return "neutral"

# ---------------------------
# 5. تحليل السهم (فريم ساعتين)
# ---------------------------

def analyze_stock_2h(ticker):
    """
    تحليل فني كامل على فريم الساعتين
    يعيد: {'signal': 'CALL'/'PUT'/None, 'confidence': 0-100, 'details': {...}}
    """
    try:
        stock = yf.Ticker(ticker)
        hist = stock.history(period="1mo", interval="2h")  # آخر شهر، فريم ساعتين
        if len(hist) < 200:
            return None

        # حساب المتوسطات
        hist['SMA50'] = hist['Close'].rolling(50).mean()
        hist['SMA100'] = hist['Close'].rolling(100).mean()
        hist['SMA200'] = hist['Close'].rolling(200).mean()
        hist['Volume_SMA20'] = hist['Volume'].rolling(20).mean()

        last = hist.iloc[-1]
        current_price = last['Close']
        sma50 = last['SMA50']
        sma100 = last['SMA100']
        sma200 = last['SMA200']

        # قمم وقيعان آخر 3 شموع
        high_last_3 = hist['High'].iloc[-3:].max()
        low_last_3 = hist['Low'].iloc[-3:].min()

        # قمة/قاع اليوم (آخر 24 ساعة = 12 شمعة)
        high_today = hist['High'].iloc[-12:].max()
        low_today = hist['Low'].iloc[-12:].min()

        volume_ok = last['Volume'] > last['Volume_SMA20']

        # شروط CALL
        call_condition1 = (sma50 > sma100) or (sma50 > sma200)
        call_condition2 = current_price > sma200
        call_condition3 = current_price > high_last_3 or current_price > high_today
        call_condition4 = volume_ok

        # شروط PUT
        put_condition1 = (sma50 < sma100) or (sma50 < sma200)
        put_condition2 = current_price < sma200
        put_condition3 = current_price < low_last_3 or current_price < low_today
        put_condition4 = volume_ok

        signal = None
        confidence = 0
        details = {}

        if call_condition1 and call_condition2 and call_condition3 and call_condition4:
            signal = "CALL"
            # حساب الثقة: نضيف 25 لكل شرط متحقق
            confidence = 25 * sum([call_condition1, call_condition2, call_condition3, call_condition4])
            details = {
                "sma50": sma50,
                "sma100": sma100,
                "sma200": sma200,
                "price": current_price,
                "high_3": high_last_3,
                "volume_ratio": last['Volume'] / last['Volume_SMA20']
            }
        elif put_condition1 and put_condition2 and put_condition3 and put_condition4:
            signal = "PUT"
            confidence = 25 * sum([put_condition1, put_condition2, put_condition3, put_condition4])
            details = {
                "sma50": sma50,
                "sma100": sma100,
                "sma200": sma200,
                "price": current_price,
                "low_3": low_last_3,
                "volume_ratio": last['Volume'] / last['Volume_SMA20']
            }

        return {"signal": signal, "confidence": int(confidence), "details": details}

    except Exception as e:
        logging.error(f"خطأ في تحليل {ticker}: {e}")
        return None

# ---------------------------
# 6. تصنيف الصفقة (سكالب، قصيرة، سوينق)
# ---------------------------

def classify_trade(signal, confidence, details):
    """
    تصنيف حسب قوة الإشارة وترتيب المتوسطات
    """
    if not signal:
        return None

    sma50 = details.get('sma50')
    sma100 = details.get('sma100')
    sma200 = details.get('sma200')

    # ترتيب المتوسطات
    if signal == "CALL":
        if sma50 > sma100 > sma200:
            return "📈 سوينق 3-7 أيام"
        elif confidence >= 75:
            return "⏳ صفقة قصيرة 1-2 يوم"
        else:
            return "⚡ سكالب"
    else:  # PUT
        if sma50 < sma100 < sma200:
            return "📈 سوينق 3-7 أيام"
        elif confidence >= 75:
            return "⏳ صفقة قصيرة 1-2 يوم"
        else:
            return "⚡ سكالب"

# ---------------------------
# 7. اختيار العقد المناسب (بيانات الخيارات)
# ---------------------------

def get_option_contract(ticker, signal, classification):
    """
    البحث عن عقد خيار مناسب حسب التصنيف.
    """
    try:
        stock = yf.Ticker(ticker)
        # جلب تواريخ انتهاء الصلاحية المتاحة
        expirations = stock.options
        if not expirations:
            return "غير متوفر"

        # اختيار التاريخ المناسب حسب التصنيف
        today = datetime.now().date()
        if "سكالب" in classification:
            days = 7
        elif "قصيرة" in classification:
            days = 14
        else:  # سوينق
            days = 30

        target_date = today + timedelta(days=days)
        # البحث عن أقرب تاريخ انتهاء للعدد المطلوب من الأيام
        exp_date = None
        for exp in expirations:
            exp_dt = datetime.strptime(exp, '%Y-%m-%d').date()
            if exp_dt >= target_date:
                exp_date = exp
                break
        if not exp_date:
            exp_date = expirations[0]  # أقرب تاريخ

        # جلب عقود الخيارات لذلك التاريخ
        opt_chain = stock.option_chain(exp_date)
        if signal == "CALL":
            contracts = opt_chain.calls
        else:
            contracts = opt_chain.puts

        if contracts.empty:
            return "غير متوفر"

        # فلترة العقود: Open Interest عالي (أعلى 10%)، Spread ضيق
        current_price = stock.history(period="1d")['Close'].iloc[-1]
        contracts['strike_diff'] = abs(contracts['strike'] - current_price)
        # فلتر OI
        oi_threshold = contracts['openInterest'].quantile(0.7)  # أعلى 30%
        eligible = contracts[contracts['openInterest'] > oi_threshold]

        if eligible.empty:
            eligible = contracts

        # نختار العقد الأقرب للسعر الحالي (للدلتا التقريبية)
        best = eligible.nsmallest(1, 'strike_diff').iloc[0]

        # حساب Spread تقريبي (ask-bid)/ask
        spread = (best['ask'] - best['bid']) / best['ask'] if best['ask'] > 0 else 1
        if spread > 0.5:  # spread كبير جداً
            return "سبريد واسع"

        # بناء نص العقد المقترح
        delta_approx = 0.5 - (best['strike_diff'] / current_price)  # تقريب بسيط
        delta_approx = max(0.1, min(0.9, delta_approx))

        return f"Delta {delta_approx:.2f} – Exp {exp_date} (أيام: {days})"

    except Exception as e:
        logging.error(f"خطأ في جلب الخيارات لـ {ticker}: {e}")
        return "غير متوفر"

# ---------------------------
# 8. بناء رسالة التنبيه النهائية
# ---------------------------

def build_alert_message(ticker, signal, confidence, classification, details, spx_trend, option_text):
    """
    تنسيق الرسالة كما طلبت تماماً.
    """
    # ترجمة الاتجاه
    if signal == "CALL":
        direction = "Call"
        cross = "50 > 100" if details.get('sma50', 0) > details.get('sma100', 0) else "50 > 200"
        price_status = f"السعر فوق {details.get('sma200', 0):.2f}"
        event = "اختراق قمة 2H"
    else:
        direction = "Put"
        cross = "50 < 100" if details.get('sma50', 0) < details.get('sma100', 0) else "50 < 200"
        price_status = f"السعر تحت {details.get('sma200', 0):.2f}"
        event = "كسر قاع 2H"

    # توافق SPX
    if "uptrend" in spx_trend and signal == "CALL":
        spx_compat = "✅ متوافق"
    elif "downtrend" in spx_trend and signal == "PUT":
        spx_compat = "✅ متوافق"
    else:
        spx_compat = "⚠️ غير متوافق"

    message = f"""
🔔 {ticker}
الاتجاه: {direction}
التصنيف: {classification}
التقاطع: {cross} (إيجابي)
الحالة: {price_status}
الحدث: {event}
SPX: {spx_compat}
الثقة: {confidence}/100
العقد المقترح: {option_text}
"""
    return message

# ---------------------------
# 9. دالة المسح الرئيسية
# ---------------------------

def scan_all_stocks():
    """تمر على كل الأسهم وتفحصها وترسل التنبيهات"""
    alerts = []
    for ticker in TICKERS:
        logging.info(f"فحص {ticker}...")
        time.sleep(0.5)  # تأخير نصف ثانية لتجنب حظر Yahoo

        # 1. الفلتر الأساسي
        if not passes_basic_filters(ticker):
            continue

        # 2. تحليل السهم
        result = analyze_stock_2h(ticker)
        if not result or not result['signal']:
            continue

        signal = result['signal']
        confidence = result['confidence']
        details = result['details']

        # 3. تحليل SPX
        spx_15m = analyze_spx("15m")

        # 4. تصنيف الصفقة
        classification = classify_trade(signal, confidence, details)

        # 5. اختيار العقد
        option = get_option_contract(ticker, signal, classification)

        # 6. بناء الرسالة
        msg = build_alert_message(ticker, signal, confidence, classification, details, spx_15m, option)
        alerts.append(msg)

    return alerts

# ---------------------------
# 10. دوال بوت تليجرام
# ---------------------------

async def send_alerts(context):
    """إرسال التنبيهات إلى تليجرام"""
    alerts = scan_all_stocks()
    if not alerts:
        await context.bot.send_message(chat_id=CHAT_ID, text="لا توجد إشارات جديدة حالياً.")
    else:
        for alert in alerts:
            await context.bot.send_message(chat_id=CHAT_ID, text=alert, parse_mode='Markdown')
            await asyncio.sleep(1)  # تجنب سبام تليجرام

async def start(update, context):
    """معالج أمر /start"""
    await update.message.reply_text(
        "مرحباً! بوت توصيات شاكر 1.0 يعمل.\n"
        "سأرسل لك التنبيهات تلقائياً كل ساعة.\n"
        "لبدء مسح فوري أرسل /scan"
    )

async def scan_now(update, context):
    """معالج أمر /scan لمسح فوري"""
    await update.message.reply_text("جاري المسح، انتظر قليلاً...")
    alerts = scan_all_stocks()
    if not alerts:
        await update.message.reply_text("لا توجد إشارات جديدة.")
    else:
        for alert in alerts:
            await update.message.reply_text(alert, parse_mode='Markdown')
            await asyncio.sleep(1)

def main():
    """الدالة الرئيسية لتشغيل البوت"""
    # إنشاء تطبيق البوت
    application = Application.builder().token(TELEGRAM_TOKEN).build()

    # إضافة الأوامر
    application.add_handler(CommandHandler("start", start))
    application.add_handler(CommandHandler("scan", scan_now))

    # جدولة المسح التلقائي كل ساعة
    job_queue = application.job_queue
    job_queue.run_repeating(send_alerts, interval=3600, first=10)  # كل ساعة، أول مرة بعد 10 ثواني

    # تشغيل البوت
    print("البوت يعمل الآن...")
    application.run_polling()

if __name__ == "__main__":
    main()
````

**Reasoning**:
The subtask is to execute the code in cell `eb8e4f66` to run the bot. This step assumes the user has already updated the `TELEGRAM_TOKEN` and `CHAT_ID` as instructed in the previous steps. Executing this cell will start the bot.



````python
#!/usr/bin/env python3
# -*- coding: utf-8 -*-

# بوت توصيات الأسهم - نسخة شاكر 1.0 (نسخة كاملة)
# آخر تحديث: فبراير 2026

import yfinance as yf
import pandas as pd
import numpy as np
import ta
import logging
import time
import schedule
from datetime import datetime, timedelta
from telegram import Bot, InlineKeyboardButton, InlineKeyboardMarkup
from telegram.ext import Application, CommandHandler, CallbackQueryHandler
import asyncio
import os

# ---------------------------
# 1. إعدادات البوت والتوكن
# ---------------------------
TELEGRAM_TOKEN = "YOUR_BOT_TOKEN_HERE"      # ضع التوكن الخاص بك
CHAT_ID = "YOUR_CHAT_ID_HERE"                # ضع معرف المستخدم أو القناة

# تفعيل السجلات لمشاهدة الأخطاء
logging.basicConfig(level=logging.INFO, format='%(asctime)s - %(name)s - %(levelname)s - %(message)s')

# ---------------------------
# 2. قائمة الأسهم الكاملة (مدمجة)
# ---------------------------
TICKERS = [
    # --- أسهمك القديمة (من الصور) ---
    "A", "AA", "ABEV", "ABNB", "ABT", "ABVX", "ACGL", "ACM", "ACN",
    "ADBE", "ADI", "ADM", "ADP", "ADSK", "AEE", "AEG", "AEIS", "AEM",
    "AEP", "AER", "AES", "AFG", "AFL", "AFRM", "AG", "AGI", "AGNC",
    "AIG", "AIT", "AIZ", "AJG", "AKAM", "ALAB", "ALB", "ALC", "ALGN",
    "ALL", "ALLE", "ALLY", "ALNY", "ALSN", "AM", "AMCR", "AME", "AMH",
    "AMKR", "AMP", "AMRZ", "AMT", "AMX",

    # --- الشركات العملاقة (> 1 تريليون دولار) ---
    "NVDA", "AAPL", "GOOGL", "MSFT", "AMZN", "META", "TSM", "AVGO", "TSLA",
    "BRK.A", "WMT",

    # --- الشركات الكبرى (200 مليار - 1 تريليون) ---
    "LLY", "JPM", "XOM", "V", "JNJ", "MA", "MU", "ORCL", "COST", "AMD",
    "NFLX", "PLTR", "CSCO", "MRK", "TMO", "ASML", "ABBV", "HD", "CVX",
    "PG", "CAT", "KO", "PEP", "INTC", "CMCSA", "MCD", "TXN", "VZ", "AMGN"
]

# ---------------------------
# 3. دوال التحليل الأساسي (الفلاتر)
# ---------------------------

def get_stock_info(ticker):
    """جلب معلومات الشركة الأساسية"""
    try:
        stock = yf.Ticker(ticker)
        info = stock.info
        return info
    except Exception as e:
        logging.error(f"خطأ في جلب معلومات {ticker}: {e}")
        return None

def passes_basic_filters(ticker):
    """
    تطبيق الفلتر الأساسي:
    - Market Cap > 10B (ملاحظة: أغلب الشركات في القائمة أكبر من ذلك بكثير، لكن الفلتر يضمن ذلك)
    - Average Volume > 1M (تقريبي)
    - Spread < 5%
    """
    info = get_stock_info(ticker)
    if not info:
        return False

    # 1. القيمة السوقية
    mcap = info.get('marketCap', 0)
    if mcap < 10e9:
        logging.debug(f"{ticker} فشل فلتر Market Cap: {mcap}")
        return False

    # 2. متوسط الحجم (نأخذ آخر 20 يوم)
    try:
        stock = yf.Ticker(ticker)
        hist = stock.history(period="1mo", interval="1d")
        if hist.empty:
            return False
        avg_vol = hist['Volume'].rolling(20).mean().iloc[-1]
        if avg_vol < 1_000_000:   # أقل من مليون سهم يومياً
            logging.debug(f"{ticker} فشل فلتر الحجم: {avg_vol}")
            return False
    except:
        return False

    # 3. Spread (bid/ask)
    bid = info.get('bid', 0)
    ask = info.get('ask', 0)
    if bid and ask:
        spread = (ask - bid) / bid if bid != 0 else 1
        if spread > 0.05:   # أكبر من 5%
            logging.debug(f"{ticker} فشل فلتر Spread: {spread:.2%}")
            return False

    return True

# ---------------------------
# 4. تحليل SPX (مؤشر S&P 500)
# ---------------------------

def analyze_spx(timeframe="15m"):
    """تحليل مؤشر SPX على الإطار المطلوب"""
    try:
        spx = yf.Ticker("^GSPC")  # أو "SPY"
        hist = spx.history(period="5d", interval=timeframe)
        if len(hist) < 50:
            return "neutral"

        # متوسطات بسيطة لتحديد الاتجاه
        hist['SMA20'] = hist['Close'].rolling(20).mean()
        last = hist.iloc[-1]
        prev = hist.iloc[-2]

        if last['Close'] > last['SMA20'] and last['Close'] > prev['Close']:
            return "uptrend"
        elif last['Close'] < last['SMA20'] and last['Close'] < prev['Close']:
            return "downtrend"
        else:
            return "neutral"
    except:
        return "neutral"

# ---------------------------
# 5. تحليل السهم (فريم ساعتين)
# ---------------------------

def analyze_stock_2h(ticker):
    """
    تحليل فني كامل على فريم الساعتين
    يعيد: {'signal': 'CALL'/'PUT'/None, 'confidence': 0-100, 'details': {...}}
    """
    try:
        stock = yf.Ticker(ticker)
        hist = stock.history(period="1mo", interval="2h")  # آخر شهر، فريم ساعتين
        if len(hist) < 200:
            return None

        # حساب المتوسطات
        hist['SMA50'] = hist['Close'].rolling(50).mean()
        hist['SMA100'] = hist['Close'].rolling(100).mean()
        hist['SMA200'] = hist['Close'].rolling(200).mean()
        hist['Volume_SMA20'] = hist['Volume'].rolling(20).mean()

        last = hist.iloc[-1]
        current_price = last['Close']
        sma50 = last['SMA50']
        sma100 = last['SMA100']
        sma200 = last['SMA200']

        # قمم وقيعان آخر 3 شموع
        high_last_3 = hist['High'].iloc[-3:].max()
        low_last_3 = hist['Low'].iloc[-3:].min()

        # قمة/قاع اليوم (آخر 24 ساعة = 12 شمعة)
        high_today = hist['High'].iloc[-12:].max()
        low_today = hist['Low'].iloc[-12:].min()

        volume_ok = last['Volume'] > last['Volume_SMA20']

        # شروط CALL
        call_condition1 = (sma50 > sma100) or (sma50 > sma200)
        call_condition2 = current_price > sma200
        call_condition3 = current_price > high_last_3 or current_price > high_today
        call_condition4 = volume_ok

        # شروط PUT
        put_condition1 = (sma50 < sma100) or (sma50 < sma200)
        put_condition2 = current_price < sma200
        put_condition3 = current_price < low_last_3 or current_price < low_today
        put_condition4 = volume_ok

        signal = None
        confidence = 0
        details = {}

        if call_condition1 and call_condition2 and call_condition3 and call_condition4:
            signal = "CALL"
            # حساب الثقة: نضيف 25 لكل شرط متحقق
            confidence = 25 * sum([call_condition1, call_condition2, call_condition3, call_condition4])
            details = {
                "sma50": sma50,
                "sma100": sma100,
                "sma200": sma200,
                "price": current_price,
                "high_3": high_last_3,
                "volume_ratio": last['Volume'] / last['Volume_SMA20']
            }
        elif put_condition1 and put_condition2 and put_condition3 and put_condition4:
            signal = "PUT"
            confidence = 25 * sum([put_condition1, put_condition2, put_condition3, put_condition4])
            details = {
                "sma50": sma50,
                "sma100": sma100,
                "sma200": sma200,
                "price": current_price,
                "low_3": low_last_3,
                "volume_ratio": last['Volume'] / last['Volume_SMA20']
            }

        return {"signal": signal, "confidence": int(confidence), "details": details}

    except Exception as e:
        logging.error(f"خطأ في تحليل {ticker}: {e}")
        return None

# ---------------------------
# 6. تصنيف الصفقة (سكالب، قصيرة، سوينق)
# ---------------------------

def classify_trade(signal, confidence, details):
    """
    تصنيف حسب قوة الإشارة وترتيب المتوسطات
    """
    if not signal:
        return None

    sma50 = details.get('sma50')
    sma100 = details.get('sma100')
    sma200 = details.get('sma200')

    # ترتيب المتوسطات
    if signal == "CALL":
        if sma50 > sma100 > sma200:
            return "📈 سوينق 3-7 أيام"
        elif confidence >= 75:
            return "⏳ صفقة قصيرة 1-2 يوم"
        else:
            return "⚡ سكالب"
    else:  # PUT
        if sma50 < sma100 < sma200:
            return "📈 سوينق 3-7 أيام"
        elif confidence >= 75:
            return "⏳ صفقة قصيرة 1-2 يوم"
        else:
            return "⚡ سكالب"

# ---------------------------
# 7. اختيار العقد المناسب (بيانات الخيارات)
# ---------------------------

def get_option_contract(ticker, signal, classification):
    """
    البحث عن عقد خيار مناسب حسب التصنيف.
    """
    try:
        stock = yf.Ticker(ticker)
        # جلب تواريخ انتهاء الصلاحية المتاحة
        expirations = stock.options
        if not expirations:
            return "غير متوفر"

        # اختيار التاريخ المناسب حسب التصنيف
        today = datetime.now().date()
        if "سكالب" in classification:
            days = 7
        elif "قصيرة" in classification:
            days = 14
        else:  # سوينق
            days = 30

        target_date = today + timedelta(days=days)
        # البحث عن أقرب تاريخ انتهاء للعدد المطلوب من الأيام
        exp_date = None
        for exp in expirations:
            exp_dt = datetime.strptime(exp, '%Y-%m-%d').date()
            if exp_dt >= target_date:
                exp_date = exp
                break
        if not exp_date:
            exp_date = expirations[0]  # أقرب تاريخ

        # جلب عقود الخيارات لذلك التاريخ
        opt_chain = stock.option_chain(exp_date)
        if signal == "CALL":
            contracts = opt_chain.calls
        else:
            contracts = opt_chain.puts

        if contracts.empty:
            return "غير متوفر"

        # فلترة العقود: Open Interest عالي (أعلى 10%)، Spread ضيق
        current_price = stock.history(period="1d")['Close'].iloc[-1]
        contracts['strike_diff'] = abs(contracts['strike'] - current_price)
        # فلتر OI
        oi_threshold = contracts['openInterest'].quantile(0.7)  # أعلى 30%
        eligible = contracts[contracts['openInterest'] > oi_threshold]

        if eligible.empty:
            eligible = contracts

        # نختار العقد الأقرب للسعر الحالي (للدلتا التقريبية)
        best = eligible.nsmallest(1, 'strike_diff').iloc[0]

        # حساب Spread تقريبي (ask-bid)/ask
        spread = (best['ask'] - best['bid']) / best['ask'] if best['ask'] > 0 else 1
        if spread > 0.5:  # spread كبير جداً
            return "سبريد واسع"

        # بناء نص العقد المقترح
        delta_approx = 0.5 - (best['strike_diff'] / current_price)  # تقريب بسيط
        delta_approx = max(0.1, min(0.9, delta_approx))

        return f"Delta {delta_approx:.2f} – Exp {exp_date} (أيام: {days})"

    except Exception as e:
        logging.error(f"خطأ في جلب الخيارات لـ {ticker}: {e}")
        return "غير متوفر"

# ---------------------------
# 8. بناء رسالة التنبيه النهائية
# ---------------------------

def build_alert_message(ticker, signal, confidence, classification, details, spx_trend, option_text):
    """
    تنسيق الرسالة كما طلبت تماماً.
    """
    # ترجمة الاتجاه
    if signal == "CALL":
        direction = "Call"
        cross = "50 > 100" if details.get('sma50', 0) > details.get('sma100', 0) else "50 > 200"
        price_status = f"السعر فوق {details.get('sma200', 0):.2f}"
        event = "اختراق قمة 2H"
    else:
        direction = "Put"
        cross = "50 < 100" if details.get('sma50', 0) < details.get('sma100', 0) else "50 < 200"
        price_status = f"السعر تحت {details.get('sma200', 0):.2f}"
        event = "كسر قاع 2H"

    # توافق SPX
    if "uptrend" in spx_trend and signal == "CALL":
        spx_compat = "✅ متوافق"
    elif "downtrend" in spx_trend and signal == "PUT":
        spx_compat = "✅ متوافق"
    else:
        spx_compat = "⚠️ غير متوافق"

    message = f"""
🔔 {ticker}
الاتجاه: {direction}
التصنيف: {classification}
التقاطع: {cross} (إيجابي)
الحالة: {price_status}
الحدث: {event}
SPX: {spx_compat}
الثقة: {confidence}/100
العقد المقترح: {option_text}
"""
    return message

# ---------------------------
# 9. دالة المسح الرئيسية
# ---------------------------

def scan_all_stocks():
    """تمر على كل الأسهم وتفحصها وترسل التنبيهات"""
    alerts = []
    for ticker in TICKERS:
        logging.info(f"فحص {ticker}...")
        time.sleep(0.5)  # تأخير نصف ثانية لتجنب حظر Yahoo

        # 1. الفلتر الأساسي
        if not passes_basic_filters(ticker):
            continue

        # 2. تحليل السهم
        result = analyze_stock_2h(ticker)
        if not result or not result['signal']:
            continue

        signal = result['signal']
        confidence = result['confidence']
        details = result['details']

        # 3. تحليل SPX
        spx_15m = analyze_spx("15m")

        # 4. تصنيف الصفقة
        classification = classify_trade(signal, confidence, details)

        # 5. اختيار العقد
        option = get_option_contract(ticker, signal, classification)

        # 6. بناء الرسالة
        msg = build_alert_message(ticker, signal, confidence, classification, details, spx_15m, option)
        alerts.append(msg)

    return alerts

# ---------------------------
# 10. دوال بوت تليجرام
# ---------------------------

async def send_alerts(context):
    """إرسال التنبيهات إلى تليجرام"""
    alerts = scan_all_stocks()
    if not alerts:
        await context.bot.send_message(chat_id=CHAT_ID, text="لا توجد إشارات جديدة حالياً.")
    else:
        for alert in alerts:
            await context.bot.send_message(chat_id=CHAT_ID, text=alert, parse_mode='Markdown')
            await asyncio.sleep(1)  # تجنب سبام تليجرام

async def start(update, context):
    """معالج أمر /start"""
    await update.message.reply_text(
        "مرحباً! بوت توصيات شاكر 1.0 يعمل.\n"
        "سأرسل لك التنبيهات تلقائياً كل ساعة.\n"
        "لبدء مسح فوري أرسل /scan"
    )

async def scan_now(update, context):
    """معالج أمر /scan لمسح فوري"""
    await update.message.reply_text("جاري المسح، انتظر قليلاً...")
    alerts = scan_all_stocks()
    if not alerts:
        await update.message.reply_text("لا توجد إشارات جديدة.")
    else:
        for alert in alerts:
            await update.message.reply_text(alert, parse_mode='Markdown')
            await asyncio.sleep(1)

def main():
    """الدالة الرئيسية لتشغيل البوت"""
    # إنشاء تطبيق البوت
    application = Application.builder().token(TELEGRAM_TOKEN).build()

    # إضافة الأوامر
    application.add_handler(CommandHandler("start", start))
    application.add_handler(CommandHandler("scan", scan_now))

    # جدولة المسح التلقائي كل ساعة
    job_queue = application.job_queue
    job_queue.run_repeating(send_alerts, interval=3600, first=10)  # كل ساعة، أول مرة بعد 10 ثواني

    # تشغيل البوت
    print("البوت يعمل الآن...")
    application.run_polling()

if __name__ == "__main__":
    main()
````

````python
#!/usr/bin/env python3
# -*- coding: utf-8 -*-

# بوت توصيات الأسهم - نسخة شاكر 1.0 (نسخة كاملة)
# آخر تحديث: فبراير 2026

import yfinance as yf
import pandas as pd
import numpy as np
import ta
import logging
import time
import schedule
from datetime import datetime, timedelta
from telegram import Bot, InlineKeyboardButton, InlineKeyboardMarkup
from telegram.ext import Application, CommandHandler, CallbackQueryHandler
import asyncio
import os

# ---------------------------
# 1. إعدادات البوت والتوكن
# ---------------------------
TELEGRAM_TOKEN = "8699821370:AAEQUSbLTgf7MmWqo5vV5LHPOz30wfqOfqw"      # ضع التوكن الخاص بك
CHAT_ID = "8699821370:AAEQUSbLTgf7MmWqo5vV5LHPOz30wfqOfqw"                # ضع معرف المستخدم أو القناة

# تفعيل السجلات لمشاهدة الأخطاء
logging.basicConfig(level=logging.INFO, format='%(asctime)s - %(name)s - %(levelname)s - %(message)s')

# ---------------------------
# 2. قائمة الأسهم الكاملة (مدمجة)
# ---------------------------
TICKERS = [
    # --- أسهمك القديمة (من الصور) ---
    "A", "AA", "ABEV", "ABNB", "ABT", "ABVX", "ACGL", "ACM", "ACN",
    "ADBE", "ADI", "ADM", "ADP", "ADSK", "AEE", "AEG", "AEIS", "AEM",
    "AEP", "AER", "AES", "AFG", "AFL", "AFRM", "AG", "AGI", "AGNC",
    "AIG", "AIT", "AIZ", "AJG", "AKAM", "ALAB", "ALB", "ALC", "ALGN",
    "ALL", "ALLE", "ALLY", "ALNY", "ALSN", "AM", "AMCR", "AME", "AMH",
    "AMKR", "AMP", "AMRZ", "AMT", "AMX",

    # --- الشركات العملاقة (> 1 تريليون دولار) ---
    "NVDA", "AAPL", "GOOGL", "MSFT", "AMZN", "META", "TSM", "AVGO", "TSLA",
    "BRK.A", "WMT",

    # --- الشركات الكبرى (200 مليار - 1 تريليون) ---
    "LLY", "JPM", "XOM", "V", "JNJ", "MA", "MU", "ORCL", "COST", "AMD",
    "NFLX", "PLTR", "CSCO", "MRK", "TMO", "ASML", "ABBV", "HD", "CVX",
    "PG", "CAT", "KO", "PEP", "INTC", "CMCSA", "MCD", "TXN", "VZ", "AMGN"
]

# ---------------------------
# 3. دوال التحليل الأساسي (الفلاتر)
# ---------------------------

def get_stock_info(ticker):
    """جلب معلومات الشركة الأساسية"""
    try:
        stock = yf.Ticker(ticker)
        info = stock.info
        return info
    except Exception as e:
        logging.error(f"خطأ في جلب معلومات {ticker}: {e}")
        return None

def passes_basic_filters(ticker):
    """
    تطبيق الفلتر الأساسي:
    - Market Cap > 10B (ملاحظة: أغلب الشركات في القائمة أكبر من ذلك بكثير، لكن الفلتر يضمن ذلك)
    - Average Volume > 1M (تقريبي)
    - Spread < 5%
    """
    info = get_stock_info(ticker)
    if not info:
        return False

    # 1. القيمة السوقية
    mcap = info.get('marketCap', 0)
    if mcap < 10e9:
        logging.debug(f"{ticker} فشل فلتر Market Cap: {mcap}")
        return False

    # 2. متوسط الحجم (نأخذ آخر 20 يوم)
    try:
        stock = yf.Ticker(ticker)
        hist = stock.history(period="1mo", interval="1d")
        if hist.empty:
            return False
        avg_vol = hist['Volume'].rolling(20).mean().iloc[-1]
        if avg_vol < 1_000_000:   # أقل من مليون سهم يومياً
            logging.debug(f"{ticker} فشل فلتر الحجم: {avg_vol}")
            return False
    except:
        return False

    # 3. Spread (bid/ask)
    bid = info.get('bid', 0)
    ask = info.get('ask', 0)
    if bid and ask:
        spread = (ask - bid) / bid if bid != 0 else 1
        if spread > 0.05:   # أكبر من 5%
            logging.debug(f"{ticker} فشل فلتر Spread: {spread:.2%}")
            return False

    return True

# ---------------------------
# 4. تحليل SPX (مؤشر S&P 500)
# ---------------------------

def analyze_spx(timeframe="15m"):
    """تحليل مؤشر SPX على الإطار المطلوب"""
    try:
        spx = yf.Ticker("^GSPC")  # أو "SPY"
        hist = spx.history(period="5d", interval=timeframe)
        if len(hist) < 50:
            return "neutral"

        # متوسطات بسيطة لتحديد الاتجاه
        hist['SMA20'] = hist['Close'].rolling(20).mean()
        last = hist.iloc[-1]
        prev = hist.iloc[-2]

        if last['Close'] > last['SMA20'] and last['Close'] > prev['Close']:
            return "uptrend"
        elif last['Close'] < last['SMA20'] and last['Close'] < prev['Close']:
            return "downtrend"
        else:
            return "neutral"
    except:
        return "neutral"

# ---------------------------
# 5. تحليل السهم (فريم ساعتين)
# ---------------------------

def analyze_stock_2h(ticker):
    """
    تحليل فني كامل على فريم الساعتين
    يعيد: {'signal': 'CALL'/'PUT'/None, 'confidence': 0-100, 'details': {...}}
    """
    try:
        stock = yf.Ticker(ticker)
        hist = stock.history(period="1mo", interval="2h")  # آخر شهر، فريم ساعتين
        if len(hist) < 200:
            return None

        # حساب المتوسطات
        hist['SMA50'] = hist['Close'].rolling(50).mean()
        hist['SMA100'] = hist['Close'].rolling(100).mean()
        hist['SMA200'] = hist['Close'].rolling(200).mean()
        hist['Volume_SMA20'] = hist['Volume'].rolling(20).mean()

        last = hist.iloc[-1]
        current_price = last['Close']
        sma50 = last['SMA50']
        sma100 = last['SMA100']
        sma200 = last['SMA200']

        # قمم وقيعان آخر 3 شموع
        high_last_3 = hist['High'].iloc[-3:].max()
        low_last_3 = hist['Low'].iloc[-3:].min()

        # قمة/قاع اليوم (آخر 24 ساعة = 12 شمعة)
        high_today = hist['High'].iloc[-12:].max()
        low_today = hist['Low'].iloc[-12:].min()

        volume_ok = last['Volume'] > last['Volume_SMA20']

        # شروط CALL
        call_condition1 = (sma50 > sma100) or (sma50 > sma200)
        call_condition2 = current_price > sma200
        call_condition3 = current_price > high_last_3 or current_price > high_today
        call_condition4 = volume_ok

        # شروط PUT
        put_condition1 = (sma50 < sma100) or (sma50 < sma200)
        put_condition2 = current_price < sma200
        put_condition3 = current_price < low_last_3 or current_price < low_today
        put_condition4 = volume_ok

        signal = None
        confidence = 0
        details = {}

        if call_condition1 and call_condition2 and call_condition3 and call_condition4:
            signal = "CALL"
            # حساب الثقة: نضيف 25 لكل شرط متحقق
            confidence = 25 * sum([call_condition1, call_condition2, call_condition3, call_condition4])
            details = {
                "sma50": sma50,
                "sma100": sma100,
                "sma200": sma200,
                "price": current_price,
                "high_3": high_last_3,
                "volume_ratio": last['Volume'] / last['Volume_SMA20']
            }
        elif put_condition1 and put_condition2 and put_condition3 and put_condition4:
            signal = "PUT"
            confidence = 25 * sum([put_condition1, put_condition2, put_condition3, put_condition4])
            details = {
                "sma50": sma50,
                "sma100": sma100,
                "sma200": sma200,
                "price": current_price,
                "low_3": low_last_3,
                "volume_ratio": last['Volume'] / last['Volume_SMA20']
            }

        return {"signal": signal, "confidence": int(confidence), "details": details}

    except Exception as e:
        logging.error(f"خطأ في تحليل {ticker}: {e}")
        return None

# ---------------------------
# 6. تصنيف الصفقة (سكالب، قصيرة، سوينق)
# ---------------------------

def classify_trade(signal, confidence, details):
    """
    تصنيف حسب قوة الإشارة وترتيب المتوسطات
    """
    if not signal:
        return None

    sma50 = details.get('sma50')
    sma100 = details.get('sma100')
    sma200 = details.get('sma200')

    # ترتيب المتوسطات
    if signal == "CALL":
        if sma50 > sma100 > sma200:
            return "📈 سوينق 3-7 أيام"
        elif confidence >= 75:
            return "⏳ صفقة قصيرة 1-2 يوم"
        else:
            return "⚡ سكالب"
    else:  # PUT
        if sma50 < sma100 < sma200:
            return "📈 سوينق 3-7 أيام"
        elif confidence >= 75:
            return "⏳ صفقة قصيرة 1-2 يوم"
        else:
            return "⚡ سكالب"

# ---------------------------
# 7. اختيار العقد المناسب (بيانات الخيارات)
# ---------------------------

def get_option_contract(ticker, signal, classification):
    """
    البحث عن عقد خيار مناسب حسب التصنيف.
    """
    try:
        stock = yf.Ticker(ticker)
        # جلب تواريخ انتهاء الصلاحية المتاحة
        expirations = stock.options
        if not expirations:
            return "غير متوفر"

        # اختيار التاريخ المناسب حسب التصنيف
        today = datetime.now().date()
        if "سكالب" in classification:
            days = 7
        elif "قصيرة" in classification:
            days = 14
        else:  # سوينق
            days = 30

        target_date = today + timedelta(days=days)
        # البحث عن أقرب تاريخ انتهاء للعدد المطلوب من الأيام
        exp_date = None
        for exp in expirations:
            exp_dt = datetime.strptime(exp, '%Y-%m-%d').date()
            if exp_dt >= target_date:
                exp_date = exp
                break
        if not exp_date:
            exp_date = expirations[0]  # أقرب تاريخ

        # جلب عقود الخيارات لذلك التاريخ
        opt_chain = stock.option_chain(exp_date)
        if signal == "CALL":
            contracts = opt_chain.calls
        else:
            contracts = opt_chain.puts

        if contracts.empty:
            return "غير متوفر"

        # فلترة العقود: Open Interest عالي (أعلى 10%)، Spread ضيق
        current_price = stock.history(period="1d")['Close'].iloc[-1]
        contracts['strike_diff'] = abs(contracts['strike'] - current_price)
        # فلتر OI
        oi_threshold = contracts['openInterest'].quantile(0.7)  # أعلى 30%
        eligible = contracts[contracts['openInterest'] > oi_threshold]

        if eligible.empty:
            eligible = contracts

        # نختار العقد الأقرب للسعر الحالي (للدلتا التقريبية)
        best = eligible.nsmallest(1, 'strike_diff').iloc[0]

        # حساب Spread تقريبي (ask-bid)/ask
        spread = (best['ask'] - best['bid']) / best['ask'] if best['ask'] > 0 else 1
        if spread > 0.5:  # spread كبير جداً
            return "سبريد واسع"

        # بناء نص العقد المقترح
        delta_approx = 0.5 - (best['strike_diff'] / current_price)  # تقريب بسيط
        delta_approx = max(0.1, min(0.9, delta_approx))

        return f"Delta {delta_approx:.2f} – Exp {exp_date} (أيام: {days})"

    except Exception as e:
        logging.error(f"خطأ في جلب الخيارات لـ {ticker}: {e}")
        return "غير متوفر"

# ---------------------------
# 8. بناء رسالة التنبيه النهائية
# ---------------------------

def build_alert_message(ticker, signal, confidence, classification, details, spx_trend, option_text):
    """
    تنسيق الرسالة كما طلبت تماماً.
    """
    # ترجمة الاتجاه
    if signal == "CALL":
        direction = "Call"
        cross = "50 > 100" if details.get('sma50', 0) > details.get('sma100', 0) else "50 > 200"
        price_status = f"السعر فوق {details.get('sma200', 0):.2f}"
        event = "اختراق قمة 2H"
    else:
        direction = "Put"
        cross = "50 < 100" if details.get('sma50', 0) < details.get('sma100', 0) else "50 < 200"
        price_status = f"السعر تحت {details.get('sma200', 0):.2f}"
        event = "كسر قاع 2H"

    # توافق SPX
    if "uptrend" in spx_trend and signal == "CALL":
        spx_compat = "✅ متوافق"
    elif "downtrend" in spx_trend and signal == "PUT":
        spx_compat = "✅ متوافق"
    else:
        spx_compat = "⚠️ غير متوافق"

    message = f"""
🔔 {ticker}
الاتجاه: {direction}
التصنيف: {classification}
التقاطع: {cross} (إيجابي)
الحالة: {price_status}
الحدث: {event}
SPX: {spx_compat}
الثقة: {confidence}/100
العقد المقترح: {option_text}
"""
    return message

# ---------------------------
# 9. دالة المسح الرئيسية
# ---------------------------

def scan_all_stocks():
    """تمر على كل الأسهم وتفحصها وترسل التنبيهات"""
    alerts = []
    for ticker in TICKERS:
        logging.info(f"فحص {ticker}...")
        time.sleep(0.5)  # تأخير نصف ثانية لتجنب حظر Yahoo

        # 1. الفلتر الأساسي
        if not passes_basic_filters(ticker):
            continue

        # 2. تحليل السهم
        result = analyze_stock_2h(ticker)
        if not result or not result['signal']:
            continue

        signal = result['signal']
        confidence = result['confidence']
        details = result['details']

        # 3. تحليل SPX
        spx_15m = analyze_spx("15m")

        # 4. تصنيف الصفقة
        classification = classify_trade(signal, confidence, details)

        # 5. اختيار العقد
        option = get_option_contract(ticker, signal, classification)

        # 6. بناء الرسالة
        msg = build_alert_message(ticker, signal, confidence, classification, details, spx_15m, option)
        alerts.append(msg)

    return alerts

# ---------------------------
# 10. دوال بوت تليجرام
# ---------------------------

async def send_alerts(context):
    """إرسال التنبيهات إلى تليجرام"""
    alerts = scan_all_stocks()
    if not alerts:
        await context.bot.send_message(chat_id=CHAT_ID, text="لا توجد إشارات جديدة حالياً.")
    else:
        for alert in alerts:
            await context.bot.send_message(chat_id=CHAT_ID, text=alert, parse_mode='Markdown')
            await asyncio.sleep(1)  # تجنب سبام تليجرام

async def start(update, context):
    """معالج أمر /start"""
    await update.message.reply_text(
        "مرحباً! بوت توصيات شاكر 1.0 يعمل.\n"
        "سأرسل لك التنبيهات تلقائياً كل ساعة.\n"
        "لبدء مسح فوري أرسل /scan"
    )

async def scan_now(update, context):
    """معالج أمر /scan لمسح فوري"""
    await update.message.reply_text("جاري المسح، انتظر قليلاً...")
    alerts = scan_all_stocks()
    if not alerts:
        await update.message.reply_text("لا توجد إشارات جديدة.")
    else:
        for alert in alerts:
            await update.message.reply_text(alert, parse_mode='Markdown')
            await asyncio.sleep(1)

def main():
    """الدالة الرئيسية لتشغيل البوت"""
    # إنشاء تطبيق البوت
    application = Application.builder().token(TELEGRAM_TOKEN).build()

    # إضافة الأوامر
    application.add_handler(CommandHandler("start", start))
    application.add_handler(CommandHandler("scan", scan_now))

    # جدولة المسح التلقائي كل ساعة
    job_queue = application.job_queue
    job_queue.run_repeating(send_alerts, interval=3600, first=10)  # كل ساعة، أول مرة بعد 10 ثواني

    # تشغيل البوت
    print("البوت يعمل الآن...")
    application.run_polling()

if __name__ == "__main__":
    main()
````

````python
#!/usr/bin/env python3
# -*- coding: utf-8 -*-

# بوت توصيات الأسهم - نسخة شاكر 1.0 (نسخة كاملة)
# آخر تحديث: فبراير 2026

import yfinance as yf
import pandas as pd
import numpy as np
import ta
import logging
import time
import schedule
from datetime import datetime, timedelta
from telegram import Bot, InlineKeyboardButton, InlineKeyboardMarkup
from telegram.ext import Application, CommandHandler, CallbackQueryHandler
import asyncio
import os

# ---------------------------
# 1. إعدادات البوت والتوكن
# ---------------------------
TELEGRAM_TOKEN = "YOUR_BOT_TOKEN_HERE"      # ضع التوكن الخاص بك
CHAT_ID = "YOUR_CHAT_ID_HERE"                # ضع معرف المستخدم أو القناة

# تفعيل السجلات لمشاهدة الأخطاء
logging.basicConfig(level=logging.INFO, format='%(asctime)s - %(name)s - %(levelname)s - %(message)s')

# ---------------------------
# 2. قائمة الأسهم الكاملة (مدمجة)
# ---------------------------
TICKERS = [
    # --- أسهمك القديمة (من الصور) ---
    "A", "AA", "ABEV", "ABNB", "ABT", "ABVX", "ACGL", "ACM", "ACN",
    "ADBE", "ADI", "ADM", "ADP", "ADSK", "AEE", "AEG", "AEIS", "AEM",
    "AEP", "AER", "AES", "AFG", "AFL", "AFRM", "AG", "AGI", "AGNC",
    "AIG", "AIT", "AIZ", "AJG", "AKAM", "ALAB", "ALB", "ALC", "ALGN",
    "ALL", "ALLE", "ALLY", "ALNY", "ALSN", "AM", "AMCR", "AME", "AMH",
    "AMKR", "AMP", "AMRZ", "AMT", "AMX",

    # --- الشركات العملاقة (> 1 تريليون دولار) ---
    "NVDA", "AAPL", "GOOGL", "MSFT", "AMZN", "META", "TSM", "AVGO", "TSLA",
    "BRK.A", "WMT",

    # --- الشركات الكبرى (200 مليار - 1 تريليون) ---
    "LLY", "JPM", "XOM", "V", "JNJ", "MA", "MU", "ORCL", "COST", "AMD",
    "NFLX", "PLTR", "CSCO", "MRK", "TMO", "ASML", "ABBV", "HD", "CVX",
    "PG", "CAT", "KO", "PEP", "INTC", "CMCSA", "MCD", "TXN", "VZ", "AMGN"
]

# ---------------------------
# 3. دوال التحليل الأساسي (الفلاتر)
# ---------------------------

def get_stock_info(ticker):
    """جلب معلومات الشركة الأساسية"""
    try:
        stock = yf.Ticker(ticker)
        info = stock.info
        return info
    except Exception as e:
        logging.error(f"خطأ في جلب معلومات {ticker}: {e}")
        return None

def passes_basic_filters(ticker):
    """
    تطبيق الفلتر الأساسي:
    - Market Cap > 10B (ملاحظة: أغلب الشركات في القائمة أكبر من ذلك بكثير، لكن الفلتر يضمن ذلك)
    - Average Volume > 1M (تقريبي)
    - Spread < 5%
    """
    info = get_stock_info(ticker)
    if not info:
        return False

    # 1. القيمة السوقية
    mcap = info.get('marketCap', 0)
    if mcap < 10e9:
        logging.debug(f"{ticker} فشل فلتر Market Cap: {mcap}")
        return False

    # 2. متوسط الحجم (نأخذ آخر 20 يوم)
    try:
        stock = yf.Ticker(ticker)
        hist = stock.history(period="1mo", interval="1d")
        if hist.empty:
            return False
        avg_vol = hist['Volume'].rolling(20).mean().iloc[-1]
        if avg_vol < 1_000_000:   # أقل من مليون سهم يومياً
            logging.debug(f"{ticker} فشل فلتر الحجم: {avg_vol}")
            return False
    except:
        return False

    # 3. Spread (bid/ask)
    bid = info.get('bid', 0)
    ask = info.get('ask', 0)
    if bid and ask:
        spread = (ask - bid) / bid if bid != 0 else 1
        if spread > 0.05:   # أكبر من 5%
            logging.debug(f"{ticker} فشل فلتر Spread: {spread:.2%}")
            return False

    return True

# ---------------------------
# 4. تحليل SPX (مؤشر S&P 500)
# ---------------------------

def analyze_spx(timeframe="15m"):
    """تحليل مؤشر SPX على الإطار المطلوب"""
    try:
        spx = yf.Ticker("^GSPC")  # أو "SPY"
        hist = spx.history(period="5d", interval=timeframe)
        if len(hist) < 50:
            return "neutral"

        # متوسطات بسيطة لتحديد الاتجاه
        hist['SMA20'] = hist['Close'].rolling(20).mean()
        last = hist.iloc[-1]
        prev = hist.iloc[-2]

        if last['Close'] > last['SMA20'] and last['Close'] > prev['Close']:
            return "uptrend"
        elif last['Close'] < last['SMA20'] and last['Close'] < prev['Close']:
            return "downtrend"
        else:
            return "neutral"
    except:
        return "neutral"

# ---------------------------
# 5. تحليل السهم (فريم ساعتين)
# ---------------------------

def analyze_stock_2h(ticker):
    """
    تحليل فني كامل على فريم الساعتين
    يعيد: {'signal': 'CALL'/'PUT'/None, 'confidence': 0-100, 'details': {...}}
    """
    try:
        stock = yf.Ticker(ticker)
        hist = stock.history(period="1mo", interval="2h")  # آخر شهر، فريم ساعتين
        if len(hist) < 200:
            return None

        # حساب المتوسطات
        hist['SMA50'] = hist['Close'].rolling(50).mean()
        hist['SMA100'] = hist['Close'].rolling(100).mean()
        hist['SMA200'] = hist['Close'].rolling(200).mean()
        hist['Volume_SMA20'] = hist['Volume'].rolling(20).mean()

        last = hist.iloc[-1]
        current_price = last['Close']
        sma50 = last['SMA50']
        sma100 = last['SMA100']
        sma200 = last['SMA200']

        # قمم وقيعان آخر 3 شموع
        high_last_3 = hist['High'].iloc[-3:].max()
        low_last_3 = hist['Low'].iloc[-3:].min()

        # قمة/قاع اليوم (آخر 24 ساعة = 12 شمعة)
        high_today = hist['High'].iloc[-12:].max()
        low_today = hist['Low'].iloc[-12:].min()

        volume_ok = last['Volume'] > last['Volume_SMA20']

        # شروط CALL
        call_condition1 = (sma50 > sma100) or (sma50 > sma200)
        call_condition2 = current_price > sma200
        call_condition3 = current_price > high_last_3 or current_price > high_today
        call_condition4 = volume_ok

        # شروط PUT
        put_condition1 = (sma50 < sma100) or (sma50 < sma200)
        put_condition2 = current_price < sma200
        put_condition3 = current_price < low_last_3 or current_price < low_today
        put_condition4 = volume_ok

        signal = None
        confidence = 0
        details = {}

        if call_condition1 and call_condition2 and call_condition3 and call_condition4:
            signal = "CALL"
            # حساب الثقة: نضيف 25 لكل شرط متحقق
            confidence = 25 * sum([call_condition1, call_condition2, call_condition3, call_condition4])
            details = {
                "sma50": sma50,
                "sma100": sma100,
                "sma200": sma200,
                "price": current_price,
                "high_3": high_last_3,
                "volume_ratio": last['Volume'] / last['Volume_SMA20']
            }
        elif put_condition1 and put_condition2 and put_condition3 and put_condition4:
            signal = "PUT"
            confidence = 25 * sum([put_condition1, put_condition2, put_condition3, put_condition4])
            details = {
                "sma50": sma50,
                "sma100": sma100,
                "sma200": sma200,
                "price": current_price,
                "low_3": low_last_3,
                "volume_ratio": last['Volume'] / last['Volume_SMA20']
            }

        return {"signal": signal, "confidence": int(confidence), "details": details}

    except Exception as e:
        logging.error(f"خطأ في تحليل {ticker}: {e}")
        return None

# ---------------------------
# 6. تصنيف الصفقة (سكالب، قصيرة، سوينق)
# ---------------------------

def classify_trade(signal, confidence, details):
    """
    تصنيف حسب قوة الإشارة وترتيب المتوسطات
    """
    if not signal:
        return None

    sma50 = details.get('sma50')
    sma100 = details.get('sma100')
    sma200 = details.get('sma200')

    # ترتيب المتوسطات
    if signal == "CALL":
        if sma50 > sma100 > sma200:
            return "📈 سوينق 3-7 أيام"
        elif confidence >= 75:
            return "⏳ صفقة قصيرة 1-2 يوم"
        else:
            return "⚡ سكالب"
    else:  # PUT
        if sma50 < sma100 < sma200:
            return "📈 سوينق 3-7 أيام"
        elif confidence >= 75:
            return "⏳ صفقة قصيرة 1-2 يوم"
        else:
            return "⚡ سكالب"

# ---------------------------
# 7. اختيار العقد المناسب (بيانات الخيارات)
# ---------------------------

def get_option_contract(ticker, signal, classification):
    """
    البحث عن عقد خيار مناسب حسب التصنيف.
    """
    try:
        stock = yf.Ticker(ticker)
        # جلب تواريخ انتهاء الصلاحية المتاحة
        expirations = stock.options
        if not expirations:
            return "غير متوفر"

        # اختيار التاريخ المناسب حسب التصنيف
        today = datetime.now().date()
        if "سكالب" in classification:
            days = 7
        elif "قصيرة" in classification:
            days = 14
        else:  # سوينق
            days = 30

        target_date = today + timedelta(days=days)
        # البحث عن أقرب تاريخ انتهاء للعدد المطلوب من الأيام
        exp_date = None
        for exp in expirations:
            exp_dt = datetime.strptime(exp, '%Y-%m-%d').date()
            if exp_dt >= target_date:
                exp_date = exp
                break
        if not exp_date:
            exp_date = expirations[0]  # أقرب تاريخ

        # جلب عقود الخيارات لذلك التاريخ
        opt_chain = stock.option_chain(exp_date)
        if signal == "CALL":
            contracts = opt_chain.calls
        else:
            contracts = opt_chain.puts

        if contracts.empty:
            return "غير متوفر"

        # فلترة العقود: Open Interest عالي (أعلى 10%)، Spread ضيق
        current_price = stock.history(period="1d")['Close'].iloc[-1]
        contracts['strike_diff'] = abs(contracts['strike'] - current_price)
        # فلتر OI
        oi_threshold = contracts['openInterest'].quantile(0.7)  # أعلى 30%
        eligible = contracts[contracts['openInterest'] > oi_threshold]

        if eligible.empty:
            eligible = contracts

        # نختار العقد الأقرب للسعر الحالي (للدلتا التقريبية)
        best = eligible.nsmallest(1, 'strike_diff').iloc[0]

        # حساب Spread تقريبي (ask-bid)/ask
        spread = (best['ask'] - best['bid']) / best['ask'] if best['ask'] > 0 else 1
        if spread > 0.5:  # spread كبير جداً
            return "سبريد واسع"

        # بناء نص العقد المقترح
        delta_approx = 0.5 - (best['strike_diff'] / current_price)  # تقريب بسيط
        delta_approx = max(0.1, min(0.9, delta_approx))

        return f"Delta {delta_approx:.2f} – Exp {exp_date} (أيام: {days})"

    except Exception as e:
        logging.error(f"خطأ في جلب الخيارات لـ {ticker}: {e}")
        return "غير متوفر"

# ---------------------------
# 8. بناء رسالة التنبيه النهائية
# ---------------------------

def build_alert_message(ticker, signal, confidence, classification, details, spx_trend, option_text):
    """
    تنسيق الرسالة كما طلبت تماماً.
    """
    # ترجمة الاتجاه
    if signal == "CALL":
        direction = "Call"
        cross = "50 > 100" if details.get('sma50', 0) > details.get('sma100', 0) else "50 > 200"
        price_status = f"السعر فوق {details.get('sma200', 0):.2f}"
        event = "اختراق قمة 2H"
    else:
        direction = "Put"
        cross = "50 < 100" if details.get('sma50', 0) < details.get('sma100', 0) else "50 < 200"
        price_status = f"السعر تحت {details.get('sma200', 0):.2f}"
        event = "كسر قاع 2H"

    # توافق SPX
    if "uptrend" in spx_trend and signal == "CALL":
        spx_compat = "✅ متوافق"
    elif "downtrend" in spx_trend and signal == "PUT":
        spx_compat = "✅ متوافق"
    else:
        spx_compat = "⚠️ غير متوافق"

    message = f"""
🔔 {ticker}
الاتجاه: {direction}
التصنيف: {classification}
التقاطع: {cross} (إيجابي)
الحالة: {price_status}
الحدث: {event}
SPX: {spx_compat}
الثقة: {confidence}/100
العقد المقترح: {option_text}
"""
    return message

# ---------------------------
# 9. دالة المسح الرئيسية
# ---------------------------

def scan_all_stocks():
    """تمر على كل الأسهم وتفحصها وترسل التنبيهات"""
    alerts = []
    for ticker in TICKERS:
        logging.info(f"فحص {ticker}...")
        time.sleep(0.5)  # تأخير نصف ثانية لتجنب حظر Yahoo

        # 1. الفلتر الأساسي
        if not passes_basic_filters(ticker):
            continue

        # 2. تحليل السهم
        result = analyze_stock_2h(ticker)
        if not result or not result['signal']:
            continue

        signal = result['signal']
        confidence = result['confidence']
        details = result['details']

        # 3. تحليل SPX
        spx_15m = analyze_spx("15m")

        # 4. تصنيف الصفقة
        classification = classify_trade(signal, confidence, details)

        # 5. اختيار العقد
        option = get_option_contract(ticker, signal, classification)

        # 6. بناء الرسالة
        msg = build_alert_message(ticker, signal, confidence, classification, details, spx_15m, option)
        alerts.append(msg)

    return alerts

# ---------------------------
# 10. دوال بوت تليجرام
# ---------------------------

async def send_alerts(context):
    """إرسال التنبيهات إلى تليجرام"""
    alerts = scan_all_stocks()
    if not alerts:
        await context.bot.send_message(chat_id=CHAT_ID, text="لا توجد إشارات جديدة حالياً.")
    else:
        for alert in alerts:
            await context.bot.send_message(chat_id=CHAT_ID, text=alert, parse_mode='Markdown')
            await asyncio.sleep(1)  # تجنب سبام تليجرام

async def start(update, context):
    """معالج أمر /start"""
    await update.message.reply_text(
        "مرحباً! بوت توصيات شاكر 1.0 يعمل.\n"
        "سأرسل لك التنبيهات تلقائياً كل ساعة.\n"
        "لبدء مسح فوري أرسل /scan"
    )

async def scan_now(update, context):
    """معالج أمر /scan لمسح فوري"""
    await update.message.reply_text("جاري المسح، انتظر قليلاً...")
    alerts = scan_all_stocks()
    if not alerts:
        await update.message.reply_text("لا توجد إشارات جديدة.")
    else:
        for alert in alerts:
            await update.message.reply_text(alert, parse_mode='Markdown')
            await asyncio.sleep(1)

def main():
    """الدالة الرئيسية لتشغيل البوت"""
    # إنشاء تطبيق البوت
    application = Application.builder().token(TELEGRAM_TOKEN).build()

    # إضافة الأوامر
    application.add_handler(CommandHandler("start", start))
    application.add_handler(CommandHandler("scan", scan_now))

    # جدولة المسح التلقائي كل ساعة
    job_queue = application.job_queue
    job_queue.run_repeating(send_alerts, interval=3600, first=10)  # كل ساعة، أول مرة بعد 10 ثواني

    # تشغيل البوت
    print("البوت يعمل الآن...")
    application.run_polling()

if __name__ == "__main__":
    main()
````

````python
#!/usr/bin/env python3
# -*- coding: utf-8 -*-

# بوت توصيات الأسهم - نسخة شاكر 1.0 (نسخة كاملة)
# آخر تحديث: فبراير 2026

import yfinance as yf
import pandas as pd
import numpy as np
import ta
import logging
import time
import schedule
from datetime import datetime, timedelta
from telegram import Bot, InlineKeyboardButton, InlineKeyboardMarkup
from telegram.ext import Application, CommandHandler, CallbackQueryHandler
import asyncio
import os

# ---------------------------
# 1. إعدادات البوت والتوكن
# ---------------------------
TELEGRAM_TOKEN = "YOUR_BOT_TOKEN_HERE"      # ضع التوكن الخاص بك
CHAT_ID = "YOUR_CHAT_ID_HERE"                # ضع معرف المستخدم أو القناة

# تفعيل السجلات لمشاهدة الأخطاء
logging.basicConfig(level=logging.INFO, format='%(asctime)s - %(name)s - %(levelname)s - %(message)s')

# ---------------------------
# 2. قائمة الأسهم الكاملة (مدمجة)
# ---------------------------
TICKERS = [
    # --- أسهمك القديمة (من الصور) ---
    "A", "AA", "ABEV", "ABNB", "ABT", "ABVX", "ACGL", "ACM", "ACN",
    "ADBE", "ADI", "ADM", "ADP", "ADSK", "AEE", "AEG", "AEIS", "AEM",
    "AEP", "AER", "AES", "AFG", "AFL", "AFRM", "AG", "AGI", "AGNC",
    "AIG", "AIT", "AIZ", "AJG", "AKAM", "ALAB", "ALB", "ALC", "ALGN",
    "ALL", "ALLE", "ALLY", "ALNY", "ALSN", "AM", "AMCR", "AME", "AMH",
    "AMKR", "AMP", "AMRZ", "AMT", "AMX",

    # --- الشركات العملاقة (> 1 تريليون دولار) ---
    "NVDA", "AAPL", "GOOGL", "MSFT", "AMZN", "META", "TSM", "AVGO", "TSLA",
    "BRK.A", "WMT",

    # --- الشركات الكبرى (200 مليار - 1 تريليون) ---
    "LLY", "JPM", "XOM", "V", "JNJ", "MA", "MU", "ORCL", "COST", "AMD",
    "NFLX", "PLTR", "CSCO", "MRK", "TMO", "ASML", "ABBV", "HD", "CVX",
    "PG", "CAT", "KO", "PEP", "INTC", "CMCSA", "MCD", "TXN", "VZ", "AMGN"
]

# ---------------------------
# 3. دوال التحليل الأساسي (الفلاتر)
# ---------------------------

def get_stock_info(ticker):
    """جلب معلومات الشركة الأساسية"""
    try:
        stock = yf.Ticker(ticker)
        info = stock.info
        return info
    except Exception as e:
        logging.error(f"خطأ في جلب معلومات {ticker}: {e}")
        return None

def passes_basic_filters(ticker):
    """
    تطبيق الفلتر الأساسي:
    - Market Cap > 10B (ملاحظة: أغلب الشركات في القائمة أكبر من ذلك بكثير، لكن الفلتر يضمن ذلك)
    - Average Volume > 1M (تقريبي)
    - Spread < 5%
    """
    info = get_stock_info(ticker)
    if not info:
        return False

    # 1. القيمة السوقية
    mcap = info.get('marketCap', 0)
    if mcap < 10e9:
        logging.debug(f"{ticker} فشل فلتر Market Cap: {mcap}")
        return False

    # 2. متوسط الحجم (نأخذ آخر 20 يوم)
    try:
        stock = yf.Ticker(ticker)
        hist = stock.history(period="1mo", interval="1d")
        if hist.empty:
            return False
        avg_vol = hist['Volume'].rolling(20).mean().iloc[-1]
        if avg_vol < 1_000_000:   # أقل من مليون سهم يومياً
            logging.debug(f"{ticker} فشل فلتر الحجم: {avg_vol}")
            return False
    except:
        return False

    # 3. Spread (bid/ask)
    bid = info.get('bid', 0)
    ask = info.get('ask', 0)
    if bid and ask:
        spread = (ask - bid) / bid if bid != 0 else 1
        if spread > 0.05:   # أكبر من 5%
            logging.debug(f"{ticker} فشل فلتر Spread: {spread:.2%}")
            return False

    return True

# ---------------------------
# 4. تحليل SPX (مؤشر S&P 500)
# ---------------------------

def analyze_spx(timeframe="15m"):
    """تحليل مؤشر SPX على الإطار المطلوب"""
    try:
        spx = yf.Ticker("^GSPC")  # أو "SPY"
        hist = spx.history(period="5d", interval=timeframe)
        if len(hist) < 50:
            return "neutral"

        # متوسطات بسيطة لتحديد الاتجاه
        hist['SMA20'] = hist['Close'].rolling(20).mean()
        last = hist.iloc[-1]
        prev = hist.iloc[-2]

        if last['Close'] > last['SMA20'] and last['Close'] > prev['Close']:
            return "uptrend"
        elif last['Close'] < last['SMA20'] and last['Close'] < prev['Close']:
            return "downtrend"
        else:
            return "neutral"
    except:
        return "neutral"

# ---------------------------
# 5. تحليل السهم (فريم ساعتين)
# ---------------------------

def analyze_stock_2h(ticker):
    """
    تحليل فني كامل على فريم الساعتين
    يعيد: {'signal': 'CALL'/'PUT'/None, 'confidence': 0-100, 'details': {...}}
    """
    try:
        stock = yf.Ticker(ticker)
        hist = stock.history(period="1mo", interval="2h")  # آخر شهر، فريم ساعتين
        if len(hist) < 200:
            return None

        # حساب المتوسطات
        hist['SMA50'] = hist['Close'].rolling(50).mean()
        hist['SMA100'] = hist['Close'].rolling(100).mean()
        hist['SMA200'] = hist['Close'].rolling(200).mean()
        hist['Volume_SMA20'] = hist['Volume'].rolling(20).mean()

        last = hist.iloc[-1]
        current_price = last['Close']
        sma50 = last['SMA50']
        sma100 = last['SMA100']
        sma200 = last['SMA200']

        # قمم وقيعان آخر 3 شموع
        high_last_3 = hist['High'].iloc[-3:].max()
        low_last_3 = hist['Low'].iloc[-3:].min()

        # قمة/قاع اليوم (آخر 24 ساعة = 12 شمعة)
        high_today = hist['High'].iloc[-12:].max()
        low_today = hist['Low'].iloc[-12:].min()

        volume_ok = last['Volume'] > last['Volume_SMA20']

        # شروط CALL
        call_condition1 = (sma50 > sma100) or (sma50 > sma200)
        call_condition2 = current_price > sma200
        call_condition3 = current_price > high_last_3 or current_price > high_today
        call_condition4 = volume_ok

        # شروط PUT
        put_condition1 = (sma50 < sma100) or (sma50 < sma200)
        put_condition2 = current_price < sma200
        put_condition3 = current_price < low_last_3 or current_price < low_today
        put_condition4 = volume_ok

        signal = None
        confidence = 0
        details = {}

        if call_condition1 and call_condition2 and call_condition3 and call_condition4:
            signal = "CALL"
            # حساب الثقة: نضيف 25 لكل شرط متحقق
            confidence = 25 * sum([call_condition1, call_condition2, call_condition3, call_condition4])
            details = {
                "sma50": sma50,
                "sma100": sma100,
                "sma200": sma200,
                "price": current_price,
                "high_3": high_last_3,
                "volume_ratio": last['Volume'] / last['Volume_SMA20']
            }
        elif put_condition1 and put_condition2 and put_condition3 and put_condition4:
            signal = "PUT"
            confidence = 25 * sum([put_condition1, put_condition2, put_condition3, put_condition4])
            details = {
                "sma50": sma50,
                "sma100": sma100,
                "sma200": sma200,
                "price": current_price,
                "low_3": low_last_3,
                "volume_ratio": last['Volume'] / last['Volume_SMA20']
            }

        return {"signal": signal, "confidence": int(confidence), "details": details}

    except Exception as e:
        logging.error(f"خطأ في تحليل {ticker}: {e}")
        return None

# ---------------------------
# 6. تصنيف الصفقة (سكالب، قصيرة، سوينق)
# ---------------------------

def classify_trade(signal, confidence, details):
    """
    تصنيف حسب قوة الإشارة وترتيب المتوسطات
    """
    if not signal:
        return None

    sma50 = details.get('sma50')
    sma100 = details.get('sma100')
    sma200 = details.get('sma200')

    # ترتيب المتوسطات
    if signal == "CALL":
        if sma50 > sma100 > sma200:
            return "📈 سوينق 3-7 أيام"
        elif confidence >= 75:
            return "⏳ صفقة قصيرة 1-2 يوم"
        else:
            return "⚡ سكالب"
    else:  # PUT
        if sma50 < sma100 < sma200:
            return "📈 سوينق 3-7 أيام"
        elif confidence >= 75:
            return "⏳ صفقة قصيرة 1-2 يوم"
        else:
            return "⚡ سكالب"

# ---------------------------
# 7. اختيار العقد المناسب (بيانات الخيارات)
# ---------------------------

def get_option_contract(ticker, signal, classification):
    """
    البحث عن عقد خيار مناسب حسب التصنيف.
    """
    try:
        stock = yf.Ticker(ticker)
        # جلب تواريخ انتهاء الصلاحية المتاحة
        expirations = stock.options
        if not expirations:
            return "غير متوفر"

        # اختيار التاريخ المناسب حسب التصنيف
        today = datetime.now().date()
        if "سكالب" in classification:
            days = 7
        elif "قصيرة" in classification:
            days = 14
        else:  # سوينق
            days = 30

        target_date = today + timedelta(days=days)
        # البحث عن أقرب تاريخ انتهاء للعدد المطلوب من الأيام
        exp_date = None
        for exp in expirations:
            exp_dt = datetime.strptime(exp, '%Y-%m-%d').date()
            if exp_dt >= target_date:
                exp_date = exp
                break
        if not exp_date:
            exp_date = expirations[0]  # أقرب تاريخ

        # جلب عقود الخيارات لذلك التاريخ
        opt_chain = stock.option_chain(exp_date)
        if signal == "CALL":
            contracts = opt_chain.calls
        else:
            contracts = opt_chain.puts

        if contracts.empty:
            return "غير متوفر"

        # فلترة العقود: Open Interest عالي (أعلى 10%)، Spread ضيق
        current_price = stock.history(period="1d")['Close'].iloc[-1]
        contracts['strike_diff'] = abs(contracts['strike'] - current_price)
        # فلتر OI
        oi_threshold = contracts['openInterest'].quantile(0.7)  # أعلى 30%
        eligible = contracts[contracts['openInterest'] > oi_threshold]

        if eligible.empty:
            eligible = contracts

        # نختار العقد الأقرب للسعر الحالي (للدلتا التقريبية)
        best = eligible.nsmallest(1, 'strike_diff').iloc[0]

        # حساب Spread تقريبي (ask-bid)/ask
        spread = (best['ask'] - best['bid']) / best['ask'] if best['ask'] > 0 else 1
        if spread > 0.5:  # spread كبير جداً
            return "سبريد واسع"

        # بناء نص العقد المقترح
        delta_approx = 0.5 - (best['strike_diff'] / current_price)  # تقريب بسيط
        delta_approx = max(0.1, min(0.9, delta_approx))

        return f"Delta {delta_approx:.2f} – Exp {exp_date} (أيام: {days})"

    except Exception as e:
        logging.error(f"خطأ في جلب الخيارات لـ {ticker}: {e}")
        return "غير متوفر"

# ---------------------------
# 8. بناء رسالة التنبيه النهائية
# ---------------------------

def build_alert_message(ticker, signal, confidence, classification, details, spx_trend, option_text):
    """
    تنسيق الرسالة كما طلبت تماماً.
    """
    # ترجمة الاتجاه
    if signal == "CALL":
        direction = "Call"
        cross = "50 > 100" if details.get('sma50', 0) > details.get('sma100', 0) else "50 > 200"
        price_status = f"السعر فوق {details.get('sma200', 0):.2f}"
        event = "اختراق قمة 2H"
    else:
        direction = "Put"
        cross = "50 < 100" if details.get('sma50', 0) < details.get('sma100', 0) else "50 < 200"
        price_status = f"السعر تحت {details.get('sma200', 0):.2f}"
        event = "كسر قاع 2H"

    # توافق SPX
    if "uptrend" in spx_trend and signal == "CALL":
        spx_compat = "✅ متوافق"
    elif "downtrend" in spx_trend and signal == "PUT":
        spx_compat = "✅ متوافق"
    else:
        spx_compat = "⚠️ غير متوافق"

    message = f"""
🔔 {ticker}
الاتجاه: {direction}
التصنيف: {classification}
التقاطع: {cross} (إيجابي)
الحالة: {price_status}
الحدث: {event}
SPX: {spx_compat}
الثقة: {confidence}/100
العقد المقترح: {option_text}
"""
    return message

# ---------------------------
# 9. دالة المسح الرئيسية
# ---------------------------

def scan_all_stocks():
    """تمر على كل الأسهم وتفحصها وترسل التنبيهات"""
    alerts = []
    for ticker in TICKERS:
        logging.info(f"فحص {ticker}...")
        time.sleep(0.5)  # تأخير نصف ثانية لتجنب حظر Yahoo

        # 1. الفلتر الأساسي
        if not passes_basic_filters(ticker):
            continue

        # 2. تحليل السهم
        result = analyze_stock_2h(ticker)
        if not result or not result['signal']:
            continue

        signal = result['signal']
        confidence = result['confidence']
        details = result['details']

        # 3. تحليل SPX
        spx_15m = analyze_spx("15m")

        # 4. تصنيف الصفقة
        classification = classify_trade(signal, confidence, details)

        # 5. اختيار العقد
        option = get_option_contract(ticker, signal, classification)

        # 6. بناء الرسالة
        msg = build_alert_message(ticker, signal, confidence, classification, details, spx_15m, option)
        alerts.append(msg)

    return alerts

# ---------------------------
# 10. دوال بوت تليجرام
# ---------------------------

async def send_alerts(context):
    """إرسال التنبيهات إلى تليجرام"""
    alerts = scan_all_stocks()
    if not alerts:
        await context.bot.send_message(chat_id=CHAT_ID, text="لا توجد إشارات جديدة حالياً.")
    else:
        for alert in alerts:
            await context.bot.send_message(chat_id=CHAT_ID, text=alert, parse_mode='Markdown')
            await asyncio.sleep(1)  # تجنب سبام تليجرام

async def start(update, context):
    """معالج أمر /start"""
    await update.message.reply_text(
        "مرحباً! بوت توصيات شاكر 1.0 يعمل.\n"
        "سأرسل لك التنبيهات تلقائياً كل ساعة.\n"
        "لبدء مسح فوري أرسل /scan"
    )

async def scan_now(update, context):
    """معالج أمر /scan لمسح فوري"""
    await update.message.reply_text("جاري المسح، انتظر قليلاً...")
    alerts = scan_all_stocks()
    if not alerts:
        await update.message.reply_text("لا توجد إشارات جديدة.")
    else:
        for alert in alerts:
            await update.message.reply_text(alert, parse_mode='Markdown')
            await asyncio.sleep(1)

def main():
    """الدالة الرئيسية لتشغيل البوت"""
    # إنشاء تطبيق البوت
    application = Application.builder().token(TELEGRAM_TOKEN).build()

    # إضافة الأوامر
    application.add_handler(CommandHandler("start", start))
    application.add_handler(CommandHandler("scan", scan_now))

    # جدولة المسح التلقائي كل ساعة
    job_queue = application.job_queue
    job_queue.run_repeating(send_alerts, interval=3600, first=10)  # كل ساعة، أول مرة بعد 10 ثواني

    # تشغيل البوت
    print("البوت يعمل الآن...")
    application.run_polling()

if __name__ == "__main__":
    main()
````

**Reasoning**:
The previous error indicated that the `JobQueue` was not properly set up, even after installing the `job-queue` extra. To fix this, I will explicitly enable the `JobQueue` when building the `Application` by adding `.job_queue()` to the `Application.builder()` chain in the `main` function.



````python
#!/usr/bin/env python3
# -*- coding: utf-8 -*-

# بوت توصيات الأسهم - نسخة شاكر 1.0 (نسخة كاملة)
# آخر تحديث: فبراير 2026

import yfinance as yf
import pandas as pd
import numpy as np
import ta
import logging
import time
import schedule
from datetime import datetime, timedelta
from telegram import Bot, InlineKeyboardButton, InlineKeyboardMarkup
from telegram.ext import Application, CommandHandler, CallbackQueryHandler
import asyncio
import os

# ---------------------------
# 1. إعدادات البوت والتوكن
# ---------------------------
TELEGRAM_TOKEN = "8699821370:AAEQUSbLTgf7MmWqo5vV5LHPOz30wfqOfqw"      # ضع التوكن الخاص بك
CHAT_ID = "YOUR_CHAT_ID_HERE"                # ضع معرف المستخدم أو القناة

# تفعيل السجلات لمشاهدة الأخطاء
logging.basicConfig(level=logging.INFO, format='%(asctime)s - %(name)s - %(levelname)s - %(message)s')

# ---------------------------
# 2. قائمة الأسهم الكاملة (مدمجة)
# ---------------------------
TICKERS = [
    # --- أسهمك القديمة (من الصور) ---
    "A", "AA", "ABEV", "ABNB", "ABT", "ABVX", "ACGL", "ACM", "ACN",
    "ADBE", "ADI", "ADM", "ADP", "ADSK", "AEE", "AEG", "AEIS", "AEM",
    "AEP", "AER", "AES", "AFG", "AFL", "AFRM", "AG", "AGI", "AGNC",
    "AIG", "AIT", "AIZ", "AJG", "AKAM", "ALAB", "ALB", "ALC", "ALGN",
    "ALL", "ALLE", "ALLY", "ALNY", "ALSN", "AM", "AMCR", "AME", "AMH",
    "AMKR", "AMP", "AMRZ", "AMT", "AMX",

    # --- الشركات العملاقة (> 1 تريليون دولار) ---
    "NVDA", "AAPL", "GOOGL", "MSFT", "AMZN", "META", "TSM", "AVGO", "TSLA",
    "BRK.A", "WMT",

    # --- الشركات الكبرى (200 مليار - 1 تريليون) ---
    "LLY", "JPM", "XOM", "V", "JNJ", "MA", "MU", "ORCL", "COST", "AMD",
    "NFLX", "PLTR", "CSCO", "MRK", "TMO", "ASML", "ABBV", "HD", "CVX",
    "PG", "CAT", "KO", "PEP", "INTC", "CMCSA", "MCD", "TXN", "VZ", "AMGN"
]

# ---------------------------
# 3. دوال التحليل الأساسي (الفلاتر)
# ---------------------------

def get_stock_info(ticker):
    """جلب معلومات الشركة الأساسية"""
    try:
        stock = yf.Ticker(ticker)
        info = stock.info
        return info
    except Exception as e:
        logging.error(f"خطأ في جلب معلومات {ticker}: {e}")
        return None

def passes_basic_filters(ticker):
    """
    تطبيق الفلتر الأساسي:
    - Market Cap > 10B (ملاحظة: أغلب الشركات في القائمة أكبر من ذلك بكثير، لكن الفلتر يضمن ذلك)
    - Average Volume > 1M (تقريبي)
    - Spread < 5%
    """
    info = get_stock_info(ticker)
    if not info:
        return False

    # 1. القيمة السوقية
    mcap = info.get('marketCap', 0)
    if mcap < 10e9:
        logging.debug(f"{ticker} فشل فلتر Market Cap: {mcap}")
        return False

    # 2. متوسط الحجم (نأخذ آخر 20 يوم)
    try:
        stock = yf.Ticker(ticker)
        hist = stock.history(period="1mo", interval="1d")
        if hist.empty:
            return False
        avg_vol = hist['Volume'].rolling(20).mean().iloc[-1]
        if avg_vol < 1_000_000:   # أقل من مليون سهم يومياً
            logging.debug(f"{ticker} فشل فلتر الحجم: {avg_vol}")
            return False
    except:
        return False

    # 3. Spread (bid/ask)
    bid = info.get('bid', 0)
    ask = info.get('ask', 0)
    if bid and ask:
        spread = (ask - bid) / bid if bid != 0 else 1
        if spread > 0.05:   # أكبر من 5%
            logging.debug(f"{ticker} فشل فلتر Spread: {spread:.2%}")
            return False

    return True

# ---------------------------
# 4. تحليل SPX (مؤشر S&P 500)
# ---------------------------

def analyze_spx(timeframe="15m"):
    """تحليل مؤشر SPX على الإطار المطلوب"""
    try:
        spx = yf.Ticker("^GSPC")  # أو "SPY"
        hist = spx.history(period="5d", interval=timeframe)
        if len(hist) < 50:
            return "neutral"

        # متوسطات بسيطة لتحديد الاتجاه
        hist['SMA20'] = hist['Close'].rolling(20).mean()
        last = hist.iloc[-1]
        prev = hist.iloc[-2]

        if last['Close'] > last['SMA20'] and last['Close'] > prev['Close']:
            return "uptrend"
        elif last['Close'] < last['SMA20'] and last['Close'] < prev['Close']:
            return "downtrend"
        else:
            return "neutral"
    except:
        return "neutral"

# ---------------------------
# 5. تحليل السهم (فريم ساعتين)
# ---------------------------

def analyze_stock_2h(ticker):
    """
    تحليل فني كامل على فريم الساعتين
    يعيد: {'signal': 'CALL'/'PUT'/None, 'confidence': 0-100, 'details': {...}}
    """
    try:
        stock = yf.Ticker(ticker)
        hist = stock.history(period="1mo", interval="2h")  # آخر شهر، فريم ساعتين
        if len(hist) < 200:
            return None

        # حساب المتوسطات
        hist['SMA50'] = hist['Close'].rolling(50).mean()
        hist['SMA100'] = hist['Close'].rolling(100).mean()
        hist['SMA200'] = hist['Close'].rolling(200).mean()
        hist['Volume_SMA20'] = hist['Volume'].rolling(20).mean()

        last = hist.iloc[-1]
        current_price = last['Close']
        sma50 = last['SMA50']
        sma100 = last['SMA100']
        sma200 = last['SMA200']

        # قمم وقيعان آخر 3 شموع
        high_last_3 = hist['High'].iloc[-3:].max()
        low_last_3 = hist['Low'].iloc[-3:].min()

        # قمة/قاع اليوم (آخر 24 ساعة = 12 شمعة)
        high_today = hist['High'].iloc[-12:].max()
        low_today = hist['Low'].iloc[-12:].min()

        volume_ok = last['Volume'] > last['Volume_SMA20']

        # شروط CALL
        call_condition1 = (sma50 > sma100) or (sma50 > sma200)
        call_condition2 = current_price > sma200
        call_condition3 = current_price > high_last_3 or current_price > high_today
        call_condition4 = volume_ok

        # شروط PUT
        put_condition1 = (sma50 < sma100) or (sma50 < sma200)
        put_condition2 = current_price < sma200
        put_condition3 = current_price < low_last_3 or current_price < low_today
        put_condition4 = volume_ok

        signal = None
        confidence = 0
        details = {}

        if call_condition1 and call_condition2 and call_condition3 and call_condition4:
            signal = "CALL"
            # حساب الثقة: نضيف 25 لكل شرط متحقق
            confidence = 25 * sum([call_condition1, call_condition2, call_condition3, call_condition4])
            details = {
                "sma50": sma50,
                "sma100": sma100,
                "sma200": sma200,
                "price": current_price,
                "high_3": high_last_3,
                "volume_ratio": last['Volume'] / last['Volume_SMA20']
            }
        elif put_condition1 and put_condition2 and put_condition3 and put_condition4:
            signal = "PUT"
            confidence = 25 * sum([put_condition1, put_condition2, put_condition3, put_condition4])
            details = {
                "sma50": sma50,
                "sma100": sma100,
                "sma200": sma200,
                "price": current_price,
                "low_3": low_last_3,
                "volume_ratio": last['Volume'] / last['Volume_SMA20']
            }

        return {"signal": signal, "confidence": int(confidence), "details": details}

    except Exception as e:
        logging.error(f"خطأ في تحليل {ticker}: {e}")
        return None

# ---------------------------
# 6. تصنيف الصفقة (سكالب، قصيرة، سوينق)
# ---------------------------

def classify_trade(signal, confidence, details):
    """
    تصنيف حسب قوة الإشارة وترتيب المتوسطات
    """
    if not signal:
        return None

    sma50 = details.get('sma50')
    sma100 = details.get('sma100')
    sma200 = details.get('sma200')

    # ترتيب المتوسطات
    if signal == "CALL":
        if sma50 > sma100 > sma200:
            return "📈 سوينق 3-7 أيام"
        elif confidence >= 75:
            return "⏳ صفقة قصيرة 1-2 يوم"
        else:
            return "⚡ سكالب"
    else:  # PUT
        if sma50 < sma100 < sma200:
            return "📈 سوينق 3-7 أيام"
        elif confidence >= 75:
            return "⏳ صفقة قصيرة 1-2 يوم"
        else:
            return "⚡ سكالب"

# ---------------------------
# 7. اختيار العقد المناسب (بيانات الخيارات)
# ---------------------------

def get_option_contract(ticker, signal, classification):
    """
    البحث عن عقد خيار مناسب حسب التصنيف.
    """
    try:
        stock = yf.Ticker(ticker)
        # جلب تواريخ انتهاء الصلاحية المتاحة
        expirations = stock.options
        if not expirations:
            return "غير متوفر"

        # اختيار التاريخ المناسب حسب التصنيف
        today = datetime.now().date()
        if "سكالب" in classification:
            days = 7
        elif "قصيرة" in classification:
            days = 14
        else:  # سوينق
            days = 30

        target_date = today + timedelta(days=days)
        # البحث عن أقرب تاريخ انتهاء للعدد المطلوب من الأيام
        exp_date = None
        for exp in expirations:
            exp_dt = datetime.strptime(exp, '%Y-%m-%d').date()
            if exp_dt >= target_date:
                exp_date = exp
                break
        if not exp_date:
            exp_date = expirations[0]  # أقرب تاريخ

        # جلب عقود الخيارات لذلك التاريخ
        opt_chain = stock.option_chain(exp_date)
        if signal == "CALL":
            contracts = opt_chain.calls
        else:
            contracts = opt_chain.puts

        if contracts.empty:
            return "غير متوفر"

        # فلترة العقود: Open Interest عالي (أعلى 10%)، Spread ضيق
        current_price = stock.history(period="1d")['Close'].iloc[-1]
        contracts['strike_diff'] = abs(contracts['strike'] - current_price)
        # فلتر OI
        oi_threshold = contracts['openInterest'].quantile(0.7)  # أعلى 30%
        eligible = contracts[contracts['openInterest'] > oi_threshold]

        if eligible.empty:
            eligible = contracts

        # نختار العقد الأقرب للسعر الحالي (للدلتا التقريبية)
        best = eligible.nsmallest(1, 'strike_diff').iloc[0]

        # حساب Spread تقريبي (ask-bid)/ask
        spread = (best['ask'] - best['bid']) / best['ask'] if best['ask'] > 0 else 1
        if spread > 0.5:  # spread كبير جداً
            return "سبريد واسع"

        # بناء نص العقد المقترح
        delta_approx = 0.5 - (best['strike_diff'] / current_price)  # تقريب بسيط
        delta_approx = max(0.1, min(0.9, delta_approx))

        return f"Delta {delta_approx:.2f} – Exp {exp_date} (أيام: {days})"

    except Exception as e:
        logging.error(f"خطأ في جلب الخيارات لـ {ticker}: {e}")
        return "غير متوفر"

# ---------------------------
# 8. بناء رسالة التنبيه النهائية
# ---------------------------

def build_alert_message(ticker, signal, confidence, classification, details, spx_trend, option_text):
    """
    تنسيق الرسالة كما طلبت تماماً.
    """
    # ترجمة الاتجاه
    if signal == "CALL":
        direction = "Call"
        cross = "50 > 100" if details.get('sma50', 0) > details.get('sma100', 0) else "50 > 200"
        price_status = f"السعر فوق {details.get('sma200', 0):.2f}"
        event = "اختراق قمة 2H"
    else:
        direction = "Put"
        cross = "50 < 100" if details.get('sma50', 0) < details.get('sma100', 0) else "50 < 200"
        price_status = f"السعر تحت {details.get('sma200', 0):.2f}"
        event = "كسر قاع 2H"

    # توافق SPX
    if "uptrend" in spx_trend and signal == "CALL":
        spx_compat = "✅ متوافق"
    elif "downtrend" in spx_trend and signal == "PUT":
        spx_compat = "✅ متوافق"
    else:
        spx_compat = "⚠️ غير متوافق"

    message = f"""
🔔 {ticker}
الاتجاه: {direction}
التصنيف: {classification}
التقاطع: {cross} (إيجابي)
الحالة: {price_status}
الحدث: {event}
SPX: {spx_compat}
الثقة: {confidence}/100
العقد المقترح: {option_text}
"""
    return message

# ---------------------------
# 9. دالة المسح الرئيسية
# ---------------------------

def scan_all_stocks():
    """تمر على كل الأسهم وتفحصها وترسل التنبيهات"""
    alerts = []
    for ticker in TICKERS:
        logging.info(f"فحص {ticker}...")
        time.sleep(0.5)  # تأخير نصف ثانية لتجنب حظر Yahoo

        # 1. الفلتر الأساسي
        if not passes_basic_filters(ticker):
            continue

        # 2. تحليل السهم
        result = analyze_stock_2h(ticker)
        if not result or not result['signal']:
            continue

        signal = result['signal']
        confidence = result['confidence']
        details = result['details']

        # 3. تحليل SPX
        spx_15m = analyze_spx("15m")

        # 4. تصنيف الصفقة
        classification = classify_trade(signal, confidence, details)

        # 5. اختيار العقد
        option = get_option_contract(ticker, signal, classification)

        # 6. بناء الرسالة
        msg = build_alert_message(ticker, signal, confidence, classification, details, spx_15m, option)
        alerts.append(msg)

    return alerts

# ---------------------------
# 10. دوال بوت تليجرام
# ---------------------------

async def send_alerts(context):
    """إرسال التنبيهات إلى تليجرام"""
    alerts = scan_all_stocks()
    if not alerts:
        await context.bot.send_message(chat_id=CHAT_ID, text="لا توجد إشارات جديدة حالياً.")
    else:
        for alert in alerts:
            await context.bot.send_message(chat_id=CHAT_ID, text=alert, parse_mode='Markdown')
            await asyncio.sleep(1)  # تجنب سبام تليجرام

async def start(update, context):
    """معالج أمر /start"""
    await update.message.reply_text(
        "مرحباً! بوت توصيات شاكر 1.0 يعمل.\n"
        "سأرسل لك التنبيهات تلقائياً كل ساعة.\n"
        "لبدء مسح فوري أرسل /scan"
    )

async def scan_now(update, context):
    """معالج أمر /scan لمسح فوري"""
    await update.message.reply_text("جاري المسح، انتظر قليلاً...")
    alerts = scan_all_stocks()
    if not alerts:
        await update.message.reply_text("لا توجد إشارات جديدة.")
    else:
        for alert in alerts:
            await update.message.reply_text(alert, parse_mode='Markdown')
            await asyncio.sleep(1)

def main():
    """الدالة الرئيسية لتشغيل البوت"""
    # إنشاء تطبيق البوت
    application = Application.builder().token(TELEGRAM_TOKEN).job_queue().build()

    # إضافة الأوامر
    application.add_handler(CommandHandler("start", start))
    application.add_handler(CommandHandler("scan", scan_now))

    # جدولة المسح التلقائي كل ساعة
    job_queue = application.job_queue
    job_queue.run_repeating(send_alerts, interval=3600, first=10)  # كل ساعة، أول مرة بعد 10 ثواني

    # تشغيل البوت
    print("البوت يعمل الآن...")
    application.run_polling()

if __name__ == "__main__":
    main()
````

**Reasoning**:
The previous `TypeError` occurred because `Application.builder().job_queue()` was called as a method expecting an argument, which is incorrect for enabling the JobQueue. The actual issue is that `application.job_queue` might be `None` even after installing the `job-queue` extra. To correctly initialize the `JobQueue`, I will explicitly create a `JobQueue` instance and assign it to `application.job_queue` after building the `Application`.



````python
#!/usr/bin/env python3
# -*- coding: utf-8 -*-

# بوت توصيات الأسهم - نسخة شاكر 1.0 (نسخة كاملة)
# آخر تحديث: فبراير 2026

import yfinance as yf
import pandas as pd
import numpy as np
import ta
import logging
import time
import schedule
from datetime import datetime, timedelta
from telegram import Bot, InlineKeyboardButton, InlineKeyboardMarkup
from telegram.ext import Application, CommandHandler, CallbackQueryHandler, JobQueue
import asyncio
import os

# ---------------------------
# 1. إعدادات البوت والتوكن
# ---------------------------
TELEGRAM_TOKEN = "8699821370:AAEQUSbLTgf7MmWqo5vV5LHPOz30wfqOfqw"      # ضع التوكن الخاص بك
CHAT_ID = "7854020427"                # ضع معرف المستخدم أو القناة

# تفعيل السجلات لمشاهدة الأخطاء
logging.basicConfig(level=logging.INFO, format='%(asctime)s - %(name)s - %(levelname)s - %(message)s')

# ---------------------------
# 2. قائمة الأسهم الكاملة (مدمجة)
# ---------------------------
TICKERS = [
    # --- أسهمك القديمة (من الصور) ---
    "A", "AA", "ABEV", "ABNB", "ABT", "ABVX", "ACGL", "ACM", "ACN",
    "ADBE", "ADI", "ADM", "ADP", "ADSK", "AEE", "AEG", "AEIS", "AEM",
    "AEP", "AER", "AES", "AFG", "AFL", "AFRM", "AG", "AGI", "AGNC",
    "AIG", "AIT", "AIZ", "AJG", "AKAM", "ALAB", "ALB", "ALC", "ALGN",
    "ALL", "ALLE", "ALLY", "ALNY", "ALSN", "AM", "AMCR", "AME", "AMH",
    "AMKR", "AMP", "AMRZ", "AMT", "AMX",

    # --- الشركات العملاقة (> 1 تريليون دولار) ---
    "NVDA", "AAPL", "GOOGL", "MSFT", "AMZN", "META", "TSM", "AVGO", "TSLA",
    "BRK.A", "WMT",

    # --- الشركات الكبرى (200 مليار - 1 تريليون) ---
    "LLY", "JPM", "XOM", "V", "JNJ", "MA", "MU", "ORCL", "COST", "AMD",
    "NFLX", "PLTR", "CSCO", "MRK", "TMO", "ASML", "ABBV", "HD", "CVX",
    "PG", "CAT", "KO", "PEP", "INTC", "CMCSA", "MCD", "TXN", "VZ", "AMGN"
]

# ---------------------------
# 3. دوال التحليل الأساسي (الفلاتر)
# ---------------------------

def get_stock_info(ticker):
    """جلب معلومات الشركة الأساسية"""
    try:
        stock = yf.Ticker(ticker)
        info = stock.info
        return info
    except Exception as e:
        logging.error(f"خطأ في جلب معلومات {ticker}: {e}")
        return None

def passes_basic_filters(ticker):
    """
    تطبيق الفلتر الأساسي:
    - Market Cap > 10B (ملاحظة: أغلب الشركات في القائمة أكبر من ذلك بكثير، لكن الفلتر يضمن ذلك)
    - Average Volume > 1M (تقريبي)
    - Spread < 5%
    """
    info = get_stock_info(ticker)
    if not info:
        return False

    # 1. القيمة السوقية
    mcap = info.get('marketCap', 0)
    if mcap < 10e9:
        logging.debug(f"{ticker} فشل فلتر Market Cap: {mcap}")
        return False

    # 2. متوسط الحجم (نأخذ آخر 20 يوم)
    try:
        stock = yf.Ticker(ticker)
        hist = stock.history(period="1mo", interval="1d")
        if hist.empty:
            return False
        avg_vol = hist['Volume'].rolling(20).mean().iloc[-1]
        if avg_vol < 1_000_000:   # أقل من مليون سهم يومياً
            logging.debug(f"{ticker} فشل فلتر الحجم: {avg_vol}")
            return False
    except:
        return False

    # 3. Spread (bid/ask)
    bid = info.get('bid', 0)
    ask = info.get('ask', 0)
    if bid and ask:
        spread = (ask - bid) / bid if bid != 0 else 1
        if spread > 0.05:   # أكبر من 5%
            logging.debug(f"{ticker} فشل فلتر Spread: {spread:.2%}")
            return False

    return True

# ---------------------------
# 4. تحليل SPX (مؤشر S&P 500)
# ---------------------------

def analyze_spx(timeframe="15m"):
    """تحليل مؤشر SPX على الإطار المطلوب"""
    try:
        spx = yf.Ticker("^GSPC")  # أو "SPY"
        hist = spx.history(period="5d", interval=timeframe)
        if len(hist) < 50:
            return "neutral"

        # متوسطات بسيطة لتحديد الاتجاه
        hist['SMA20'] = hist['Close'].rolling(20).mean()
        last = hist.iloc[-1]
        prev = hist.iloc[-2]

        if last['Close'] > last['SMA20'] and last['Close'] > prev['Close']:
            return "uptrend"
        elif last['Close'] < last['SMA20'] and last['Close'] < prev['Close']:
            return "downtrend"
        else:
            return "neutral"
    except:
        return "neutral"

# ---------------------------
# 5. تحليل السهم (فريم ساعتين)
# ---------------------------

def analyze_stock_2h(ticker):
    """
    تحليل فني كامل على فريم الساعتين
    يعيد: {'signal': 'CALL'/'PUT'/None, 'confidence': 0-100, 'details': {...}}
    """
    try:
        stock = yf.Ticker(ticker)
        hist = stock.history(period="1mo", interval="2h")  # آخر شهر، فريم ساعتين
        if len(hist) < 200:
            return None

        # حساب المتوسطات
        hist['SMA50'] = hist['Close'].rolling(50).mean()
        hist['SMA100'] = hist['Close'].rolling(100).mean()
        hist['SMA200'] = hist['Close'].rolling(200).mean()
        hist['Volume_SMA20'] = hist['Volume'].rolling(20).mean()

        last = hist.iloc[-1]
        current_price = last['Close']
        sma50 = last['SMA50']
        sma100 = last['SMA100']
        sma200 = last['SMA200']

        # قمم وقيعان آخر 3 شموع
        high_last_3 = hist['High'].iloc[-3:].max()
        low_last_3 = hist['Low'].iloc[-3:].min()

        # قمة/قاع اليوم (آخر 24 ساعة = 12 شمعة)
        high_today = hist['High'].iloc[-12:].max()
        low_today = hist['Low'].iloc[-12:].min()

        volume_ok = last['Volume'] > last['Volume_SMA20']

        # شروط CALL
        call_condition1 = (sma50 > sma100) or (sma50 > sma200)
        call_condition2 = current_price > sma200
        call_condition3 = current_price > high_last_3 or current_price > high_today
        call_condition4 = volume_ok

        # شروط PUT
        put_condition1 = (sma50 < sma100) or (sma50 < sma200)
        put_condition2 = current_price < sma200
        put_condition3 = current_price < low_last_3 or current_price < low_today
        put_condition4 = volume_ok

        signal = None
        confidence = 0
        details = {}

        if call_condition1 and call_condition2 and call_condition3 and call_condition4:
            signal = "CALL"
            # حساب الثقة: نضيف 25 لكل شرط متحقق
            confidence = 25 * sum([call_condition1, call_condition2, call_condition3, call_condition4])
            details = {
                "sma50": sma50,
                "sma100": sma100,
                "sma200": sma200,
                "price": current_price,
                "high_3": high_last_3,
                "volume_ratio": last['Volume'] / last['Volume_SMA20']
            }
        elif put_condition1 and put_condition2 and put_condition3 and put_condition4:
            signal = "PUT"
            confidence = 25 * sum([put_condition1, put_condition2, put_condition3, put_condition4])
            details = {
                "sma50": sma50,
                "sma100": sma100,
                "sma200": sma200,
                "price": current_price,
                "low_3": low_last_3,
                "volume_ratio": last['Volume'] / last['Volume_SMA20']
            }

        return {"signal": signal, "confidence": int(confidence), "details": details}

    except Exception as e:
        logging.error(f"خطأ في تحليل {ticker}: {e}")
        return None

# ---------------------------
# 6. تصنيف الصفقة (سكالب، قصيرة، سوينق)
# ---------------------------

def classify_trade(signal, confidence, details):
    """
    تصنيف حسب قوة الإشارة وترتيب المتوسطات
    """
    if not signal:
        return None

    sma50 = details.get('sma50')
    sma100 = details.get('sma100')
    sma200 = details.get('sma200')

    # ترتيب المتوسطات
    if signal == "CALL":
        if sma50 > sma100 > sma200:
            return "📈 سوينق 3-7 أيام"
        elif confidence >= 75:
            return "⏳ صفقة قصيرة 1-2 يوم"
        else:
            return "⚡ سكالب"
    else:  # PUT
        if sma50 < sma100 < sma200:
            return "📈 سوينق 3-7 أيام"
        elif confidence >= 75:
            return "⏳ صفقة قصيرة 1-2 يوم"
        else:
            return "⚡ سكالب"

# ---------------------------
# 7. اختيار العقد المناسب (بيانات الخيارات)
# ---------------------------

def get_option_contract(ticker, signal, classification):
    """
    البحث عن عقد خيار مناسب حسب التصنيف.
    """
    try:
        stock = yf.Ticker(ticker)
        # جلب تواريخ انتهاء الصلاحية المتاحة
        expirations = stock.options
        if not expirations:
            return "غير متوفر"

        # اختيار التاريخ المناسب حسب التصنيف
        today = datetime.now().date()
        if "سكالب" in classification:
            days = 7
        elif "قصيرة" in classification:
            days = 14
        else:  # سوينق
            days = 30

        target_date = today + timedelta(days=days)
        # البحث عن أقرب تاريخ انتهاء للعدد المطلوب من الأيام
        exp_date = None
        for exp in expirations:
            exp_dt = datetime.strptime(exp, '%Y-%m-%d').date()
            if exp_dt >= target_date:
                exp_date = exp
                break
        if not exp_date:
            exp_date = expirations[0]  # أقرب تاريخ

        # جلب عقود الخيارات لذلك التاريخ
        opt_chain = stock.option_chain(exp_date)
        if signal == "CALL":
            contracts = opt_chain.calls
        else:
            contracts = opt_chain.puts

        if contracts.empty:
            return "غير متوفر"

        # فلترة العقود: Open Interest عالي (أعلى 10%)، Spread ضيق
        current_price = stock.history(period="1d")['Close'].iloc[-1]
        contracts['strike_diff'] = abs(contracts['strike'] - current_price)
        # فلتر OI
        oi_threshold = contracts['openInterest'].quantile(0.7)  # أعلى 30%
        eligible = contracts[contracts['openInterest'] > oi_threshold]

        if eligible.empty:
            eligible = contracts

        # نختار العقد الأقرب للسعر الحالي (للدلتا التقريبية)
        best = eligible.nsmallest(1, 'strike_diff').iloc[0]

        # حساب Spread تقريبي (ask-bid)/ask
        spread = (best['ask'] - best['bid']) / best['ask'] if best['ask'] > 0 else 1
        if spread > 0.5:  # spread كبير جداً
            return "سبريد واسع"

        # بناء نص العقد المقترح
        delta_approx = 0.5 - (best['strike_diff'] / current_price)  # تقريب بسيط
        delta_approx = max(0.1, min(0.9, delta_approx))

        return f"Delta {delta_approx:.2f} – Exp {exp_date} (أيام: {days})"

    except Exception as e:
        logging.error(f"خطأ في جلب الخيارات لـ {ticker}: {e}")
        return "غير متوفر"

# ---------------------------
# 8. بناء رسالة التنبيه النهائية
# ---------------------------

def build_alert_message(ticker, signal, confidence, classification, details, spx_trend, option_text):
    """
    تنسيق الرسالة كما طلبت تماماً.
    """
    # ترجمة الاتجاه
    if signal == "CALL":
        direction = "Call"
        cross = "50 > 100" if details.get('sma50', 0) > details.get('sma100', 0) else "50 > 200"
        price_status = f"السعر فوق {details.get('sma200', 0):.2f}"
        event = "اختراق قمة 2H"
    else:
        direction = "Put"
        cross = "50 < 100" if details.get('sma50', 0) < details.get('sma100', 0) else "50 < 200"
        price_status = f"السعر تحت {details.get('sma200', 0):.2f}"
        event = "كسر قاع 2H"

    # توافق SPX
    if "uptrend" in spx_trend and signal == "CALL":
        spx_compat = "✅ متوافق"
    elif "downtrend" in spx_trend and signal == "PUT":
        spx_compat = "✅ متوافق"
    else:
        spx_compat = "⚠️ غير متوافق"

    message = f"""
🔔 {ticker}
الاتجاه: {direction}
التصنيف: {classification}
التقاطع: {cross} (إيجابي)
الحالة: {price_status}
الحدث: {event}
SPX: {spx_compat}
الثقة: {confidence}/100
العقد المقترح: {option_text}
"""
    return message

# ---------------------------
# 9. دالة المسح الرئيسية
# ---------------------------

def scan_all_stocks():
    """تمر على كل الأسهم وتفحصها وترسل التنبيهات"""
    alerts = []
    for ticker in TICKERS:
        logging.info(f"فحص {ticker}...")
        time.sleep(0.5)  # تأخير نصف ثانية لتجنب حظر Yahoo

        # 1. الفلتر الأساسي
        if not passes_basic_filters(ticker):
            continue

        # 2. تحليل السهم
        result = analyze_stock_2h(ticker)
        if not result or not result['signal']:
            continue

        signal = result['signal']
        confidence = result['confidence']
        details = result['details']

        # 3. تحليل SPX
        spx_15m = analyze_spx("15m")

        # 4. تصنيف الصفقة
        classification = classify_trade(signal, confidence, details)

        # 5. اختيار العقد
        option = get_option_contract(ticker, signal, classification)

        # 6. بناء الرسالة
        msg = build_alert_message(ticker, signal, confidence, classification, details, spx_15m, option)
        alerts.append(msg)

    return alerts

# ---------------------------
# 10. دوال بوت تليجرام
# ---------------------------

async def send_alerts(context):
    """إرسال التنبيهات إلى تليجرام"""
    alerts = scan_all_stocks()
    if not alerts:
        await context.bot.send_message(chat_id=CHAT_ID, text="لا توجد إشارات جديدة حالياً.")
    else:
        for alert in alerts:
            await context.bot.send_message(chat_id=CHAT_ID, text=alert, parse_mode='Markdown')
            await asyncio.sleep(1)  # تجنب سبام تليجرام

async def start(update, context):
    """معالج أمر /start"""
    await update.message.reply_text(
        "مرحباً! بوت توصيات شاكر 1.0 يعمل.\n"
        "سأرسل لك التنبيهات تلقائياً كل ساعة.\n"
        "لبدء مسح فوري أرسل /scan"
    )

async def scan_now(update, context):
    """معالج أمر /scan لمسح فوري"""
    await update.message.reply_text("جاري المسح، انتظر قليلاً...")
    alerts = scan_all_stocks()
    if not alerts:
        await update.message.reply_text("لا توجد إشارات جديدة.")
    else:
        for alert in alerts:
            await update.message.reply_text(alert, parse_mode='Markdown')
            await asyncio.sleep(1)

def main():
    """الدالة الرئيسية لتشغيل البوت"""
    # إنشاء تطبيق البوت
    application = Application.builder().token(TELEGRAM_TOKEN).build()

    # Explicitly create and assign JobQueue if it's not already set
    # The JobQueue constructor requires the Application instance
    if application.job_queue is None:
        application.job_queue = JobQueue(application=application)

    # إضافة الأوامر
    application.add_handler(CommandHandler("start", start))
    application.add_handler(CommandHandler("scan", scan_now))

    # جدولة المسح التلقائي كل ساعة
    job_queue = application.job_queue
    job_queue.run_repeating(send_alerts, interval=3600, first=10)  # كل ساعة، أول مرة بعد 10 ثواني

    # تشغيل البوت
    print("البوت يعمل الآن...\n")
    application.run_polling()

if __name__ == "__main__":
    main()
````

**Reasoning**:
The previous `TypeError` indicated that `Application.builder().job_queue()` required a `JobQueue` instance as an argument, and the direct assignment `application.job_queue = JobQueue(application=application)` failed because `JobQueue` constructor does not accept `application` directly. To correctly set up the `JobQueue`, I will explicitly create a `JobQueue` instance and pass it to the `Application.builder().job_queue()` method.



````python
#!/usr/bin/env python3
# -*- coding: utf-8 -*-

# بوت توصيات الأسهم - نسخة شاكر 1.0 (نسخة كاملة)
# آخر تحديث: فبراير 2026

import yfinance as yf
import pandas as pd
import numpy as np
import ta
import logging
import time
import schedule
from datetime import datetime, timedelta
from telegram import Bot, InlineKeyboardButton, InlineKeyboardMarkup
from telegram.ext import Application, CommandHandler, CallbackQueryHandler, JobQueue
import asyncio
import os

# ---------------------------
# 1. إعدادات البوت والتوكن
# ---------------------------
TELEGRAM_TOKEN = "8699821370:AAEQUSbLTgf7MmWqo5vV5LHPOz30wfqOfqw"      # ضع التوكن الخاص بك
CHAT_ID = "ID: 7854020427"                # ضع معرف المستخدم أو القناة

# تفعيل السجلات لمشاهدة الأخطاء
logging.basicConfig(level=logging.INFO, format='%(asctime)s - %(name)s - %(levelname)s - %(message)s')

# ---------------------------
# 2. قائمة الأسهم الكاملة (مدمجة)
# ---------------------------
TICKERS = [
    # --- أسهمك القديمة (من الصور) ---
    "A", "AA", "ABEV", "ABNB", "ABT", "ABVX", "ACGL", "ACM", "ACN",
    "ADBE", "ADI", "ADM", "ADP", "ADSK", "AEE", "AEG", "AEIS", "AEM",
    "AEP", "AER", "AES", "AFG", "AFL", "AFRM", "AG", "AGI", "AGNC",
    "AIG", "AIT", "AIZ", "AJG", "AKAM", "ALAB", "ALB", "ALC", "ALGN",
    "ALL", "ALLE", "ALLY", "ALNY", "ALSN", "AM", "AMCR", "AME", "AMH",
    "AMKR", "AMP", "AMRZ", "AMT", "AMX",

    # --- الشركات العملاقة (> 1 تريليون دولار) ---
    "NVDA", "AAPL", "GOOGL", "MSFT", "AMZN", "META", "TSM", "AVGO", "TSLA",
    "BRK.A", "WMT",

    # --- الشركات الكبرى (200 مليار - 1 تريليون) ---
    "LLY", "JPM", "XOM", "V", "JNJ", "MA", "MU", "ORCL", "COST", "AMD",
    "NFLX", "PLTR", "CSCO", "MRK", "TMO", "ASML", "ABBV", "HD", "CVX",
    "PG", "CAT", "KO", "PEP", "INTC", "CMCSA", "MCD", "TXN", "VZ", "AMGN"
]

# ---------------------------
# 3. دوال التحليل الأساسي (الفلاتر)
# ---------------------------

def get_stock_info(ticker):
    """جلب معلومات الشركة الأساسية"""
    try:
        stock = yf.Ticker(ticker)
        info = stock.info
        return info
    except Exception as e:
        logging.error(f"خطأ في جلب معلومات {ticker}: {e}")
        return None

def passes_basic_filters(ticker):
    """
    تطبيق الفلتر الأساسي:
    - Market Cap > 10B (ملاحظة: أغلب الشركات في القائمة أكبر من ذلك بكثير، لكن الفلتر يضمن ذلك)
    - Average Volume > 1M (تقريبي)
    - Spread < 5%
    """
    info = get_stock_info(ticker)
    if not info:
        return False

    # 1. القيمة السوقية
    mcap = info.get('marketCap', 0)
    if mcap < 10e9:
        logging.debug(f"{ticker} فشل فلتر Market Cap: {mcap}")
        return False

    # 2. متوسط الحجم (نأخذ آخر 20 يوم)
    try:
        stock = yf.Ticker(ticker)
        hist = stock.history(period="1mo", interval="1d")
        if hist.empty:
            return False
        avg_vol = hist['Volume'].rolling(20).mean().iloc[-1]
        if avg_vol < 1_000_000:   # أقل من مليون سهم يومياً
            logging.debug(f"{ticker} فشل فلتر الحجم: {avg_vol}")
            return False
    except:
        return False

    # 3. Spread (bid/ask)
    bid = info.get('bid', 0)
    ask = info.get('ask', 0)
    if bid and ask:
        spread = (ask - bid) / bid if bid != 0 else 1
        if spread > 0.05:   # أكبر من 5%
            logging.debug(f"{ticker} فشل فلتر Spread: {spread:.2%}")
            return False

    return True

# ---------------------------
# 4. تحليل SPX (مؤشر S&P 500)
# ---------------------------

def analyze_spx(timeframe="15m"):
    """تحليل مؤشر SPX على الإطار المطلوب"""
    try:
        spx = yf.Ticker("^GSPC")  # أو "SPY"
        hist = spx.history(period="5d", interval=timeframe)
        if len(hist) < 50:
            return "neutral"

        # متوسطات بسيطة لتحديد الاتجاه
        hist['SMA20'] = hist['Close'].rolling(20).mean()
        last = hist.iloc[-1]
        prev = hist.iloc[-2]

        if last['Close'] > last['SMA20'] and last['Close'] > prev['Close']:
            return "uptrend"
        elif last['Close'] < last['SMA20'] and last['Close'] < prev['Close']:
            return "downtrend"
        else:
            return "neutral"
    except:
        return "neutral"

# ---------------------------
# 5. تحليل السهم (فريم ساعتين)
# ---------------------------

def analyze_stock_2h(ticker):
    """
    تحليل فني كامل على فريم الساعتين
    يعيد: {'signal': 'CALL'/'PUT'/None, 'confidence': 0-100, 'details': {...}}
    """
    try:
        stock = yf.Ticker(ticker)
        hist = stock.history(period="1mo", interval="2h")  # آخر شهر، فريم ساعتين
        if len(hist) < 200:
            return None

        # حساب المتوسطات
        hist['SMA50'] = hist['Close'].rolling(50).mean()
        hist['SMA100'] = hist['Close'].rolling(100).mean()
        hist['SMA200'] = hist['Close'].rolling(200).mean()
        hist['Volume_SMA20'] = hist['Volume'].rolling(20).mean()

        last = hist.iloc[-1]
        current_price = last['Close']
        sma50 = last['SMA50']
        sma100 = last['SMA100']
        sma200 = last['SMA200']

        # قمم وقيعان آخر 3 شموع
        high_last_3 = hist['High'].iloc[-3:].max()
        low_last_3 = hist['Low'].iloc[-3:].min()

        # قمة/قاع اليوم (آخر 24 ساعة = 12 شمعة)
        high_today = hist['High'].iloc[-12:].max()
        low_today = hist['Low'].iloc[-12:].min()

        volume_ok = last['Volume'] > last['Volume_SMA20']

        # شروط CALL
        call_condition1 = (sma50 > sma100) or (sma50 > sma200)
        call_condition2 = current_price > sma200
        call_condition3 = current_price > high_last_3 or current_price > high_today
        call_condition4 = volume_ok

        # شروط PUT
        put_condition1 = (sma50 < sma100) or (sma50 < sma200)
        put_condition2 = current_price < sma200
        put_condition3 = current_price < low_last_3 or current_price < low_today
        put_condition4 = volume_ok

        signal = None
        confidence = 0
        details = {}

        if call_condition1 and call_condition2 and call_condition3 and call_condition4:
            signal = "CALL"
            # حساب الثقة: نضيف 25 لكل شرط متحقق
            confidence = 25 * sum([call_condition1, call_condition2, call_condition3, call_condition4])
            details = {
                "sma50": sma50,
                "sma100": sma100,
                "sma200": sma200,
                "price": current_price,
                "high_3": high_last_3,
                "volume_ratio": last['Volume'] / last['Volume_SMA20']
            }
        elif put_condition1 and put_condition2 and put_condition3 and put_condition4:
            signal = "PUT"
            confidence = 25 * sum([put_condition1, put_condition2, put_condition3, put_condition4])
            details = {
                "sma50": sma50,
                "sma100": sma100,
                "sma200": sma200,
                "price": current_price,
                "low_3": low_last_3,
                "volume_ratio": last['Volume'] / last['Volume_SMA20']
            }

        return {"signal": signal, "confidence": int(confidence), "details": details}

    except Exception as e:
        logging.error(f"خطأ في تحليل {ticker}: {e}")
        return None

# ---------------------------
# 6. تصنيف الصفقة (سكالب، قصيرة، سوينق)
# ---------------------------

def classify_trade(signal, confidence, details):
    """
    تصنيف حسب قوة الإشارة وترتيب المتوسطات
    """
    if not signal:
        return None

    sma50 = details.get('sma50')
    sma100 = details.get('sma100')
    sma200 = details.get('sma200')

    # ترتيب المتوسطات
    if signal == "CALL":
        if sma50 > sma100 > sma200:
            return "📈 سوينق 3-7 أيام"
        elif confidence >= 75:
            return "⏳ صفقة قصيرة 1-2 يوم"
        else:
            return "⚡ سكالب"
    else:  # PUT
        if sma50 < sma100 < sma200:
            return "📈 سوينق 3-7 أيام"
        elif confidence >= 75:
            return "⏳ صفقة قصيرة 1-2 يوم"
        else:
            return "⚡ سكالب"

# ---------------------------
# 7. اختيار العقد المناسب (بيانات الخيارات)
# ---------------------------

def get_option_contract(ticker, signal, classification):
    """
    البحث عن عقد خيار مناسب حسب التصنيف.
    """
    try:
        stock = yf.Ticker(ticker)
        # جلب تواريخ انتهاء الصلاحية المتاحة
        expirations = stock.options
        if not expirations:
            return "غير متوفر"

        # اختيار التاريخ المناسب حسب التصنيف
        today = datetime.now().date()
        if "سكالب" in classification:
            days = 7
        elif "قصيرة" in classification:
            days = 14
        else:  # سوينق
            days = 30

        target_date = today + timedelta(days=days)
        # البحث عن أقرب تاريخ انتهاء للعدد المطلوب من الأيام
        exp_date = None
        for exp in expirations:
            exp_dt = datetime.strptime(exp, '%Y-%m-%d').date()
            if exp_dt >= target_date:
                exp_date = exp
                break
        if not exp_date:
            exp_date = expirations[0]  # أقرب تاريخ

        # جلب عقود الخيارات لذلك التاريخ
        opt_chain = stock.option_chain(exp_date)
        if signal == "CALL":
            contracts = opt_chain.calls
        else:
            contracts = opt_chain.puts

        if contracts.empty:
            return "غير متوفر"

        # فلترة العقود: Open Interest عالي (أعلى 10%)، Spread ضيق
        current_price = stock.history(period="1d")['Close'].iloc[-1]
        contracts['strike_diff'] = abs(contracts['strike'] - current_price)
        # فلتر OI
        oi_threshold = contracts['openInterest'].quantile(0.7)  # أعلى 30%
        eligible = contracts[contracts['openInterest'] > oi_threshold]

        if eligible.empty:
            eligible = contracts

        # نختار العقد الأقرب للسعر الحالي (للدلتا التقريبية)
        best = eligible.nsmallest(1, 'strike_diff').iloc[0]

        # حساب Spread تقريبي (ask-bid)/ask
        spread = (best['ask'] - best['bid']) / best['ask'] if best['ask'] > 0 else 1
        if spread > 0.5:  # spread كبير جداً
            return "سبريد واسع"

        # بناء نص العقد المقترح
        delta_approx = 0.5 - (best['strike_diff'] / current_price)  # تقريب بسيط
        delta_approx = max(0.1, min(0.9, delta_approx))

        return f"Delta {delta_approx:.2f} – Exp {exp_date} (أيام: {days})"

    except Exception as e:
        logging.error(f"خطأ في جلب الخيارات لـ {ticker}: {e}")
        return "غير متوفر"

# ---------------------------
# 8. بناء رسالة التنبيه النهائية
# ---------------------------

def build_alert_message(ticker, signal, confidence, classification, details, spx_trend, option_text):
    """
    تنسيق الرسالة كما طلبت تماماً.
    """
    # ترجمة الاتجاه
    if signal == "CALL":
        direction = "Call"
        cross = "50 > 100" if details.get('sma50', 0) > details.get('sma100', 0) else "50 > 200"
        price_status = f"السعر فوق {details.get('sma200', 0):.2f}"
        event = "اختراق قمة 2H"
    else:
        direction = "Put"
        cross = "50 < 100" if details.get('sma50', 0) < details.get('sma100', 0) else "50 < 200"
        price_status = f"السعر تحت {details.get('sma200', 0):.2f}"
        event = "كسر قاع 2H"

    # توافق SPX
    if "uptrend" in spx_trend and signal == "CALL":
        spx_compat = "✅ متوافق"
    elif "downtrend" in spx_trend and signal == "PUT":
        spx_compat = "✅ متوافق"
    else:
        spx_compat = "⚠️ غير متوافق"

    message = f"""
🔔 {ticker}
الاتجاه: {direction}
التصنيف: {classification}
التقاطع: {cross} (إيجابي)
الحالة: {price_status}
الحدث: {event}
SPX: {spx_compat}
الثقة: {confidence}/100
العقد المقترح: {option_text}
"""
    return message

# ---------------------------
# 9. دالة المسح الرئيسية
# ---------------------------

def scan_all_stocks():
    """تمر على كل الأسهم وتفحصها وترسل التنبيهات"""
    alerts = []
    for ticker in TICKERS:
        logging.info(f"فحص {ticker}...")
        time.sleep(0.5)  # تأخير نصف ثانية لتجنب حظر Yahoo

        # 1. الفلتر الأساسي
        if not passes_basic_filters(ticker):
            continue

        # 2. تحليل السهم
        result = analyze_stock_2h(ticker)
        if not result or not result['signal']:
            continue

        signal = result['signal']
        confidence = result['confidence']
        details = result['details']

        # 3. تحليل SPX
        spx_15m = analyze_spx("15m")

        # 4. تصنيف الصفقة
        classification = classify_trade(signal, confidence, details)

        # 5. اختيار العقد
        option = get_option_contract(ticker, signal, classification)

        # 6. بناء الرسالة
        msg = build_alert_message(ticker, signal, confidence, classification, details, spx_15m, option)
        alerts.append(msg)

    return alerts

# ---------------------------
# 10. دوال بوت تليجرام
# ---------------------------

async def send_alerts(context):
    """إرسال التنبيهات إلى تليجرام"""
    alerts = scan_all_stocks()
    if not alerts:
        await context.bot.send_message(chat_id=CHAT_ID, text="لا توجد إشارات جديدة حالياً.")
    else:
        for alert in alerts:
            await context.bot.send_message(chat_id=CHAT_ID, text=alert, parse_mode='Markdown')
            await asyncio.sleep(1)  # تجنب سبام تليجرام

async def start(update, context):
    """معالج أمر /start"""
    await update.message.reply_text(
        "مرحباً! بوت توصيات شاكر 1.0 يعمل.\n"
        "سأرسل لك التنبيهات تلقائياً كل ساعة.\n"
        "لبدء مسح فوري أرسل /scan"
    )

async def scan_now(update, context):
    """معالج أمر /scan لمسح فوري"""
    await update.message.reply_text("جاري المسح، انتظر قليلاً...")
    alerts = scan_all_stocks()
    if not alerts:
        await update.message.reply_text("لا توجد إشارات جديدة.")
    else:
        for alert in alerts:
            await update.message.reply_text(alert, parse_mode='Markdown')
            await asyncio.sleep(1)

def main():
    """الدالة الرئيسية لتشغيل البوت"""
    # إنشاء تطبيق البوت
    # Create a JobQueue instance first
    job_queue_instance = JobQueue()

    # Pass the JobQueue instance to the ApplicationBuilder
    application = Application.builder().token(TELEGRAM_TOKEN).job_queue(job_queue_instance).build()

    # إضافة المعالجات للأوامر والرسائل
    application.add_handler(CommandHandler("start", start))
    application.add_handler(CommandHandler("scan", scan_now))

    # جدولة المسح التلقائي كل ساعة
    job_queue_instance.run_repeating(send_alerts, interval=3600, first=10)  # كل ساعة، أول مرة بعد 10 ثواني

    # تشغيل البوت
    print("البوت يعمل الآن...\n")
    application.run_polling()

if __name__ == "__main__":
    main()
````

````python
#!/usr/bin/env python3
# -*- coding: utf-8 -*-

# بوت توصيات الأسهم - نسخة شاكر 1.0 (نسخة كاملة)
# آخر تحديث: فبراير 2026

import yfinance as yf
import pandas as pd
import numpy as np
import ta
import logging
import time
import schedule
from datetime import datetime, timedelta
from telegram import Bot, InlineKeyboardButton, InlineKeyboardMarkup, Update
from telegram.ext import Application, CommandHandler, CallbackQueryHandler, ContextTypes, JobQueue, ChatMemberHandler # أضفنا ChatMemberHandler
import asyncio
import os

# ---------------------------
# 1. إعدادات البوت والتوكن
# ---------------------------
TELEGRAM_TOKEN = "8699821370:AAEQUSbLTgf7MmWqo5vV5LHPOz30wfqOfqw"      # ضع التوكن الخاص بك
CHAT_ID = "7854020427"                # ضع معرف المستخدم أو القناة

# تفعيل السجلات لمشاهدة الأخطاء
logging.basicConfig(level=logging.INFO, format='%(asctime)s - %(name)s - %(levelname)s - %(message)s')

# ---------------------------
# 2. قائمة الأسهم الكاملة (مدمجة)
# ---------------------------
TICKERS = [
    # --- أسهمك القديمة (من الصور) ---
    "A", "AA", "ABEV", "ABNB", "ABT", "ABVX", "ACGL", "ACM", "ACN",
    "ADBE", "ADI", "ADM", "ADP", "ADSK", "AEE", "AEG", "AEIS", "AEM",
    "AEP", "AER", "AES", "AFG", "AFL", "AFRM", "AG", "AGI", "AGNC",
    "AIG", "AIT", "AIZ", "AJG", "AKAM", "ALAB", "ALB", "ALC", "ALGN",
    "ALL", "ALLE", "ALLY", "ALNY", "ALSN", "AM", "AMCR", "AME", "AMH",
    "AMKR", "AMP", "AMRZ", "AMT", "AMX",

    # --- الشركات العملاقة (> 1 تريليون دولار) ---
    "NVDA", "AAPL", "GOOGL", "MSFT", "AMZN", "META", "TSM", "AVGO", "TSLA",
    "BRK.A", "WMT",

    # --- الشركات الكبرى (200 مليار - 1 تريليون) ---
    "LLY", "JPM", "XOM", "V", "JNJ", "MA", "MU", "ORCL", "COST", "AMD",
    "NFLX", "PLTR", "CSCO", "MRK", "TMO", "ASML", "ABBV", "HD", "CVX",
    "PG", "CAT", "KO", "PEP", "INTC", "CMCSA", "MCD", "TXN", "VZ", "AMGN"
]

# ---------------------------
# 3. دوال التحليل الأساسي (الفلاتر)
# ---------------------------

def get_stock_info(ticker):
    """جلب معلومات الشركة الأساسية"""
    try:
        stock = yf.Ticker(ticker)
        info = stock.info
        return info
    except Exception as e:
        logging.error(f"خطأ في جلب معلومات {ticker}: {e}")
        return None

def passes_basic_filters(ticker):
    """
    تطبيق الفلتر الأساسي:
    - Market Cap > 10B (ملاحظة: أغلب الشركات في القائمة أكبر من ذلك بكثير، لكن الفلتر يضمن ذلك)
    - Average Volume > 1M (تقريبي)
    - Spread < 5%
    """
    info = get_stock_info(ticker)
    if not info:
        return False

    # 1. القيمة السوقية
    mcap = info.get('marketCap', 0)
    if mcap < 10e9:
        print(f"{ticker} فشل فلتر Market Cap: {mcap}") # Changed from logging.debug
        return False

    # 2. متوسط الحجم (نأخذ آخر 20 يوم)
    try:
        stock = yf.Ticker(ticker)
        hist = stock.history(period="1mo", interval="1d")
        if hist.empty:
            return False
        avg_vol = hist['Volume'].rolling(20).mean().iloc[-1]
        if avg_vol < 1_000_000:   # أقل من مليون سهم يومياً
            print(f"{ticker} فشل فلتر الحجم: {avg_vol}") # Changed from logging.debug
            return False
    except:
        return False

    # 3. Spread (bid/ask)
    bid = info.get('bid', 0)
    ask = info.get('ask', 0)
    if bid and ask:
        spread = (ask - bid) / bid if bid != 0 else 1
        if spread > 0.05:   # أكبر من 5%
            print(f"{ticker} فشل فلتر Spread: {spread:.2%}") # Changed from logging.debug
            return False

    return True

# ---------------------------
# 4. تحليل SPX (مؤشر S&P 500)
# ---------------------------

def analyze_spx(timeframe="15m"):
    """تحليل مؤشر SPX على الإطار المطلوب"""
    try:
        spx = yf.Ticker("^GSPC")  # أو "SPY"
        hist = spx.history(period="5d", interval=timeframe)
        if len(hist) < 50:
            return "neutral"

        # متوسطات بسيطة لتحديد الاتجاه
        hist['SMA20'] = hist['Close'].rolling(20).mean()
        last = hist.iloc[-1]
        prev = hist.iloc[-2]

        if last['Close'] > last['SMA20'] and last['Close'] > prev['Close']:
            return "uptrend"
        elif last['Close'] < last['SMA20'] and last['Close'] < prev['Close']:
            return "downtrend"
        else:
            return "neutral"
    except:
        return "neutral"

# ---------------------------
# 5. تحليل السهم (فريم ساعتين)
# ---------------------------

def analyze_stock_2h(ticker):
    """
    تحليل فني كامل على فريم الساعتين
    يعيد: {'signal': 'CALL'/'PUT'/None, 'confidence': 0-100, 'details': {...}}
    """
    try:
        print(f"Analyzing {ticker} for 2h interval.") # Added print statement
        stock = yf.Ticker(ticker)
        hist = stock.history(period="1mo", interval="2h")  # آخر شهر، فريم ساعتين
        if len(hist) < 200:
            print(f"Not enough historical data for {ticker} in 2h interval.") # Added print statement
            return None

        # حساب المتوسطات
        hist['SMA50'] = hist['Close'].rolling(50).mean()
        hist['SMA100'] = hist['Close'].rolling(100).mean()
        hist['SMA200'] = hist['Close'].rolling(200).mean()
        hist['Volume_SMA20'] = hist['Volume'].rolling(20).mean()

        last = hist.iloc[-1]
        current_price = last['Close']
        sma50 = last['SMA50']
        sma100 = last['SMA100']
        sma200 = last['SMA200']

        # قمم وقيعان آخر 3 شموع
        high_last_3 = hist['High'].iloc[-3:].max()
        low_last_3 = hist['Low'].iloc[-3:].min()

        # قمة/قاع اليوم (آخر 24 ساعة = 12 شمعة)
        high_today = hist['High'].iloc[-12:].max()
        low_today = hist['Low'].iloc[-12:].min()

        volume_ok = last['Volume'] > last['Volume_SMA20']

        # شروط CALL
        call_condition1 = (sma50 > sma100) or (sma50 > sma200)
        call_condition2 = current_price > sma200
        call_condition3 = current_price > high_last_3 or current_price > high_today
        call_condition4 = volume_ok

        # شروط PUT
        put_condition1 = (sma50 < sma100) or (sma50 < sma200)
        put_condition2 = current_price < sma200
        put_condition3 = current_price < low_last_3 or current_price < low_today
        put_condition4 = volume_ok

        signal = None
        confidence = 0
        details = {}

        if call_condition1 and call_condition2 and call_condition3 and call_condition4:
            signal = "CALL"
            # حساب الثقة: نضيف 25 لكل شرط متحقق
            confidence = 25 * sum([call_condition1, call_condition2, call_condition3, call_condition4])
            details = {
                "sma50": sma50,
                "sma100": sma100,
                "sma200": sma200,
                "price": current_price,
                "high_3": high_last_3,
                "volume_ratio": last['Volume'] / last['Volume_SMA20']
            }
        elif put_condition1 and put_condition2 and put_condition3 and put_condition4:
            signal = "PUT"
            confidence = 25 * sum([put_condition1, put_condition2, put_condition3, put_condition4])
            details = {
                "sma50": sma50,
                "sma100": sma100,
                "sma200": sma200,
                "price": current_price,
                "low_3": low_last_3,
                "volume_ratio": last['Volume'] / last['Volume_SMA20']
            }

        if signal: # Added print statement
            print(f"Signal for {ticker}: {signal}, Confidence: {confidence}")
        return {"signal": signal, "confidence": int(confidence), "details": details}

    except Exception as e:
        logging.error(f"خطأ في تحليل {ticker}: {e}")
        return None

# ---------------------------
# 6. تصنيف الصفقة (سكالب، قصيرة، سوينق)
# ---------------------------

def classify_trade(signal, confidence, details):
    """
    تصنيف حسب قوة الإشارة وترتيب المتوسطات
    """
    if not signal:
        return None

    sma50 = details.get('sma50')
    sma100 = details.get('sma100')
    sma200 = details.get('sma200')

    # ترتيب المتوسطات
    if signal == "CALL":
        if sma50 > sma100 > sma200:
            return "📈 سوينق 3-7 أيام"
        elif confidence >= 75:
            return "⏳ صفقة قصيرة 1-2 يوم"
        else:
            return "⚡ سكالب"
    else:  # PUT
        if sma50 < sma100 < sma200:
            return "📈 سوينق 3-7 أيام"
        elif confidence >= 75:
            return "⏳ صفقة قصيرة 1-2 يوم"
        else:
            return "⚡ سكالب"

# ---------------------------
# 7. اختيار العقد المناسب (بيانات الخيارات)
# ---------------------------

def get_option_contract(ticker, signal, classification):
    """
    البحث عن عقد خيار مناسب حسب التصنيف.
    """
    try:
        stock = yf.Ticker(ticker)
        # جلب تواريخ انتهاء الصلاحية المتاحة
        expirations = stock.options
        if not expirations:
            return "غير متوفر"

        # اختيار التاريخ المناسب حسب التصنيف
        today = datetime.now().date()
        if "سكالب" in classification:
            days = 7
        elif "قصيرة" in classification:
            days = 14
        else:  # سوينق
            days = 30

        target_date = today + timedelta(days=days)
        # البحث عن أقرب تاريخ انتهاء للعدد المطلوب من الأيام
        exp_date = None
        for exp in expirations:
            exp_dt = datetime.strptime(exp, '%Y-%m-%d').date()
            if exp_dt >= target_date:
                exp_date = exp
                break
        if not exp_date:
            exp_date = expirations[0]  # أقرب تاريخ

        # جلب عقود الخيارات لذلك التاريخ
        opt_chain = stock.option_chain(exp_date)
        if signal == "CALL":
            contracts = opt_chain.calls
        else:
            contracts = opt_chain.puts

        if contracts.empty:
            return "غير متوفر"

        # فلترة العقود: Open Interest عالي (أعلى 10%)، Spread ضيق
        current_price = stock.history(period="1d")['Close'].iloc[-1]
        contracts['strike_diff'] = abs(contracts['strike'] - current_price)
        # فلتر OI
        oi_threshold = contracts['openInterest'].quantile(0.7)  # أعلى 30%
        eligible = contracts[contracts['openInterest'] > oi_threshold]

        if eligible.empty:
            eligible = contracts

        # نختار العقد الأقرب للسعر الحالي (للدلتا التقريبية)
        best = eligible.nsmallest(1, 'strike_diff').iloc[0]

        # حساب Spread تقريبي (ask-bid)/ask
        spread = (best['ask'] - best['bid']) / best['ask'] if best['ask'] > 0 else 1
        if spread > 0.05:   # أكبر من 5%
            return "سبريد واسع"

        # بناء نص العقد المقترح
        delta_approx = 0.5 - (best['strike_diff'] / current_price)  # تقريب بسيط
        delta_approx = max(0.1, min(0.9, delta_approx))

        return f"Delta {delta_approx:.2f} – Exp {exp_date} (أيام: {days})"

    except Exception as e:
        logging.error(f"خطأ في جلب الخيارات لـ {ticker}: {e}")
        return "غير متوفر"

# ---------------------------
# 8. بناء رسالة التنبيه النهائية
# ---------------------------

def build_alert_message(ticker, signal, confidence, classification, details, spx_trend, option_text):
    """
    تنسيق الرسالة كما طلبت تماماً.
    """
    # ترجمة الاتجاه
    if signal == "CALL":
        direction = "Call"
        cross = "50 > 100" if details.get('sma50', 0) > details.get('sma100', 0) else "50 > 200"
        price_status = f"السعر فوق {details.get('sma200', 0):.2f}"
        event = "اختراق قمة 2H"
    else:
        direction = "Put"
        cross = "50 < 100" if details.get('sma50', 0) < details.get('sma100', 0) else "50 < 200"
        price_status = f"السعر تحت {details.get('sma200', 0):.2f}"
        event = "كسر قاع 2H"

    # توافق SPX
    if "uptrend" in spx_trend and signal == "CALL":
        spx_compat = "✅ متوافق"
    elif "downtrend" in spx_trend and signal == "PUT":
        spx_compat = "✅ متوافق"
    else:
        spx_compat = "⚠️ غير متوافق"

    message = f"""
🔔 {ticker}
الاتجاه: {direction}
التصنيف: {classification}
التقاطع: {cross} (إيجابي)
الحالة: {price_status}
الحدث: {event}
SPX: {spx_compat}
الثقة: {confidence}/100
العقد المقترح: {option_text}
"""
    return message

# ---------------------------
# 9. دالة المسح الرئيسية
# ---------------------------

def scan_all_stocks():
    """تمر على كل الأسهم وتفحصها وترسل التنبيهات"""
    alerts = []
    print("Starting stock scan...") # Added print statement
    for ticker in TICKERS:
        logging.info(f"فحص {ticker}...")
        time.sleep(0.5)  # تأخير نصف ثانية لتجنب حظر Yahoo

        # 1. الفلتر الأساسي
        if not passes_basic_filters(ticker):
            print(f"Ticker {ticker} did not pass basic filters.") # Added print statement
            continue

        # 2. تحليل السهم
        result = analyze_stock_2h(ticker)
        if not result or not result['signal']:
            print(f"No signal found for {ticker} after 2h analysis.") # Added print statement
            continue

        signal = result['signal']
        confidence = result['confidence']
        details = result['details']

        # 3. تحليل SPX
        spx_15m = analyze_spx("15m")

        # 4. تصنيف الصفقة
        classification = classify_trade(signal, confidence, details)

        # 5. اختيار العقد
        option = get_option_contract(ticker, signal, classification)

        # 6. بناء الرسالة
        msg = build_alert_message(ticker, signal, confidence, classification, details, spx_15m, option)
        alerts.append(msg)
    print(f"Scan finished. Found {len(alerts)} alerts.") # Added print statement
    return alerts

# ---------------------------
# 10. دوال بوت تليجرام
# ---------------------------

async def send_alerts(context):
    """إرسال التنبيهات إلى تليجرام"""
    print("send_alerts function called.") # Added print statement
    alerts = scan_all_stocks()
    if not alerts:
        print("No alerts found to send.") # Added print statement
        await context.bot.send_message(chat_id=CHAT_ID, text="لا توجد إشارات جديدة حالياً.")
    else:
        print(f"Sending {len(alerts)} alerts.") # Added print statement
        for alert in alerts:
            await context.bot.send_message(chat_id=CHAT_ID, text=alert, parse_mode='Markdown')
            await asyncio.sleep(1)  # تجنب سبام تليجرام

# دالة للترحيب بالأعضاء الجدد وإرسال إخلاء المسؤولية
async def greet_chat_members(update: Update, context: ContextTypes.DEFAULT_TYPE) -> None:
    print(f"New chat member event: {update.chat_member}") # Added print
    if update.chat_member.new_chat_member.status == 'member':
        user = update.chat_member.new_chat_member.user
        chat_title = update.chat_member.chat.title or "هذه المجموعة"
        welcome_new_member_message = f"""
أهلاً وسهلاً بك {user.full_name} في {chat_title}! 👋

⚠️ **إخلاء مسؤولية:**
توصيات هذا البوت هي لأغراض تعليمية وتحليلية فقط، ولا تشكل نصيحة استثمارية بأي شكل من الأشكال. تداول العقود والمشتقات المالية ينطوي على مخاطر عالية وقد يؤدي إلى خسارة رأس المال. أنت وحدك المسؤول عن قراراتك الاستثمارية. يرجى استشارة مستشار مالي متخصص قبل اتخاذ أي قرارات استثمارية.
"""
        print(f"Sending welcome message to {user.full_name} in chat {update.chat_member.chat.id}") # Added print
        await context.bot.send_message(chat_id=update.chat_member.chat.id, text=welcome_new_member_message, parse_mode='Markdown')


async def start(update: Update, context: ContextTypes.DEFAULT_TYPE) -> None:
    """معالج أمر /start"""
    print(f"Start command received from {update.effective_user.id} in chat {update.effective_chat.id}") # Added print
    welcome_message = """
👋 مرحباً بك في بوت توصيات الأسهم!

هذا البوت يقدم توصيات تحليل فني للأسهم بناءً على استراتيجيات محددة.

⚠️ **إخلاء مسؤولية:**
توصيات هذا البوت هي لأغراض تعليمية وتحليلية فقط، ولا تشكل نصيحة استثمارية بأي شكل من الأشكال. تداول العقود والمشتقات المالية ينطوي على مخاطر عالية وقد يؤدي إلى خسارة رأس المال. أنت وحدك المسؤول عن قراراتك الاستثمارية. يرجى استشارة مستشار مالي متخصص قبل اتخاذ أي قرارات استثمارية.

اختر نوع التوصية التي ترغب في الحصول عليها:
"""

    keyboard = [
        [InlineKeyboardButton("📈 عقود CALL (شراء)", callback_data='recommend_call')],
        [InlineKeyboardButton("📉 عقود PUT (بيع)", callback_data='recommend_put')],
        [InlineKeyboardButton("🔄 مسح الآن", callback_data='scan_now_manual')]
    ]
    reply_markup = InlineKeyboardMarkup(keyboard)

    await update.message.reply_text(welcome_message, reply_markup=reply_markup, parse_mode='Markdown')

async def button_callback(update: Update, context: ContextTypes.DEFAULT_TYPE) -> None:
    query = update.callback_query
    print(f"Callback query received: {query.data} from {query.from_user.id}") # Added print
    await query.answer()  # Always answer callback queries

    if query.data == 'recommend_call':
        await query.edit_message_text(text="جاري البحث عن توصيات CALL...")
        # Logic to scan for CALL recommendations
        # For now, let's just send a placeholder message
        await asyncio.sleep(2) # Simulate work
        await query.edit_message_text(text="لم يتم العثور على توصيات CALL حالياً.") # Replace with actual results
    elif query.data == 'recommend_put':
        await query.edit_message_text(text="جاري البحث عن توصيات PUT...")
        # Logic to scan for PUT recommendations
        # For now, let's just send a placeholder message
        await asyncio.sleep(2) # Simulate work
        await query.edit_message_text(text="لم يتم العثور على توصيات PUT حالياً.") # Replace with actual results
    elif query.data == 'scan_now_manual':
        await query.edit_message_text(text="جاري المسح الفوري، انتظر قليلاً...")
        alerts = scan_all_stocks()
        if not alerts:
            await query.edit_message_text("لا توجد إشارات جديدة.")
        else:
            for alert in alerts:
                await query.message.reply_text(alert, parse_mode='Markdown')
                await asyncio.sleep(1)
            await query.edit_message_text(text="تم الانتهاء من المسح الفوري.")


async def scan_now(update, context):
    """معالج أمر /scan لمسح فوري"""
    print("scan_now command called.") # Added print statement
    await update.message.reply_text("جاري المسح، انتظر قليلاً...")
    alerts = scan_all_stocks()
    if not alerts:
        print("No alerts found for manual scan.") # Added print statement
        await update.message.reply_text("لا توجد إشارات جديدة.")
    else:
        print(f"Sending {len(alerts)} alerts for manual scan.") # Added print statement
        for alert in alerts:
            await update.message.reply_text(alert, parse_mode='Markdown')
            await asyncio.sleep(1)

def main():
    """الدالة الرئيسية لتشغيل البوت"""
    # إنشاء تطبيق البوت
    # Create a JobQueue instance first
    job_queue_instance = JobQueue()

    # Pass the JobQueue instance to the ApplicationBuilder
    application = Application.builder().token(TELEGRAM_TOKEN).job_queue(job_queue_instance).build()

    # إضافة المعالجات للأوامر والرسائل
    application.add_handler(CommandHandler("start", start))
    application.add_handler(CommandHandler("scan", scan_now))
    application.add_handler(CallbackQueryHandler(button_callback)) # إضافة معالج لأزرار الـ Inline Keyboard
    application.add_handler(ChatMemberHandler(greet_chat_members, ChatMemberHandler.CHAT_MEMBER)) # إضافة معالج للأعضاء الجدد

    # جدولة المسح التلقائي كل ساعة
    job_queue_instance.run_repeating(send_alerts, interval=3600, first=10)  # كل ساعة، أول مرة بعد 10 ثواني

    # تشغيل البوت
    print("البوت يعمل الآن... اذهب لتجربته!")
    application.run_polling()
    # application.idle() is not necessary after run_polling() as it's a blocking call

# استدعاء الدالة الرئيسية
if __name__ == '__main__':
    main()
````

````python
#!/usr/bin/env python3
# -*- coding: utf-8 -*-

# بوت توصيات الأسهم - نسخة شاكر 1.0 (نسخة كاملة)
# آخر تحديث: فبراير 2026

import yfinance as yf
import pandas as pd
import numpy as np
import ta
import logging
import time
import schedule
from datetime import datetime, timedelta
from telegram import Bot, InlineKeyboardButton, InlineKeyboardMarkup, Update
from telegram.ext import Application, CommandHandler, CallbackQueryHandler, ContextTypes, JobQueue, ChatMemberHandler # أضفنا ChatMemberHandler
import asyncio
import os

# ---------------------------
# 1. إعدادات البوت والتوكن
# ---------------------------
TELEGRAM_TOKEN = "8699821370:AAEQUSbLTgf7MmWqo5vV5LHPOz30wfqOfqw"      # ضع التوكن الخاص بك
CHAT_ID = "7854020427"                # ضع معرف المستخدم أو القناة

# تفعيل السجلات لمشاهدة الأخطاء
logging.basicConfig(level=logging.INFO, format='%(asctime)s - %(name)s - %(levelname)s - %(message)s')

# ---------------------------
# 2. قائمة الأسهم الكاملة (مدمجة)
# ---------------------------
TICKERS = [
    # --- أسهمك القديمة (من الصور) ---
    "A", "AA", "ABEV", "ABNB", "ABT", "ABVX", "ACGL", "ACM", "ACN",
    "ADBE", "ADI", "ADM", "ADP", "ADSK", "AEE", "AEG", "AEIS", "AEM",
    "AEP", "AER", "AES", "AFG", "AFL", "AFRM", "AG", "AGI", "AGNC",
    "AIG", "AIT", "AIZ", "AJG", "AKAM", "ALAB", "ALB", "ALC", "ALGN",
    "ALL", "ALLE", "ALLY", "ALNY", "ALSN", "AM", "AMCR", "AME", "AMH",
    "AMKR", "AMP", "AMRZ", "AMT", "AMX",

    # --- الشركات العملاقة (> 1 تريليون دولار) ---
    "NVDA", "AAPL", "GOOGL", "MSFT", "AMZN", "META", "TSM", "AVGO", "TSLA",
    "BRK.A", "WMT",

    # --- الشركات الكبرى (200 مليار - 1 تريليون) ---
    "LLY", "JPM", "XOM", "V", "JNJ", "MA", "MU", "ORCL", "COST", "AMD",
    "NFLX", "PLTR", "CSCO", "MRK", "TMO", "ASML", "ABBV", "HD", "CVX",
    "PG", "CAT", "KO", "PEP", "INTC", "CMCSA", "MCD", "TXN", "VZ", "AMGN"
]

# ---------------------------
# 3. دوال التحليل الأساسي (الفلاتر)
# ---------------------------

def get_stock_info(ticker):
    """جلب معلومات الشركة الأساسية"""
    try:
        stock = yf.Ticker(ticker)
        info = stock.info
        return info
    except Exception as e:
        logging.error(f"خطأ في جلب معلومات {ticker}: {e}")
        return None

def passes_basic_filters(ticker):
    """
    تطبيق الفلتر الأساسي:
    - Market Cap > 10B (ملاحظة: أغلب الشركات في القائمة أكبر من ذلك بكثير، لكن الفلتر يضمن ذلك)
    - Average Volume > 1M (تقريبي)
    - Spread < 5%
    """
    info = get_stock_info(ticker)
    if not info:
        return False

    # 1. القيمة السوقية
    mcap = info.get('marketCap', 0)
    if mcap < 10e9:
        print(f"{ticker} فشل فلتر Market Cap: {mcap}") # Changed from logging.debug
        return False

    # 2. متوسط الحجم (نأخذ آخر 20 يوم)
    try:
        stock = yf.Ticker(ticker)
        hist = stock.history(period="1mo", interval="1d")
        if hist.empty:
            return False
        avg_vol = hist['Volume'].rolling(20).mean().iloc[-1]
        if avg_vol < 1_000_000:   # أقل من مليون سهم يومياً
            print(f"{ticker} فشل فلتر الحجم: {avg_vol}") # Changed from logging.debug
            return False
    except:
        return False

    # 3. Spread (bid/ask)
    bid = info.get('bid', 0)
    ask = info.get('ask', 0)
    if bid and ask:
        spread = (ask - bid) / bid if bid != 0 else 1
        if spread > 0.05:   # أكبر من 5%
            print(f"{ticker} فشل فلتر Spread: {spread:.2%}") # Changed from logging.debug
            return False

    return True

# ---------------------------
# 4. تحليل SPX (مؤشر S&P 500)
# ---------------------------

def analyze_spx(timeframe="15m"):
    """تحليل مؤشر SPX على الإطار المطلوب"""
    try:
        spx = yf.Ticker("^GSPC")  # أو "SPY"
        hist = spx.history(period="5d", interval=timeframe)
        if len(hist) < 50:
            return "neutral"

        # متوسطات بسيطة لتحديد الاتجاه
        hist['SMA20'] = hist['Close'].rolling(20).mean()
        last = hist.iloc[-1]
        prev = hist.iloc[-2]

        if last['Close'] > last['SMA20'] and last['Close'] > prev['Close']:
            return "uptrend"
        elif last['Close'] < last['SMA20'] and last['Close'] < prev['Close']:
            return "downtrend"
        else:
            return "neutral"
    except:
        return "neutral"

# ---------------------------
# 5. تحليل السهم (فريم ساعتين)
# ---------------------------

def analyze_stock_2h(ticker):
    """
    تحليل فني كامل على فريم الساعتين
    يعيد: {'signal': 'CALL'/'PUT'/None, 'confidence': 0-100, 'details': {...}}
    """
    try:
        print(f"Analyzing {ticker} for 2h interval.") # Added print statement
        stock = yf.Ticker(ticker)
        hist = stock.history(period="1mo", interval="2h")  # آخر شهر، فريم ساعتين
        if len(hist) < 200:
            print(f"Not enough historical data for {ticker} in 2h interval.") # Added print statement
            return None

        # حساب المتوسطات
        hist['SMA50'] = hist['Close'].rolling(50).mean()
        hist['SMA100'] = hist['Close'].rolling(100).mean()
        hist['SMA200'] = hist['Close'].rolling(200).mean()
        hist['Volume_SMA20'] = hist['Volume'].rolling(20).mean()

        last = hist.iloc[-1]
        current_price = last['Close']
        sma50 = last['SMA50']
        sma100 = last['SMA100']
        sma200 = last['SMA200']

        # قمم وقيعان آخر 3 شموع
        high_last_3 = hist['High'].iloc[-3:].max()
        low_last_3 = hist['Low'].iloc[-3:].min()

        # قمة/قاع اليوم (آخر 24 ساعة = 12 شمعة)
        high_today = hist['High'].iloc[-12:].max()
        low_today = hist['Low'].iloc[-12:].min()

        volume_ok = last['Volume'] > last['Volume_SMA20']

        # شروط CALL
        call_condition1 = (sma50 > sma100) or (sma50 > sma200)
        call_condition2 = current_price > sma200
        call_condition3 = current_price > high_last_3 or current_price > high_today
        call_condition4 = volume_ok

        # شروط PUT
        put_condition1 = (sma50 < sma100) or (sma50 < sma200)
        put_condition2 = current_price < sma200
        put_condition3 = current_price < low_last_3 or current_price < low_today
        put_condition4 = volume_ok

        signal = None
        confidence = 0
        details = {}

        if call_condition1 and call_condition2 and call_condition3 and call_condition4:
            signal = "CALL"
            # حساب الثقة: نضيف 25 لكل شرط متحقق
            confidence = 25 * sum([call_condition1, call_condition2, call_condition3, call_condition4])
            details = {
                "sma50": sma50,
                "sma100": sma100,
                "sma200": sma200,
                "price": current_price,
                "high_3": high_last_3,
                "volume_ratio": last['Volume'] / last['Volume_SMA20']
            }
        elif put_condition1 and put_condition2 and put_condition3 and put_condition4:
            signal = "PUT"
            confidence = 25 * sum([put_condition1, put_condition2, put_condition3, put_condition4])
            details = {
                "sma50": sma50,
                "sma100": sma100,
                "sma200": sma200,
                "price": current_price,
                "low_3": low_last_3,
                "volume_ratio": last['Volume'] / last['Volume_SMA20']
            }

        if signal: # Added print statement
            print(f"Signal for {ticker}: {signal}, Confidence: {confidence}")
        return {"signal": signal, "confidence": int(confidence), "details": details}

    except Exception as e:
        logging.error(f"خطأ في تحليل {ticker}: {e}")
        return None

# ---------------------------
# 6. تصنيف الصفقة (سكالب، قصيرة، سوينق)
# ---------------------------

def classify_trade(signal, confidence, details):
    """
    تصنيف حسب قوة الإشارة وترتيب المتوسطات
    """
    if not signal:
        return None

    sma50 = details.get('sma50')
    sma100 = details.get('sma100')
    sma200 = details.get('sma200')

    # ترتيب المتوسطات
    if signal == "CALL":
        if sma50 > sma100 > sma200:
            return "📈 سوينق 3-7 أيام"
        elif confidence >= 75:
            return "⏳ صفقة قصيرة 1-2 يوم"
        else:
            return "⚡ سكالب"
    else:  # PUT
        if sma50 < sma100 < sma200:
            return "📈 سوينق 3-7 أيام"
        elif confidence >= 75:
            return "⏳ صفقة قصيرة 1-2 يوم"
        else:
            return "⚡ سكالب"

# ---------------------------
# 7. اختيار العقد المناسب (بيانات الخيارات)
# ---------------------------

def get_option_contract(ticker, signal, classification):
    """
    البحث عن عقد خيار مناسب حسب التصنيف.
    """
    try:
        stock = yf.Ticker(ticker)
        # جلب تواريخ انتهاء الصلاحية المتاحة
        expirations = stock.options
        if not expirations:
            return "غير متوفر"

        # اختيار التاريخ المناسب حسب التصنيف
        today = datetime.now().date()
        if "سكالب" in classification:
            days = 7
        elif "قصيرة" in classification:
            days = 14
        else:  # سوينق
            days = 30

        target_date = today + timedelta(days=days)
        # البحث عن أقرب تاريخ انتهاء للعدد المطلوب من الأيام
        exp_date = None
        for exp in expirations:
            exp_dt = datetime.strptime(exp, '%Y-%m-%d').date()
            if exp_dt >= target_date:
                exp_date = exp
                break
        if not exp_date:
            exp_date = expirations[0]  # أقرب تاريخ

        # جلب عقود الخيارات لذلك التاريخ
        opt_chain = stock.option_chain(exp_date)
        if signal == "CALL":
            contracts = opt_chain.calls
        else:
            contracts = opt_chain.puts

        if contracts.empty:
            return "غير متوفر"

        # فلترة العقود: Open Interest عالي (أعلى 10%)، Spread ضيق
        current_price = stock.history(period="1d")['Close'].iloc[-1]
        contracts['strike_diff'] = abs(contracts['strike'] - current_price)
        # فلتر OI
        oi_threshold = contracts['openInterest'].quantile(0.7)  # أعلى 30%
        eligible = contracts[contracts['openInterest'] > oi_threshold]

        if eligible.empty:
            eligible = contracts

        # نختار العقد الأقرب للسعر الحالي (للدلتا التقريبية)
        best = eligible.nsmallest(1, 'strike_diff').iloc[0]

        # حساب Spread تقريبي (ask-bid)/ask
        spread = (best['ask'] - best['bid']) / best['ask'] if best['ask'] > 0 else 1
        if spread > 0.05:   # أكبر من 5%
            return "سبريد واسع"

        # بناء نص العقد المقترح
        delta_approx = 0.5 - (best['strike_diff'] / current_price)  # تقريب بسيط
        delta_approx = max(0.1, min(0.9, delta_approx))

        return f"Delta {delta_approx:.2f} – Exp {exp_date} (أيام: {days})"

    except Exception as e:
        logging.error(f"خطأ في جلب الخيارات لـ {ticker}: {e}")
        return "غير متوفر"

# ---------------------------
# 8. بناء رسالة التنبيه النهائية
# ---------------------------

def build_alert_message(ticker, signal, confidence, classification, details, spx_trend, option_text):
    """
    تنسيق الرسالة كما طلبت تماماً.
    """
    # ترجمة الاتجاه
    if signal == "CALL":
        direction = "Call"
        cross = "50 > 100" if details.get('sma50', 0) > details.get('sma100', 0) else "50 > 200"
        price_status = f"السعر فوق {details.get('sma200', 0):.2f}"
        event = "اختراق قمة 2H"
    else:
        direction = "Put"
        cross = "50 < 100" if details.get('sma50', 0) < details.get('sma100', 0) else "50 < 200"
        price_status = f"السعر تحت {details.get('sma200', 0):.2f}"
        event = "كسر قاع 2H"

    # توافق SPX
    if "uptrend" in spx_trend and signal == "CALL":
        spx_compat = "✅ متوافق"
    elif "downtrend" in spx_trend and signal == "PUT":
        spx_compat = "✅ متوافق"
    else:
        spx_compat = "⚠️ غير متوافق"

    message = f"""
🔔 {ticker}
الاتجاه: {direction}
التصنيف: {classification}
التقاطع: {cross} (إيجابي)
الحالة: {price_status}
الحدث: {event}
SPX: {spx_compat}
الثقة: {confidence}/100
العقد المقترح: {option_text}
"""
    return message

# ---------------------------
# 9. دالة المسح الرئيسية
# ---------------------------

def scan_all_stocks():
    """تمر على كل الأسهم وتفحصها وترسل التنبيهات"""
    alerts = []
    print("Starting stock scan...") # Added print statement
    for ticker in TICKERS:
        logging.info(f"فحص {ticker}...")
        time.sleep(0.5)  # تأخير نصف ثانية لتجنب حظر Yahoo

        # 1. الفلتر الأساسي
        if not passes_basic_filters(ticker):
            print(f"Ticker {ticker} did not pass basic filters.") # Added print statement
            continue

        # 2. تحليل السهم
        result = analyze_stock_2h(ticker)
        if not result or not result['signal']:
            print(f"No signal found for {ticker} after 2h analysis.") # Added print statement
            continue

        signal = result['signal']
        confidence = result['confidence']
        details = result['details']

        # 3. تحليل SPX
        spx_15m = analyze_spx("15m")

        # 4. تصنيف الصفقة
        classification = classify_trade(signal, confidence, details)

        # 5. اختيار العقد
        option = get_option_contract(ticker, signal, classification)

        # 6. بناء الرسالة
        msg = build_alert_message(ticker, signal, confidence, classification, details, spx_15m, option)
        alerts.append(msg)
    print(f"Scan finished. Found {len(alerts)} alerts.") # Added print statement
    return alerts

# ---------------------------
# 10. دوال بوت تليجرام
# ---------------------------

async def send_alerts(context):
    """إرسال التنبيهات إلى تليجرام"""
    print("send_alerts function called.") # Added print statement
    alerts = scan_all_stocks()
    if not alerts:
        print("No alerts found to send.") # Added print statement
        await context.bot.send_message(chat_id=CHAT_ID, text="لا توجد إشارات جديدة حالياً.")
    else:
        print(f"Sending {len(alerts)} alerts.") # Added print statement
        for alert in alerts:
            await context.bot.send_message(chat_id=CHAT_ID, text=alert, parse_mode='Markdown')
            await asyncio.sleep(1)  # تجنب سبام تليجرام

# دالة للترحيب بالأعضاء الجدد وإرسال إخلاء المسؤولية
async def greet_chat_members(update: Update, context: ContextTypes.DEFAULT_TYPE) -> None:
    print(f"New chat member event: {update.chat_member}") # Added print
    if update.chat_member.new_chat_member.status == 'member':
        user = update.chat_member.new_chat_member.user
        chat_title = update.chat_member.chat.title or "هذه المجموعة"
        welcome_new_member_message = f"""
أهلاً وسهلاً بك {user.full_name} في {chat_title}! 👋

⚠️ **إخلاء مسؤولية:**
توصيات هذا البوت هي لأغراض تعليمية وتحليلية فقط، ولا تشكل نصيحة استثمارية بأي شكل من الأشكال. تداول العقود والمشتقات المالية ينطوي على مخاطر عالية وقد يؤدي إلى خسارة رأس المال. أنت وحدك المسؤول عن قراراتك الاستثمارية. يرجى استشارة مستشار مالي متخصص قبل اتخاذ أي قرارات استثمارية.
"""
        print(f"Sending welcome message to {user.full_name} in chat {update.chat_member.chat.id}") # Added print
        await context.bot.send_message(chat_id=update.chat_member.chat.id, text=welcome_new_member_message, parse_mode='Markdown')


async def start(update: Update, context: ContextTypes.DEFAULT_TYPE) -> None:
    """معالج أمر /start"""
    print(f"Start command received from {update.effective_user.id} in chat {update.effective_chat.id}") # Added print
    welcome_message = """
👋 مرحباً بك في بوت توصيات الأسهم!

هذا البوت يقدم توصيات تحليل فني للأسهم بناءً على استراتيجيات محددة.

⚠️ **إخلاء مسؤولية:**
توصيات هذا البوت هي لأغراض تعليمية وتحليلية فقط، ولا تشكل نصيحة استثمارية بأي شكل من الأشكال. تداول العقود والمشتقات المالية ينطوي على مخاطر عالية وقد يؤدي إلى خسارة رأس المال. أنت وحدك المسؤول عن قراراتك الاستثمارية. يرجى استشارة مستشار مالي متخصص قبل اتخاذ أي قرارات استثمارية.

اختر نوع التوصية التي ترغب في الحصول عليها:
"""

    keyboard = [
        [InlineKeyboardButton("📈 عقود CALL (شراء)", callback_data='recommend_call')],
        [InlineKeyboardButton("📉 عقود PUT (بيع)", callback_data='recommend_put')],
        [InlineKeyboardButton("🔄 مسح الآن", callback_data='scan_now_manual')]
    ]
    reply_markup = InlineKeyboardMarkup(keyboard)

    await update.message.reply_text(welcome_message, reply_markup=reply_markup, parse_mode='Markdown')

async def button_callback(update: Update, context: ContextTypes.DEFAULT_TYPE) -> None:
    query = update.callback_query
    print(f"Callback query received: {query.data} from {query.from_user.id}") # Added print
    await query.answer()  # Always answer callback queries

    if query.data == 'recommend_call':
        await query.edit_message_text(text="جاري البحث عن توصيات CALL...")
        # Logic to scan for CALL recommendations
        # For now, let's just send a placeholder message
        await asyncio.sleep(2) # Simulate work
        await query.edit_message_text(text="لم يتم العثور على توصيات CALL حالياً.") # Replace with actual results
    elif query.data == 'recommend_put':
        await query.edit_message_text(text="جاري البحث عن توصيات PUT...")
        # Logic to scan for PUT recommendations
        # For now, let's just send a placeholder message
        await asyncio.sleep(2) # Simulate work
        await query.edit_message_text(text="لم يتم العثور على توصيات PUT حالياً.") # Replace with actual results
    elif query.data == 'scan_now_manual':
        await query.edit_message_text(text="جاري المسح الفوري، انتظر قليلاً...")
        alerts = scan_all_stocks()
        if not alerts:
            await query.edit_message_text("لا توجد إشارات جديدة.")
        else:
            for alert in alerts:
                await query.message.reply_text(alert, parse_mode='Markdown')
                await asyncio.sleep(1)
            await query.edit_message_text(text="تم الانتهاء من المسح الفوري.")


async def scan_now(update, context):
    """معالج أمر /scan لمسح فوري"""
    print("scan_now command called.") # Added print statement
    await update.message.reply_text("جاري المسح، انتظر قليلاً...")
    alerts = scan_all_stocks()
    if not alerts:
        print("No alerts found for manual scan.") # Added print statement
        await update.message.reply_text("لا توجد إشارات جديدة.")
    else:
        print(f"Sending {len(alerts)} alerts for manual scan.") # Added print statement
        for alert in alerts:
            await update.message.reply_text(alert, parse_mode='Markdown')
            await asyncio.sleep(1)

def main():
    """الدالة الرئيسية لتشغيل البوت"""
    # إنشاء تطبيق البوت
    # Create a JobQueue instance first
    job_queue_instance = JobQueue()

    # Pass the JobQueue instance to the ApplicationBuilder
    application = Application.builder().token(TELEGRAM_TOKEN).job_queue(job_queue_instance).build()

    # إضافة المعالجات للأوامر والرسائل
    application.add_handler(CommandHandler("start", start))
    application.add_handler(CommandHandler("scan", scan_now))
    application.add_handler(CallbackQueryHandler(button_callback)) # إضافة معالج لأزرار الـ Inline Keyboard
    application.add_handler(ChatMemberHandler(greet_chat_members, ChatMemberHandler.CHAT_MEMBER)) # إضافة معالج للأعضاء الجدد

    # جدولة المسح التلقائي كل ساعة
    job_queue_instance.run_repeating(send_alerts, interval=3600, first=10)  # كل ساعة، أول مرة بعد 10 ثواني

    # تشغيل البوت
    print("البوت يعمل الآن... اذهب لتجربته!")
    application.run_polling()
    # application.idle() is not necessary after run_polling() as it's a blocking call

# استدعاء الدالة الرئيسية
if __name__ == '__main__':
    main()
````

````python
#!/usr/bin/env python3
# -*- coding: utf-8 -*-

# بوت توصيات الأسهم - نسخة شاكر 1.0 (نسخة كاملة)
# آخر تحديث: فبراير 2026

import yfinance as yf
import pandas as pd
import numpy as np
import ta
import logging
import time
import schedule
from datetime import datetime, timedelta
from telegram import Bot, InlineKeyboardButton, InlineKeyboardMarkup, Update
from telegram.ext import Application, CommandHandler, CallbackQueryHandler, ContextTypes, JobQueue, ChatMemberHandler # أضفنا ChatMemberHandler
import asyncio
import os

# ---------------------------
# 1. إعدادات البوت والتوكن
# ---------------------------
TELEGRAM_TOKEN = "8699821370:AAEQUSbLTgf7MmWqo5vV5LHPOz30wfqOfqw"      # ضع التوكن الخاص بك
CHAT_ID = "7854020427"                # ضع معرف المستخدم أو القناة

# تفعيل السجلات لمشاهدة الأخطاء
logging.basicConfig(level=logging.INFO, format='%(asctime)s - %(name)s - %(levelname)s - %(message)s')

# ---------------------------
# 2. قائمة الأسهم الكاملة (مدمجة)
# ---------------------------
TICKERS = [
    # --- أسهمك القديمة (من الصور) ---
    "A", "AA", "ABEV", "ABNB", "ABT", "ABVX", "ACGL", "ACM", "ACN",
    "ADBE", "ADI", "ADM", "ADP", "ADSK", "AEE", "AEG", "AEIS", "AEM",
    "AEP", "AER", "AES", "AFG", "AFL", "AFRM", "AG", "AGI", "AGNC",
    "AIG", "AIT", "AIZ", "AJG", "AKAM", "ALAB", "ALB", "ALC", "ALGN",
    "ALL", "ALLE", "ALLY", "ALNY", "ALSN", "AM", "AMCR", "AME", "AMH",
    "AMKR", "AMP", "AMRZ", "AMT", "AMX",

    # --- الشركات العملاقة (> 1 تريليون دولار) ---
    "NVDA", "AAPL", "GOOGL", "MSFT", "AMZN", "META", "TSM", "AVGO", "TSLA",
    "BRK.A", "WMT",

    # --- الشركات الكبرى (200 مليار - 1 تريليون) ---
    "LLY", "JPM", "XOM", "V", "JNJ", "MA", "MU", "ORCL", "COST", "AMD",
    "NFLX", "PLTR", "CSCO", "MRK", "TMO", "ASML", "ABBV", "HD", "CVX",
    "PG", "CAT", "KO", "PEP", "INTC", "CMCSA", "MCD", "TXN", "VZ", "AMGN"
]

# ---------------------------
# 3. دوال التحليل الأساسي (الفلاتر)
# ---------------------------

def get_stock_info(ticker):
    """جلب معلومات الشركة الأساسية"""
    try:
        stock = yf.Ticker(ticker)
        info = stock.info
        return info
    except Exception as e:
        logging.error(f"خطأ في جلب معلومات {ticker}: {e}")
        return None

def passes_basic_filters(ticker):
    """
    تطبيق الفلتر الأساسي:
    - Market Cap > 10B (ملاحظة: أغلب الشركات في القائمة أكبر من ذلك بكثير، لكن الفلتر يضمن ذلك)
    - Average Volume > 1M (تقريبي)
    - Spread < 5%
    """
    info = get_stock_info(ticker)
    if not info:
        return False

    # 1. القيمة السوقية
    mcap = info.get('marketCap', 0)
    if mcap < 10e9:
        print(f"{ticker} فشل فلتر Market Cap: {mcap}") # Changed from logging.debug
        return False

    # 2. متوسط الحجم (نأخذ آخر 20 يوم)
    try:
        stock = yf.Ticker(ticker)
        hist = stock.history(period="1mo", interval="1d")
        if hist.empty:
            return False
        avg_vol = hist['Volume'].rolling(20).mean().iloc[-1]
        if avg_vol < 1_000_000:   # أقل من مليون سهم يومياً
            print(f"{ticker} فشل فلتر الحجم: {avg_vol}") # Changed from logging.debug
            return False
    except:
        return False

    # 3. Spread (bid/ask)
    bid = info.get('bid', 0)
    ask = info.get('ask', 0)
    if bid and ask:
        spread = (ask - bid) / bid if bid != 0 else 1
        if spread > 0.05:   # أكبر من 5%
            print(f"{ticker} فشل فلتر Spread: {spread:.2%}") # Changed from logging.debug
            return False

    return True

# ---------------------------
# 4. تحليل SPX (مؤشر S&P 500)
# ---------------------------

def analyze_spx(timeframe="15m"):
    """تحليل مؤشر SPX على الإطار المطلوب"""
    try:
        spx = yf.Ticker("^GSPC")  # أو "SPY"
        hist = spx.history(period="5d", interval=timeframe)
        if len(hist) < 50:
            return "neutral"

        # متوسطات بسيطة لتحديد الاتجاه
        hist['SMA20'] = hist['Close'].rolling(20).mean()
        last = hist.iloc[-1]
        prev = hist.iloc[-2]

        if last['Close'] > last['SMA20'] and last['Close'] > prev['Close']:
            return "uptrend"
        elif last['Close'] < last['SMA20'] and last['Close'] < prev['Close']:
            return "downtrend"
        else:
            return "neutral"
    except:
        return "neutral"

# ---------------------------
# 5. تحليل السهم (فريم ساعتين)
# ---------------------------

def analyze_stock_2h(ticker):
    """
    تحليل فني كامل على فريم الساعتين
    يعيد: {'signal': 'CALL'/'PUT'/None, 'confidence': 0-100, 'details': {...}}
    """
    try:
        print(f"Analyzing {ticker} for 2h interval.") # Added print statement
        stock = yf.Ticker(ticker)
        hist = stock.history(period="1mo", interval="2h")  # آخر شهر، فريم ساعتين
        if len(hist) < 200:
            print(f"Not enough historical data for {ticker} in 2h interval.") # Added print statement
            return None

        # حساب المتوسطات
        hist['SMA50'] = hist['Close'].rolling(50).mean()
        hist['SMA100'] = hist['Close'].rolling(100).mean()
        hist['SMA200'] = hist['Close'].rolling(200).mean()
        hist['Volume_SMA20'] = hist['Volume'].rolling(20).mean()

        last = hist.iloc[-1]
        current_price = last['Close']
        sma50 = last['SMA50']
        sma100 = last['SMA100']
        sma200 = last['SMA200']

        # قمم وقيعان آخر 3 شموع
        high_last_3 = hist['High'].iloc[-3:].max()
        low_last_3 = hist['Low'].iloc[-3:].min()

        # قمة/قاع اليوم (آخر 24 ساعة = 12 شمعة)
        high_today = hist['High'].iloc[-12:].max()
        low_today = hist['Low'].iloc[-12:].min()

        volume_ok = last['Volume'] > last['Volume_SMA20']

        # شروط CALL
        call_condition1 = (sma50 > sma100) or (sma50 > sma200)
        call_condition2 = current_price > sma200
        call_condition3 = current_price > high_last_3 or current_price > high_today
        call_condition4 = volume_ok

        # شروط PUT
        put_condition1 = (sma50 < sma100) or (sma50 < sma200)
        put_condition2 = current_price < sma200
        put_condition3 = current_price < low_last_3 or current_price < low_today
        put_condition4 = volume_ok

        signal = None
        confidence = 0
        details = {}

        if call_condition1 and call_condition2 and call_condition3 and call_condition4:
            signal = "CALL"
            # حساب الثقة: نضيف 25 لكل شرط متحقق
            confidence = 25 * sum([call_condition1, call_condition2, call_condition3, call_condition4])
            details = {
                "sma50": sma50,
                "sma100": sma100,
                "sma200": sma200,
                "price": current_price,
                "high_3": high_last_3,
                "volume_ratio": last['Volume'] / last['Volume_SMA20']
            }
        elif put_condition1 and put_condition2 and put_condition3 and put_condition4:
            signal = "PUT"
            confidence = 25 * sum([put_condition1, put_condition2, put_condition3, put_condition4])
            details = {
                "sma50": sma50,
                "sma100": sma100,
                "sma200": sma200,
                "price": current_price,
                "low_3": low_last_3,
                "volume_ratio": last['Volume'] / last['Volume_SMA20']
            }

        if signal: # Added print statement
            print(f"Signal for {ticker}: {signal}, Confidence: {confidence}")
        return {"signal": signal, "confidence": int(confidence), "details": details}

    except Exception as e:
        logging.error(f"خطأ في تحليل {ticker}: {e}")
        return None

# ---------------------------
# 6. تصنيف الصفقة (سكالب، قصيرة، سوينق)
# ---------------------------

def classify_trade(signal, confidence, details):
    """
    تصنيف حسب قوة الإشارة وترتيب المتوسطات
    """
    if not signal:
        return None

    sma50 = details.get('sma50')
    sma100 = details.get('sma100')
    sma200 = details.get('sma200')

    # ترتيب المتوسطات
    if signal == "CALL":
        if sma50 > sma100 > sma200:
            return "📈 سوينق 3-7 أيام"
        elif confidence >= 75:
            return "⏳ صفقة قصيرة 1-2 يوم"
        else:
            return "⚡ سكالب"
    else:  # PUT
        if sma50 < sma100 < sma200:
            return "📈 سوينق 3-7 أيام"
        elif confidence >= 75:
            return "⏳ صفقة قصيرة 1-2 يوم"
        else:
            return "⚡ سكالب"

# ---------------------------
# 7. اختيار العقد المناسب (بيانات الخيارات)
# ---------------------------

def get_option_contract(ticker, signal, classification):
    """
    البحث عن عقد خيار مناسب حسب التصنيف.
    """
    try:
        stock = yf.Ticker(ticker)
        # جلب تواريخ انتهاء الصلاحية المتاحة
        expirations = stock.options
        if not expirations:
            return "غير متوفر"

        # اختيار التاريخ المناسب حسب التصنيف
        today = datetime.now().date()
        if "سكالب" in classification:
            days = 7
        elif "قصيرة" in classification:
            days = 14
        else:  # سوينق
            days = 30

        target_date = today + timedelta(days=days)
        # البحث عن أقرب تاريخ انتهاء للعدد المطلوب من الأيام
        exp_date = None
        for exp in expirations:
            exp_dt = datetime.strptime(exp, '%Y-%m-%d').date()
            if exp_dt >= target_date:
                exp_date = exp
                break
        if not exp_date:
            exp_date = expirations[0]  # أقرب تاريخ

        # جلب عقود الخيارات لذلك التاريخ
        opt_chain = stock.option_chain(exp_date)
        if signal == "CALL":
            contracts = opt_chain.calls
        else:
            contracts = opt_chain.puts

        if contracts.empty:
            return "غير متوفر"

        # فلترة العقود: Open Interest عالي (أعلى 10%)، Spread ضيق
        current_price = stock.history(period="1d")['Close'].iloc[-1]
        contracts['strike_diff'] = abs(contracts['strike'] - current_price)
        # فلتر OI
        oi_threshold = contracts['openInterest'].quantile(0.7)  # أعلى 30%
        eligible = contracts[contracts['openInterest'] > oi_threshold]

        if eligible.empty:
            eligible = contracts

        # نختار العقد الأقرب للسعر الحالي (للدلتا التقريبية)
        best = eligible.nsmallest(1, 'strike_diff').iloc[0]

        # حساب Spread تقريبي (ask-bid)/ask
        spread = (best['ask'] - best['bid']) / best['ask'] if best['ask'] > 0 else 1
        if spread > 0.05:   # أكبر من 5%
            return "سبريد واسع"

        # بناء نص العقد المقترح
        delta_approx = 0.5 - (best['strike_diff'] / current_price)  # تقريب بسيط
        delta_approx = max(0.1, min(0.9, delta_approx))

        return f"Delta {delta_approx:.2f} – Exp {exp_date} (أيام: {days})"

    except Exception as e:
        logging.error(f"خطأ في جلب الخيارات لـ {ticker}: {e}")
        return "غير متوفر"

# ---------------------------
# 8. بناء رسالة التنبيه النهائية
# ---------------------------

def build_alert_message(ticker, signal, confidence, classification, details, spx_trend, option_text):
    """
    تنسيق الرسالة كما طلبت تماماً.
    """
    # ترجمة الاتجاه
    if signal == "CALL":
        direction = "Call"
        cross = "50 > 100" if details.get('sma50', 0) > details.get('sma100', 0) else "50 > 200"
        price_status = f"السعر فوق {details.get('sma200', 0):.2f}"
        event = "اختراق قمة 2H"
    else:
        direction = "Put"
        cross = "50 < 100" if details.get('sma50', 0) < details.get('sma100', 0) else "50 < 200"
        price_status = f"السعر تحت {details.get('sma200', 0):.2f}"
        event = "كسر قاع 2H"

    # توافق SPX
    if "uptrend" in spx_trend and signal == "CALL":
        spx_compat = "✅ متوافق"
    elif "downtrend" in spx_trend and signal == "PUT":
        spx_compat = "✅ متوافق"
    else:
        spx_compat = "⚠️ غير متوافق"

    message = f"""
🔔 {ticker}
الاتجاه: {direction}
التصنيف: {classification}
التقاطع: {cross} (إيجابي)
الحالة: {price_status}
الحدث: {event}
SPX: {spx_compat}
الثقة: {confidence}/100
العقد المقترح: {option_text}
"""
    return message

# ---------------------------
# 9. دالة المسح الرئيسية
# ---------------------------

def scan_all_stocks():
    """تمر على كل الأسهم وتفحصها وترسل التنبيهات"""
    alerts = []
    print("Starting stock scan...") # Added print statement
    for ticker in TICKERS:
        logging.info(f"فحص {ticker}...")
        time.sleep(0.5)  # تأخير نصف ثانية لتجنب حظر Yahoo

        # 1. الفلتر الأساسي
        if not passes_basic_filters(ticker):
            print(f"Ticker {ticker} did not pass basic filters.") # Added print statement
            continue

        # 2. تحليل السهم
        result = analyze_stock_2h(ticker)
        if not result or not result['signal']:
            print(f"No signal found for {ticker} after 2h analysis.") # Added print statement
            continue

        signal = result['signal']
        confidence = result['confidence']
        details = result['details']

        # 3. تحليل SPX
        spx_15m = analyze_spx("15m")

        # 4. تصنيف الصفقة
        classification = classify_trade(signal, confidence, details)

        # 5. اختيار العقد
        option = get_option_contract(ticker, signal, classification)

        # 6. بناء الرسالة
        msg = build_alert_message(ticker, signal, confidence, classification, details, spx_15m, option)
        alerts.append(msg)
    print(f"Scan finished. Found {len(alerts)} alerts.") # Added print statement
    return alerts

# ---------------------------
# 10. دوال بوت تليجرام
# ---------------------------

async def send_alerts(context):
    """إرسال التنبيهات إلى تليجرام"""
    print("send_alerts function called.") # Added print statement
    alerts = scan_all_stocks()
    if not alerts:
        print("No alerts found to send.") # Added print statement
        await context.bot.send_message(chat_id=CHAT_ID, text="لا توجد إشارات جديدة حالياً.")
    else:
        print(f"Sending {len(alerts)} alerts.") # Added print statement
        for alert in alerts:
            await context.bot.send_message(chat_id=CHAT_ID, text=alert, parse_mode='Markdown')
            await asyncio.sleep(1)  # تجنب سبام تليجرام

# دالة للترحيب بالأعضاء الجدد وإرسال إخلاء المسؤولية
async def greet_chat_members(update: Update, context: ContextTypes.DEFAULT_TYPE) -> None:
    print(f"New chat member event: {update.chat_member}") # Added print
    if update.chat_member.new_chat_member.status == 'member':
        user = update.chat_member.new_chat_member.user
        chat_title = update.chat_member.chat.title or "هذه المجموعة"
        welcome_new_member_message = f"""
أهلاً وسهلاً بك {user.full_name} في {chat_title}! 👋

⚠️ **إخلاء مسؤولية:**
توصيات هذا البوت هي لأغراض تعليمية وتحليلية فقط، ولا تشكل نصيحة استثمارية بأي شكل من الأشكال. تداول العقود والمشتقات المالية ينطوي على مخاطر عالية وقد يؤدي إلى خسارة رأس المال. أنت وحدك المسؤول عن قراراتك الاستثمارية. يرجى استشارة مستشار مالي متخصص قبل اتخاذ أي قرارات استثمارية.
"""
        print(f"Sending welcome message to {user.full_name} in chat {update.chat_member.chat.id}") # Added print
        await context.bot.send_message(chat_id=update.chat_member.chat.id, text=welcome_new_member_message, parse_mode='Markdown')


async def start(update: Update, context: ContextTypes.DEFAULT_TYPE) -> None:
    """معالج أمر /start"""
    print(f"Start command received from {update.effective_user.id} in chat {update.effective_chat.id}") # Added print
    welcome_message = """
👋 مرحباً بك في بوت توصيات الأسهم!

هذا البوت يقدم توصيات تحليل فني للأسهم بناءً على استراتيجيات محددة.

⚠️ **إخلاء مسؤولية:**
توصيات هذا البوت هي لأغراض تعليمية وتحليلية فقط، ولا تشكل نصيحة استثمارية بأي شكل من الأشكال. تداول العقود والمشتقات المالية ينطوي على مخاطر عالية وقد يؤدي إلى خسارة رأس المال. أنت وحدك المسؤول عن قراراتك الاستثمارية. يرجى استشارة مستشار مالي متخصص قبل اتخاذ أي قرارات استثمارية.

اختر نوع التوصية التي ترغب في الحصول عليها:
"""

    keyboard = [
        [InlineKeyboardButton("📈 عقود CALL (شراء)", callback_data='recommend_call')],
        [InlineKeyboardButton("📉 عقود PUT (بيع)", callback_data='recommend_put')],
        [InlineKeyboardButton("🔄 مسح الآن", callback_data='scan_now_manual')]
    ]
    reply_markup = InlineKeyboardMarkup(keyboard)

    await update.message.reply_text(welcome_message, reply_markup=reply_markup, parse_mode='Markdown')

async def button_callback(update: Update, context: ContextTypes.DEFAULT_TYPE) -> None:
    query = update.callback_query
    print(f"Callback query received: {query.data} from {query.from_user.id}") # Added print
    await query.answer()  # Always answer callback queries

    if query.data == 'recommend_call':
        await query.edit_message_text(text="جاري البحث عن توصيات CALL...")
        # Logic to scan for CALL recommendations
        # For now, let's just send a placeholder message
        await asyncio.sleep(2) # Simulate work
        await query.edit_message_text(text="لم يتم العثور على توصيات CALL حالياً.") # Replace with actual results
    elif query.data == 'recommend_put':
        await query.edit_message_text(text="جاري البحث عن توصيات PUT...")
        # Logic to scan for PUT recommendations
        # For now, let's just send a placeholder message
        await asyncio.sleep(2) # Simulate work
        await query.edit_message_text(text="لم يتم العثور على توصيات PUT حالياً.") # Replace with actual results
    elif query.data == 'scan_now_manual':
        await query.edit_message_text(text="جاري المسح الفوري، انتظر قليلاً...")
        alerts = scan_all_stocks()
        if not alerts:
            await query.edit_message_text("لا توجد إشارات جديدة.")
        else:
            for alert in alerts:
                await query.message.reply_text(alert, parse_mode='Markdown')
                await asyncio.sleep(1)
            await query.edit_message_text(text="تم الانتهاء من المسح الفوري.")


async def scan_now(update, context):
    """معالج أمر /scan لمسح فوري"""
    print("scan_now command called.") # Added print statement
    await update.message.reply_text("جاري المسح، انتظر قليلاً...")
    alerts = scan_all_stocks()
    if not alerts:
        print("No alerts found for manual scan.") # Added print statement
        await update.message.reply_text("لا توجد إشارات جديدة.")
    else:
        print(f"Sending {len(alerts)} alerts for manual scan.") # Added print statement
        for alert in alerts:
            await update.message.reply_text(alert, parse_mode='Markdown')
            await asyncio.sleep(1)

def main():
    """الدالة الرئيسية لتشغيل البوت"""
    # إنشاء تطبيق البوت
    # Create a JobQueue instance first
    job_queue_instance = JobQueue()

    # Pass the JobQueue instance to the ApplicationBuilder
    application = Application.builder().token(TELEGRAM_TOKEN).job_queue(job_queue_instance).build()

    # إضافة المعالجات للأوامر والرسائل
    application.add_handler(CommandHandler("start", start))
    application.add_handler(CommandHandler("scan", scan_now))
    application.add_handler(CallbackQueryHandler(button_callback)) # إضافة معالج لأزرار الـ Inline Keyboard
    application.add_handler(ChatMemberHandler(greet_chat_members, ChatMemberHandler.CHAT_MEMBER)) # إضافة معالج للأعضاء الجدد

    # جدولة المسح التلقائي كل ساعة
    job_queue_instance.run_repeating(send_alerts, interval=3600, first=10)  # كل ساعة، أول مرة بعد 10 ثواني

    # تشغيل البوت
    print("البوت يعمل الآن... اذهب لتجربته!")
    application.run_polling()
    # application.idle() is not necessary after run_polling() as it's a blocking call

# استدعاء الدالة الرئيسية
if __name__ == '__main__':
    main()
````

````python
#!/usr/bin/env python3
# -*- coding: utf-8 -*-

# بوت توصيات الأسهم - نسخة شاكر 1.0 (نسخة كاملة)
# آخر تحديث: فبراير 2026

import yfinance as yf
import pandas as pd
import numpy as np
import ta
import logging
import time
import schedule
from datetime import datetime, timedelta
from telegram import Bot, InlineKeyboardButton, InlineKeyboardMarkup, Update
from telegram.ext import Application, CommandHandler, CallbackQueryHandler, ContextTypes, JobQueue, ChatMemberHandler # أضفنا ChatMemberHandler
import asyncio
import os

# ---------------------------
# 1. إعدادات البوت والتوكن
# ---------------------------
TELEGRAM_TOKEN = "8699821370:AAEQUSbLTgf7MmWqo5vV5LHPOz30wfqOfqw"      # ضع التوكن الخاص بك
CHAT_ID = "7854020427"                # ضع معرف المستخدم أو القناة

# تفعيل السجلات لمشاهدة الأخطاء
logging.basicConfig(level=logging.INFO, format='%(asctime)s - %(name)s - %(levelname)s - %(message)s')

# ---------------------------
# 2. قائمة الأسهم الكاملة (مدمجة)
# ---------------------------
TICKERS = [
    # --- أسهمك القديمة (من الصور) ---
    "A", "AA", "ABEV", "ABNB", "ABT", "ABVX", "ACGL", "ACM", "ACN",
    "ADBE", "ADI", "ADM", "ADP", "ADSK", "AEE", "AEG", "AEIS", "AEM",
    "AEP", "AER", "AES", "AFG", "AFL", "AFRM", "AG", "AGI", "AGNC",
    "AIG", "AIT", "AIZ", "AJG", "AKAM", "ALAB", "ALB", "ALC", "ALGN",
    "ALL", "ALLE", "ALLY", "ALNY", "ALSN", "AM", "AMCR", "AME", "AMH",
    "AMKR", "AMP", "AMRZ", "AMT", "AMX",

    # --- الشركات العملاقة (> 1 تريليون دولار) ---
    "NVDA", "AAPL", "GOOGL", "MSFT", "AMZN", "META", "TSM", "AVGO", "TSLA",
    "BRK.A", "WMT",

    # --- الشركات الكبرى (200 مليار - 1 تريليون) ---
    "LLY", "JPM", "XOM", "V", "JNJ", "MA", "MU", "ORCL", "COST", "AMD",
    "NFLX", "PLTR", "CSCO", "MRK", "TMO", "ASML", "ABBV", "HD", "CVX",
    "PG", "CAT", "KO", "PEP", "INTC", "CMCSA", "MCD", "TXN", "VZ", "AMGN"
]

# ---------------------------
# 3. دوال التحليل الأساسي (الفلاتر)
# ---------------------------

def get_stock_info(ticker):
    """جلب معلومات الشركة الأساسية"""
    try:
        stock = yf.Ticker(ticker)
        info = stock.info
        return info
    except Exception as e:
        logging.error(f"خطأ في جلب معلومات {ticker}: {e}")
        return None

def passes_basic_filters(ticker):
    """
    تطبيق الفلتر الأساسي:
    - Market Cap > 10B (ملاحظة: أغلب الشركات في القائمة أكبر من ذلك بكثير، لكن الفلتر يضمن ذلك)
    - Average Volume > 1M (تقريبي)
    - Spread < 5%
    """
    info = get_stock_info(ticker)
    if not info:
        return False

    # 1. القيمة السوقية
    mcap = info.get('marketCap', 0)
    if mcap < 10e9:
        print(f"{ticker} فشل فلتر Market Cap: {mcap}") # Changed from logging.debug
        return False

    # 2. متوسط الحجم (نأخذ آخر 20 يوم)
    try:
        stock = yf.Ticker(ticker)
        hist = stock.history(period="1mo", interval="1d")
        if hist.empty:
            return False
        avg_vol = hist['Volume'].rolling(20).mean().iloc[-1]
        if avg_vol < 1_000_000:   # أقل من مليون سهم يومياً
            print(f"{ticker} فشل فلتر الحجم: {avg_vol}") # Changed from logging.debug
            return False
    except:
        return False

    # 3. Spread (bid/ask)
    bid = info.get('bid', 0)
    ask = info.get('ask', 0)
    if bid and ask:
        spread = (ask - bid) / bid if bid != 0 else 1
        if spread > 0.05:   # أكبر من 5%
            print(f"{ticker} فشل فلتر Spread: {spread:.2%}") # Changed from logging.debug
            return False

    return True

# ---------------------------
# 4. تحليل SPX (مؤشر S&P 500)
# ---------------------------

def analyze_spx(timeframe="15m"):
    """تحليل مؤشر SPX على الإطار المطلوب"""
    try:
        spx = yf.Ticker("^GSPC")  # أو "SPY"
        hist = spx.history(period="5d", interval=timeframe)
        if len(hist) < 50:
            return "neutral"

        # متوسطات بسيطة لتحديد الاتجاه
        hist['SMA20'] = hist['Close'].rolling(20).mean()
        last = hist.iloc[-1]
        prev = hist.iloc[-2]

        if last['Close'] > last['SMA20'] and last['Close'] > prev['Close']:
            return "uptrend"
        elif last['Close'] < last['SMA20'] and last['Close'] < prev['Close']:
            return "downtrend"
        else:
            return "neutral"
    except:
        return "neutral"

# ---------------------------
# 5. تحليل السهم (فريم ساعتين)
# ---------------------------

def analyze_stock_2h(ticker):
    """
    تحليل فني كامل على فريم الساعتين
    يعيد: {'signal': 'CALL'/'PUT'/None, 'confidence': 0-100, 'details': {...}}
    """
    try:
        print(f"Analyzing {ticker} for 2h interval.") # Added print statement
        stock = yf.Ticker(ticker)
        hist = stock.history(period="1mo", interval="2h")  # آخر شهر، فريم ساعتين
        if len(hist) < 200:
            print(f"Not enough historical data for {ticker} in 2h interval.") # Added print statement
            return None

        # حساب المتوسطات
        hist['SMA50'] = hist['Close'].rolling(50).mean()
        hist['SMA100'] = hist['Close'].rolling(100).mean()
        hist['SMA200'] = hist['Close'].rolling(200).mean()
        hist['Volume_SMA20'] = hist['Volume'].rolling(20).mean()

        last = hist.iloc[-1]
        current_price = last['Close']
        sma50 = last['SMA50']
        sma100 = last['SMA100']
        sma200 = last['SMA200']

        # قمم وقيعان آخر 3 شموع
        high_last_3 = hist['High'].iloc[-3:].max()
        low_last_3 = hist['Low'].iloc[-3:].min()

        # قمة/قاع اليوم (آخر 24 ساعة = 12 شمعة)
        high_today = hist['High'].iloc[-12:].max()
        low_today = hist['Low'].iloc[-12:].min()

        volume_ok = last['Volume'] > last['Volume_SMA20']

        # شروط CALL
        call_condition1 = (sma50 > sma100) or (sma50 > sma200)
        call_condition2 = current_price > sma200
        call_condition3 = current_price > high_last_3 or current_price > high_today
        call_condition4 = volume_ok

        # شروط PUT
        put_condition1 = (sma50 < sma100) or (sma50 < sma200)
        put_condition2 = current_price < sma200
        put_condition3 = current_price < low_last_3 or current_price < low_today
        put_condition4 = volume_ok

        signal = None
        confidence = 0
        details = {}

        if call_condition1 and call_condition2 and call_condition3 and call_condition4:
            signal = "CALL"
            # حساب الثقة: نضيف 25 لكل شرط متحقق
            confidence = 25 * sum([call_condition1, call_condition2, call_condition3, call_condition4])
            details = {
                "sma50": sma50,
                "sma100": sma100,
                "sma200": sma200,
                "price": current_price,
                "high_3": high_last_3,
                "volume_ratio": last['Volume'] / last['Volume_SMA20']
            }
        elif put_condition1 and put_condition2 and put_condition3 and put_condition4:
            signal = "PUT"
            confidence = 25 * sum([put_condition1, put_condition2, put_condition3, put_condition4])
            details = {
                "sma50": sma50,
                "sma100": sma100,
                "sma200": sma200,
                "price": current_price,
                "low_3": low_last_3,
                "volume_ratio": last['Volume'] / last['Volume_SMA20']
            }

        if signal: # Added print statement
            print(f"Signal for {ticker}: {signal}, Confidence: {confidence}")
        return {"signal": signal, "confidence": int(confidence), "details": details}

    except Exception as e:
        logging.error(f"خطأ في تحليل {ticker}: {e}")
        return None

# ---------------------------
# 6. تصنيف الصفقة (سكالب، قصيرة، سوينق)
# ---------------------------

def classify_trade(signal, confidence, details):
    """
    تصنيف حسب قوة الإشارة وترتيب المتوسطات
    """
    if not signal:
        return None

    sma50 = details.get('sma50')
    sma100 = details.get('sma100')
    sma200 = details.get('sma200')

    # ترتيب المتوسطات
    if signal == "CALL":
        if sma50 > sma100 > sma200:
            return "📈 سوينق 3-7 أيام"
        elif confidence >= 75:
            return "⏳ صفقة قصيرة 1-2 يوم"
        else:
            return "⚡ سكالب"
    else:  # PUT
        if sma50 < sma100 < sma200:
            return "📈 سوينق 3-7 أيام"
        elif confidence >= 75:
            return "⏳ صفقة قصيرة 1-2 يوم"
        else:
            return "⚡ سكالب"

# ---------------------------
# 7. اختيار العقد المناسب (بيانات الخيارات)
# ---------------------------

def get_option_contract(ticker, signal, classification):
    """
    البحث عن عقد خيار مناسب حسب التصنيف.
    """
    try:
        stock = yf.Ticker(ticker)
        # جلب تواريخ انتهاء الصلاحية المتاحة
        expirations = stock.options
        if not expirations:
            return "غير متوفر"

        # اختيار التاريخ المناسب حسب التصنيف
        today = datetime.now().date()
        if "سكالب" in classification:
            days = 7
        elif "قصيرة" in classification:
            days = 14
        else:  # سوينق
            days = 30

        target_date = today + timedelta(days=days)
        # البحث عن أقرب تاريخ انتهاء للعدد المطلوب من الأيام
        exp_date = None
        for exp in expirations:
            exp_dt = datetime.strptime(exp, '%Y-%m-%d').date()
            if exp_dt >= target_date:
                exp_date = exp
                break
        if not exp_date:
            exp_date = expirations[0]  # أقرب تاريخ

        # جلب عقود الخيارات لذلك التاريخ
        opt_chain = stock.option_chain(exp_date)
        if signal == "CALL":
            contracts = opt_chain.calls
        else:
            contracts = opt_chain.puts

        if contracts.empty:
            return "غير متوفر"

        # فلترة العقود: Open Interest عالي (أعلى 10%)، Spread ضيق
        current_price = stock.history(period="1d")['Close'].iloc[-1]
        contracts['strike_diff'] = abs(contracts['strike'] - current_price)
        # فلتر OI
        oi_threshold = contracts['openInterest'].quantile(0.7)  # أعلى 30%
        eligible = contracts[contracts['openInterest'] > oi_threshold]

        if eligible.empty:
            eligible = contracts

        # نختار العقد الأقرب للسعر الحالي (للدلتا التقريبية)
        best = eligible.nsmallest(1, 'strike_diff').iloc[0]

        # حساب Spread تقريبي (ask-bid)/ask
        spread = (best['ask'] - best['bid']) / best['ask'] if best['ask'] > 0 else 1
        if spread > 0.05:   # أكبر من 5%
            return "سبريد واسع"

        # بناء نص العقد المقترح
        delta_approx = 0.5 - (best['strike_diff'] / current_price)  # تقريب بسيط
        delta_approx = max(0.1, min(0.9, delta_approx))

        return f"Delta {delta_approx:.2f} – Exp {exp_date} (أيام: {days})"

    except Exception as e:
        logging.error(f"خطأ في جلب الخيارات لـ {ticker}: {e}")
        return "غير متوفر"

# ---------------------------
# 8. بناء رسالة التنبيه النهائية
# ---------------------------

def build_alert_message(ticker, signal, confidence, classification, details, spx_trend, option_text):
    """
    تنسيق الرسالة كما طلبت تماماً.
    """
    # ترجمة الاتجاه
    if signal == "CALL":
        direction = "Call"
        cross = "50 > 100" if details.get('sma50', 0) > details.get('sma100', 0) else "50 > 200"
        price_status = f"السعر فوق {details.get('sma200', 0):.2f}"
        event = "اختراق قمة 2H"
    else:
        direction = "Put"
        cross = "50 < 100" if details.get('sma50', 0) < details.get('sma100', 0) else "50 < 200"
        price_status = f"السعر تحت {details.get('sma200', 0):.2f}"
        event = "كسر قاع 2H"

    # توافق SPX
    if "uptrend" in spx_trend and signal == "CALL":
        spx_compat = "✅ متوافق"
    elif "downtrend" in spx_trend and signal == "PUT":
        spx_compat = "✅ متوافق"
    else:
        spx_compat = "⚠️ غير متوافق"

    message = f"""
🔔 {ticker}
الاتجاه: {direction}
التصنيف: {classification}
التقاطع: {cross} (إيجابي)
الحالة: {price_status}
الحدث: {event}
SPX: {spx_compat}
الثقة: {confidence}/100
العقد المقترح: {option_text}
"""
    return message

# ---------------------------
# 9. دالة المسح الرئيسية
# ---------------------------

def scan_all_stocks():
    """تمر على كل الأسهم وتفحصها وترسل التنبيهات"""
    alerts = []
    print("Starting stock scan...") # Added print statement
    for ticker in TICKERS:
        logging.info(f"فحص {ticker}...")
        time.sleep(0.5)  # تأخير نصف ثانية لتجنب حظر Yahoo

        # 1. الفلتر الأساسي
        if not passes_basic_filters(ticker):
            print(f"Ticker {ticker} did not pass basic filters.") # Added print statement
            continue

        # 2. تحليل السهم
        result = analyze_stock_2h(ticker)
        if not result or not result['signal']:
            print(f"No signal found for {ticker} after 2h analysis.") # Added print statement
            continue

        signal = result['signal']
        confidence = result['confidence']
        details = result['details']

        # 3. تحليل SPX
        spx_15m = analyze_spx("15m")

        # 4. تصنيف الصفقة
        classification = classify_trade(signal, confidence, details)

        # 5. اختيار العقد
        option = get_option_contract(ticker, signal, classification)

        # 6. بناء الرسالة
        msg = build_alert_message(ticker, signal, confidence, classification, details, spx_15m, option)
        alerts.append(msg)
    print(f"Scan finished. Found {len(alerts)} alerts.") # Added print statement
    return alerts

# ---------------------------
# 10. دوال بوت تليجرام
# ---------------------------

async def send_alerts(context):
    """إرسال التنبيهات إلى تليجرام"""
    print("send_alerts function called.") # Added print statement
    alerts = scan_all_stocks()
    if not alerts:
        print("No alerts found to send.") # Added print statement
        await context.bot.send_message(chat_id=CHAT_ID, text="لا توجد إشارات جديدة حالياً.")
    else:
        print(f"Sending {len(alerts)} alerts.") # Added print statement
        for alert in alerts:
            await context.bot.send_message(chat_id=CHAT_ID, text=alert, parse_mode='Markdown')
            await asyncio.sleep(1)  # تجنب سبام تليجرام

# دالة للترحيب بالأعضاء الجدد وإرسال إخلاء المسؤولية
async def greet_chat_members(update: Update, context: ContextTypes.DEFAULT_TYPE) -> None:
    print(f"New chat member event: {update.chat_member}") # Added print
    if update.chat_member.new_chat_member.status == 'member':
        user = update.chat_member.new_chat_member.user
        chat_title = update.chat_member.chat.title or "هذه المجموعة"
        welcome_new_member_message = f"""
أهلاً وسهلاً بك {user.full_name} في {chat_title}! 👋

⚠️ **إخلاء مسؤولية:**
توصيات هذا البوت هي لأغراض تعليمية وتحليلية فقط، ولا تشكل نصيحة استثمارية بأي شكل من الأشكال. تداول العقود والمشتقات المالية ينطوي على مخاطر عالية وقد يؤدي إلى خسارة رأس المال. أنت وحدك المسؤول عن قراراتك الاستثمارية. يرجى استشارة مستشار مالي متخصص قبل اتخاذ أي قرارات استثمارية.
"""
        print(f"Sending welcome message to {user.full_name} in chat {update.chat_member.chat.id}") # Added print
        await context.bot.send_message(chat_id=update.chat_member.chat.id, text=welcome_new_member_message, parse_mode='Markdown')


async def start(update: Update, context: ContextTypes.DEFAULT_TYPE) -> None:
    """معالج أمر /start"""
    print(f"Start command received from {update.effective_user.id} in chat {update.effective_chat.id}") # Added print
    welcome_message = """
👋 مرحباً بك في بوت توصيات الأسهم!

هذا البوت يقدم توصيات تحليل فني للأسهم بناءً على استراتيجيات محددة.

⚠️ **إخلاء مسؤولية:**
توصيات هذا البوت هي لأغراض تعليمية وتحليلية فقط، ولا تشكل نصيحة استثمارية بأي شكل من الأشكال. تداول العقود والمشتقات المالية ينطوي على مخاطر عالية وقد يؤدي إلى خسارة رأس المال. أنت وحدك المسؤول عن قراراتك الاستثمارية. يرجى استشارة مستشار مالي متخصص قبل اتخاذ أي قرارات استثمارية.

اختر نوع التوصية التي ترغب في الحصول عليها:
"""

    keyboard = [
        [InlineKeyboardButton("📈 عقود CALL (شراء)", callback_data='recommend_call')],
        [InlineKeyboardButton("📉 عقود PUT (بيع)", callback_data='recommend_put')],
        [InlineKeyboardButton("🔄 مسح الآن", callback_data='scan_now_manual')]
    ]
    reply_markup = InlineKeyboardMarkup(keyboard)

    await update.message.reply_text(welcome_message, reply_markup=reply_markup, parse_mode='Markdown')

async def button_callback(update: Update, context: ContextTypes.DEFAULT_TYPE) -> None:
    query = update.callback_query
    print(f"Callback query received: {query.data} from {query.from_user.id}") # Added print
    await query.answer()  # Always answer callback queries

    if query.data == 'recommend_call':
        await query.edit_message_text(text="جاري البحث عن توصيات CALL...")
        # Logic to scan for CALL recommendations
        # For now, let's just send a placeholder message
        await asyncio.sleep(2) # Simulate work
        await query.edit_message_text(text="لم يتم العثور على توصيات CALL حالياً.") # Replace with actual results
    elif query.data == 'recommend_put':
        await query.edit_message_text(text="جاري البحث عن توصيات PUT...")
        # Logic to scan for PUT recommendations
        # For now, let's just send a placeholder message
        await asyncio.sleep(2) # Simulate work
        await query.edit_message_text(text="لم يتم العثور على توصيات PUT حالياً.") # Replace with actual results
    elif query.data == 'scan_now_manual':
        await query.edit_message_text(text="جاري المسح الفوري، انتظر قليلاً...")
        alerts = scan_all_stocks()
        if not alerts:
            await query.edit_message_text("لا توجد إشارات جديدة.")
        else:
            for alert in alerts:
                await query.message.reply_text(alert, parse_mode='Markdown')
                await asyncio.sleep(1)
            await query.edit_message_text(text="تم الانتهاء من المسح الفوري.")


async def scan_now(update, context):
    """معالج أمر /scan لمسح فوري"""
    print("scan_now command called.") # Added print statement
    await update.message.reply_text("جاري المسح، انتظر قليلاً...")
    alerts = scan_all_stocks()
    if not alerts:
        print("No alerts found for manual scan.") # Added print statement
        await update.message.reply_text("لا توجد إشارات جديدة.")
    else:
        print(f"Sending {len(alerts)} alerts for manual scan.") # Added print statement
        for alert in alerts:
            await update.message.reply_text(alert, parse_mode='Markdown')
            await asyncio.sleep(1)

def main():
    """الدالة الرئيسية لتشغيل البوت"""
    # إنشاء تطبيق البوت
    # Create a JobQueue instance first
    job_queue_instance = JobQueue()

    # Pass the JobQueue instance to the ApplicationBuilder
    application = Application.builder().token(TELEGRAM_TOKEN).job_queue(job_queue_instance).build()

    # إضافة المعالجات للأوامر والرسائل
    application.add_handler(CommandHandler("start", start))
    application.add_handler(CommandHandler("scan", scan_now))
    application.add_handler(CallbackQueryHandler(button_callback)) # إضافة معالج لأزرار الـ Inline Keyboard
    application.add_handler(ChatMemberHandler(greet_chat_members, ChatMemberHandler.CHAT_MEMBER)) # إضافة معالج للأعضاء الجدد

    # جدولة المسح التلقائي كل ساعة
    job_queue_instance.run_repeating(send_alerts, interval=3600, first=10)  # كل ساعة، أول مرة بعد 10 ثواني

    # تشغيل البوت
    print("البوت يعمل الآن... اذهب لتجربته!")
    application.run_polling()
    # application.idle() is not necessary after run_polling() as it's a blocking call

# استدعاء الدالة الرئيسية
if __name__ == '__main__':
    main()
````

````python
#!/usr/bin/env python3
# -*- coding: utf-8 -*-

# بوت توصيات الأسهم - نسخة شاكر 1.0 (نسخة كاملة)
# آخر تحديث: فبراير 2026

import yfinance as yf
import pandas as pd
import numpy as np
import ta
import logging
import time
import schedule
from datetime import datetime, timedelta
from telegram import Bot, InlineKeyboardButton, InlineKeyboardMarkup, Update
from telegram.ext import Application, CommandHandler, CallbackQueryHandler, ContextTypes, JobQueue, ChatMemberHandler # أضفنا ChatMemberHandler
import asyncio
import os

# ---------------------------
# 1. إعدادات البوت والتوكن
# ---------------------------
TELEGRAM_TOKEN = "8699821370:AAEQUSbLTgf7MmWqo5vV5LHPOz30wfqOfqw"      # ضع التوكن الخاص بك
CHAT_ID = "7854020427"                # ضع معرف المستخدم أو القناة

# تفعيل السجلات لمشاهدة الأخطاء
logging.basicConfig(level=logging.INFO, format='%(asctime)s - %(name)s - %(levelname)s - %(message)s')

# ---------------------------
# 2. قائمة الأسهم الكاملة (مدمجة)
# ---------------------------
TICKERS = [
    # --- أسهمك القديمة (من الصور) ---
    "A", "AA", "ABEV", "ABNB", "ABT", "ABVX", "ACGL", "ACM", "ACN",
    "ADBE", "ADI", "ADM", "ADP", "ADSK", "AEE", "AEG", "AEIS", "AEM",
    "AEP", "AER", "AES", "AFG", "AFL", "AFRM", "AG", "AGI", "AGNC",
    "AIG", "AIT", "AIZ", "AJG", "AKAM", "ALAB", "ALB", "ALC", "ALGN",
    "ALL", "ALLE", "ALLY", "ALNY", "ALSN", "AM", "AMCR", "AME", "AMH",
    "AMKR", "AMP", "AMRZ", "AMT", "AMX",

    # --- الشركات العملاقة (> 1 تريليون دولار) ---
    "NVDA", "AAPL", "GOOGL", "MSFT", "AMZN", "META", "TSM", "AVGO", "TSLA",
    "BRK.A", "WMT",

    # --- الشركات الكبرى (200 مليار - 1 تريليون) ---
    "LLY", "JPM", "XOM", "V", "JNJ", "MA", "MU", "ORCL", "COST", "AMD",
    "NFLX", "PLTR", "CSCO", "MRK", "TMO", "ASML", "ABBV", "HD", "CVX",
    "PG", "CAT", "KO", "PEP", "INTC", "CMCSA", "MCD", "TXN", "VZ", "AMGN"
]

# ---------------------------
# 3. دوال التحليل الأساسي (الفلاتر)
# ---------------------------

def get_stock_info(ticker):
    """جلب معلومات الشركة الأساسية"""
    try:
        stock = yf.Ticker(ticker)
        info = stock.info
        return info
    except Exception as e:
        logging.error(f"خطأ في جلب معلومات {ticker}: {e}")
        return None

def passes_basic_filters(ticker):
    """
    تطبيق الفلتر الأساسي:
    - Market Cap > 10B (ملاحظة: أغلب الشركات في القائمة أكبر من ذلك بكثير، لكن الفلتر يضمن ذلك)
    - Average Volume > 1M (تقريبي)
    - Spread < 5%
    """
    info = get_stock_info(ticker)
    if not info:
        return False

    # 1. القيمة السوقية
    mcap = info.get('marketCap', 0)
    if mcap < 10e9:
        print(f"{ticker} فشل فلتر Market Cap: {mcap}") # Changed from logging.debug
        return False

    # 2. متوسط الحجم (نأخذ آخر 20 يوم)
    try:
        stock = yf.Ticker(ticker)
        hist = stock.history(period="1mo", interval="1d")
        if hist.empty:
            return False
        avg_vol = hist['Volume'].rolling(20).mean().iloc[-1]
        if avg_vol < 1_000_000:   # أقل من مليون سهم يومياً
            print(f"{ticker} فشل فلتر الحجم: {avg_vol}") # Changed from logging.debug
            return False
    except:
        return False

    # 3. Spread (bid/ask)
    bid = info.get('bid', 0)
    ask = info.get('ask', 0)
    if bid and ask:
        spread = (ask - bid) / bid if bid != 0 else 1
        if spread > 0.05:   # أكبر من 5%
            print(f"{ticker} فشل فلتر Spread: {spread:.2%}") # Changed from logging.debug
            return False

    return True

# ---------------------------
# 4. تحليل SPX (مؤشر S&P 500)
# ---------------------------

def analyze_spx(timeframe="15m"):
    """تحليل مؤشر SPX على الإطار المطلوب"""
    try:
        spx = yf.Ticker("^GSPC")  # أو "SPY"
        hist = spx.history(period="5d", interval=timeframe)
        if len(hist) < 50:
            return "neutral"

        # متوسطات بسيطة لتحديد الاتجاه
        hist['SMA20'] = hist['Close'].rolling(20).mean()
        last = hist.iloc[-1]
        prev = hist.iloc[-2]

        if last['Close'] > last['SMA20'] and last['Close'] > prev['Close']:
            return "uptrend"
        elif last['Close'] < last['SMA20'] and last['Close'] < prev['Close']:
            return "downtrend"
        else:
            return "neutral"
    except:
        return "neutral"

# ---------------------------
# 5. تحليل السهم (فريم ساعتين)
# ---------------------------

def analyze_stock_2h(ticker):
    """
    تحليل فني كامل على فريم الساعتين
    يعيد: {'signal': 'CALL'/'PUT'/None, 'confidence': 0-100, 'details': {...}}
    """
    try:
        print(f"Analyzing {ticker} for 2h interval.") # Added print statement
        stock = yf.Ticker(ticker)
        hist = stock.history(period="1mo", interval="2h")  # آخر شهر، فريم ساعتين
        if len(hist) < 200:
            print(f"Not enough historical data for {ticker} in 2h interval.") # Added print statement
            return None

        # حساب المتوسطات
        hist['SMA50'] = hist['Close'].rolling(50).mean()
        hist['SMA100'] = hist['Close'].rolling(100).mean()
        hist['SMA200'] = hist['Close'].rolling(200).mean()
        hist['Volume_SMA20'] = hist['Volume'].rolling(20).mean()

        last = hist.iloc[-1]
        current_price = last['Close']
        sma50 = last['SMA50']
        sma100 = last['SMA100']
        sma200 = last['SMA200']

        # قمم وقيعان آخر 3 شموع
        high_last_3 = hist['High'].iloc[-3:].max()
        low_last_3 = hist['Low'].iloc[-3:].min()

        # قمة/قاع اليوم (آخر 24 ساعة = 12 شمعة)
        high_today = hist['High'].iloc[-12:].max()
        low_today = hist['Low'].iloc[-12:].min()

        volume_ok = last['Volume'] > last['Volume_SMA20']

        # شروط CALL
        call_condition1 = (sma50 > sma100) or (sma50 > sma200)
        call_condition2 = current_price > sma200
        call_condition3 = current_price > high_last_3 or current_price > high_today
        call_condition4 = volume_ok

        # شروط PUT
        put_condition1 = (sma50 < sma100) or (sma50 < sma200)
        put_condition2 = current_price < sma200
        put_condition3 = current_price < low_last_3 or current_price < low_today
        put_condition4 = volume_ok

        signal = None
        confidence = 0
        details = {}

        if call_condition1 and call_condition2 and call_condition3 and call_condition4:
            signal = "CALL"
            # حساب الثقة: نضيف 25 لكل شرط متحقق
            confidence = 25 * sum([call_condition1, call_condition2, call_condition3, call_condition4])
            details = {
                "sma50": sma50,
                "sma100": sma100,
                "sma200": sma200,
                "price": current_price,
                "high_3": high_last_3,
                "volume_ratio": last['Volume'] / last['Volume_SMA20']
            }
        elif put_condition1 and put_condition2 and put_condition3 and put_condition4:
            signal = "PUT"
            confidence = 25 * sum([put_condition1, put_condition2, put_condition3, put_condition4])
            details = {
                "sma50": sma50,
                "sma100": sma100,
                "sma200": sma200,
                "price": current_price,
                "low_3": low_last_3,
                "volume_ratio": last['Volume'] / last['Volume_SMA20']
            }

        if signal: # Added print statement
            print(f"Signal for {ticker}: {signal}, Confidence: {confidence}")
        return {"signal": signal, "confidence": int(confidence), "details": details}

    except Exception as e:
        logging.error(f"خطأ في تحليل {ticker}: {e}")
        return None

# ---------------------------
# 6. تصنيف الصفقة (سكالب، قصيرة، سوينق)
# ---------------------------

def classify_trade(signal, confidence, details):
    """
    تصنيف حسب قوة الإشارة وترتيب المتوسطات
    """
    if not signal:
        return None

    sma50 = details.get('sma50')
    sma100 = details.get('sma100')
    sma200 = details.get('sma200')

    # ترتيب المتوسطات
    if signal == "CALL":
        if sma50 > sma100 > sma200:
            return "📈 سوينق 3-7 أيام"
        elif confidence >= 75:
            return "⏳ صفقة قصيرة 1-2 يوم"
        else:
            return "⚡ سكالب"
    else:  # PUT
        if sma50 < sma100 < sma200:
            return "📈 سوينق 3-7 أيام"
        elif confidence >= 75:
            return "⏳ صفقة قصيرة 1-2 يوم"
        else:
            return "⚡ سكالب"

# ---------------------------
# 7. اختيار العقد المناسب (بيانات الخيارات)
# ---------------------------

def get_option_contract(ticker, signal, classification):
    """
    البحث عن عقد خيار مناسب حسب التصنيف.
    """
    try:
        stock = yf.Ticker(ticker)
        # جلب تواريخ انتهاء الصلاحية المتاحة
        expirations = stock.options
        if not expirations:
            return "غير متوفر"

        # اختيار التاريخ المناسب حسب التصنيف
        today = datetime.now().date()
        if "سكالب" in classification:
            days = 7
        elif "قصيرة" in classification:
            days = 14
        else:  # سوينق
            days = 30

        target_date = today + timedelta(days=days)
        # البحث عن أقرب تاريخ انتهاء للعدد المطلوب من الأيام
        exp_date = None
        for exp in expirations:
            exp_dt = datetime.strptime(exp, '%Y-%m-%d').date()
            if exp_dt >= target_date:
                exp_date = exp
                break
        if not exp_date:
            exp_date = expirations[0]  # أقرب تاريخ

        # جلب عقود الخيارات لذلك التاريخ
        opt_chain = stock.option_chain(exp_date)
        if signal == "CALL":
            contracts = opt_chain.calls
        else:
            contracts = opt_chain.puts

        if contracts.empty:
            return "غير متوفر"

        # فلترة العقود: Open Interest عالي (أعلى 10%)، Spread ضيق
        current_price = stock.history(period="1d")['Close'].iloc[-1]
        contracts['strike_diff'] = abs(contracts['strike'] - current_price)
        # فلتر OI
        oi_threshold = contracts['openInterest'].quantile(0.7)  # أعلى 30%
        eligible = contracts[contracts['openInterest'] > oi_threshold]

        if eligible.empty:
            eligible = contracts

        # نختار العقد الأقرب للسعر الحالي (للدلتا التقريبية)
        best = eligible.nsmallest(1, 'strike_diff').iloc[0]

        # حساب Spread تقريبي (ask-bid)/ask
        spread = (best['ask'] - best['bid']) / best['ask'] if best['ask'] > 0 else 1
        if spread > 0.05:   # أكبر من 5%
            return "سبريد واسع"

        # بناء نص العقد المقترح
        delta_approx = 0.5 - (best['strike_diff'] / current_price)  # تقريب بسيط
        delta_approx = max(0.1, min(0.9, delta_approx))

        return f"Delta {delta_approx:.2f} – Exp {exp_date} (أيام: {days})"

    except Exception as e:
        logging.error(f"خطأ في جلب الخيارات لـ {ticker}: {e}")
        return "غير متوفر"

# ---------------------------
# 8. بناء رسالة التنبيه النهائية
# ---------------------------

def build_alert_message(ticker, signal, confidence, classification, details, spx_trend, option_text):
    """
    تنسيق الرسالة كما طلبت تماماً.
    """
    # ترجمة الاتجاه
    if signal == "CALL":
        direction = "Call"
        cross = "50 > 100" if details.get('sma50', 0) > details.get('sma100', 0) else "50 > 200"
        price_status = f"السعر فوق {details.get('sma200', 0):.2f}"
        event = "اختراق قمة 2H"
    else:
        direction = "Put"
        cross = "50 < 100" if details.get('sma50', 0) < details.get('sma100', 0) else "50 < 200"
        price_status = f"السعر تحت {details.get('sma200', 0):.2f}"
        event = "كسر قاع 2H"

    # توافق SPX
    if "uptrend" in spx_trend and signal == "CALL":
        spx_compat = "✅ متوافق"
    elif "downtrend" in spx_trend and signal == "PUT":
        spx_compat = "✅ متوافق"
    else:
        spx_compat = "⚠️ غير متوافق"

    message = f"""
🔔 {ticker}
الاتجاه: {direction}
التصنيف: {classification}
التقاطع: {cross} (إيجابي)
الحالة: {price_status}
الحدث: {event}
SPX: {spx_compat}
الثقة: {confidence}/100
العقد المقترح: {option_text}
"""
    return message

# ---------------------------
# 9. دالة المسح الرئيسية
# ---------------------------

def scan_all_stocks():
    """تمر على كل الأسهم وتفحصها وترسل التنبيهات"""
    alerts = []
    print("Starting stock scan...") # Added print statement
    for ticker in TICKERS:
        logging.info(f"فحص {ticker}...")
        time.sleep(0.5)  # تأخير نصف ثانية لتجنب حظر Yahoo

        # 1. الفلتر الأساسي
        if not passes_basic_filters(ticker):
            print(f"Ticker {ticker} did not pass basic filters.") # Added print statement
            continue

        # 2. تحليل السهم
        result = analyze_stock_2h(ticker)
        if not result or not result['signal']:
            print(f"No signal found for {ticker} after 2h analysis.") # Added print statement
            continue

        signal = result['signal']
        confidence = result['confidence']
        details = result['details']

        # 3. تحليل SPX
        spx_15m = analyze_spx("15m")

        # 4. تصنيف الصفقة
        classification = classify_trade(signal, confidence, details)

        # 5. اختيار العقد
        option = get_option_contract(ticker, signal, classification)

        # 6. بناء الرسالة
        msg = build_alert_message(ticker, signal, confidence, classification, details, spx_15m, option)
        alerts.append(msg)
    print(f"Scan finished. Found {len(alerts)} alerts.") # Added print statement
    return alerts

# ---------------------------
# 10. دوال بوت تليجرام
# ---------------------------

async def send_alerts(context):
    """إرسال التنبيهات إلى تليجرام"""
    print("send_alerts function called.") # Added print statement
    alerts = scan_all_stocks()
    if not alerts:
        print("No alerts found to send.") # Added print statement
        await context.bot.send_message(chat_id=CHAT_ID, text="لا توجد إشارات جديدة حالياً.")
    else:
        print(f"Sending {len(alerts)} alerts.") # Added print statement
        for alert in alerts:
            await context.bot.send_message(chat_id=CHAT_ID, text=alert, parse_mode='Markdown')
            await asyncio.sleep(1)  # تجنب سبام تليجرام

# دالة للترحيب بالأعضاء الجدد وإرسال إخلاء المسؤولية
async def greet_chat_members(update: Update, context: ContextTypes.DEFAULT_TYPE) -> None:
    print(f"New chat member event: {update.chat_member}") # Added print
    if update.chat_member.new_chat_member.status == 'member':
        user = update.chat_member.new_chat_member.user
        chat_title = update.chat_member.chat.title or "هذه المجموعة"
        welcome_new_member_message = f"""
أهلاً وسهلاً بك {user.full_name} في {chat_title}! 👋

⚠️ **إخلاء مسؤولية:**
توصيات هذا البوت هي لأغراض تعليمية وتحليلية فقط، ولا تشكل نصيحة استثمارية بأي شكل من الأشكال. تداول العقود والمشتقات المالية ينطوي على مخاطر عالية وقد يؤدي إلى خسارة رأس المال. أنت وحدك المسؤول عن قراراتك الاستثمارية. يرجى استشارة مستشار مالي متخصص قبل اتخاذ أي قرارات استثمارية.
"""
        print(f"Sending welcome message to {user.full_name} in chat {update.chat_member.chat.id}") # Added print
        await context.bot.send_message(chat_id=update.chat_member.chat.id, text=welcome_new_member_message, parse_mode='Markdown')


async def start(update: Update, context: ContextTypes.DEFAULT_TYPE) -> None:
    """معالج أمر /start"""
    print(f"Start command received from {update.effective_user.id} in chat {update.effective_chat.id}") # Added print
    welcome_message = """
👋 مرحباً بك في بوت توصيات الأسهم!

هذا البوت يقدم توصيات تحليل فني للأسهم بناءً على استراتيجيات محددة.

⚠️ **إخلاء مسؤولية:**
توصيات هذا البوت هي لأغراض تعليمية وتحليلية فقط، ولا تشكل نصيحة استثمارية بأي شكل من الأشكال. تداول العقود والمشتقات المالية ينطوي على مخاطر عالية وقد يؤدي إلى خسارة رأس المال. أنت وحدك المسؤول عن قراراتك الاستثمارية. يرجى استشارة مستشار مالي متخصص قبل اتخاذ أي قرارات استثمارية.

اختر نوع التوصية التي ترغب في الحصول عليها:
"""

    keyboard = [
        [InlineKeyboardButton("📈 عقود CALL (شراء)", callback_data='recommend_call')],
        [InlineKeyboardButton("📉 عقود PUT (بيع)", callback_data='recommend_put')],
        [InlineKeyboardButton("🔄 مسح الآن", callback_data='scan_now_manual')]
    ]
    reply_markup = InlineKeyboardMarkup(keyboard)

    await update.message.reply_text(welcome_message, reply_markup=reply_markup, parse_mode='Markdown')

async def button_callback(update: Update, context: ContextTypes.DEFAULT_TYPE) -> None:
    query = update.callback_query
    print(f"Callback query received: {query.data} from {query.from_user.id}") # Added print
    await query.answer()  # Always answer callback queries

    if query.data == 'recommend_call':
        await query.edit_message_text(text="جاري البحث عن توصيات CALL...")
        # Logic to scan for CALL recommendations
        # For now, let's just send a placeholder message
        await asyncio.sleep(2) # Simulate work
        await query.edit_message_text(text="لم يتم العثور على توصيات CALL حالياً.") # Replace with actual results
    elif query.data == 'recommend_put':
        await query.edit_message_text(text="جاري البحث عن توصيات PUT...")
        # Logic to scan for PUT recommendations
        # For now, let's just send a placeholder message
        await asyncio.sleep(2) # Simulate work
        await query.edit_message_text(text="لم يتم العثور على توصيات PUT حالياً.") # Replace with actual results
    elif query.data == 'scan_now_manual':
        await query.edit_message_text(text="جاري المسح الفوري، انتظر قليلاً...")
        alerts = scan_all_stocks()
        if not alerts:
            await query.edit_message_text("لا توجد إشارات جديدة.")
        else:
            for alert in alerts:
                await query.message.reply_text(alert, parse_mode='Markdown')
                await asyncio.sleep(1)
            await query.edit_message_text(text="تم الانتهاء من المسح الفوري.")


async def scan_now(update, context):
    """معالج أمر /scan لمسح فوري"""
    print("scan_now command called.") # Added print statement
    await update.message.reply_text("جاري المسح، انتظر قليلاً...")
    alerts = scan_all_stocks()
    if not alerts:
        print("No alerts found for manual scan.") # Added print statement
        await update.message.reply_text("لا توجد إشارات جديدة.")
    else:
        print(f"Sending {len(alerts)} alerts for manual scan.") # Added print statement
        for alert in alerts:
            await update.message.reply_text(alert, parse_mode='Markdown')
            await asyncio.sleep(1)

def main():
    """الدالة الرئيسية لتشغيل البوت"""
    # إنشاء تطبيق البوت
    # Create a JobQueue instance first
    job_queue_instance = JobQueue()

    # Pass the JobQueue instance to the ApplicationBuilder
    application = Application.builder().token(TELEGRAM_TOKEN).job_queue(job_queue_instance).build()

    # إضافة المعالجات للأوامر والرسائل
    application.add_handler(CommandHandler("start", start))
    application.add_handler(CommandHandler("scan", scan_now))
    application.add_handler(CallbackQueryHandler(button_callback)) # إضافة معالج لأزرار الـ Inline Keyboard
    application.add_handler(ChatMemberHandler(greet_chat_members, ChatMemberHandler.CHAT_MEMBER)) # إضافة معالج للأعضاء الجدد

    # جدولة المسح التلقائي كل ساعة
    job_queue_instance.run_repeating(send_alerts, interval=3600, first=10)  # كل ساعة، أول مرة بعد 10 ثواني

    # تشغيل البوت
    print("البوت يعمل الآن... اذهب لتجربته!")
    application.run_polling()
    # application.idle() is not necessary after run_polling() as it's a blocking call

# استدعاء الدالة الرئيسية
if __name__ == '__main__':
    main()
````

````python
#!/usr/bin/env python3
# -*- coding: utf-8 -*-

# بوت توصيات الأسهم - نسخة شاكر 1.0 (نسخة كاملة)
# آخر تحديث: فبراير 2026

import yfinance as yf
import pandas as pd
import numpy as np
import ta
import logging
import time
import schedule
from datetime import datetime, timedelta
from telegram import Bot, InlineKeyboardButton, InlineKeyboardMarkup, Update
from telegram.ext import Application, CommandHandler, CallbackQueryHandler, ContextTypes, JobQueue, ChatMemberHandler # أضفنا ChatMemberHandler
import asyncio
import os

# ---------------------------
# 1. إعدادات البوت والتوكن
# ---------------------------
TELEGRAM_TOKEN = "8699821370:AAEQUSbLTgf7MmWqo5vV5LHPOz30wfqOfqw"      # ضع التوكن الخاص بك
CHAT_ID = "7854020427"                # ضع معرف المستخدم أو القناة

# تفعيل السجلات لمشاهدة الأخطاء
logging.basicConfig(level=logging.INFO, format='%(asctime)s - %(name)s - %(levelname)s - %(message)s')

# ---------------------------
# 2. قائمة الأسهم الكاملة (مدمجة)
# ---------------------------
TICKERS = [
    # --- أسهمك القديمة (من الصور) ---
    "A", "AA", "ABEV", "ABNB", "ABT", "ABVX", "ACGL", "ACM", "ACN",
    "ADBE", "ADI", "ADM", "ADP", "ADSK", "AEE", "AEG", "AEIS", "AEM",
    "AEP", "AER", "AES", "AFG", "AFL", "AFRM", "AG", "AGI", "AGNC",
    "AIG", "AIT", "AIZ", "AJG", "AKAM", "ALAB", "ALB", "ALC", "ALGN",
    "ALL", "ALLE", "ALLY", "ALNY", "ALSN", "AM", "AMCR", "AME", "AMH",
    "AMKR", "AMP", "AMRZ", "AMT", "AMX",

    # --- الشركات العملاقة (> 1 تريليون دولار) ---
    "NVDA", "AAPL", "GOOGL", "MSFT", "AMZN", "META", "TSM", "AVGO", "TSLA",
    "BRK.A", "WMT",

    # --- الشركات الكبرى (200 مليار - 1 تريليون) ---
    "LLY", "JPM", "XOM", "V", "JNJ", "MA", "MU", "ORCL", "COST", "AMD",
    "NFLX", "PLTR", "CSCO", "MRK", "TMO", "ASML", "ABBV", "HD", "CVX",
    "PG", "CAT", "KO", "PEP", "INTC", "CMCSA", "MCD", "TXN", "VZ", "AMGN"
]

# ---------------------------
# 3. دوال التحليل الأساسي (الفلاتر)
# ---------------------------

def get_stock_info(ticker):
    """جلب معلومات الشركة الأساسية"""
    try:
        stock = yf.Ticker(ticker)
        info = stock.info
        return info
    except Exception as e:
        logging.error(f"خطأ في جلب معلومات {ticker}: {e}")
        return None

def passes_basic_filters(ticker):
    """
    تطبيق الفلتر الأساسي:
    - Market Cap > 10B (ملاحظة: أغلب الشركات في القائمة أكبر من ذلك بكثير، لكن الفلتر يضمن ذلك)
    - Average Volume > 1M (تقريبي)
    - Spread < 5%
    """
    info = get_stock_info(ticker)
    if not info:
        return False

    # 1. القيمة السوقية
    mcap = info.get('marketCap', 0)
    if mcap < 10e9:
        print(f"{ticker} فشل فلتر Market Cap: {mcap}") # Changed from logging.debug
        return False

    # 2. متوسط الحجم (نأخذ آخر 20 يوم)
    try:
        stock = yf.Ticker(ticker)
        hist = stock.history(period="1mo", interval="1d")
        if hist.empty:
            return False
        avg_vol = hist['Volume'].rolling(20).mean().iloc[-1]
        if avg_vol < 1_000_000:   # أقل من مليون سهم يومياً
            print(f"{ticker} فشل فلتر الحجم: {avg_vol}") # Changed from logging.debug
            return False
    except:
        return False

    # 3. Spread (bid/ask)
    bid = info.get('bid', 0)
    ask = info.get('ask', 0)
    if bid and ask:
        spread = (ask - bid) / bid if bid != 0 else 1
        if spread > 0.05:   # أكبر من 5%
            print(f"{ticker} فشل فلتر Spread: {spread:.2%}") # Changed from logging.debug
            return False

    return True

# ---------------------------
# 4. تحليل SPX (مؤشر S&P 500)
# ---------------------------

def analyze_spx(timeframe="15m"):
    """تحليل مؤشر SPX على الإطار المطلوب"""
    try:
        spx = yf.Ticker("^GSPC")  # أو "SPY"
        hist = spx.history(period="5d", interval=timeframe)
        if len(hist) < 50:
            return "neutral"

        # متوسطات بسيطة لتحديد الاتجاه
        hist['SMA20'] = hist['Close'].rolling(20).mean()
        last = hist.iloc[-1]
        prev = hist.iloc[-2]

        if last['Close'] > last['SMA20'] and last['Close'] > prev['Close']:
            return "uptrend"
        elif last['Close'] < last['SMA20'] and last['Close'] < prev['Close']:
            return "downtrend"
        else:
            return "neutral"
    except:
        return "neutral"

# ---------------------------
# 5. تحليل السهم (فريم ساعتين)
# ---------------------------

def analyze_stock_2h(ticker):
    """
    تحليل فني كامل على فريم الساعتين
    يعيد: {'signal': 'CALL'/'PUT'/None, 'confidence': 0-100, 'details': {...}}
    """
    try:
        print(f"Analyzing {ticker} for 2h interval.") # Added print statement
        stock = yf.Ticker(ticker)
        hist = stock.history(period="1mo", interval="2h")  # آخر شهر، فريم ساعتين
        if len(hist) < 200:
            print(f"Not enough historical data for {ticker} in 2h interval.") # Added print statement
            return None

        # حساب المتوسطات
        hist['SMA50'] = hist['Close'].rolling(50).mean()
        hist['SMA100'] = hist['Close'].rolling(100).mean()
        hist['SMA200'] = hist['Close'].rolling(200).mean()
        hist['Volume_SMA20'] = hist['Volume'].rolling(20).mean()

        last = hist.iloc[-1]
        current_price = last['Close']
        sma50 = last['SMA50']
        sma100 = last['SMA100']
        sma200 = last['SMA200']

        # قمم وقيعان آخر 3 شموع
        high_last_3 = hist['High'].iloc[-3:].max()
        low_last_3 = hist['Low'].iloc[-3:].min()

        # قمة/قاع اليوم (آخر 24 ساعة = 12 شمعة)
        high_today = hist['High'].iloc[-12:].max()
        low_today = hist['Low'].iloc[-12:].min()

        volume_ok = last['Volume'] > last['Volume_SMA20']

        # شروط CALL
        call_condition1 = (sma50 > sma100) or (sma50 > sma200)
        call_condition2 = current_price > sma200
        call_condition3 = current_price > high_last_3 or current_price > high_today
        call_condition4 = volume_ok

        # شروط PUT
        put_condition1 = (sma50 < sma100) or (sma50 < sma200)
        put_condition2 = current_price < sma200
        put_condition3 = current_price < low_last_3 or current_price < low_today
        put_condition4 = volume_ok

        signal = None
        confidence = 0
        details = {}

        if call_condition1 and call_condition2 and call_condition3 and call_condition4:
            signal = "CALL"
            # حساب الثقة: نضيف 25 لكل شرط متحقق
            confidence = 25 * sum([call_condition1, call_condition2, call_condition3, call_condition4])
            details = {
                "sma50": sma50,
                "sma100": sma100,
                "sma200": sma200,
                "price": current_price,
                "high_3": high_last_3,
                "volume_ratio": last['Volume'] / last['Volume_SMA20']
            }
        elif put_condition1 and put_condition2 and put_condition3 and put_condition4:
            signal = "PUT"
            confidence = 25 * sum([put_condition1, put_condition2, put_condition3, put_condition4])
            details = {
                "sma50": sma50,
                "sma100": sma100,
                "sma200": sma200,
                "price": current_price,
                "low_3": low_last_3,
                "volume_ratio": last['Volume'] / last['Volume_SMA20']
            }

        if signal: # Added print statement
            print(f"Signal for {ticker}: {signal}, Confidence: {confidence}")
        return {"signal": signal, "confidence": int(confidence), "details": details}

    except Exception as e:
        logging.error(f"خطأ في تحليل {ticker}: {e}")
        return None

# ---------------------------
# 6. تصنيف الصفقة (سكالب، قصيرة، سوينق)
# ---------------------------

def classify_trade(signal, confidence, details):
    """
    تصنيف حسب قوة الإشارة وترتيب المتوسطات
    """
    if not signal:
        return None

    sma50 = details.get('sma50')
    sma100 = details.get('sma100')
    sma200 = details.get('sma200')

    # ترتيب المتوسطات
    if signal == "CALL":
        if sma50 > sma100 > sma200:
            return "📈 سوينق 3-7 أيام"
        elif confidence >= 75:
            return "⏳ صفقة قصيرة 1-2 يوم"
        else:
            return "⚡ سكالب"
    else:  # PUT
        if sma50 < sma100 < sma200:
            return "📈 سوينق 3-7 أيام"
        elif confidence >= 75:
            return "⏳ صفقة قصيرة 1-2 يوم"
        else:
            return "⚡ سكالب"

# ---------------------------
# 7. اختيار العقد المناسب (بيانات الخيارات)
# ---------------------------

def get_option_contract(ticker, signal, classification):
    """
    البحث عن عقد خيار مناسب حسب التصنيف.
    """
    try:
        stock = yf.Ticker(ticker)
        # جلب تواريخ انتهاء الصلاحية المتاحة
        expirations = stock.options
        if not expirations:
            return "غير متوفر"

        # اختيار التاريخ المناسب حسب التصنيف
        today = datetime.now().date()
        if "سكالب" in classification:
            days = 7
        elif "قصيرة" in classification:
            days = 14
        else:  # سوينق
            days = 30

        target_date = today + timedelta(days=days)
        # البحث عن أقرب تاريخ انتهاء للعدد المطلوب من الأيام
        exp_date = None
        for exp in expirations:
            exp_dt = datetime.strptime(exp, '%Y-%m-%d').date()
            if exp_dt >= target_date:
                exp_date = exp
                break
        if not exp_date:
            exp_date = expirations[0]  # أقرب تاريخ

        # جلب عقود الخيارات لذلك التاريخ
        opt_chain = stock.option_chain(exp_date)
        if signal == "CALL":
            contracts = opt_chain.calls
        else:
            contracts = opt_chain.puts

        if contracts.empty:
            return "غير متوفر"

        # فلترة العقود: Open Interest عالي (أعلى 10%)، Spread ضيق
        current_price = stock.history(period="1d")['Close'].iloc[-1]
        contracts['strike_diff'] = abs(contracts['strike'] - current_price)
        # فلتر OI
        oi_threshold = contracts['openInterest'].quantile(0.7)  # أعلى 30%
        eligible = contracts[contracts['openInterest'] > oi_threshold]

        if eligible.empty:
            eligible = contracts

        # نختار العقد الأقرب للسعر الحالي (للدلتا التقريبية)
        best = eligible.nsmallest(1, 'strike_diff').iloc[0]

        # حساب Spread تقريبي (ask-bid)/ask
        spread = (best['ask'] - best['bid']) / best['ask'] if best['ask'] > 0 else 1
        if spread > 0.05:   # أكبر من 5%
            return "سبريد واسع"

        # بناء نص العقد المقترح
        delta_approx = 0.5 - (best['strike_diff'] / current_price)  # تقريب بسيط
        delta_approx = max(0.1, min(0.9, delta_approx))

        return f"Delta {delta_approx:.2f} – Exp {exp_date} (أيام: {days})"

    except Exception as e:
        logging.error(f"خطأ في جلب الخيارات لـ {ticker}: {e}")
        return "غير متوفر"

# ---------------------------
# 8. بناء رسالة التنبيه النهائية
# ---------------------------

def build_alert_message(ticker, signal, confidence, classification, details, spx_trend, option_text):
    """
    تنسيق الرسالة كما طلبت تماماً.
    """
    # ترجمة الاتجاه
    if signal == "CALL":
        direction = "Call"
        cross = "50 > 100" if details.get('sma50', 0) > details.get('sma100', 0) else "50 > 200"
        price_status = f"السعر فوق {details.get('sma200', 0):.2f}"
        event = "اختراق قمة 2H"
    else:
        direction = "Put"
        cross = "50 < 100" if details.get('sma50', 0) < details.get('sma100', 0) else "50 < 200"
        price_status = f"السعر تحت {details.get('sma200', 0):.2f}"
        event = "كسر قاع 2H"

    # توافق SPX
    if "uptrend" in spx_trend and signal == "CALL":
        spx_compat = "✅ متوافق"
    elif "downtrend" in spx_trend and signal == "PUT":
        spx_compat = "✅ متوافق"
    else:
        spx_compat = "⚠️ غير متوافق"

    message = f"""
🔔 {ticker}
الاتجاه: {direction}
التصنيف: {classification}
التقاطع: {cross} (إيجابي)
الحالة: {price_status}
الحدث: {event}
SPX: {spx_compat}
الثقة: {confidence}/100
العقد المقترح: {option_text}
"""
    return message

# ---------------------------
# 9. دالة المسح الرئيسية
# ---------------------------

def scan_all_stocks():
    """تمر على كل الأسهم وتفحصها وترسل التنبيهات"""
    alerts = []
    print("Starting stock scan...") # Added print statement
    for ticker in TICKERS:
        logging.info(f"فحص {ticker}...")
        time.sleep(0.5)  # تأخير نصف ثانية لتجنب حظر Yahoo

        # 1. الفلتر الأساسي
        if not passes_basic_filters(ticker):
            print(f"Ticker {ticker} did not pass basic filters.") # Added print statement
            continue

        # 2. تحليل السهم
        result = analyze_stock_2h(ticker)
        if not result or not result['signal']:
            print(f"No signal found for {ticker} after 2h analysis.") # Added print statement
            continue

        signal = result['signal']
        confidence = result['confidence']
        details = result['details']

        # 3. تحليل SPX
        spx_15m = analyze_spx("15m")

        # 4. تصنيف الصفقة
        classification = classify_trade(signal, confidence, details)

        # 5. اختيار العقد
        option = get_option_contract(ticker, signal, classification)

        # 6. بناء الرسالة
        msg = build_alert_message(ticker, signal, confidence, classification, details, spx_15m, option)
        alerts.append(msg)
    print(f"Scan finished. Found {len(alerts)} alerts.") # Added print statement
    return alerts

# ---------------------------
# 10. دوال بوت تليجرام
# ---------------------------

async def send_alerts(context):
    """إرسال التنبيهات إلى تليجرام"""
    print("send_alerts function called.") # Added print statement
    alerts = scan_all_stocks()
    if not alerts:
        print("No alerts found to send.") # Added print statement
        await context.bot.send_message(chat_id=CHAT_ID, text="لا توجد إشارات جديدة حالياً.")
    else:
        print(f"Sending {len(alerts)} alerts.") # Added print statement
        for alert in alerts:
            await context.bot.send_message(chat_id=CHAT_ID, text=alert, parse_mode='Markdown')
            await asyncio.sleep(1)  # تجنب سبام تليجرام

# دالة للترحيب بالأعضاء الجدد وإرسال إخلاء المسؤولية
async def greet_chat_members(update: Update, context: ContextTypes.DEFAULT_TYPE) -> None:
    print(f"New chat member event: {update.chat_member}") # Added print
    if update.chat_member.new_chat_member.status == 'member':
        user = update.chat_member.new_chat_member.user
        chat_title = update.chat_member.chat.title or "هذه المجموعة"
        welcome_new_member_message = f"""
أهلاً وسهلاً بك {user.full_name} في {chat_title}! 👋

⚠️ **إخلاء مسؤولية:**
توصيات هذا البوت هي لأغراض تعليمية وتحليلية فقط، ولا تشكل نصيحة استثمارية بأي شكل من الأشكال. تداول العقود والمشتقات المالية ينطوي على مخاطر عالية وقد يؤدي إلى خسارة رأس المال. أنت وحدك المسؤول عن قراراتك الاستثمارية. يرجى استشارة مستشار مالي متخصص قبل اتخاذ أي قرارات استثمارية.
"""
        print(f"Sending welcome message to {user.full_name} in chat {update.chat_member.chat.id}") # Added print
        await context.bot.send_message(chat_id=update.chat_member.chat.id, text=welcome_new_member_message, parse_mode='Markdown')


async def start(update: Update, context: ContextTypes.DEFAULT_TYPE) -> None:
    """معالج أمر /start"""
    print(f"Start command received from {update.effective_user.id} in chat {update.effective_chat.id}") # Added print
    welcome_message = """
👋 مرحباً بك في بوت توصيات الأسهم!

هذا البوت يقدم توصيات تحليل فني للأسهم بناءً على استراتيجيات محددة.

⚠️ **إخلاء مسؤولية:**
توصيات هذا البوت هي لأغراض تعليمية وتحليلية فقط، ولا تشكل نصيحة استثمارية بأي شكل من الأشكال. تداول العقود والمشتقات المالية ينطوي على مخاطر عالية وقد يؤدي إلى خسارة رأس المال. أنت وحدك المسؤول عن قراراتك الاستثمارية. يرجى استشارة مستشار مالي متخصص قبل اتخاذ أي قرارات استثمارية.

اختر نوع التوصية التي ترغب في الحصول عليها:
"""

    keyboard = [
        [InlineKeyboardButton("📈 عقود CALL (شراء)", callback_data='recommend_call')],
        [InlineKeyboardButton("📉 عقود PUT (بيع)", callback_data='recommend_put')],
        [InlineKeyboardButton("🔄 مسح الآن", callback_data='scan_now_manual')]
    ]
    reply_markup = InlineKeyboardMarkup(keyboard)

    await update.message.reply_text(welcome_message, reply_markup=reply_markup, parse_mode='Markdown')

async def button_callback(update: Update, context: ContextTypes.DEFAULT_TYPE) -> None:
    query = update.callback_query
    print(f"Callback query received: {query.data} from {query.from_user.id}") # Added print
    await query.answer()  # Always answer callback queries

    if query.data == 'recommend_call':
        await query.edit_message_text(text="جاري البحث عن توصيات CALL...")
        # Logic to scan for CALL recommendations
        # For now, let's just send a placeholder message
        await asyncio.sleep(2) # Simulate work
        await query.edit_message_text(text="لم يتم العثور على توصيات CALL حالياً.") # Replace with actual results
    elif query.data == 'recommend_put':
        await query.edit_message_text(text="جاري البحث عن توصيات PUT...")
        # Logic to scan for PUT recommendations
        # For now, let's just send a placeholder message
        await asyncio.sleep(2) # Simulate work
        await query.edit_message_text(text="لم يتم العثور على توصيات PUT حالياً.") # Replace with actual results
    elif query.data == 'scan_now_manual':
        await query.edit_message_text(text="جاري المسح الفوري، انتظر قليلاً...")
        alerts = scan_all_stocks()
        if not alerts:
            await query.edit_message_text("لا توجد إشارات جديدة.")
        else:
            for alert in alerts:
                await query.message.reply_text(alert, parse_mode='Markdown')
                await asyncio.sleep(1)
            await query.edit_message_text(text="تم الانتهاء من المسح الفوري.")


async def scan_now(update, context):
    """معالج أمر /scan لمسح فوري"""
    print("scan_now command called.") # Added print statement
    await update.message.reply_text("جاري المسح، انتظر قليلاً...")
    alerts = scan_all_stocks()
    if not alerts:
        print("No alerts found for manual scan.") # Added print statement
        await update.message.reply_text("لا توجد إشارات جديدة.")
    else:
        print(f"Sending {len(alerts)} alerts for manual scan.") # Added print statement
        for alert in alerts:
            await update.message.reply_text(alert, parse_mode='Markdown')
            await asyncio.sleep(1)

def main():
    """الدالة الرئيسية لتشغيل البوت"""
    # إنشاء تطبيق البوت
    # Create a JobQueue instance first
    job_queue_instance = JobQueue()

    # Pass the JobQueue instance to the ApplicationBuilder
    application = Application.builder().token(TELEGRAM_TOKEN).job_queue(job_queue_instance).build()

    # إضافة المعالجات للأوامر والرسائل
    application.add_handler(CommandHandler("start", start))
    application.add_handler(CommandHandler("scan", scan_now))
    application.add_handler(CallbackQueryHandler(button_callback)) # إضافة معالج لأزرار الـ Inline Keyboard
    application.add_handler(ChatMemberHandler(greet_chat_members, ChatMemberHandler.CHAT_MEMBER)) # إضافة معالج للأعضاء الجدد

    # جدولة المسح التلقائي كل ساعة
    job_queue_instance.run_repeating(send_alerts, interval=3600, first=10)  # كل ساعة، أول مرة بعد 10 ثواني

    # تشغيل البوت
    print("البوت يعمل الآن... اذهب لتجربته!")
    application.run_polling()
    # application.idle() is not necessary after run_polling() as it's a blocking call

# استدعاء الدالة الرئيسية
if __name__ == '__main__':
    main()
````

````python
#!/usr/bin/env python3
# -*- coding: utf-8 -*-

# بوت توصيات الأسهم - نسخة شاكر 1.0 (نسخة كاملة)
# آخر تحديث: فبراير 2026

import yfinance as yf
import pandas as pd
import numpy as np
import ta
import logging
import time
import schedule
from datetime import datetime, timedelta
from telegram import Bot, InlineKeyboardButton, InlineKeyboardMarkup, Update
from telegram.ext import Application, CommandHandler, CallbackQueryHandler, ContextTypes, JobQueue, ChatMemberHandler # أضفنا ChatMemberHandler
import asyncio
import os

# ---------------------------
# 1. إعدادات البوت والتوكن
# ---------------------------
TELEGRAM_TOKEN = "8699821370:AAEQUSbLTgf7MmWqo5vV5LHPOz30wfqOfqw"      # ضع التوكن الخاص بك
CHAT_ID = "7854020427"                # ضع معرف المستخدم أو القناة

# تفعيل السجلات لمشاهدة الأخطاء
logging.basicConfig(level=logging.INFO, format='%(asctime)s - %(name)s - %(levelname)s - %(message)s')

# ---------------------------
# 2. قائمة الأسهم الكاملة (مدمجة)
# ---------------------------
TICKERS = [
    # --- أسهمك القديمة (من الصور) ---
    "A", "AA", "ABEV", "ABNB", "ABT", "ABVX", "ACGL", "ACM", "ACN",
    "ADBE", "ADI", "ADM", "ADP", "ADSK", "AEE", "AEG", "AEIS", "AEM",
    "AEP", "AER", "AES", "AFG", "AFL", "AFRM", "AG", "AGI", "AGNC",
    "AIG", "AIT", "AIZ", "AJG", "AKAM", "ALAB", "ALB", "ALC", "ALGN",
    "ALL", "ALLE", "ALLY", "ALNY", "ALSN", "AM", "AMCR", "AME", "AMH",
    "AMKR", "AMP", "AMRZ", "AMT", "AMX",

    # --- الشركات العملاقة (> 1 تريليون دولار) ---
    "NVDA", "AAPL", "GOOGL", "MSFT", "AMZN", "META", "TSM", "AVGO", "TSLA",
    "BRK.A", "WMT",

    # --- الشركات الكبرى (200 مليار - 1 تريليون) ---
    "LLY", "JPM", "XOM", "V", "JNJ", "MA", "MU", "ORCL", "COST", "AMD",
    "NFLX", "PLTR", "CSCO", "MRK", "TMO", "ASML", "ABBV", "HD", "CVX",
    "PG", "CAT", "KO", "PEP", "INTC", "CMCSA", "MCD", "TXN", "VZ", "AMGN"
]

# ---------------------------
# 3. دوال التحليل الأساسي (الفلاتر)
# ---------------------------

def get_stock_info(ticker):
    """جلب معلومات الشركة الأساسية"""
    try:
        stock = yf.Ticker(ticker)
        info = stock.info
        return info
    except Exception as e:
        logging.error(f"خطأ في جلب معلومات {ticker}: {e}")
        return None

def passes_basic_filters(ticker):
    """
    تطبيق الفلتر الأساسي:
    - Market Cap > 10B (ملاحظة: أغلب الشركات في القائمة أكبر من ذلك بكثير، لكن الفلتر يضمن ذلك)
    - Average Volume > 1M (تقريبي)
    - Spread < 5%
    """
    info = get_stock_info(ticker)
    if not info:
        return False

    # 1. القيمة السوقية
    mcap = info.get('marketCap', 0)
    if mcap < 10e9:
        print(f"{ticker} فشل فلتر Market Cap: {mcap}") # Changed from logging.debug
        return False

    # 2. متوسط الحجم (نأخذ آخر 20 يوم)
    try:
        stock = yf.Ticker(ticker)
        hist = stock.history(period="1mo", interval="1d")
        if hist.empty:
            return False
        avg_vol = hist['Volume'].rolling(20).mean().iloc[-1]
        if avg_vol < 1_000_000:   # أقل من مليون سهم يومياً
            print(f"{ticker} فشل فلتر الحجم: {avg_vol}") # Changed from logging.debug
            return False
    except:
        return False

    # 3. Spread (bid/ask)
    bid = info.get('bid', 0)
    ask = info.get('ask', 0)
    if bid and ask:
        spread = (ask - bid) / bid if bid != 0 else 1
        if spread > 0.05:   # أكبر من 5%
            print(f"{ticker} فشل فلتر Spread: {spread:.2%}") # Changed from logging.debug
            return False

    return True

# ---------------------------
# 4. تحليل SPX (مؤشر S&P 500)
# ---------------------------

def analyze_spx(timeframe="15m"):
    """تحليل مؤشر SPX على الإطار المطلوب"""
    try:
        spx = yf.Ticker("^GSPC")  # أو "SPY"
        hist = spx.history(period="5d", interval=timeframe)
        if len(hist) < 50:
            return "neutral"

        # متوسطات بسيطة لتحديد الاتجاه
        hist['SMA20'] = hist['Close'].rolling(20).mean()
        last = hist.iloc[-1]
        prev = hist.iloc[-2]

        if last['Close'] > last['SMA20'] and last['Close'] > prev['Close']:
            return "uptrend"
        elif last['Close'] < last['SMA20'] and last['Close'] < prev['Close']:
            return "downtrend"
        else:
            return "neutral"
    except:
        return "neutral"

# ---------------------------
# 5. تحليل السهم (فريم ساعتين)
# ---------------------------

def analyze_stock_2h(ticker):
    """
    تحليل فني كامل على فريم الساعتين
    يعيد: {'signal': 'CALL'/'PUT'/None, 'confidence': 0-100, 'details': {...}}
    """
    try:
        print(f"Analyzing {ticker} for 2h interval.") # Added print statement
        stock = yf.Ticker(ticker)
        hist = stock.history(period="1mo", interval="2h")  # آخر شهر، فريم ساعتين
        if len(hist) < 200:
            print(f"Not enough historical data for {ticker} in 2h interval.") # Added print statement
            return None

        # حساب المتوسطات
        hist['SMA50'] = hist['Close'].rolling(50).mean()
        hist['SMA100'] = hist['Close'].rolling(100).mean()
        hist['SMA200'] = hist['Close'].rolling(200).mean()
        hist['Volume_SMA20'] = hist['Volume'].rolling(20).mean()

        last = hist.iloc[-1]
        current_price = last['Close']
        sma50 = last['SMA50']
        sma100 = last['SMA100']
        sma200 = last['SMA200']

        # قمم وقيعان آخر 3 شموع
        high_last_3 = hist['High'].iloc[-3:].max()
        low_last_3 = hist['Low'].iloc[-3:].min()

        # قمة/قاع اليوم (آخر 24 ساعة = 12 شمعة)
        high_today = hist['High'].iloc[-12:].max()
        low_today = hist['Low'].iloc[-12:].min()

        volume_ok = last['Volume'] > last['Volume_SMA20']

        # شروط CALL
        call_condition1 = (sma50 > sma100) or (sma50 > sma200)
        call_condition2 = current_price > sma200
        call_condition3 = current_price > high_last_3 or current_price > high_today
        call_condition4 = volume_ok

        # شروط PUT
        put_condition1 = (sma50 < sma100) or (sma50 < sma200)
        put_condition2 = current_price < sma200
        put_condition3 = current_price < low_last_3 or current_price < low_today
        put_condition4 = volume_ok

        signal = None
        confidence = 0
        details = {}

        if call_condition1 and call_condition2 and call_condition3 and call_condition4:
            signal = "CALL"
            # حساب الثقة: نضيف 25 لكل شرط متحقق
            confidence = 25 * sum([call_condition1, call_condition2, call_condition3, call_condition4])
            details = {
                "sma50": sma50,
                "sma100": sma100,
                "sma200": sma200,
                "price": current_price,
                "high_3": high_last_3,
                "volume_ratio": last['Volume'] / last['Volume_SMA20']
            }
        elif put_condition1 and put_condition2 and put_condition3 and put_condition4:
            signal = "PUT"
            confidence = 25 * sum([put_condition1, put_condition2, put_condition3, put_condition4])
            details = {
                "sma50": sma50,
                "sma100": sma100,
                "sma200": sma200,
                "price": current_price,
                "low_3": low_last_3,
                "volume_ratio": last['Volume'] / last['Volume_SMA20']
            }

        if signal: # Added print statement
            print(f"Signal for {ticker}: {signal}, Confidence: {confidence}")
        return {"signal": signal, "confidence": int(confidence), "details": details}

    except Exception as e:
        logging.error(f"خطأ في تحليل {ticker}: {e}")
        return None

# ---------------------------
# 6. تصنيف الصفقة (سكالب، قصيرة، سوينق)
# ---------------------------

def classify_trade(signal, confidence, details):
    """
    تصنيف حسب قوة الإشارة وترتيب المتوسطات
    """
    if not signal:
        return None

    sma50 = details.get('sma50')
    sma100 = details.get('sma100')
    sma200 = details.get('sma200')

    # ترتيب المتوسطات
    if signal == "CALL":
        if sma50 > sma100 > sma200:
            return "📈 سوينق 3-7 أيام"
        elif confidence >= 75:
            return "⏳ صفقة قصيرة 1-2 يوم"
        else:
            return "⚡ سكالب"
    else:  # PUT
        if sma50 < sma100 < sma200:
            return "📈 سوينق 3-7 أيام"
        elif confidence >= 75:
            return "⏳ صفقة قصيرة 1-2 يوم"
        else:
            return "⚡ سكالب"

# ---------------------------
# 7. اختيار العقد المناسب (بيانات الخيارات)
# ---------------------------

def get_option_contract(ticker, signal, classification):
    """
    البحث عن عقد خيار مناسب حسب التصنيف.
    """
    try:
        stock = yf.Ticker(ticker)
        # جلب تواريخ انتهاء الصلاحية المتاحة
        expirations = stock.options
        if not expirations:
            return "غير متوفر"

        # اختيار التاريخ المناسب حسب التصنيف
        today = datetime.now().date()
        if "سكالب" in classification:
            days = 7
        elif "قصيرة" in classification:
            days = 14
        else:  # سوينق
            days = 30

        target_date = today + timedelta(days=days)
        # البحث عن أقرب تاريخ انتهاء للعدد المطلوب من الأيام
        exp_date = None
        for exp in expirations:
            exp_dt = datetime.strptime(exp, '%Y-%m-%d').date()
            if exp_dt >= target_date:
                exp_date = exp
                break
        if not exp_date:
            exp_date = expirations[0]  # أقرب تاريخ

        # جلب عقود الخيارات لذلك التاريخ
        opt_chain = stock.option_chain(exp_date)
        if signal == "CALL":
            contracts = opt_chain.calls
        else:
            contracts = opt_chain.puts

        if contracts.empty:
            return "غير متوفر"

        # فلترة العقود: Open Interest عالي (أعلى 10%)، Spread ضيق
        current_price = stock.history(period="1d")['Close'].iloc[-1]
        contracts['strike_diff'] = abs(contracts['strike'] - current_price)
        # فلتر OI
        oi_threshold = contracts['openInterest'].quantile(0.7)  # أعلى 30%
        eligible = contracts[contracts['openInterest'] > oi_threshold]

        if eligible.empty:
            eligible = contracts

        # نختار العقد الأقرب للسعر الحالي (للدلتا التقريبية)
        best = eligible.nsmallest(1, 'strike_diff').iloc[0]

        # حساب Spread تقريبي (ask-bid)/ask
        spread = (best['ask'] - best['bid']) / best['ask'] if best['ask'] > 0 else 1
        if spread > 0.05:   # أكبر من 5%
            return "سبريد واسع"

        # بناء نص العقد المقترح
        delta_approx = 0.5 - (best['strike_diff'] / current_price)  # تقريب بسيط
        delta_approx = max(0.1, min(0.9, delta_approx))

        return f"Delta {delta_approx:.2f} – Exp {exp_date} (أيام: {days})"

    except Exception as e:
        logging.error(f"خطأ في جلب الخيارات لـ {ticker}: {e}")
        return "غير متوفر"

# ---------------------------
# 8. بناء رسالة التنبيه النهائية
# ---------------------------

def build_alert_message(ticker, signal, confidence, classification, details, spx_trend, option_text):
    """
    تنسيق الرسالة كما طلبت تماماً.
    """
    # ترجمة الاتجاه
    if signal == "CALL":
        direction = "Call"
        cross = "50 > 100" if details.get('sma50', 0) > details.get('sma100', 0) else "50 > 200"
        price_status = f"السعر فوق {details.get('sma200', 0):.2f}"
        event = "اختراق قمة 2H"
    else:
        direction = "Put"
        cross = "50 < 100" if details.get('sma50', 0) < details.get('sma100', 0) else "50 < 200"
        price_status = f"السعر تحت {details.get('sma200', 0):.2f}"
        event = "كسر قاع 2H"

    # توافق SPX
    if "uptrend" in spx_trend and signal == "CALL":
        spx_compat = "✅ متوافق"
    elif "downtrend" in spx_trend and signal == "PUT":
        spx_compat = "✅ متوافق"
    else:
        spx_compat = "⚠️ غير متوافق"

    message = f"""
🔔 {ticker}
الاتجاه: {direction}
التصنيف: {classification}
التقاطع: {cross} (إيجابي)
الحالة: {price_status}
الحدث: {event}
SPX: {spx_compat}
الثقة: {confidence}/100
العقد المقترح: {option_text}
"""
    return message

# ---------------------------
# 9. دالة المسح الرئيسية
# ---------------------------

def scan_all_stocks():
    """تمر على كل الأسهم وتفحصها وترسل التنبيهات"""
    alerts = []
    print("Starting stock scan...") # Added print statement
    for ticker in TICKERS:
        logging.info(f"فحص {ticker}...")
        time.sleep(0.5)  # تأخير نصف ثانية لتجنب حظر Yahoo

        # 1. الفلتر الأساسي
        if not passes_basic_filters(ticker):
            print(f"Ticker {ticker} did not pass basic filters.") # Added print statement
            continue

        # 2. تحليل السهم
        result = analyze_stock_2h(ticker)
        if not result or not result['signal']:
            print(f"No signal found for {ticker} after 2h analysis.") # Added print statement
            continue

        signal = result['signal']
        confidence = result['confidence']
        details = result['details']

        # 3. تحليل SPX
        spx_15m = analyze_spx("15m")

        # 4. تصنيف الصفقة
        classification = classify_trade(signal, confidence, details)

        # 5. اختيار العقد
        option = get_option_contract(ticker, signal, classification)

        # 6. بناء الرسالة
        msg = build_alert_message(ticker, signal, confidence, classification, details, spx_15m, option)
        alerts.append(msg)
    print(f"Scan finished. Found {len(alerts)} alerts.") # Added print statement
    return alerts

# ---------------------------
# 10. دوال بوت تليجرام
# ---------------------------

async def send_alerts(context):
    """إرسال التنبيهات إلى تليجرام"""
    print("send_alerts function called.") # Added print statement
    alerts = scan_all_stocks()
    if not alerts:
        print("No alerts found to send.") # Added print statement
        await context.bot.send_message(chat_id=CHAT_ID, text="لا توجد إشارات جديدة حالياً.")
    else:
        print(f"Sending {len(alerts)} alerts.") # Added print statement
        for alert in alerts:
            await context.bot.send_message(chat_id=CHAT_ID, text=alert, parse_mode='Markdown')
            await asyncio.sleep(1)  # تجنب سبام تليجرام

# دالة للترحيب بالأعضاء الجدد وإرسال إخلاء المسؤولية
async def greet_chat_members(update: Update, context: ContextTypes.DEFAULT_TYPE) -> None:
    print(f"New chat member event: {update.chat_member}") # Added print
    if update.chat_member.new_chat_member.status == 'member':
        user = update.chat_member.new_chat_member.user
        chat_title = update.chat_member.chat.title or "هذه المجموعة"
        welcome_new_member_message = f"""
أهلاً وسهلاً بك {user.full_name} في {chat_title}! 👋

⚠️ **إخلاء مسؤولية:**
توصيات هذا البوت هي لأغراض تعليمية وتحليلية فقط، ولا تشكل نصيحة استثمارية بأي شكل من الأشكال. تداول العقود والمشتقات المالية ينطوي على مخاطر عالية وقد يؤدي إلى خسارة رأس المال. أنت وحدك المسؤول عن قراراتك الاستثمارية. يرجى استشارة مستشار مالي متخصص قبل اتخاذ أي قرارات استثمارية.
"""
        print(f"Sending welcome message to {user.full_name} in chat {update.chat_member.chat.id}") # Added print
        await context.bot.send_message(chat_id=update.chat_member.chat.id, text=welcome_new_member_message, parse_mode='Markdown')


async def start(update: Update, context: ContextTypes.DEFAULT_TYPE) -> None:
    """معالج أمر /start"""
    print(f"Start command received from {update.effective_user.id} in chat {update.effective_chat.id}") # Added print
    welcome_message = """
👋 مرحباً بك في بوت توصيات الأسهم!

هذا البوت يقدم توصيات تحليل فني للأسهم بناءً على استراتيجيات محددة.

⚠️ **إخلاء مسؤولية:**
توصيات هذا البوت هي لأغراض تعليمية وتحليلية فقط، ولا تشكل نصيحة استثمارية بأي شكل من الأشكال. تداول العقود والمشتقات المالية ينطوي على مخاطر عالية وقد يؤدي إلى خسارة رأس المال. أنت وحدك المسؤول عن قراراتك الاستثمارية. يرجى استشارة مستشار مالي متخصص قبل اتخاذ أي قرارات استثمارية.

اختر نوع التوصية التي ترغب في الحصول عليها:
"""

    keyboard = [
        [InlineKeyboardButton("📈 عقود CALL (شراء)", callback_data='recommend_call')],
        [InlineKeyboardButton("📉 عقود PUT (بيع)", callback_data='recommend_put')],
        [InlineKeyboardButton("🔄 مسح الآن", callback_data='scan_now_manual')]
    ]
    reply_markup = InlineKeyboardMarkup(keyboard)

    await update.message.reply_text(welcome_message, reply_markup=reply_markup, parse_mode='Markdown')

async def button_callback(update: Update, context: ContextTypes.DEFAULT_TYPE) -> None:
    query = update.callback_query
    print(f"Callback query received: {query.data} from {query.from_user.id}") # Added print
    await query.answer()  # Always answer callback queries

    if query.data == 'recommend_call':
        await query.edit_message_text(text="جاري البحث عن توصيات CALL...")
        # Logic to scan for CALL recommendations
        # For now, let's just send a placeholder message
        await asyncio.sleep(2) # Simulate work
        await query.edit_message_text(text="لم يتم العثور على توصيات CALL حالياً.") # Replace with actual results
    elif query.data == 'recommend_put':
        await query.edit_message_text(text="جاري البحث عن توصيات PUT...")
        # Logic to scan for PUT recommendations
        # For now, let's just send a placeholder message
        await asyncio.sleep(2) # Simulate work
        await query.edit_message_text(text="لم يتم العثور على توصيات PUT حالياً.") # Replace with actual results
    elif query.data == 'scan_now_manual':
        await query.edit_message_text(text="جاري المسح الفوري، انتظر قليلاً...")
        alerts = scan_all_stocks()
        if not alerts:
            await query.edit_message_text("لا توجد إشارات جديدة.")
        else:
            for alert in alerts:
                await query.message.reply_text(alert, parse_mode='Markdown')
                await asyncio.sleep(1)
            await query.edit_message_text(text="تم الانتهاء من المسح الفوري.")


async def scan_now(update, context):
    """معالج أمر /scan لمسح فوري"""
    print("scan_now command called.") # Added print statement
    await update.message.reply_text("جاري المسح، انتظر قليلاً...")
    alerts = scan_all_stocks()
    if not alerts:
        print("No alerts found for manual scan.") # Added print statement
        await update.message.reply_text("لا توجد إشارات جديدة.")
    else:
        print(f"Sending {len(alerts)} alerts for manual scan.") # Added print statement
        for alert in alerts:
            await update.message.reply_text(alert, parse_mode='Markdown')
            await asyncio.sleep(1)

def main():
    """الدالة الرئيسية لتشغيل البوت"""
    # إنشاء تطبيق البوت
    # Create a JobQueue instance first
    job_queue_instance = JobQueue()

    # Pass the JobQueue instance to the ApplicationBuilder
    application = Application.builder().token(TELEGRAM_TOKEN).job_queue(job_queue_instance).build()

    # إضافة المعالجات للأوامر والرسائل
    application.add_handler(CommandHandler("start", start))
    application.add_handler(CommandHandler("scan", scan_now))
    application.add_handler(CallbackQueryHandler(button_callback)) # إضافة معالج لأزرار الـ Inline Keyboard
    application.add_handler(ChatMemberHandler(greet_chat_members, ChatMemberHandler.CHAT_MEMBER)) # إضافة معالج للأعضاء الجدد

    # جدولة المسح التلقائي كل ساعة
    job_queue_instance.run_repeating(send_alerts, interval=3600, first=10)  # كل ساعة، أول مرة بعد 10 ثواني

    # تشغيل البوت
    print("البوت يعمل الآن... اذهب لتجربته!")
    application.run_polling()
    # application.idle() is not necessary after run_polling() as it's a blocking call

# استدعاء الدالة الرئيسية
if __name__ == '__main__':
    main()
````

````python
#!/usr/bin/env python3
# -*- coding: utf-8 -*-

# بوت توصيات الأسهم - نسخة شاكر 1.0 (نسخة كاملة)
# آخر تحديث: فبراير 2026

import yfinance as yf
import pandas as pd
import numpy as np
import ta
import logging
import time
import schedule
from datetime import datetime, timedelta
from telegram import Bot, InlineKeyboardButton, InlineKeyboardMarkup, Update
from telegram.ext import Application, CommandHandler, CallbackQueryHandler, ContextTypes, JobQueue, ChatMemberHandler # أضفنا ChatMemberHandler
import asyncio
import os

# ---------------------------
# 1. إعدادات البوت والتوكن
# ---------------------------
TELEGRAM_TOKEN = "8699821370:AAEQUSbLTgf7MmWqo5vV5LHPOz30wfqOfqw"      # ضع التوكن الخاص بك
CHAT_ID = "7854020427"                # ضع معرف المستخدم أو القناة

# تفعيل السجلات لمشاهدة الأخطاء
logging.basicConfig(level=logging.INFO, format='%(asctime)s - %(name)s - %(levelname)s - %(message)s')

# ---------------------------
# 2. قائمة الأسهم الكاملة (مدمجة)
# ---------------------------
TICKERS = [
    # --- أسهمك القديمة (من الصور) ---
    "A", "AA", "ABEV", "ABNB", "ABT", "ABVX", "ACGL", "ACM", "ACN",
    "ADBE", "ADI", "ADM", "ADP", "ADSK", "AEE", "AEG", "AEIS", "AEM",
    "AEP", "AER", "AES", "AFG", "AFL", "AFRM", "AG", "AGI", "AGNC",
    "AIG", "AIT", "AIZ", "AJG", "AKAM", "ALAB", "ALB", "ALC", "ALGN",
    "ALL", "ALLE", "ALLY", "ALNY", "ALSN", "AM", "AMCR", "AME", "AMH",
    "AMKR", "AMP", "AMRZ", "AMT", "AMX",

    # --- الشركات العملاقة (> 1 تريليون دولار) ---
    "NVDA", "AAPL", "GOOGL", "MSFT", "AMZN", "META", "TSM", "AVGO", "TSLA",
    "BRK.A", "WMT",

    # --- الشركات الكبرى (200 مليار - 1 تريليون) ---
    "LLY", "JPM", "XOM", "V", "JNJ", "MA", "MU", "ORCL", "COST", "AMD",
    "NFLX", "PLTR", "CSCO", "MRK", "TMO", "ASML", "ABBV", "HD", "CVX",
    "PG", "CAT", "KO", "PEP", "INTC", "CMCSA", "MCD", "TXN", "VZ", "AMGN"
]

# ---------------------------
# 3. دوال التحليل الأساسي (الفلاتر)
# ---------------------------

def get_stock_info(ticker):
    """جلب معلومات الشركة الأساسية"""
    try:
        stock = yf.Ticker(ticker)
        info = stock.info
        return info
    except Exception as e:
        logging.error(f"خطأ في جلب معلومات {ticker}: {e}")
        return None

def passes_basic_filters(ticker):
    """
    تطبيق الفلتر الأساسي:
    - Market Cap > 10B (ملاحظة: أغلب الشركات في القائمة أكبر من ذلك بكثير، لكن الفلتر يضمن ذلك)
    - Average Volume > 1M (تقريبي)
    - Spread < 5%
    """
    info = get_stock_info(ticker)
    if not info:
        return False

    # 1. القيمة السوقية
    mcap = info.get('marketCap', 0)
    if mcap < 10e9:
        print(f"{ticker} فشل فلتر Market Cap: {mcap}") # Changed from logging.debug
        return False

    # 2. متوسط الحجم (نأخذ آخر 20 يوم)
    try:
        stock = yf.Ticker(ticker)
        hist = stock.history(period="1mo", interval="1d")
        if hist.empty:
            return False
        avg_vol = hist['Volume'].rolling(20).mean().iloc[-1]
        if avg_vol < 1_000_000:   # أقل من مليون سهم يومياً
            print(f"{ticker} فشل فلتر الحجم: {avg_vol}") # Changed from logging.debug
            return False
    except:
        return False

    # 3. Spread (bid/ask)
    bid = info.get('bid', 0)
    ask = info.get('ask', 0)
    if bid and ask:
        spread = (ask - bid) / bid if bid != 0 else 1
        if spread > 0.05:   # أكبر من 5%
            print(f"{ticker} فشل فلتر Spread: {spread:.2%}") # Changed from logging.debug
            return False

    return True

# ---------------------------
# 4. تحليل SPX (مؤشر S&P 500)
# ---------------------------

def analyze_spx(timeframe="15m"):
    """تحليل مؤشر SPX على الإطار المطلوب"""
    try:
        spx = yf.Ticker("^GSPC")  # أو "SPY"
        hist = spx.history(period="5d", interval=timeframe)
        if len(hist) < 50:
            return "neutral"

        # متوسطات بسيطة لتحديد الاتجاه
        hist['SMA20'] = hist['Close'].rolling(20).mean()
        last = hist.iloc[-1]
        prev = hist.iloc[-2]

        if last['Close'] > last['SMA20'] and last['Close'] > prev['Close']:
            return "uptrend"
        elif last['Close'] < last['SMA20'] and last['Close'] < prev['Close']:
            return "downtrend"
        else:
            return "neutral"
    except:
        return "neutral"

# ---------------------------
# 5. تحليل السهم (فريم ساعتين)
# ---------------------------

def analyze_stock_2h(ticker):
    """
    تحليل فني كامل على فريم الساعتين
    يعيد: {'signal': 'CALL'/'PUT'/None, 'confidence': 0-100, 'details': {...}}
    """
    try:
        print(f"Analyzing {ticker} for 2h interval.") # Added print statement
        stock = yf.Ticker(ticker)
        hist = stock.history(period="1mo", interval="2h")  # آخر شهر، فريم ساعتين
        if len(hist) < 200:
            print(f"Not enough historical data for {ticker} in 2h interval.") # Added print statement
            return None

        # حساب المتوسطات
        hist['SMA50'] = hist['Close'].rolling(50).mean()
        hist['SMA100'] = hist['Close'].rolling(100).mean()
        hist['SMA200'] = hist['Close'].rolling(200).mean()
        hist['Volume_SMA20'] = hist['Volume'].rolling(20).mean()

        last = hist.iloc[-1]
        current_price = last['Close']
        sma50 = last['SMA50']
        sma100 = last['SMA100']
        sma200 = last['SMA200']

        # قمم وقيعان آخر 3 شموع
        high_last_3 = hist['High'].iloc[-3:].max()
        low_last_3 = hist['Low'].iloc[-3:].min()

        # قمة/قاع اليوم (آخر 24 ساعة = 12 شمعة)
        high_today = hist['High'].iloc[-12:].max()
        low_today = hist['Low'].iloc[-12:].min()

        volume_ok = last['Volume'] > last['Volume_SMA20']

        # شروط CALL
        call_condition1 = (sma50 > sma100) or (sma50 > sma200)
        call_condition2 = current_price > sma200
        call_condition3 = current_price > high_last_3 or current_price > high_today
        call_condition4 = volume_ok

        # شروط PUT
        put_condition1 = (sma50 < sma100) or (sma50 < sma200)
        put_condition2 = current_price < sma200
        put_condition3 = current_price < low_last_3 or current_price < low_today
        put_condition4 = volume_ok

        signal = None
        confidence = 0
        details = {}

        if call_condition1 and call_condition2 and call_condition3 and call_condition4:
            signal = "CALL"
            # حساب الثقة: نضيف 25 لكل شرط متحقق
            confidence = 25 * sum([call_condition1, call_condition2, call_condition3, call_condition4])
            details = {
                "sma50": sma50,
                "sma100": sma100,
                "sma200": sma200,
                "price": current_price,
                "high_3": high_last_3,
                "volume_ratio": last['Volume'] / last['Volume_SMA20']
            }
        elif put_condition1 and put_condition2 and put_condition3 and put_condition4:
            signal = "PUT"
            confidence = 25 * sum([put_condition1, put_condition2, put_condition3, put_condition4])
            details = {
                "sma50": sma50,
                "sma100": sma100,
                "sma200": sma200,
                "price": current_price,
                "low_3": low_last_3,
                "volume_ratio": last['Volume'] / last['Volume_SMA20']
            }

        if signal: # Added print statement
            print(f"Signal for {ticker}: {signal}, Confidence: {confidence}")
        return {"signal": signal, "confidence": int(confidence), "details": details}

    except Exception as e:
        logging.error(f"خطأ في تحليل {ticker}: {e}")
        return None

# ---------------------------
# 6. تصنيف الصفقة (سكالب، قصيرة، سوينق)
# ---------------------------

def classify_trade(signal, confidence, details):
    """
    تصنيف حسب قوة الإشارة وترتيب المتوسطات
    """
    if not signal:
        return None

    sma50 = details.get('sma50')
    sma100 = details.get('sma100')
    sma200 = details.get('sma200')

    # ترتيب المتوسطات
    if signal == "CALL":
        if sma50 > sma100 > sma200:
            return "📈 سوينق 3-7 أيام"
        elif confidence >= 75:
            return "⏳ صفقة قصيرة 1-2 يوم"
        else:
            return "⚡ سكالب"
    else:  # PUT
        if sma50 < sma100 < sma200:
            return "📈 سوينق 3-7 أيام"
        elif confidence >= 75:
            return "⏳ صفقة قصيرة 1-2 يوم"
        else:
            return "⚡ سكالب"

# ---------------------------
# 7. اختيار العقد المناسب (بيانات الخيارات)
# ---------------------------

def get_option_contract(ticker, signal, classification):
    """
    البحث عن عقد خيار مناسب حسب التصنيف.
    """
    try:
        stock = yf.Ticker(ticker)
        # جلب تواريخ انتهاء الصلاحية المتاحة
        expirations = stock.options
        if not expirations:
            return "غير متوفر"

        # اختيار التاريخ المناسب حسب التصنيف
        today = datetime.now().date()
        if "سكالب" in classification:
            days = 7
        elif "قصيرة" in classification:
            days = 14
        else:  # سوينق
            days = 30

        target_date = today + timedelta(days=days)
        # البحث عن أقرب تاريخ انتهاء للعدد المطلوب من الأيام
        exp_date = None
        for exp in expirations:
            exp_dt = datetime.strptime(exp, '%Y-%m-%d').date()
            if exp_dt >= target_date:
                exp_date = exp
                break
        if not exp_date:
            exp_date = expirations[0]  # أقرب تاريخ

        # جلب عقود الخيارات لذلك التاريخ
        opt_chain = stock.option_chain(exp_date)
        if signal == "CALL":
            contracts = opt_chain.calls
        else:
            contracts = opt_chain.puts

        if contracts.empty:
            return "غير متوفر"

        # فلترة العقود: Open Interest عالي (أعلى 10%)، Spread ضيق
        current_price = stock.history(period="1d")['Close'].iloc[-1]
        contracts['strike_diff'] = abs(contracts['strike'] - current_price)
        # فلتر OI
        oi_threshold = contracts['openInterest'].quantile(0.7)  # أعلى 30%
        eligible = contracts[contracts['openInterest'] > oi_threshold]

        if eligible.empty:
            eligible = contracts

        # نختار العقد الأقرب للسعر الحالي (للدلتا التقريبية)
        best = eligible.nsmallest(1, 'strike_diff').iloc[0]

        # حساب Spread تقريبي (ask-bid)/ask
        spread = (best['ask'] - best['bid']) / best['ask'] if best['ask'] > 0 else 1
        if spread > 0.05:   # أكبر من 5%
            return "سبريد واسع"

        # بناء نص العقد المقترح
        delta_approx = 0.5 - (best['strike_diff'] / current_price)  # تقريب بسيط
        delta_approx = max(0.1, min(0.9, delta_approx))

        return f"Delta {delta_approx:.2f} – Exp {exp_date} (أيام: {days})"

    except Exception as e:
        logging.error(f"خطأ في جلب الخيارات لـ {ticker}: {e}")
        return "غير متوفر"

# ---------------------------
# 8. بناء رسالة التنبيه النهائية
# ---------------------------

def build_alert_message(ticker, signal, confidence, classification, details, spx_trend, option_text):
    """
    تنسيق الرسالة كما طلبت تماماً.
    """
    # ترجمة الاتجاه
    if signal == "CALL":
        direction = "Call"
        cross = "50 > 100" if details.get('sma50', 0) > details.get('sma100', 0) else "50 > 200"
        price_status = f"السعر فوق {details.get('sma200', 0):.2f}"
        event = "اختراق قمة 2H"
    else:
        direction = "Put"
        cross = "50 < 100" if details.get('sma50', 0) < details.get('sma100', 0) else "50 < 200"
        price_status = f"السعر تحت {details.get('sma200', 0):.2f}"
        event = "كسر قاع 2H"

    # توافق SPX
    if "uptrend" in spx_trend and signal == "CALL":
        spx_compat = "✅ متوافق"
    elif "downtrend" in spx_trend and signal == "PUT":
        spx_compat = "✅ متوافق"
    else:
        spx_compat = "⚠️ غير متوافق"

    message = f"""
🔔 {ticker}
الاتجاه: {direction}
التصنيف: {classification}
التقاطع: {cross} (إيجابي)
الحالة: {price_status}
الحدث: {event}
SPX: {spx_compat}
الثقة: {confidence}/100
العقد المقترح: {option_text}
"""
    return message

# ---------------------------
# 9. دالة المسح الرئيسية
# ---------------------------

def scan_all_stocks():
    """تمر على كل الأسهم وتفحصها وترسل التنبيهات"""
    alerts = []
    print("Starting stock scan...") # Added print statement
    for ticker in TICKERS:
        logging.info(f"فحص {ticker}...")
        time.sleep(0.5)  # تأخير نصف ثانية لتجنب حظر Yahoo

        # 1. الفلتر الأساسي
        if not passes_basic_filters(ticker):
            print(f"Ticker {ticker} did not pass basic filters.") # Added print statement
            continue

        # 2. تحليل السهم
        result = analyze_stock_2h(ticker)
        if not result or not result['signal']:
            print(f"No signal found for {ticker} after 2h analysis.") # Added print statement
            continue

        signal = result['signal']
        confidence = result['confidence']
        details = result['details']

        # 3. تحليل SPX
        spx_15m = analyze_spx("15m")

        # 4. تصنيف الصفقة
        classification = classify_trade(signal, confidence, details)

        # 5. اختيار العقد
        option = get_option_contract(ticker, signal, classification)

        # 6. بناء الرسالة
        msg = build_alert_message(ticker, signal, confidence, classification, details, spx_15m, option)
        alerts.append(msg)
    print(f"Scan finished. Found {len(alerts)} alerts.") # Added print statement
    return alerts

# ---------------------------
# 10. دوال بوت تليجرام
# ---------------------------

async def send_alerts(context):
    """إرسال التنبيهات إلى تليجرام"""
    print("send_alerts function called.") # Added print statement
    alerts = scan_all_stocks()
    if not alerts:
        print("No alerts found to send.") # Added print statement
        await context.bot.send_message(chat_id=CHAT_ID, text="لا توجد إشارات جديدة حالياً.")
    else:
        print(f"Sending {len(alerts)} alerts.") # Added print statement
        for alert in alerts:
            await context.bot.send_message(chat_id=CHAT_ID, text=alert, parse_mode='Markdown')
            await asyncio.sleep(1)  # تجنب سبام تليجرام

# دالة للترحيب بالأعضاء الجدد وإرسال إخلاء المسؤولية
async def greet_chat_members(update: Update, context: ContextTypes.DEFAULT_TYPE) -> None:
    print(f"New chat member event: {update.chat_member}") # Added print
    if update.chat_member.new_chat_member.status == 'member':
        user = update.chat_member.new_chat_member.user
        chat_title = update.chat_member.chat.title or "هذه المجموعة"
        welcome_new_member_message = f"""
أهلاً وسهلاً بك {user.full_name} في {chat_title}! 👋

⚠️ **إخلاء مسؤولية:**
توصيات هذا البوت هي لأغراض تعليمية وتحليلية فقط، ولا تشكل نصيحة استثمارية بأي شكل من الأشكال. تداول العقود والمشتقات المالية ينطوي على مخاطر عالية وقد يؤدي إلى خسارة رأس المال. أنت وحدك المسؤول عن قراراتك الاستثمارية. يرجى استشارة مستشار مالي متخصص قبل اتخاذ أي قرارات استثمارية.
"""
        print(f"Sending welcome message to {user.full_name} in chat {update.chat_member.chat.id}") # Added print
        await context.bot.send_message(chat_id=update.chat_member.chat.id, text=welcome_new_member_message, parse_mode='Markdown')


async def start(update: Update, context: ContextTypes.DEFAULT_TYPE) -> None:
    """معالج أمر /start"""
    print(f"Start command received from {update.effective_user.id} in chat {update.effective_chat.id}") # Added print
    welcome_message = """
👋 مرحباً بك في بوت توصيات الأسهم!

هذا البوت يقدم توصيات تحليل فني للأسهم بناءً على استراتيجيات محددة.

⚠️ **إخلاء مسؤولية:**
توصيات هذا البوت هي لأغراض تعليمية وتحليلية فقط، ولا تشكل نصيحة استثمارية بأي شكل من الأشكال. تداول العقود والمشتقات المالية ينطوي على مخاطر عالية وقد يؤدي إلى خسارة رأس المال. أنت وحدك المسؤول عن قراراتك الاستثمارية. يرجى استشارة مستشار مالي متخصص قبل اتخاذ أي قرارات استثمارية.

اختر نوع التوصية التي ترغب في الحصول عليها:
"""

    keyboard = [
        [InlineKeyboardButton("📈 عقود CALL (شراء)", callback_data='recommend_call')],
        [InlineKeyboardButton("📉 عقود PUT (بيع)", callback_data='recommend_put')],
        [InlineKeyboardButton("🔄 مسح الآن", callback_data='scan_now_manual')]
    ]
    reply_markup = InlineKeyboardMarkup(keyboard)

    await update.message.reply_text(welcome_message, reply_markup=reply_markup, parse_mode='Markdown')

async def button_callback(update: Update, context: ContextTypes.DEFAULT_TYPE) -> None:
    query = update.callback_query
    print(f"Callback query received: {query.data} from {query.from_user.id}") # Added print
    await query.answer()  # Always answer callback queries

    if query.data == 'recommend_call':
        await query.edit_message_text(text="جاري البحث عن توصيات CALL...")
        # Logic to scan for CALL recommendations
        # For now, let's just send a placeholder message
        await asyncio.sleep(2) # Simulate work
        await query.edit_message_text(text="لم يتم العثور على توصيات CALL حالياً.") # Replace with actual results
    elif query.data == 'recommend_put':
        await query.edit_message_text(text="جاري البحث عن توصيات PUT...")
        # Logic to scan for PUT recommendations
        # For now, let's just send a placeholder message
        await asyncio.sleep(2) # Simulate work
        await query.edit_message_text(text="لم يتم العثور على توصيات PUT حالياً.") # Replace with actual results
    elif query.data == 'scan_now_manual':
        await query.edit_message_text(text="جاري المسح الفوري، انتظر قليلاً...")
        alerts = scan_all_stocks()
        if not alerts:
            await query.edit_message_text("لا توجد إشارات جديدة.")
        else:
            for alert in alerts:
                await query.message.reply_text(alert, parse_mode='Markdown')
                await asyncio.sleep(1)
            await query.edit_message_text(text="تم الانتهاء من المسح الفوري.")


async def scan_now(update, context):
    """معالج أمر /scan لمسح فوري"""
    print("scan_now command called.") # Added print statement
    await update.message.reply_text("جاري المسح، انتظر قليلاً...")
    alerts = scan_all_stocks()
    if not alerts:
        print("No alerts found for manual scan.") # Added print statement
        await update.message.reply_text("لا توجد إشارات جديدة.")
    else:
        print(f"Sending {len(alerts)} alerts for manual scan.") # Added print statement
        for alert in alerts:
            await update.message.reply_text(alert, parse_mode='Markdown')
            await asyncio.sleep(1)

def main():
    """الدالة الرئيسية لتشغيل البوت"""
    # إنشاء تطبيق البوت
    # Create a JobQueue instance first
    job_queue_instance = JobQueue()

    # Pass the JobQueue instance to the ApplicationBuilder
    application = Application.builder().token(TELEGRAM_TOKEN).job_queue(job_queue_instance).build()

    # إضافة المعالجات للأوامر والرسائل
    application.add_handler(CommandHandler("start", start))
    application.add_handler(CommandHandler("scan", scan_now))
    application.add_handler(CallbackQueryHandler(button_callback)) # إضافة معالج لأزرار الـ Inline Keyboard
    application.add_handler(ChatMemberHandler(greet_chat_members, ChatMemberHandler.CHAT_MEMBER)) # إضافة معالج للأعضاء الجدد

    # جدولة المسح التلقائي كل ساعة
    job_queue_instance.run_repeating(send_alerts, interval=3600, first=10)  # كل ساعة، أول مرة بعد 10 ثواني

    # تشغيل البوت
    print("البوت يعمل الآن... اذهب لتجربته!")
    application.run_polling()
    # application.idle() is not necessary after run_polling() as it's a blocking call

# استدعاء الدالة الرئيسية
if __name__ == '__main__':
    main()
````

The repeated `RuntimeError` regarding `JobQueue` even after reinstalling indicates that the Python runtime might not be fully recognizing the newly installed packages.
````python
#!/usr/bin/env python3
# -*- coding: utf-8 -*-

# بوت توصيات الأسهم - نسخة شاكر 1.0 (نسخة كاملة)
# آخر تحديث: فبراير 2026

import yfinance as yf
import pandas as pd
import numpy as np
import ta
import logging
import time
import schedule
from datetime import datetime, timedelta
from telegram import Bot, InlineKeyboardButton, InlineKeyboardMarkup, Update
from telegram.ext import Application, CommandHandler, CallbackQueryHandler, ContextTypes, JobQueue, ChatMemberHandler # أضفنا ChatMemberHandler
import asyncio
import os

# ---------------------------
# 1. إعدادات البوت والتوكن
# ---------------------------
TELEGRAM_TOKEN = "8699821370:AAEQUSbLTgf7MmWqo5vV5LHPOz30wfqOfqw"      # ضع التوكن الخاص بك
CHAT_ID = "7854020427"                # ضع معرف المستخدم أو القناة

# تفعيل السجلات لمشاهدة الأخطاء
logging.basicConfig(level=logging.INFO, format='%(asctime)s - %(name)s - %(levelname)s - %(message)s')

# ---------------------------
# 2. قائمة الأسهم الكاملة (مدمجة)
# ---------------------------
TICKERS = [
    # --- أسهمك القديمة (من الصور) ---
    "A", "AA", "ABEV", "ABNB", "ABT", "ABVX", "ACGL", "ACM", "ACN",
    "ADBE", "ADI", "ADM", "ADP", "ADSK", "AEE", "AEG", "AEIS", "AEM",
    "AEP", "AER", "AES", "AFG", "AFL", "AFRM", "AG", "AGI", "AGNC",
    "AIG", "AIT", "AIZ", "AJG", "AKAM", "ALAB", "ALB", "ALC", "ALGN",
    "ALL", "ALLE", "ALLY", "ALNY", "ALSN", "AM", "AMCR", "AME", "AMH",
    "AMKR", "AMP", "AMRZ", "AMT", "AMX",

    # --- الشركات العملاقة (> 1 تريليون دولار) ---
    "NVDA", "AAPL", "GOOGL", "MSFT", "AMZN", "META", "TSM", "AVGO", "TSLA",
    "BRK.A", "WMT",

    # --- الشركات الكبرى (200 مليار - 1 تريليون) ---
    "LLY", "JPM", "XOM", "V", "JNJ", "MA", "MU", "ORCL", "COST", "AMD",
    "NFLX", "PLTR", "CSCO", "MRK", "TMO", "ASML", "ABBV", "HD", "CVX",
    "PG", "CAT", "KO", "PEP", "INTC", "CMCSA", "MCD", "TXN", "VZ", "AMGN"
]

# ---------------------------
# 3. دوال التحليل الأساسي (الفلاتر)
# ---------------------------

def get_stock_info(ticker):
    """جلب معلومات الشركة الأساسية"""
    try:
        stock = yf.Ticker(ticker)
        info = stock.info
        return info
    except Exception as e:
        logging.error(f"خطأ في جلب معلومات {ticker}: {e}")
        return None

def passes_basic_filters(ticker):
    """
    تطبيق الفلتر الأساسي:
    - Market Cap > 10B (ملاحظة: أغلب الشركات في القائمة أكبر من ذلك بكثير، لكن الفلتر يضمن ذلك)
    - Average Volume > 1M (تقريبي)
    - Spread < 5%
    """
    info = get_stock_info(ticker)
    if not info:
        return False

    # 1. القيمة السوقية
    mcap = info.get('marketCap', 0)
    if mcap < 10e9:
        print(f"{ticker} فشل فلتر Market Cap: {mcap}") # Changed from logging.debug
        return False

    # 2. متوسط الحجم (نأخذ آخر 20 يوم)
    try:
        stock = yf.Ticker(ticker)
        hist = stock.history(period="1mo", interval="1d")
        if hist.empty:
            return False
        avg_vol = hist['Volume'].rolling(20).mean().iloc[-1]
        if avg_vol < 1_000_000:   # أقل من مليون سهم يومياً
            print(f"{ticker} فشل فلتر الحجم: {avg_vol}") # Changed from logging.debug
            return False
    except:
        return False

    # 3. Spread (bid/ask)
    bid = info.get('bid', 0)
    ask = info.get('ask', 0)
    if bid and ask:
        spread = (ask - bid) / bid if bid != 0 else 1
        if spread > 0.05:   # أكبر من 5%
            print(f"{ticker} فشل فلتر Spread: {spread:.2%}") # Changed from logging.debug
            return False

    return True

# ---------------------------
# 4. تحليل SPX (مؤشر S&P 500)
# ---------------------------

def analyze_spx(timeframe="15m"):
    """تحليل مؤشر SPX على الإطار المطلوب"""
    try:
        spx = yf.Ticker("^GSPC")  # أو "SPY"
        hist = spx.history(period="5d", interval=timeframe)
        if len(hist) < 50:
            return "neutral"

        # متوسطات بسيطة لتحديد الاتجاه
        hist['SMA20'] = hist['Close'].rolling(20).mean()
        last = hist.iloc[-1]
        prev = hist.iloc[-2]

        if last['Close'] > last['SMA20'] and last['Close'] > prev['Close']:
            return "uptrend"
        elif last['Close'] < last['SMA20'] and last['Close'] < prev['Close']:
            return "downtrend"
        else:
            return "neutral"
    except:
        return "neutral"

# ---------------------------
# 5. تحليل السهم (فريم ساعتين)
# ---------------------------

def analyze_stock_2h(ticker):
    """
    تحليل فني كامل على فريم الساعتين
    يعيد: {'signal': 'CALL'/'PUT'/None, 'confidence': 0-100, 'details': {...}}
    """
    try:
        print(f"Analyzing {ticker} for 2h interval.") # Added print statement
        stock = yf.Ticker(ticker)
        hist = stock.history(period="1mo", interval="2h")  # آخر شهر، فريم ساعتين
        if len(hist) < 200:
            print(f"Not enough historical data for {ticker} in 2h interval.") # Added print statement
            return None

        # حساب المتوسطات
        hist['SMA50'] = hist['Close'].rolling(50).mean()
        hist['SMA100'] = hist['Close'].rolling(100).mean()
        hist['SMA200'] = hist['Close'].rolling(200).mean()
        hist['Volume_SMA20'] = hist['Volume'].rolling(20).mean()

        last = hist.iloc[-1]
        current_price = last['Close']
        sma50 = last['SMA50']
        sma100 = last['SMA100']
        sma200 = last['SMA200']

        # قمم وقيعان آخر 3 شموع
        high_last_3 = hist['High'].iloc[-3:].max()
        low_last_3 = hist['Low'].iloc[-3:].min()

        # قمة/قاع اليوم (آخر 24 ساعة = 12 شمعة)
        high_today = hist['High'].iloc[-12:].max()
        low_today = hist['Low'].iloc[-12:].min()

        volume_ok = last['Volume'] > last['Volume_SMA20']

        # شروط CALL
        call_condition1 = (sma50 > sma100) or (sma50 > sma200)
        call_condition2 = current_price > sma200
        call_condition3 = current_price > high_last_3 or current_price > high_today
        call_condition4 = volume_ok

        # شروط PUT
        put_condition1 = (sma50 < sma100) or (sma50 < sma200)
        put_condition2 = current_price < sma200
        put_condition3 = current_price < low_last_3 or current_price < low_today
        put_condition4 = volume_ok

        signal = None
        confidence = 0
        details = {}

        if call_condition1 and call_condition2 and call_condition3 and call_condition4:
            signal = "CALL"
            # حساب الثقة: نضيف 25 لكل شرط متحقق
            confidence = 25 * sum([call_condition1, call_condition2, call_condition3, call_condition4])
            details = {
                "sma50": sma50,
                "sma100": sma100,
                "sma200": sma200,
                "price": current_price,
                "high_3": high_last_3,
                "volume_ratio": last['Volume'] / last['Volume_SMA20']
            }
        elif put_condition1 and put_condition2 and put_condition3 and put_condition4:
            signal = "PUT"
            confidence = 25 * sum([put_condition1, put_condition2, put_condition3, put_condition4])
            details = {
                "sma50": sma50,
                "sma100": sma100,
                "sma200": sma200,
                "price": current_price,
                "low_3": low_last_3,
                "volume_ratio": last['Volume'] / last['Volume_SMA20']
            }

        if signal: # Added print statement
            print(f"Signal for {ticker}: {signal}, Confidence: {confidence}")
        return {"signal": signal, "confidence": int(confidence), "details": details}

    except Exception as e:
        logging.error(f"خطأ في تحليل {ticker}: {e}")
        return None

# ---------------------------
# 6. تصنيف الصفقة (سكالب، قصيرة، سوينق)
# ---------------------------

def classify_trade(signal, confidence, details):
    """
    تصنيف حسب قوة الإشارة وترتيب المتوسطات
    """
    if not signal:
        return None

    sma50 = details.get('sma50')
    sma100 = details.get('sma100')
    sma200 = details.get('sma200')

    # ترتيب المتوسطات
    if signal == "CALL":
        if sma50 > sma100 > sma200:
            return "📈 سوينق 3-7 أيام"
        elif confidence >= 75:
            return "⏳ صفقة قصيرة 1-2 يوم"
        else:
            return "⚡ سكالب"
    else:  # PUT
        if sma50 < sma100 < sma200:
            return "📈 سوينق 3-7 أيام"
        elif confidence >= 75:
            return "⏳ صفقة قصيرة 1-2 يوم"
        else:
            return "⚡ سكالب"

# ---------------------------
# 7. اختيار العقد المناسب (بيانات الخيارات)
# ---------------------------

def get_option_contract(ticker, signal, classification):
    """
    البحث عن عقد خيار مناسب حسب التصنيف.
    """
    try:
        stock = yf.Ticker(ticker)
        # جلب تواريخ انتهاء الصلاحية المتاحة
        expirations = stock.options
        if not expirations:
            return "غير متوفر"

        # اختيار التاريخ المناسب حسب التصنيف
        today = datetime.now().date()
        if "سكالب" in classification:
            days = 7
        elif "قصيرة" in classification:
            days = 14
        else:  # سوينق
            days = 30

        target_date = today + timedelta(days=days)
        # البحث عن أقرب تاريخ انتهاء للعدد المطلوب من الأيام
        exp_date = None
        for exp in expirations:
            exp_dt = datetime.strptime(exp, '%Y-%m-%d').date()
            if exp_dt >= target_date:
                exp_date = exp
                break
        if not exp_date:
            exp_date = expirations[0]  # أقرب تاريخ

        # جلب عقود الخيارات لذلك التاريخ
        opt_chain = stock.option_chain(exp_date)
        if signal == "CALL":
            contracts = opt_chain.calls
        else:
            contracts = opt_chain.puts

        if contracts.empty:
            return "غير متوفر"

        # فلترة العقود: Open Interest عالي (أعلى 10%)، Spread ضيق
        current_price = stock.history(period="1d")['Close'].iloc[-1]
        contracts['strike_diff'] = abs(contracts['strike'] - current_price)
        # فلتر OI
        oi_threshold = contracts['openInterest'].quantile(0.7)  # أعلى 30%
        eligible = contracts[contracts['openInterest'] > oi_threshold]

        if eligible.empty:
            eligible = contracts

        # نختار العقد الأقرب للسعر الحالي (للدلتا التقريبية)
        best = eligible.nsmallest(1, 'strike_diff').iloc[0]

        # حساب Spread تقريبي (ask-bid)/ask
        spread = (best['ask'] - best['bid']) / best['ask'] if best['ask'] > 0 else 1
        if spread > 0.05:   # أكبر من 5%
            return "سبريد واسع"

        # بناء نص العقد المقترح
        delta_approx = 0.5 - (best['strike_diff'] / current_price)  # تقريب بسيط
        delta_approx = max(0.1, min(0.9, delta_approx))

        return f"Delta {delta_approx:.2f} – Exp {exp_date} (أيام: {days})"

    except Exception as e:
        logging.error(f"خطأ في جلب الخيارات لـ {ticker}: {e}")
        return "غير متوفر"

# ---------------------------
# 8. بناء رسالة التنبيه النهائية
# ---------------------------

def build_alert_message(ticker, signal, confidence, classification, details, spx_trend, option_text):
    """
    تنسيق الرسالة كما طلبت تماماً.
    """
    # ترجمة الاتجاه
    if signal == "CALL":
        direction = "Call"
        cross = "50 > 100" if details.get('sma50', 0) > details.get('sma100', 0) else "50 > 200"
        price_status = f"السعر فوق {details.get('sma200', 0):.2f}"
        event = "اختراق قمة 2H"
    else:
        direction = "Put"
        cross = "50 < 100" if details.get('sma50', 0) < details.get('sma100', 0) else "50 < 200"
        price_status = f"السعر تحت {details.get('sma200', 0):.2f}"
        event = "كسر قاع 2H"

    # توافق SPX
    if "uptrend" in spx_trend and signal == "CALL":
        spx_compat = "✅ متوافق"
    elif "downtrend" in spx_trend and signal == "PUT":
        spx_compat = "✅ متوافق"
    else:
        spx_compat = "⚠️ غير متوافق"

    message = f"""
🔔 {ticker}
الاتجاه: {direction}
التصنيف: {classification}
التقاطع: {cross} (إيجابي)
الحالة: {price_status}
الحدث: {event}
SPX: {spx_compat}
الثقة: {confidence}/100
العقد المقترح: {option_text}
"""
    return message

# ---------------------------
# 9. دالة المسح الرئيسية
# ---------------------------

def scan_all_stocks():
    """تمر على كل الأسهم وتفحصها وترسل التنبيهات"""
    alerts = []
    print("Starting stock scan...") # Added print statement
    for ticker in TICKERS:
        logging.info(f"فحص {ticker}...")
        time.sleep(0.5)  # تأخير نصف ثانية لتجنب حظر Yahoo

        # 1. الفلتر الأساسي
        if not passes_basic_filters(ticker):
            print(f"Ticker {ticker} did not pass basic filters.") # Added print statement
            continue

        # 2. تحليل السهم
        result = analyze_stock_2h(ticker)
        if not result or not result['signal']:
            print(f"No signal found for {ticker} after 2h analysis.") # Added print statement
            continue

        signal = result['signal']
        confidence = result['confidence']
        details = result['details']

        # 3. تحليل SPX
        spx_15m = analyze_spx("15m")

        # 4. تصنيف الصفقة
        classification = classify_trade(signal, confidence, details)

        # 5. اختيار العقد
        option = get_option_contract(ticker, signal, classification)

        # 6. بناء الرسالة
        msg = build_alert_message(ticker, signal, confidence, classification, details, spx_15m, option)
        alerts.append(msg)
    print(f"Scan finished. Found {len(alerts)} alerts.") # Added print statement
    return alerts

# ---------------------------
# 10. دوال بوت تليجرام
# ---------------------------

async def send_alerts(context):
    """إرسال التنبيهات إلى تليجرام"""
    print("send_alerts function called.") # Added print statement
    alerts = scan_all_stocks()
    if not alerts:
        print("No alerts found to send.") # Added print statement
        await context.bot.send_message(chat_id=CHAT_ID, text="لا توجد إشارات جديدة حالياً.")
    else:
        print(f"Sending {len(alerts)} alerts.") # Added print statement
        for alert in alerts:
            await context.bot.send_message(chat_id=CHAT_ID, text=alert, parse_mode='Markdown')
            await asyncio.sleep(1)  # تجنب سبام تليجرام

# دالة للترحيب بالأعضاء الجدد وإرسال إخلاء المسؤولية
async def greet_chat_members(update: Update, context: ContextTypes.DEFAULT_TYPE) -> None:
    print(f"New chat member event: {update.chat_member}") # Added print
    if update.chat_member.new_chat_member.status == 'member':
        user = update.chat_member.new_chat_member.user
        chat_title = update.chat_member.chat.title or "هذه المجموعة"
        welcome_new_member_message = f"""
أهلاً وسهلاً بك {user.full_name} في {chat_title}! 👋

⚠️ **إخلاء مسؤولية:**
توصيات هذا البوت هي لأغراض تعليمية وتحليلية فقط، ولا تشكل نصيحة استثمارية بأي شكل من الأشكال. تداول العقود والمشتقات المالية ينطوي على مخاطر عالية وقد يؤدي إلى خسارة رأس المال. أنت وحدك المسؤول عن قراراتك الاستثمارية. يرجى استشارة مستشار مالي متخصص قبل اتخاذ أي قرارات استثمارية.
"""
        print(f"Sending welcome message to {user.full_name} in chat {update.chat_member.chat.id}") # Added print
        await context.bot.send_message(chat_id=update.chat_member.chat.id, text=welcome_new_member_message, parse_mode='Markdown')


async def start(update: Update, context: ContextTypes.DEFAULT_TYPE) -> None:
    """معالج أمر /start"""
    print(f"Start command received from {update.effective_user.id} in chat {update.effective_chat.id}") # Added print
    welcome_message = """
👋 مرحباً بك في بوت توصيات الأسهم!

هذا البوت يقدم توصيات تحليل فني للأسهم بناءً على استراتيجيات محددة.

⚠️ **إخلاء مسؤولية:**
توصيات هذا البوت هي لأغراض تعليمية وتحليلية فقط، ولا تشكل نصيحة استثمارية بأي شكل من الأشكال. تداول العقود والمشتقات المالية ينطوي على مخاطر عالية وقد يؤدي إلى خسارة رأس المال. أنت وحدك المسؤول عن قراراتك الاستثمارية. يرجى استشارة مستشار مالي متخصص قبل اتخاذ أي قرارات استثمارية.

اختر نوع التوصية التي ترغب في الحصول عليها:
"""

    keyboard = [
        [InlineKeyboardButton("📈 عقود CALL (شراء)", callback_data='recommend_call')],
        [InlineKeyboardButton("📉 عقود PUT (بيع)", callback_data='recommend_put')],
        [InlineKeyboardButton("🔄 مسح الآن", callback_data='scan_now_manual')]
    ]
    reply_markup = InlineKeyboardMarkup(keyboard)

    await update.message.reply_text(welcome_message, reply_markup=reply_markup, parse_mode='Markdown')

async def button_callback(update: Update, context: ContextTypes.DEFAULT_TYPE) -> None:
    query = update.callback_query
    print(f"Callback query received: {query.data} from {query.from_user.id}") # Added print
    await query.answer()  # Always answer callback queries

    if query.data == 'recommend_call':
        await query.edit_message_text(text="جاري البحث عن توصيات CALL...")
        # Logic to scan for CALL recommendations
        # For now, let's just send a placeholder message
        await asyncio.sleep(2) # Simulate work
        await query.edit_message_text(text="لم يتم العثور على توصيات CALL حالياً.") # Replace with actual results
    elif query.data == 'recommend_put':
        await query.edit_message_text(text="جاري البحث عن توصيات PUT...")
        # Logic to scan for PUT recommendations
        # For now, let's just send a placeholder message
        await asyncio.sleep(2) # Simulate work
        await query.edit_message_text(text="لم يتم العثور على توصيات PUT حالياً.") # Replace with actual results
    elif query.data == 'scan_now_manual':
        await query.edit_message_text(text="جاري المسح الفوري، انتظر قليلاً...")
        alerts = scan_all_stocks()
        if not alerts:
            await query.edit_message_text("لا توجد إشارات جديدة.")
        else:
            for alert in alerts:
                await query.message.reply_text(alert, parse_mode='Markdown')
                await asyncio.sleep(1)
            await query.edit_message_text(text="تم الانتهاء من المسح الفوري.")


async def scan_now(update, context):
    """معالج أمر /scan لمسح فوري"""
    print("scan_now command called.") # Added print statement
    await update.message.reply_text("جاري المسح، انتظر قليلاً...")
    alerts = scan_all_stocks()
    if not alerts:
        print("No alerts found for manual scan.") # Added print statement
        await update.message.reply_text("لا توجد إشارات جديدة.")
    else:
        print(f"Sending {len(alerts)} alerts.") # Added print statement
        for alert in alerts:
            await update.message.reply_text(alert, parse_mode='Markdown')
            await asyncio.sleep(1)

def main():
    """الدالة الرئيسية لتشغيل البوت"""
    # إنشاء تطبيق البوت
    # Create a JobQueue instance first
    job_queue_instance = JobQueue()

    # Pass the JobQueue instance to the ApplicationBuilder
    application = Application.builder().token(TELEGRAM_TOKEN).job_queue(job_queue_instance).build()

    # إضافة المعالجات للأوامر والرسائل
    application.add_handler(CommandHandler("start", start))
    application.add_handler(CommandHandler("scan", scan_now))
    application.add_handler(CallbackQueryHandler(button_callback)) # إضافة معالج لأزرار الـ Inline Keyboard
    application.add_handler(ChatMemberHandler(greet_chat_members, ChatMemberHandler.CHAT_MEMBER)) # إضافة معالج للأعضاء الجدد

    # جدولة المسح التلقائي كل ساعة
    job_queue_instance.run_repeating(send_alerts, interval=3600, first=10)  # كل ساعة، أول مرة بعد 10 ثواني

    # تشغيل البوت
    print("البوت يعمل الآن... اذهب لتجربته!")
    application.run_polling()
    # application.idle() is not necessary after run_polling() as it's a blocking call

# استدعاء الدالة الرئيسية
if __name__ == '__main__':
    main()
````

````python
#!/usr/bin/env python3
# -*- coding: utf-8 -*-

# بوت توصيات الأسهم - نسخة شاكر 1.0 (نسخة كاملة)
# آخر تحديث: فبراير 2026

import yfinance as yf
import pandas as pd
import numpy as np
import ta
import logging
import time
import schedule
from datetime import datetime, timedelta
from telegram import Bot, InlineKeyboardButton, InlineKeyboardMarkup, Update
from telegram.ext import Application, CommandHandler, CallbackQueryHandler, ContextTypes, JobQueue, ChatMemberHandler # أضفنا ChatMemberHandler
import asyncio
import os

# ---------------------------
# 1. إعدادات البوت والتوكن
# ---------------------------
TELEGRAM_TOKEN = "8699821370:AAEQUSbLTgf7MmWqo5vV5LHPOz30wfqOfqw"      # ضع التوكن الخاص بك
CHAT_ID = "7854020427"                # ضع معرف المستخدم أو القناة

# تفعيل السجلات لمشاهدة الأخطاء
logging.basicConfig(level=logging.INFO, format='%(asctime)s - %(name)s - %(levelname)s - %(message)s')

# ---------------------------
# 2. قائمة الأسهم الكاملة (مدمجة)
# ---------------------------
TICKERS = [
    # --- أسهمك القديمة (من الصور) ---
    "A", "AA", "ABEV", "ABNB", "ABT", "ABVX", "ACGL", "ACM", "ACN",
    "ADBE", "ADI", "ADM", "ADP", "ADSK", "AEE", "AEG", "AEIS", "AEM",
    "AEP", "AER", "AES", "AFG", "AFL", "AFRM", "AG", "AGI", "AGNC",
    "AIG", "AIT", "AIZ", "AJG", "AKAM", "ALAB", "ALB", "ALC", "ALGN",
    "ALL", "ALLE", "ALLY", "ALNY", "ALSN", "AM", "AMCR", "AME", "AMH",
    "AMKR", "AMP", "AMRZ", "AMT", "AMX",

    # --- الشركات العملاقة (> 1 تريليون دولار) ---
    "NVDA", "AAPL", "GOOGL", "MSFT", "AMZN", "META", "TSM", "AVGO", "TSLA",
    "BRK.A", "WMT",

    # --- الشركات الكبرى (200 مليار - 1 تريليون) ---
    "LLY", "JPM", "XOM", "V", "JNJ", "MA", "MU", "ORCL", "COST", "AMD",
    "NFLX", "PLTR", "CSCO", "MRK", "TMO", "ASML", "ABBV", "HD", "CVX",
    "PG", "CAT", "KO", "PEP", "INTC", "CMCSA", "MCD", "TXN", "VZ", "AMGN"
]

# ---------------------------
# 3. دوال التحليل الأساسي (الفلاتر)
# ---------------------------

def get_stock_info(ticker):
    """جلب معلومات الشركة الأساسية"""
    try:
        stock = yf.Ticker(ticker)
        info = stock.info
        return info
    except Exception as e:
        logging.error(f"خطأ في جلب معلومات {ticker}: {e}")
        return None

def passes_basic_filters(ticker):
    """
    تطبيق الفلتر الأساسي:
    - Market Cap > 10B (ملاحظة: أغلب الشركات في القائمة أكبر من ذلك بكثير، لكن الفلتر يضمن ذلك)
    - Average Volume > 1M (تقريبي)
    - Spread < 5%
    """
    info = get_stock_info(ticker)
    if not info:
        return False

    # 1. القيمة السوقية
    mcap = info.get('marketCap', 0)
    if mcap < 10e9:
        print(f"{ticker} فشل فلتر Market Cap: {mcap}") # Changed from logging.debug
        return False

    # 2. متوسط الحجم (نأخذ آخر 20 يوم)
    try:
        stock = yf.Ticker(ticker)
        hist = stock.history(period="1mo", interval="1d")
        if hist.empty:
            return False
        avg_vol = hist['Volume'].rolling(20).mean().iloc[-1]
        if avg_vol < 1_000_000:   # أقل من مليون سهم يومياً
            print(f"{ticker} فشل فلتر الحجم: {avg_vol}") # Changed from logging.debug
            return False
    except:
        return False

    # 3. Spread (bid/ask)
    bid = info.get('bid', 0)
    ask = info.get('ask', 0)
    if bid and ask:
        spread = (ask - bid) / bid if bid != 0 else 1
        if spread > 0.05:   # أكبر من 5%
            print(f"{ticker} فشل فلتر Spread: {spread:.2%}") # Changed from logging.debug
            return False

    return True

# ---------------------------
# 4. تحليل SPX (مؤشر S&P 500)
# ---------------------------

def analyze_spx(timeframe="15m"):
    """تحليل مؤشر SPX على الإطار المطلوب"""
    try:
        spx = yf.Ticker("^GSPC")  # أو "SPY"
        hist = spx.history(period="5d", interval=timeframe)
        if len(hist) < 50:
            return "neutral"

        # متوسطات بسيطة لتحديد الاتجاه
        hist['SMA20'] = hist['Close'].rolling(20).mean()
        last = hist.iloc[-1]
        prev = hist.iloc[-2]

        if last['Close'] > last['SMA20'] and last['Close'] > prev['Close']:
            return "uptrend"
        elif last['Close'] < last['SMA20'] and last['Close'] < prev['Close']:
            return "downtrend"
        else:
            return "neutral"
    except:
        return "neutral"

# ---------------------------
# 5. تحليل السهم (فريم ساعتين)
# ---------------------------

def analyze_stock_2h(ticker):
    """
    تحليل فني كامل على فريم الساعتين
    يعيد: {'signal': 'CALL'/'PUT'/None, 'confidence': 0-100, 'details': {...}}
    """
    try:
        print(f"Analyzing {ticker} for 2h interval.") # Added print statement
        stock = yf.Ticker(ticker)
        hist = stock.history(period="1mo", interval="2h")  # آخر شهر، فريم ساعتين
        if len(hist) < 200:
            print(f"Not enough historical data for {ticker} in 2h interval.") # Added print statement
            return None

        # حساب المتوسطات
        hist['SMA50'] = hist['Close'].rolling(50).mean()
        hist['SMA100'] = hist['Close'].rolling(100).mean()
        hist['SMA200'] = hist['Close'].rolling(200).mean()
        hist['Volume_SMA20'] = hist['Volume'].rolling(20).mean()

        last = hist.iloc[-1]
        current_price = last['Close']
        sma50 = last['SMA50']
        sma100 = last['SMA100']
        sma200 = last['SMA200']

        # قمم وقيعان آخر 3 شموع
        high_last_3 = hist['High'].iloc[-3:].max()
        low_last_3 = hist['Low'].iloc[-3:].min()

        # قمة/قاع اليوم (آخر 24 ساعة = 12 شمعة)
        high_today = hist['High'].iloc[-12:].max()
        low_today = hist['Low'].iloc[-12:].min()

        volume_ok = last['Volume'] > last['Volume_SMA20']

        # شروط CALL
        call_condition1 = (sma50 > sma100) or (sma50 > sma200)
        call_condition2 = current_price > sma200
        call_condition3 = current_price > high_last_3 or current_price > high_today
        call_condition4 = volume_ok

        # شروط PUT
        put_condition1 = (sma50 < sma100) or (sma50 < sma200)
        put_condition2 = current_price < sma200
        put_condition3 = current_price < low_last_3 or current_price < low_today
        put_condition4 = volume_ok

        signal = None
        confidence = 0
        details = {}

        if call_condition1 and call_condition2 and call_condition3 and call_condition4:
            signal = "CALL"
            # حساب الثقة: نضيف 25 لكل شرط متحقق
            confidence = 25 * sum([call_condition1, call_condition2, call_condition3, call_condition4])
            details = {
                "sma50": sma50,
                "sma100": sma100,
                "sma200": sma200,
                "price": current_price,
                "high_3": high_last_3,
                "volume_ratio": last['Volume'] / last['Volume_SMA20']
            }
        elif put_condition1 and put_condition2 and put_condition3 and put_condition4:
            signal = "PUT"
            confidence = 25 * sum([put_condition1, put_condition2, put_condition3, put_condition4])
            details = {
                "sma50": sma50,
                "sma100": sma100,
                "sma200": sma200,
                "price": current_price,
                "low_3": low_last_3,
                "volume_ratio": last['Volume'] / last['Volume_SMA20']
            }

        if signal: # Added print statement
            print(f"Signal for {ticker}: {signal}, Confidence: {confidence}")
        return {"signal": signal, "confidence": int(confidence), "details": details}

    except Exception as e:
        logging.error(f"خطأ في تحليل {ticker}: {e}")
        return None

# ---------------------------
# 6. تصنيف الصفقة (سكالب، قصيرة، سوينق)
# ---------------------------

def classify_trade(signal, confidence, details):
    """
    تصنيف حسب قوة الإشارة وترتيب المتوسطات
    """
    if not signal:
        return None

    sma50 = details.get('sma50')
    sma100 = details.get('sma100')
    sma200 = details.get('sma200')

    # ترتيب المتوسطات
    if signal == "CALL":
        if sma50 > sma100 > sma200:
            return "📈 سوينق 3-7 أيام"
        elif confidence >= 75:
            return "⏳ صفقة قصيرة 1-2 يوم"
        else:
            return "⚡ سكالب"
    else:  # PUT
        if sma50 < sma100 < sma200:
            return "📈 سوينق 3-7 أيام"
        elif confidence >= 75:
            return "⏳ صفقة قصيرة 1-2 يوم"
        else:
            return "⚡ سكالب"

# ---------------------------
# 7. اختيار العقد المناسب (بيانات الخيارات)
# ---------------------------

def get_option_contract(ticker, signal, classification):
    """
    البحث عن عقد خيار مناسب حسب التصنيف.
    """
    try:
        stock = yf.Ticker(ticker)
        # جلب تواريخ انتهاء الصلاحية المتاحة
        expirations = stock.options
        if not expirations:
            return "غير متوفر"

        # اختيار التاريخ المناسب حسب التصنيف
        today = datetime.now().date()
        if "سكالب" in classification:
            days = 7
        elif "قصيرة" in classification:
            days = 14
        else:  # سوينق
            days = 30

        target_date = today + timedelta(days=days)
        # البحث عن أقرب تاريخ انتهاء للعدد المطلوب من الأيام
        exp_date = None
        for exp in expirations:
            exp_dt = datetime.strptime(exp, '%Y-%m-%d').date()
            if exp_dt >= target_date:
                exp_date = exp
                break
        if not exp_date:
            exp_date = expirations[0]  # أقرب تاريخ

        # جلب عقود الخيارات لذلك التاريخ
        opt_chain = stock.option_chain(exp_date)
        if signal == "CALL":
            contracts = opt_chain.calls
        else:
            contracts = opt_chain.puts

        if contracts.empty:
            return "غير متوفر"

        # فلترة العقود: Open Interest عالي (أعلى 10%)، Spread ضيق
        current_price = stock.history(period="1d")['Close'].iloc[-1]
        contracts['strike_diff'] = abs(contracts['strike'] - current_price)
        # فلتر OI
        oi_threshold = contracts['openInterest'].quantile(0.7)  # أعلى 30%
        eligible = contracts[contracts['openInterest'] > oi_threshold]

        if eligible.empty:
            eligible = contracts

        # نختار العقد الأقرب للسعر الحالي (للدلتا التقريبية)
        best = eligible.nsmallest(1, 'strike_diff').iloc[0]

        # حساب Spread تقريبي (ask-bid)/ask
        spread = (best['ask'] - best['bid']) / best['ask'] if best['ask'] > 0 else 1
        if spread > 0.05:   # أكبر من 5%
            return "سبريد واسع"

        # بناء نص العقد المقترح
        delta_approx = 0.5 - (best['strike_diff'] / current_price)  # تقريب بسيط
        delta_approx = max(0.1, min(0.9, delta_approx))

        return f"Delta {delta_approx:.2f} – Exp {exp_date} (أيام: {days})"

    except Exception as e:
        logging.error(f"خطأ في جلب الخيارات لـ {ticker}: {e}")
        return "غير متوفر"

# ---------------------------
# 8. بناء رسالة التنبيه النهائية
# ---------------------------

def build_alert_message(ticker, signal, confidence, classification, details, spx_trend, option_text):
    """
    تنسيق الرسالة كما طلبت تماماً.
    """
    # ترجمة الاتجاه
    if signal == "CALL":
        direction = "Call"
        cross = "50 > 100" if details.get('sma50', 0) > details.get('sma100', 0) else "50 > 200"
        price_status = f"السعر فوق {details.get('sma200', 0):.2f}"
        event = "اختراق قمة 2H"
    else:
        direction = "Put"
        cross = "50 < 100" if details.get('sma50', 0) < details.get('sma100', 0) else "50 < 200"
        price_status = f"السعر تحت {details.get('sma200', 0):.2f}"
        event = "كسر قاع 2H"

    # توافق SPX
    if "uptrend" in spx_trend and signal == "CALL":
        spx_compat = "✅ متوافق"
    elif "downtrend" in spx_trend and signal == "PUT":
        spx_compat = "✅ متوافق"
    else:
        spx_compat = "⚠️ غير متوافق"

    message = f"""
🔔 {ticker}
الاتجاه: {direction}
التصنيف: {classification}
التقاطع: {cross} (إيجابي)
الحالة: {price_status}
الحدث: {event}
SPX: {spx_compat}
الثقة: {confidence}/100
العقد المقترح: {option_text}
"""
    return message

# ---------------------------
# 9. دالة المسح الرئيسية
# ---------------------------

def scan_all_stocks():
    """تمر على كل الأسهم وتفحصها وترسل التنبيهات"""
    alerts = []
    print("Starting stock scan...") # Added print statement
    for ticker in TICKERS:
        logging.info(f"فحص {ticker}...")
        time.sleep(0.5)  # تأخير نصف ثانية لتجنب حظر Yahoo

        # 1. الفلتر الأساسي
        if not passes_basic_filters(ticker):
            print(f"Ticker {ticker} did not pass basic filters.") # Added print statement
            continue

        # 2. تحليل السهم
        result = analyze_stock_2h(ticker)
        if not result or not result['signal']:
            print(f"No signal found for {ticker} after 2h analysis.") # Added print statement
            continue

        signal = result['signal']
        confidence = result['confidence']
        details = result['details']

        # 3. تحليل SPX
        spx_15m = analyze_spx("15m")

        # 4. تصنيف الصفقة
        classification = classify_trade(signal, confidence, details)

        # 5. اختيار العقد
        option = get_option_contract(ticker, signal, classification)

        # 6. بناء الرسالة
        msg = build_alert_message(ticker, signal, confidence, classification, details, spx_15m, option)
        alerts.append(msg)
    print(f"Scan finished. Found {len(alerts)} alerts.") # Added print statement
    return alerts

# ---------------------------
# 10. دوال بوت تليجرام
# ---------------------------

async def send_alerts(context):
    """إرسال التنبيهات إلى تليجرام"""
    print("send_alerts function called.") # Added print statement
    alerts = scan_all_stocks()
    if not alerts:
        print("No alerts found to send.") # Added print statement
        await context.bot.send_message(chat_id=CHAT_ID, text="لا توجد إشارات جديدة حالياً.")
    else:
        print(f"Sending {len(alerts)} alerts.") # Added print statement
        for alert in alerts:
            await context.bot.send_message(chat_id=CHAT_ID, text=alert, parse_mode='Markdown')
            await asyncio.sleep(1)  # تجنب سبام تليجرام

# دالة للترحيب بالأعضاء الجدد وإرسال إخلاء المسؤولية
async def greet_chat_members(update: Update, context: ContextTypes.DEFAULT_TYPE) -> None:
    print(f"New chat member event: {update.chat_member}") # Added print
    if update.chat_member.new_chat_member.status == 'member':
        user = update.chat_member.new_chat_member.user
        chat_title = update.chat_member.chat.title or "هذه المجموعة"
        welcome_new_member_message = f"""
أهلاً وسهلاً بك {user.full_name} في {chat_title}! 👋

⚠️ **إخلاء مسؤولية:**
توصيات هذا البوت هي لأغراض تعليمية وتحليلية فقط، ولا تشكل نصيحة استثمارية بأي شكل من الأشكال. تداول العقود والمشتقات المالية ينطوي على مخاطر عالية وقد يؤدي إلى خسارة رأس المال. أنت وحدك المسؤول عن قراراتك الاستثمارية. يرجى استشارة مستشار مالي متخصص قبل اتخاذ أي قرارات استثمارية.
"""
        print(f"Sending welcome message to {user.full_name} in chat {update.chat_member.chat.id}") # Added print
        await context.bot.send_message(chat_id=update.chat_member.chat.id, text=welcome_new_member_message, parse_mode='Markdown')


async def start(update: Update, context: ContextTypes.DEFAULT_TYPE) -> None:
    """معالج أمر /start"""
    print(f"Start command received from {update.effective_user.id} in chat {update.effective_chat.id}") # Added print
    welcome_message = """
👋 مرحباً بك في بوت توصيات الأسهم!

هذا البوت يقدم توصيات تحليل فني للأسهم بناءً على استراتيجيات محددة.

⚠️ **إخلاء مسؤولية:**
توصيات هذا البوت هي لأغراض تعليمية وتحليلية فقط، ولا تشكل نصيحة استثمارية بأي شكل من الأشكال. تداول العقود والمشتقات المالية ينطوي على مخاطر عالية وقد يؤدي إلى خسارة رأس المال. أنت وحدك المسؤول عن قراراتك الاستثمارية. يرجى استشارة مستشار مالي متخصص قبل اتخاذ أي قرارات استثمارية.

اختر نوع التوصية التي ترغب في الحصول عليها:
"""

    keyboard = [
        [InlineKeyboardButton("📈 عقود CALL (شراء)", callback_data='recommend_call')],
        [InlineKeyboardButton("📉 عقود PUT (بيع)", callback_data='recommend_put')],
        [InlineKeyboardButton("🔄 مسح الآن", callback_data='scan_now_manual')]
    ]
    reply_markup = InlineKeyboardMarkup(keyboard)

    await update.message.reply_text(welcome_message, reply_markup=reply_markup, parse_mode='Markdown')

async def button_callback(update: Update, context: ContextTypes.DEFAULT_TYPE) -> None:
    query = update.callback_query
    print(f"Callback query received: {query.data} from {query.from_user.id}") # Added print
    await query.answer()  # Always answer callback queries

    if query.data == 'recommend_call':
        await query.edit_message_text(text="جاري البحث عن توصيات CALL...")
        # Logic to scan for CALL recommendations
        # For now, let's just send a placeholder message
        await asyncio.sleep(2) # Simulate work
        await query.edit_message_text(text="لم يتم العثور على توصيات CALL حالياً.") # Replace with actual results
    elif query.data == 'recommend_put':
        await query.edit_message_text(text="جاري البحث عن توصيات PUT...")
        # Logic to scan for PUT recommendations
        # For now, let's just send a placeholder message
        await asyncio.sleep(2) # Simulate work
        await query.edit_message_text(text="لم يتم العثور على توصيات PUT حالياً.") # Replace with actual results
    elif query.data == 'scan_now_manual':
        await query.edit_message_text(text="جاري المسح الفوري، انتظر قليلاً...")
        alerts = scan_all_stocks()
        if not alerts:
            await query.edit_message_text("لا توجد إشارات جديدة.")
        else:
            for alert in alerts:
                await query.message.reply_text(alert, parse_mode='Markdown')
                await asyncio.sleep(1)
            await query.edit_message_text(text="تم الانتهاء من المسح الفوري.")


async def scan_now(update, context):
    """معالج أمر /scan لمسح فوري"""
    print("scan_now command called.") # Added print statement
    await update.message.reply_text("جاري المسح، انتظر قليلاً...")
    alerts = scan_all_stocks()
    if not alerts:
        print("No alerts found for manual scan.") # Added print statement
        await update.message.reply_text("لا توجد إشارات جديدة.")
    else:
        print(f"Sending {len(alerts)} alerts for manual scan.") # Added print statement
        for alert in alerts:
            await update.message.reply_text(alert, parse_mode='Markdown')
            await asyncio.sleep(1)

def main():
    """الدالة الرئيسية لتشغيل البوت"""
    # إنشاء تطبيق البوت
    # Create a JobQueue instance first
    job_queue_instance = JobQueue()

    # Pass the JobQueue instance to the ApplicationBuilder
    application = Application.builder().token(TELEGRAM_TOKEN).job_queue(job_queue_instance).build()

    # إضافة المعالجات للأوامر والرسائل
    application.add_handler(CommandHandler("start", start))
    application.add_handler(CommandHandler("scan", scan_now))
    application.add_handler(CallbackQueryHandler(button_callback)) # إضافة معالج لأزرار الـ Inline Keyboard
    application.add_handler(ChatMemberHandler(greet_chat_members, ChatMemberHandler.CHAT_MEMBER)) # إضافة معالج للأعضاء الجدد

    # جدولة المسح التلقائي كل ساعة
    job_queue_instance.run_repeating(send_alerts, interval=3600, first=10)  # كل ساعة، أول مرة بعد 10 ثواني

    # تشغيل البوت
    print("البوت يعمل الآن... اذهب لتجربته!")
    application.run_polling()
    # application.idle() is not necessary after run_polling() as it's a blocking call

# استدعاء الدالة الرئيسية
if __name__ == '__main__':
    main()
````

**Reasoning**:
The persistent `RuntimeError` regarding `JobQueue` indicates that the `apscheduler` dependency is not being recognized by `python-telegram-bot`, despite previous installation attempts and runtime restarts. While this is likely an environmental issue within Colab, adding `nest_asyncio.apply()` is a common best practice for running `python-telegram-bot` in Colab notebooks to prevent event loop conflicts. I will also ensure `TELEGRAM_TOKEN` and `CHAT_ID` are correctly set within the code.



````python
#!/usr/bin/env python3
# -*- coding: utf-8 -*-

# بوت توصيات الأسهم - نسخة شاكر 1.0 (نسخة كاملة)
# آخر تحديث: فبراير 2026

import yfinance as yf
import pandas as pd
import numpy as np
import ta
import logging
import time
import schedule
from datetime import datetime, timedelta
from telegram import Bot, InlineKeyboardButton, InlineKeyboardMarkup, Update
from telegram.ext import Application, CommandHandler, CallbackQueryHandler, ContextTypes, JobQueue, ChatMemberHandler # أضفنا ChatMemberHandler
import asyncio
import os
import nest_asyncio # ADDED

# Apply nest_asyncio to allow nested event loops in environments like Colab
nest_asyncio.apply() # ADDED

# ---------------------------
# 1. إعدادات البوت والتوكن
# ---------------------------
TELEGRAM_TOKEN = "8699821370:AAEQUSbLTgf7MmWqo5vV5LHPOz30wfqOfqw"      # ضع التوكن الخاص بك
CHAT_ID = "7854020427"                # ضع معرف المستخدم أو القناة

# تفعيل السجلات لمشاهدة الأخطاء
logging.basicConfig(level=logging.INFO, format='%(asctime)s - %(name)s - %(levelname)s - %(message)s')

# ---------------------------
# 2. قائمة الأسهم الكاملة (مدمجة)
# ---------------------------
TICKERS = [
    # --- أسهمك القديمة (من الصور) ---
    "A", "AA", "ABEV", "ABNB", "ABT", "ABVX", "ACGL", "ACM", "ACN",
    "ADBE", "ADI", "ADM", "ADP", "ADSK", "AEE", "AEG", "AEIS", "AEM",
    "AEP", "AER", "AES", "AFG", "AFL", "AFRM", "AG", "AGI", "AGNC",
    "AIG", "AIT", "AIZ", "AJG", "AKAM", "ALAB", "ALB", "ALC", "ALGN",
    "ALL", "ALLE", "ALLY", "ALNY", "ALSN", "AM", "AMCR", "AME", "AMH",
    "AMKR", "AMP", "AMRZ", "AMT", "AMX",

    # --- الشركات العملاقة (> 1 تريليون دولار) ---
    "NVDA", "AAPL", "GOOGL", "MSFT", "AMZN", "META", "TSM", "AVGO", "TSLA",
    "BRK.A", "WMT",

    # --- الشركات الكبرى (200 مليار - 1 تريليون) ---
    "LLY", "JPM", "XOM", "V", "JNJ", "MA", "MU", "ORCL", "COST", "AMD",
    "NFLX", "PLTR", "CSCO", "MRK", "TMO", "ASML", "ABBV", "HD", "CVX",
    "PG", "CAT", "KO", "PEP", "INTC", "CMCSA", "MCD", "TXN", "VZ", "AMGN"
]

# ---------------------------
# 3. دوال التحليل الأساسي (الفلاتر)
# ---------------------------

def get_stock_info(ticker):
    """جلب معلومات الشركة الأساسية"""
    try:
        stock = yf.Ticker(ticker)
        info = stock.info
        return info
    except Exception as e:
        logging.error(f"خطأ في جلب معلومات {ticker}: {e}")
        return None

def passes_basic_filters(ticker):
    """
    تطبيق الفلتر الأساسي:
    - Market Cap > 10B (ملاحظة: أغلب الشركات في القائمة أكبر من ذلك بكثير، لكن الفلتر يضمن ذلك)
    - Average Volume > 1M (تقريبي)
    - Spread < 5%
    """
    info = get_stock_info(ticker)
    if not info:
        return False

    # 1. القيمة السوقية
    mcap = info.get('marketCap', 0)
    if mcap < 10e9:
        print(f"{ticker} فشل فلتر Market Cap: {mcap}") # Changed from logging.debug
        return False

    # 2. متوسط الحجم (نأخذ آخر 20 يوم)
    try:
        stock = yf.Ticker(ticker)
        hist = stock.history(period="1mo", interval="1d")
        if hist.empty:
            return False
        avg_vol = hist['Volume'].rolling(20).mean().iloc[-1]
        if avg_vol < 1_000_000:   # أقل من مليون سهم يومياً
            print(f"{ticker} فشل فلتر الحجم: {avg_vol}") # Changed from logging.debug
            return False
    except:
        return False

    # 3. Spread (bid/ask)
    bid = info.get('bid', 0)
    ask = info.get('ask', 0)
    if bid and ask:
        spread = (ask - bid) / bid if bid != 0 else 1
        if spread > 0.05:   # أكبر من 5%
            print(f"{ticker} فشل فلتر Spread: {spread:.2%}") # Changed from logging.debug
            return False

    return True

# ---------------------------
# 4. تحليل SPX (مؤشر S&P 500)
# ---------------------------

def analyze_spx(timeframe="15m"):
    """تحليل مؤشر SPX على الإطار المطلوب"""
    try:
        spx = yf.Ticker("^GSPC")  # أو "SPY"
        hist = spx.history(period="5d", interval=timeframe)
        if len(hist) < 50:
            return "neutral"

        # متوسطات بسيطة لتحديد الاتجاه
        hist['SMA20'] = hist['Close'].rolling(20).mean()
        last = hist.iloc[-1]
        prev = hist.iloc[-2]

        if last['Close'] > last['SMA20'] and last['Close'] > prev['Close']:
            return "uptrend"
        elif last['Close'] < last['SMA20'] and last['Close'] < prev['Close']:
            return "downtrend"
        else:
            return "neutral"
    except:
        return "neutral"

# ---------------------------
# 5. تحليل السهم (فريم ساعتين)
# ---------------------------

def analyze_stock_2h(ticker):
    """
    تحليل فني كامل على فريم الساعتين
    يعيد: {'signal': 'CALL'/'PUT'/None, 'confidence': 0-100, 'details': {...}}
    """
    try:
        print(f"Analyzing {ticker} for 2h interval.") # Added print statement
        stock = yf.Ticker(ticker)
        hist = stock.history(period="1mo", interval="2h")  # آخر شهر، فريم ساعتين
        if len(hist) < 200:
            print(f"Not enough historical data for {ticker} in 2h interval.") # Added print statement
            return None

        # حساب المتوسطات
        hist['SMA50'] = hist['Close'].rolling(50).mean()
        hist['SMA100'] = hist['Close'].rolling(100).mean()
        hist['SMA200'] = hist['Close'].rolling(200).mean()
        hist['Volume_SMA20'] = hist['Volume'].rolling(20).mean()

        last = hist.iloc[-1]
        current_price = last['Close']
        sma50 = last['SMA50']
        sma100 = last['SMA100']
        sma200 = last['SMA200']

        # قمم وقيعان آخر 3 شموع
        high_last_3 = hist['High'].iloc[-3:].max()
        low_last_3 = hist['Low'].iloc[-3:].min()

        # قمة/قاع اليوم (آخر 24 ساعة = 12 شمعة)
        high_today = hist['High'].iloc[-12:].max()
        low_today = hist['Low'].iloc[-12:].min()

        volume_ok = last['Volume'] > last['Volume_SMA20']

        # شروط CALL
        call_condition1 = (sma50 > sma100) or (sma50 > sma200)
        call_condition2 = current_price > sma200
        call_condition3 = current_price > high_last_3 or current_price > high_today
        call_condition4 = volume_ok

        # شروط PUT
        put_condition1 = (sma50 < sma100) or (sma50 < sma200)
        put_condition2 = current_price < sma200
        put_condition3 = current_price < low_last_3 or current_price < low_today
        put_condition4 = volume_ok

        signal = None
        confidence = 0
        details = {}

        if call_condition1 and call_condition2 and call_condition3 and call_condition4:
            signal = "CALL"
            # حساب الثقة: نضيف 25 لكل شرط متحقق
            confidence = 25 * sum([call_condition1, call_condition2, call_condition3, call_condition4])
            details = {
                "sma50": sma50,
                "sma100": sma100,
                "sma200": sma200,
                "price": current_price,
                "high_3": high_last_3,
                "volume_ratio": last['Volume'] / last['Volume_SMA20']
            }
        elif put_condition1 and put_condition2 and put_condition3 and put_condition4:
            signal = "PUT"
            confidence = 25 * sum([put_condition1, put_condition2, put_condition3, put_condition4])
            details = {
                "sma50": sma50,
                "sma100": sma100,
                "sma200": sma200,
                "price": current_price,
                "low_3": low_last_3,
                "volume_ratio": last['Volume'] / last['Volume_SMA20']
            }

        if signal: # Added print statement
            print(f"Signal for {ticker}: {signal}, Confidence: {confidence}")
        return {"signal": signal, "confidence": int(confidence), "details": details}

    except Exception as e:
        logging.error(f"خطأ في تحليل {ticker}: {e}")
        return None

# ---------------------------
# 6. تصنيف الصفقة (سكالب، قصيرة، سوينق)
# ---------------------------

def classify_trade(signal, confidence, details):
    """
    تصنيف حسب قوة الإشارة وترتيب المتوسطات
    """
    if not signal:
        return None

    sma50 = details.get('sma50')
    sma100 = details.get('sma100')
    sma200 = details.get('sma200')

    # ترتيب المتوسطات
    if signal == "CALL":
        if sma50 > sma100 > sma200:
            return "📈 سوينق 3-7 أيام"
        elif confidence >= 75:
            return "⏳ صفقة قصيرة 1-2 يوم"
        else:
            return "⚡ سكالب"
    else:  # PUT
        if sma50 < sma100 < sma200:
            return "📈 سوينق 3-7 أيام"
        elif confidence >= 75:
            return "⏳ صفقة قصيرة 1-2 يوم"
        else:
            return "⚡ سكالب"

# ---------------------------
# 7. اختيار العقد المناسب (بيانات الخيارات)
# ---------------------------

def get_option_contract(ticker, signal, classification):
    """
    البحث عن عقد خيار مناسب حسب التصنيف.
    """
    try:
        stock = yf.Ticker(ticker)
        # جلب تواريخ انتهاء الصلاحية المتاحة
        expirations = stock.options
        if not expirations:
            return "غير متوفر"

        # اختيار التاريخ المناسب حسب التصنيف
        today = datetime.now().date()
        if "سكالب" in classification:
            days = 7
        elif "قصيرة" in classification:
            days = 14
        else:  # سوينق
            days = 30

        target_date = today + timedelta(days=days)
        # البحث عن أقرب تاريخ انتهاء للعدد المطلوب من الأيام
        exp_date = None
        for exp in expirations:
            exp_dt = datetime.strptime(exp, '%Y-%m-%d').date()
            if exp_dt >= target_date:
                exp_date = exp
                break
        if not exp_date:
            exp_date = expirations[0]  # أقرب تاريخ

        # جلب عقود الخيارات لذلك التاريخ
        opt_chain = stock.option_chain(exp_date)
        if signal == "CALL":
            contracts = opt_chain.calls
        else:
            contracts = opt_chain.puts

        if contracts.empty:
            return "غير متوفر"

        # فلترة العقود: Open Interest عالي (أعلى 10%)، Spread ضيق
        current_price = stock.history(period="1d")['Close'].iloc[-1]
        contracts['strike_diff'] = abs(contracts['strike'] - current_price)
        # فلتر OI
        oi_threshold = contracts['openInterest'].quantile(0.7)  # أعلى 30%
        eligible = contracts[contracts['openInterest'] > oi_threshold]

        if eligible.empty:
            eligible = contracts

        # نختار العقد الأقرب للسعر الحالي (للدلتا التقريبية)
        best = eligible.nsmallest(1, 'strike_diff').iloc[0]

        # حساب Spread تقريبي (ask-bid)/ask
        spread = (best['ask'] - best['bid']) / best['ask'] if best['ask'] > 0 else 1
        if spread > 0.05:   # أكبر من 5%
            return "سبريد واسع"

        # بناء نص العقد المقترح
        delta_approx = 0.5 - (best['strike_diff'] / current_price)  # تقريب بسيط
        delta_approx = max(0.1, min(0.9, delta_approx))

        return f"Delta {delta_approx:.2f} – Exp {exp_date} (أيام: {days})"

    except Exception as e:
        logging.error(f"خطأ في جلب الخيارات لـ {ticker}: {e}")
        return "غير متوفر"

# ---------------------------
# 8. بناء رسالة التنبيه النهائية
# ---------------------------

def build_alert_message(ticker, signal, confidence, classification, details, spx_trend, option_text):
    """
    تنسيق الرسالة كما طلبت تماماً.
    """
    # ترجمة الاتجاه
    if signal == "CALL":
        direction = "Call"
        cross = "50 > 100" if details.get('sma50', 0) > details.get('sma100', 0) else "50 > 200"
        price_status = f"السعر فوق {details.get('sma200', 0):.2f}"
        event = "اختراق قمة 2H"
    else:
        direction = "Put"
        cross = "50 < 100" if details.get('sma50', 0) < details.get('sma100', 0) else "50 < 200"
        price_status = f"السعر تحت {details.get('sma200', 0):.2f}"
        event = "كسر قاع 2H"

    # توافق SPX
    if "uptrend" in spx_trend and signal == "CALL":
        spx_compat = "✅ متوافق"
    elif "downtrend" in spx_trend and signal == "PUT":
        spx_compat = "✅ متوافق"
    else:
        spx_compat = "⚠️ غير متوافق"

    message = f"""
🔔 {ticker}
الاتجاه: {direction}
التصنيف: {classification}
التقاطع: {cross} (إيجابي)
الحالة: {price_status}
الحدث: {event}
SPX: {spx_compat}
الثقة: {confidence}/100
العقد المقترح: {option_text}
"""
    return message

# ---------------------------
# 9. دالة المسح الرئيسية
# ---------------------------

def scan_all_stocks():
    """تمر على كل الأسهم وتفحصها وترسل التنبيهات"""
    alerts = []
    print("Starting stock scan...") # Added print statement
    for ticker in TICKERS:
        logging.info(f"فحص {ticker}...")
        time.sleep(0.5)  # تأخير نصف ثانية لتجنب حظر Yahoo

        # 1. الفلتر الأساسي
        if not passes_basic_filters(ticker):
            print(f"Ticker {ticker} did not pass basic filters.") # Added print statement
            continue

        # 2. تحليل السهم
        result = analyze_stock_2h(ticker)
        if not result or not result['signal']:
            print(f"No signal found for {ticker} after 2h analysis.") # Added print statement
            continue

        signal = result['signal']
        confidence = result['confidence']
        details = result['details']

        # 3. تحليل SPX
        spx_15m = analyze_spx("15m")

        # 4. تصنيف الصفقة
        classification = classify_trade(signal, confidence, details)

        # 5. اختيار العقد
        option = get_option_contract(ticker, signal, classification)

        # 6. بناء الرسالة
        msg = build_alert_message(ticker, signal, confidence, classification, details, spx_15m, option)
        alerts.append(msg)
    print(f"Scan finished. Found {len(alerts)} alerts.") # Added print statement
    return alerts

# ---------------------------
# 10. دوال بوت تليجرام
# ---------------------------

async def send_alerts(context):
    """إرسال التنبيهات إلى تليجرام"""
    print("send_alerts function called.") # Added print statement
    alerts = scan_all_stocks()
    if not alerts:
        print("No alerts found to send.") # Added print statement
        await context.bot.send_message(chat_id=CHAT_ID, text="لا توجد إشارات جديدة حالياً.")
    else:
        print(f"Sending {len(alerts)} alerts.") # Added print statement
        for alert in alerts:
            await context.bot.send_message(chat_id=CHAT_ID, text=alert, parse_mode='Markdown')
            await asyncio.sleep(1)  # تجنب سبام تليجرام

# دالة للترحيب بالأعضاء الجدد وإرسال إخلاء المسؤولية
async def greet_chat_members(update: Update, context: ContextTypes.DEFAULT_TYPE) -> None:
    print(f"New chat member event: {update.chat_member}") # Added print
    if update.chat_member.new_chat_member.status == 'member':
        user = update.chat_member.new_chat_member.user
        chat_title = update.chat_member.chat.title or "هذه المجموعة"
        welcome_new_member_message = f"""
أهلاً وسهلاً بك {user.full_name} في {chat_title}! 👋

⚠️ **إخلاء مسؤولية:**
توصيات هذا البوت هي لأغراض تعليمية وتحليلية فقط، ولا تشكل نصيحة استثمارية بأي شكل من الأشكال. تداول العقود والمشتقات المالية ينطوي على مخاطر عالية وقد يؤدي إلى خسارة رأس المال. أنت وحدك المسؤول عن قراراتك الاستثمارية. يرجى استشارة مستشار مالي متخصص قبل اتخاذ أي قرارات استثمارية.
"""
        print(f"Sending welcome message to {user.full_name} in chat {update.chat_member.chat.id}") # Added print
        await context.bot.send_message(chat_id=update.chat_member.chat.id, text=welcome_new_member_message, parse_mode='Markdown')


async def start(update: Update, context: ContextTypes.DEFAULT_TYPE) -> None:
    """معالج أمر /start"""
    print(f"Start command received from {update.effective_user.id} in chat {update.effective_chat.id}") # Added print
    welcome_message = """
👋 مرحباً بك في بوت توصيات الأسهم!

هذا البوت يقدم توصيات تحليل فني للأسهم بناءً على استراتيجيات محددة.

⚠️ **إخلاء مسؤولية:**
توصيات هذا البوت هي لأغراض تعليمية وتحليلية فقط، ولا تشكل نصيحة استثمارية بأي شكل من الأشكال. تداول العقود والمشتقات المالية ينطوي على مخاطر عالية وقد يؤدي إلى خسارة رأس المال. أنت وحدك المسؤول عن قراراتك الاستثمارية. يرجى استشارة مستشار مالي متخصص قبل اتخاذ أي قرارات استثمارية.

اختر نوع التوصية التي ترغب في الحصول عليها:
"""

    keyboard = [
        [InlineKeyboardButton("📈 عقود CALL (شراء)", callback_data='recommend_call')],
        [InlineKeyboardButton("📉 عقود PUT (بيع)", callback_data='recommend_put')],
        [InlineKeyboardButton("🔄 مسح الآن", callback_data='scan_now_manual')]
    ]
    reply_markup = InlineKeyboardMarkup(keyboard)

    await update.message.reply_text(welcome_message, reply_markup=reply_markup, parse_mode='Markdown')

async def button_callback(update: Update, context: ContextTypes.DEFAULT_TYPE) -> None:
    query = update.callback_query
    print(f"Callback query received: {query.data} from {query.from_user.id}") # Added print
    await query.answer()  # Always answer callback queries

    if query.data == 'recommend_call':
        await query.edit_message_text(text="جاري البحث عن توصيات CALL...")
        # Logic to scan for CALL recommendations
        # For now, let's just send a placeholder message
        await asyncio.sleep(2) # Simulate work
        await query.edit_message_text(text="لم يتم العثور على توصيات CALL حالياً.") # Replace with actual results
    elif query.data == 'recommend_put':
        await query.edit_message_text(text="جاري البحث عن توصيات PUT...")
        # Logic to scan for PUT recommendations
        # For now, let's just send a placeholder message
        await asyncio.sleep(2) # Simulate work
        await query.edit_message_text(text="لم يتم العثور على توصيات PUT حالياً.") # Replace with actual results
    elif query.data == 'scan_now_manual':
        await query.edit_message_text(text="جاري المسح الفوري، انتظر قليلاً...")
        alerts = scan_all_stocks()
        if not alerts:
            await query.edit_message_text("لا توجد إشارات جديدة.")
        else:
            for alert in alerts:
                await query.message.reply_text(alert, parse_mode='Markdown')
                await asyncio.sleep(1)
            await query.edit_message_text(text="تم الانتهاء من المسح الفوري.")


async def scan_now(update, context):
    """معالج أمر /scan لمسح فوري"""
    print("scan_now command called.") # Added print statement
    await update.message.reply_text("جاري المسح، انتظر قليلاً...")
    alerts = scan_all_stocks()
    if not alerts:
        print("No alerts found for manual scan.") # Added print statement
        await update.message.reply_text("لا توجد إشارات جديدة.")
    else:
        print(f"Sending {len(alerts)} alerts for manual scan.") # Added print statement
        for alert in alerts:
            await update.message.reply_text(alert, parse_mode='Markdown')
            await asyncio.sleep(1)

def main():
    """الدالة الرئيسية لتشغيل البوت"""
    # إنشاء تطبيق البوت
    # Create a JobQueue instance first
    job_queue_instance = JobQueue()

    # Pass the JobQueue instance to the ApplicationBuilder
    application = Application.builder().token(TELEGRAM_TOKEN).job_queue(job_queue_instance).build()

    # إضافة المعالجات للأوامر والرسائل
    application.add_handler(CommandHandler("start", start))
    application.add_handler(CommandHandler("scan", scan_now))
    application.add_handler(CallbackQueryHandler(button_callback)) # إضافة معالج لأزرار الـ Inline Keyboard
    application.add_handler(ChatMemberHandler(greet_chat_members, ChatMemberHandler.CHAT_MEMBER)) # إضافة معالج للأعضاء الجدد

    # جدولة المسح التلقائي كل ساعة
    job_queue_instance.run_repeating(send_alerts, interval=3600, first=10)  # كل ساعة، أول مرة بعد 10 ثواني

    # تشغيل البوت
    print("البوت يعمل الآن... اذهب لتجربته!")
    application.run_polling()
    # application.idle() is not necessary after run_polling() as it's a blocking call

# استدعاء الدالة الرئيسية
if __name__ == '__main__':
    main()
````
